## Benchmark analytique microscopique du puits de potentiel quantique unidimensionnel

## 1. Fondements physiques

Le puits de potentiel quantique unidimensionnel est l'un des systèmes fondamentaux de la mécanique quantique. Malgré sa simplicité, il possède un spectre d'énergie discret, des fonctions d'onde analytiques et constitue un benchmark classique pour l'étude des états quantiques confinés.

Dans ce notebook, ce système est utilisé comme deuxième benchmark microscopique du framework **GVH (Geometric Vibrational Harmonization)**, après l'oscillateur harmonique quantique.

L'objectif n'est pas de modifier la mécanique quantique standard, mais d'évaluer le comportement des observables GVH sur un système dont les solutions analytiques sont parfaitement connues.

Les fonctions d'onde exactes serviront à construire un espace d'état géométrique GVH, à partir duquel seront calculées les observables de transition, les fonctions de survie et les observables angulaires. Une validation numérique complète sera ensuite réalisée afin de vérifier la robustesse des résultats vis-à-vis de la résolution spatiale et du domaine de calcul.

À la fin de ce pipeline, les observables obtenues seront comparées à celles du benchmark de l'oscillateur harmonique afin d'évaluer leur stabilité et leur capacité à caractériser différents systèmes quantiques.

### 1.1 Le puits de potentiel quantique infini

On considère une particule de masse \(m\) confinée dans un puits de potentiel unidimensionnel infini de largeur \(L\).

Le potentiel est défini par :

\[
V(x)=
\begin{cases}
0, & 0<x<L,\\
+\infty, & \text{sinon}.
\end{cases}
\]

À l'intérieur du puits, la particule évolue librement. Les parois infinies imposent que la fonction d'onde s'annule aux frontières :

\[
\psi(0)=0,
\qquad
\psi(L)=0.
\]

Ces conditions aux limites conduisent à une quantification naturelle des états stationnaires et des niveaux d'énergie.

### 1.2 Équation de Schrödinger stationnaire

Dans le cadre de la mécanique quantique non relativiste, les états stationnaires d'une particule sont décrits par l'équation de Schrödinger indépendante du temps :

$begin:math:display$
\\hat\{H\}\\psi\(x\)\=E\\psi\(x\)\,
$end:math:display$

où :

- $begin:math:text$\\hat\{H\}$end:math:text$ est l'opérateur Hamiltonien,
- $begin:math:text$\\psi\(x\)$end:math:text$ est la fonction d'onde,
- $begin:math:text$E$end:math:text$ est l'énergie de l'état stationnaire.

Pour une particule de masse $begin:math:text$m$end:math:text$ évoluant dans un potentiel $begin:math:text$V\(x\)$end:math:text$, l'Hamiltonien s'écrit :

$begin:math:display$
\\hat\{H\}
\=
\-\\frac\{\\hbar\^2\}\{2m\}
\\frac\{d\^2\}\{dx\^2\}
\+
V\(x\)\.
$end:math:display$

L'équation de Schrödinger devient alors

$begin:math:display$
\-\\frac\{\\hbar\^2\}\{2m\}
\\frac\{d\^2\\psi\(x\)\}\{dx\^2\}
\+
V\(x\)\\psi\(x\)
\=
E\\psi\(x\)\.
$end:math:display$

À l'intérieur du puits de potentiel infini, le potentiel est nul :

$begin:math:display$
V\(x\)\=0\,
\\qquad
0\<x\<L\.
$end:math:display$

L'équation se réduit donc à

$begin:math:display$
\-\\frac\{\\hbar\^2\}\{2m\}
\\frac\{d\^2\\psi\(x\)\}\{dx\^2\}
\=
E\\psi\(x\)\,
$end:math:display$

qui constitue une équation différentielle linéaire du second ordre. Les conditions aux limites imposées par les parois infinies,

$begin:math:display$
\\psi\(0\)\=0\,
\\qquad
\\psi\(L\)\=0\,
$end:math:display$

sélectionnent uniquement un ensemble discret de solutions admissibles, conduisant à la quantification des niveaux d'énergie.

Dans la section suivante, cette équation sera résolue analytiquement afin d'obtenir les fonctions d'onde propres et les valeurs propres d'énergie qui serviront de base au pipeline GVH.

### 1.3 Choix des unités naturelles

Afin de simplifier les expressions analytiques et les calculs numériques, ce pipeline adopte les unités naturelles couramment utilisées pour l'oscillateur harmonique et le puits de potentiel quantique :

$begin:math:display$
\\hbar \= 1\,
\\qquad
m \= 1\.
$end:math:display$

Ces choix n'altèrent pas la physique du système ; ils correspondent simplement à une adimensionnalisation des équations.

Dans ces unités, l'équation de Schrödinger stationnaire à l'intérieur du puits devient

$begin:math:display$
\-\\frac\{1\}\{2\}
\\frac\{d\^\{2\}\\psi\(x\)\}\{dx\^\{2\}\}
\=
E\\\,\\psi\(x\)\,
\\qquad
0\<x\<L\,
$end:math:display$

avec les conditions aux limites

$begin:math:display$
\\psi\(0\)\=0\,
\\qquad
\\psi\(L\)\=0\.
$end:math:display$

Cette forme simplifiée facilite l'obtention des solutions analytiques et constitue la base du pipeline numérique développé dans les sections suivantes.

### 1.4 Solutions analytiques et quantification de l'énergie

À l'intérieur du puits de potentiel infini, le potentiel est nul. L'équation de Schrödinger stationnaire s'écrit alors

$begin:math:display$
\-\\frac\{1\}\{2\}
\\frac\{d\^\{2\}\\psi\(x\)\}\{dx\^\{2\}\}
\=
E\\\,\\psi\(x\)\,
\\qquad
0\<x\<L\.
$end:math:display$

Cette équation admet des solutions sinusoïdales. Les conditions aux limites

$begin:math:display$
\\psi\(0\)\=0\,
\\qquad
\\psi\(L\)\=0\,
$end:math:display$

imposent que seules certaines longueurs d'onde soient autorisées à l'intérieur du puits.

Les fonctions d'onde propres normalisées sont

$begin:math:display$
\\psi\_n\(x\)
\=
\\sqrt\{\\frac\{2\}\{L\}\}
\\sin\\\!\\left\(\\frac\{n\\pi x\}\{L\}\\right\)\,
\\qquad
n\=1\,2\,3\,\\ldots
$end:math:display$

Chaque entier positif $begin:math:text$n$end:math:text$ définit un état quantique stationnaire distinct.

Les valeurs propres d'énergie associées sont

$begin:math:display$
E\_n
\=
\\frac\{n\^2\\pi\^2\}\{2L\^2\}\,
\\qquad
n\=1\,2\,3\,\\ldots
$end:math:display$

Contrairement à une particule classique, l'énergie ne peut prendre qu'un ensemble discret de valeurs. Cette quantification résulte directement des conditions aux limites imposées par le confinement de la particule.

Les états propres présentent également des propriétés caractéristiques :

- ils sont orthonormés ;
- chaque fonction d'onde possède exactement $begin:math:text$n\-1$end:math:text$ nœuds à l'intérieur du puits ;
- la densité de probabilité est donnée par

$begin:math:display$
\\rho\_n\(x\)\=\|\\psi\_n\(x\)\|\^2\.
$end:math:display$

Ces solutions analytiques constituent la référence du présent benchmark. Elles seront utilisées dans les sections suivantes pour construire la base de données quantique, l'espace d'état GVH et les différentes observables géométriques, tout en conservant une comparaison directe avec les résultats exacts de la mécanique quantique.

## 2.1 Paramètres du benchmark

Avant de générer les solutions analytiques, les paramètres numériques du benchmark sont définis.

Le puits de potentiel est caractérisé par sa largeur $begin:math:text$L$end:math:text$, tandis que l'espace est discrétisé sur une grille régulière comportant $begin:math:text$N\_x$end:math:text$ points.

Le nombre d'états quantiques étudiés est également fixé afin de construire une base de données suffisamment riche pour les analyses GVH tout en conservant un temps de calcul réduit.

Ces paramètres pourront être modifiés ultérieurement lors des tests de convergence numérique présentés dans la section de validation.

```python
# ==========================================================
# Paramètres du benchmark
# ==========================================================

import numpy as np

# Largeur du puits
L = 1.0

# Nombre de points spatiaux
Nx = 4001

# Domaine spatial
x = np.linspace(0.0, L, Nx)

# Pas spatial
dx = x[1] - x[0]

# Nombre d'états quantiques
n_states = 10

print("Largeur du puits L :", L)
print("Nombre de points Nx :", Nx)
print("Pas spatial dx :", dx)
print("Nombre d'états :", n_states)
```

## 2.2 Implémentation des solutions analytiques

Les solutions analytiques du puits de potentiel infini sont maintenant implémentées sous forme de fonctions Python.

Trois fonctions fondamentales sont définies :

- `psi_n(x, n)` : fonction d'onde normalisée de l'état $begin:math:text$n$end:math:text$ ;
- `rho_n(x, n)` : densité de probabilité associée ;
- `energy_n(n)` : niveau d'énergie analytique.

Ces fonctions seront utilisées dans l'ensemble du pipeline pour construire automatiquement les états quantiques, générer les observables GVH et effectuer les validations numériques.

```python
# ==========================================================
# Solutions analytiques du puits de potentiel infini
# ==========================================================

def psi_n(x, n, L=L):
    """
    Fonction d'onde normalisée de l'état n.
    """
    return np.sqrt(2.0 / L) * np.sin(n * np.pi * x / L)


def rho_n(x, n, L=L):
    """
    Densité de probabilité.
    """
    psi = psi_n(x, n, L)
    return psi**2


def energy_n(n, L=L):
    """
    Valeur propre d'énergie.
    """
    return (n**2 * np.pi**2) / (2.0 * L**2)
```

```python
# ==========================================================
# Vérification des fonctions analytiques
# ==========================================================

for n in range(1, 6):
    print(f"État n={n}")
    print(f"Énergie = {energy_n(n):.6f}")
    print(f"ψ(0) = {psi_n(0.0, n):.6e}")
    print(f"ψ(L) = {psi_n(L, n):.6e}")
    print("-" * 40)
```

## 2.3 Génération de la base des états quantiques

Les solutions analytiques sont maintenant utilisées pour générer automatiquement l'ensemble des états quantiques étudiés dans ce benchmark.

Pour chaque état propre, le pipeline calcule :

- la fonction d'onde normalisée ;
- la densité de probabilité ;
- le niveau d'énergie analytique.

Toutes ces informations sont ensuite regroupées dans une base de données qui servira d'entrée aux calculs des observables GVH.

Cette étape constitue le lien entre la mécanique quantique analytique et l'analyse géométrique développée dans les sections suivantes.

```python
# ==========================================================
# Génération des états quantiques
# ==========================================================

quantum_states = []

for n in range(1, n_states + 1):

    psi = psi_n(x, n)
    rho = rho_n(x, n)
    energy = energy_n(n)

    quantum_states.append({
        "n": n,
        "energy": energy,
        "psi": psi,
        "rho": rho
    })

print(f"{len(quantum_states)} états quantiques générés.")
```

```python
# Aperçu des états générés

for state in quantum_states[:5]:

    print(f"n = {state['n']}")
    print(f"Énergie = {state['energy']:.6f}")
    print(f"ψ shape = {state['psi'].shape}")
    print(f"ρ shape = {state['rho'].shape}")
    print("-"*40)
```

```python
# ==========================================================
# Tableau récapitulatif
# ==========================================================

import pandas as pd

resume_etats = pd.DataFrame({
    "Etat n": [s["n"] for s in quantum_states],
    "Energie": [s["energy"] for s in quantum_states]
})

resume_etats
```

## 2.4 Validation analytique

Les états quantiques générés sont maintenant soumis à plusieurs tests de validation afin de vérifier leur conformité avec les propriétés analytiques du puits de potentiel infini.

Les vérifications portent sur :

- la normalisation des fonctions d'onde ;
- l'orthogonalité des états propres ;
- les conditions aux limites ;
- le nombre de nœuds des fonctions d'onde.

Ces contrôles garantissent que les solutions numériques reproduisent fidèlement les solutions exactes de la mécanique quantique avant leur utilisation dans les calculs des observables GVH.

```python
# ==========================================================
# Validation de la normalisation
# ==========================================================

print("Validation de la normalisation\n")

normalisation = []

for state in quantum_states:

    integral = np.trapz(state["psi"]**2, x)

    normalisation.append({
        "Etat": state["n"],
        "Norme": integral,
        "Erreur": abs(integral - 1.0)
    })

normalisation_df = pd.DataFrame(normalisation)

normalisation_df
```

```python
# ==========================================================
# Validation de l'orthogonalité
# ==========================================================

M = np.zeros((n_states, n_states))

for i in range(n_states):
    for j in range(n_states):

        psi_i = quantum_states[i]["psi"]
        psi_j = quantum_states[j]["psi"]

        M[i, j] = np.trapz(psi_i * psi_j, x)

orthogonalite_df = pd.DataFrame(
    M,
    index=[f"n={i}" for i in range(1, n_states+1)],
    columns=[f"n={i}" for i in range(1, n_states+1)]
)

orthogonalite_df.round(8)
```

```python
# ==========================================================
# Conditions aux limites
# ==========================================================

limites = []

for state in quantum_states:

    psi = state["psi"]

    limites.append({
        "Etat": state["n"],
        "psi(0)": psi[0],
        "psi(L)": psi[-1]
    })

limites_df = pd.DataFrame(limites)

limites_df
```

```python
# ==========================================================
# Comptage des nœuds internes
# ==========================================================

def compter_noeuds(psi):
    signe = np.sign(psi)
    changement = np.diff(signe)
    indices = np.where(changement != 0)[0]

    # On retire les zéros aux frontières
    internes = [
        i for i in indices
        if 0 < x[i] < L
    ]

    return len(internes)

noeuds = []

for state in quantum_states:

    n = state["n"]

    noeuds.append({
        "Etat": n,
        "Noeuds observés": compter_noeuds(state["psi"]),
        "Noeuds théoriques": n-1
    })

noeuds_df = pd.DataFrame(noeuds)

noeuds_df
```

### Conclusion

Les solutions analytiques générées reproduisent les propriétés fondamentales du puits de potentiel quantique infini.

Les fonctions d'onde sont correctement normalisées, orthogonales et satisfont les conditions aux limites imposées par les parois du puits. Le nombre de nœuds observé est conforme à la théorie pour chaque état propre.

Cette validation confirme que la base de données quantique est cohérente avec les solutions exactes de la mécanique quantique et peut être utilisée comme entrée du pipeline GVH.

## 4.2 Calcul des transitions

Les transitions GVH sont obtenues en calculant la distance euclidienne entre deux vecteurs d'état successifs.

Pour deux états successifs $begin:math:text$X\_i$end:math:text$ et $begin:math:text$X\_\{i\+1\}$end:math:text$, la transition est définie par

$begin:math:display$
D\_T\(i\)\=\\\|X\_\{i\+1\}\-X\_i\\\|\.
$end:math:display$

Cette suite de distances constitue la base de l'ensemble des observables GVH.

```python
# ==========================================================
# Calcul des transitions GVH
# ==========================================================

# Matrice des états
X = gvh_state.to_numpy()

# Différences entre états successifs
delta_X = np.diff(X, axis=0)

# Distances euclidiennes
D_T = np.linalg.norm(delta_X, axis=1)

print("Transitions calculées.")
print("Nombre de transitions :", len(D_T))

print("\nPremières valeurs :")
print(D_T[:10])
```

```python
# ==========================================================
# Statistiques des transitions
# ==========================================================

transition_stats = pd.DataFrame({
    "Observable": [
        "Nombre",
        "Minimum",
        "Maximum",
        "Moyenne",
        "Écart-type"
    ],
    "Valeur": [
        len(D_T),
        D_T.min(),
        D_T.max(),
        D_T.mean(),
        D_T.std()
    ]
})

transition_stats
```

## 4.3 Construction des variables normalisées

Les transitions sont maintenant normalisées afin de rendre les observables indépendantes de leur échelle absolue.

Chaque transition est transformée selon :

$begin:math:display$
u\=\\frac\{D\_T\}\{D\_T\^\{\\max\}\}\.
$end:math:display$

À partir de cette variable normalisée sont ensuite calculés :

- la distance critique $begin:math:text$D\_c$end:math:text$, définie comme le quantile choisi de la distribution des transitions ;
- le seuil normalisé $begin:math:text$u\_c$end:math:text$ ;
- la fraction de transitions supérieures au seuil, notée $begin:math:text$R\_\{\\mathrm\{super\}\}$end:math:text$.

Ces quantités constituent les principales observables statistiques du pipeline GVH.

```python
# ==========================================================
# Variables normalisées GVH
# ==========================================================

# Transition maximale
D_T_max = D_T.max()

# Variable normalisée
u = D_T / D_T_max

# Quantile critique (95 %)
q = 0.95

D_c = np.quantile(D_T, q)
u_c = D_c / D_T_max

# Fraction de transitions supérieures au seuil
R_super = np.mean(u > u_c)

print(f"D_T_max = {D_T_max:.6f}")
print(f"D_c     = {D_c:.6f}")
print(f"u_c     = {u_c:.6f}")
print(f"R_super = {R_super:.6f}")
```

```python
# ==========================================================
# Résumé des observables normalisées
# ==========================================================

normalisation_gvh = pd.DataFrame({
    "Observable": [
        "D_T_mean",
        "S_T",
        "D_T_max",
        "D_c",
        "u_c",
        "R_super"
    ],
    "Valeur": [
        D_T.mean(),
        D_T.std(),
        D_T_max,
        D_c,
        u_c,
        R_super
    ]
})

normalisation_gvh
```

## 4.4 Construction de la fonction de survie

La distribution des transitions normalisées est étudiée à l'aide d'une fonction de survie empirique.

Pour un seuil \(u_0\), cette fonction est définie par

\[
R_{\mathrm{super}}(u_0)
=
P(u>u_0).
\]

Elle représente la fraction des transitions dont l'amplitude normalisée dépasse le seuil considéré.

Cette fonction sera ensuite comparée à un modèle exponentiel afin de déterminer les paramètres géométriques \(A_H\) et \(B_H\).

```python
# ==========================================================
# Construction de la fonction de survie empirique
# ==========================================================

# Seuils normalisés régulièrement espacés
u_thresholds = np.linspace(0.0, 1.0, 200)

# Fonction de survie empirique
R_survival = np.array([
    np.mean(u > threshold)
    for threshold in u_thresholds
])

survival_df = pd.DataFrame({
    "u_threshold": u_thresholds,
    "R_super": R_survival
})

print("Fonction de survie construite.")
print("Nombre de seuils :", len(u_thresholds))

survival_df.head(10)
```

```python
# ==========================================================
# Vérifications de la fonction de survie
# ==========================================================

is_monotonic = np.all(np.diff(R_survival) <= 1e-15)

print(f"R_super(u=0)   = {R_survival[0]:.6f}")
print(f"R_super(u=1)   = {R_survival[-1]:.6f}")
print(f"Décroissance monotone : {is_monotonic}")

assert is_monotonic, (
    "La fonction de survie devrait être monotone décroissante."
)

assert np.all((R_survival >= 0.0) & (R_survival <= 1.0)), (
    "Les probabilités de survie doivent rester entre 0 et 1."
)

print("\n✓ Fonction de survie conforme.")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(
    u_thresholds,
    R_survival,
    label="Survie empirique"
)

plt.xlabel("Seuil normalisé u")
plt.ylabel("R_super(u)")
plt.title("Fonction de survie des transitions GVH")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## 4.5 Ajustement exponentiel

La fonction de survie empirique est comparée au modèle exponentiel

\[
R_H(u)=A_H\,e^{-B_Hu},
\]

où :

- \(A_H\) représente l’amplitude extrapolée de la fonction de survie ;
- \(B_H\) représente le taux d’atténuation géométrique.

La qualité de l’ajustement est évaluée à l’aide du coefficient de détermination \(R_H^2\) et de la racine de l’erreur quadratique moyenne, notée \(\mathrm{RMSE}_H\).

Comme la fonction de survie empirique atteint exactement zéro au seuil maximal, seuls les points strictement positifs sont utilisés pour l’estimation des paramètres.

```python
# ==========================================================
# Modèle exponentiel GVH
# ==========================================================

from scipy.optimize import curve_fit

def exponential_model(u_value, A_H, B_H):
    """
    Modèle exponentiel de la fonction de survie GVH.
    """
    return A_H * np.exp(-B_H * u_value)
```

```python
# ==========================================================
# Sélection des points utilisables
# ==========================================================

fit_mask = (
    np.isfinite(u_thresholds)
    & np.isfinite(R_survival)
    & (R_survival > 0.0)
)

u_fit = u_thresholds[fit_mask]
R_fit = R_survival[fit_mask]

print("Nombre total de seuils :", len(u_thresholds))
print("Nombre de points ajustés :", len(u_fit))
print("Points exclus :", len(u_thresholds) - len(u_fit))
```

```python
# ==========================================================
# Ajustement exponentiel non linéaire
# ==========================================================

initial_parameters = [1.0, 1.0]

parameters_H, covariance_H = curve_fit(
    exponential_model,
    u_fit,
    R_fit,
    p0=initial_parameters,
    bounds=(
        [0.0, 0.0],
        [np.inf, np.inf]
    ),
    maxfev=20000
)

A_H, B_H = parameters_H

print(f"A_H = {A_H:.10f}")
print(f"B_H = {B_H:.10f}")
```

```python
# ==========================================================
# Prédictions exponentielles
# ==========================================================

R_pred_fit = exponential_model(
    u_fit,
    A_H,
    B_H
)

R_pred_full = exponential_model(
    u_thresholds,
    A_H,
    B_H
)
```

```python
# ==========================================================
# Diagnostics de l'ajustement
# ==========================================================

residuals_H = R_fit - R_pred_fit

SSE_H = np.sum(residuals_H**2)
SST_H = np.sum((R_fit - np.mean(R_fit))**2)

if SST_H > 0:
    R2_H = 1.0 - SSE_H / SST_H
else:
    R2_H = np.nan

RMSE_H = np.sqrt(np.mean(residuals_H**2))

print(f"A_H    = {A_H:.10f}")
print(f"B_H    = {B_H:.10f}")
print(f"R²_H   = {R2_H:.10f}")
print(f"RMSE_H = {RMSE_H:.10f}")
```

```python
# ==========================================================
# Tableau des paramètres exponentiels
# ==========================================================

exponential_fit_summary = pd.DataFrame({
    "Observable": [
        "A_H",
        "B_H",
        "R2_H",
        "RMSE_H",
        "N_fit"
    ],
    "Valeur": [
        A_H,
        B_H,
        R2_H,
        RMSE_H,
        len(u_fit)
    ]
})

exponential_fit_summary
```

```python
# ==========================================================
# Comparaison survie empirique / modèle exponentiel
# ==========================================================

plt.figure(figsize=(8, 5))

plt.plot(
    u_thresholds,
    R_survival,
    label="Survie empirique"
)

plt.plot(
    u_thresholds,
    R_pred_full,
    linestyle="--",
    label="Ajustement exponentiel"
)

plt.xlabel("Seuil normalisé u")
plt.ylabel("R_super(u)")
plt.title("Ajustement exponentiel de la fonction de survie GVH")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

### Conclusion de l’ajustement exponentiel

L’ajustement exponentiel global fournit les paramètres

\[
A_H = 1.250962,
\qquad
B_H = 1.115625.
\]

La qualité de l’ajustement reste toutefois limitée :

\[
R_H^2 = 0.571769,
\qquad
\mathrm{RMSE}_H = 0.232148.
\]

La fonction exponentielle ne reproduit pas correctement la structure complète de la fonction de survie. Elle surestime notamment la survie aux faibles seuils, ne représente pas le plateau initial et décroît trop lentement dans la région finale.

Ces paramètres sont conservés comme observables GVH de référence afin de permettre une comparaison homogène avec les autres pipelines. Cependant, ils ne doivent pas être interprétés comme une description optimale de la distribution des transitions du puits quantique infini.

```python
# ==========================================================
# Ajustement exponentiel contraint : A_H = 1
# ==========================================================

def constrained_exponential_model(u_value, B_H):
    return np.exp(-B_H * u_value)

parameters_constrained, covariance_constrained = curve_fit(
    constrained_exponential_model,
    u_fit,
    R_fit,
    p0=[1.0],
    bounds=([0.0], [np.inf]),
    maxfev=20000
)

B_H_constrained = parameters_constrained[0]

R_pred_constrained = constrained_exponential_model(
    u_fit,
    B_H_constrained
)

residuals_constrained = R_fit - R_pred_constrained

SSE_constrained = np.sum(residuals_constrained**2)
SST_constrained = np.sum((R_fit - np.mean(R_fit))**2)

R2_H_constrained = 1.0 - SSE_constrained / SST_constrained
RMSE_H_constrained = np.sqrt(
    np.mean(residuals_constrained**2)
)

print(f"A_H contraint = 1.000000")
print(f"B_H contraint = {B_H_constrained:.10f}")
print(f"R² contraint  = {R2_H_constrained:.10f}")
print(f"RMSE contrainte = {RMSE_H_constrained:.10f}")
```

```python
# ============================================================
# GVH 0.1 — Observables angulaires
# ============================================================

import numpy as np

# Vecteurs de transition
delta_vectors = delta_X

angles = []

for i in range(len(delta_vectors) - 1):

    v1 = delta_vectors[i]
    v2 = delta_vectors[i + 1]

    n1 = np.linalg.norm(v1)
    n2 = np.linalg.norm(v2)

    if n1 == 0 or n2 == 0:
        continue

    cos_theta = np.dot(v1, v2) / (n1 * n2)

    cos_theta = np.clip(cos_theta, -1.0, 1.0)

    theta = np.degrees(np.arccos(cos_theta))

    angles.append(theta)

angles = np.array(angles)

print(f"Nombre d'angles : {len(angles)}")
```
