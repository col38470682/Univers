# 6. Classement technique

````python
score_summary_df = (
    evaluation_df
    .groupby(["model_id","name"], as_index=False)
    .agg(
        weighted_score=("weighted_score","sum"),
        total_weight=("weight","sum"),
    )
)

score_summary_df["normalized_score"] = (
    score_summary_df["weighted_score"]
    / score_summary_df["total_weight"]
)

score_summary_df = (
    score_summary_df
    .sort_values("normalized_score", ascending=False)
    .reset_index(drop=True)
)

score_summary_df["rank"] = np.arange(len(score_summary_df)) + 1
score_summary_df
````

# 7. Portes dures

````python
HARD_GATE_THRESHOLD = 0.75

hard_gate_detail_df = evaluation_df.loc[
    evaluation_df["hard_gate"]
].copy()

hard_gate_detail_df["hard_gate_pass"] = (
    hard_gate_detail_df["score"] >= HARD_GATE_THRESHOLD
)

hard_gate_summary_df = (
    hard_gate_detail_df
    .groupby(["model_id","name"], as_index=False)
    .agg(
        hard_gates_passed=("hard_gate_pass","sum"),
        hard_gate_count=("hard_gate_pass","count"),
        all_hard_gates_pass=("hard_gate_pass","all"),
    )
)

hard_gate_summary_df
````
