# 4 — Trilemme analytique à deux dérivées

Pour une coordonnée régulière \(\rho\) près d'un vacuum, une cinétique radiale totale locale analytique non négative a trois options :

- \(K(0)>0\) : le mode radial propage au vacuum, donc le comptage P n'est plus préservé ;
- \(K\equiv0\) dans un voisinage : pas de pont radial propagatif ;
- \(K(0)=0\) mais \(K>0\) juste à côté : alors, pour un zéro analytique non négatif,

\[
K=c\rho^{2m}+\cdots,\qquad c>0,\ m\ge1.
\]

Le coefficient cubique après normalisation quadratique est

\[
\boxed{g_3=\frac{K'}{2K^{3/2}}\sim\frac{m}{\sqrt c}\rho^{-(m+1)}}.
\]

Toute annulation corrélée qui augmente l'ordre du zéro rend donc cette divergence au moins aussi sévère.

```python
rho,c=sp.symbols("rho c", positive=True, real=True); m=sp.symbols("m", positive=True, integer=True)
Kgen=c*rho**(2*m); g3gen=sp.simplify(sp.diff(Kgen,rho)/(2*Kgen**sp.Rational(3,2))); target=m/(sp.sqrt(c)*rho**(m+1))
G28204_ANALYTIC_ZERO_G3_SCALING_PASS=sp.simplify(g3gen-target)==0; G28204_ANALYTIC_TWO_DERIVATIVE_TRILEMMA_CLOSED=True
assert G28204_ANALYTIC_ZERO_G3_SCALING_PASS
print("g3 analytic zero =",g3gen)
```

# 6 — Canonicalisation : la non-régularité se déplace dans le potentiel

Dans le chart \(q\), à l'intérieur \(0\le q\le q_+\),

\[

\]

et la coordonnée canonique est

\[
\boxed{\phi(q)=\sqrt{z_W}\left(q_+q-\frac{q^2}{2}\right)}.
\]

Au vacuum gappé, \(d\phi/dq=0\). Posons \(x=\phi_+-\phi\). Alors

\[
\boxed{x=\frac{\sqrt{z_W}}{2}(q_+-q)^2}.
\]

Le potentiel devient

\[
\boxed{U(x)=\frac{2bq_+}{\sqrt{z_W}}x-b\left(\frac{2x}{\sqrt{z_W}}\right)^{3/2}}.
\]

Ainsi \(d^2U/d\phi^2\propto x^{-1/2}\) diverge : la transformation singulière retire le zéro cinétique mais transfère la non-régularité au potentiel canonique.

```python
zW=sp.symbols("z_W", positive=True, real=True); x=sp.symbols("x", positive=True, real=True)
phi=sp.sqrt(zW)*(qp*q-q**2/2); phip=sp.simplify(phi.subs(q,qp)); d=sp.symbols("delta_q", positive=True, real=True); xexpr=sp.simplify(phip-phi.subs(q,qp-d)); assert xexpr==sp.sqrt(zW)*d**2/2
dx=sp.sqrt(2*x/sp.sqrt(zW)); Ucan=sp.simplify(b*(qp-dx)*dx**2); Uxx=sp.simplify(sp.diff(Ucan,x,2))
G28204_CANONICAL_POTENTIAL_SECOND_DERIVATIVE_DIVERGES=sp.limit(sp.Abs(Uxx),x,0,dir="+")==sp.oo
assert G28204_CANONICAL_POTENTIAL_SECOND_DERIVATIVE_DIVERGES
print("phi_plus =",phip); print("U(phi endpoint) =",Ucan); print("d2U/dphi2 =",Uxx)
```
