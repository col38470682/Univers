# 11 — Pourquoi le comptage R ne peut pas être transporté à \(q=0\)

Il faut réconcilier deux descriptions :

\[
q=r^2.
\]

Dans le chart \(q\), Route R possède :

\[
\frac{\eta_Q}{2}\dot q^2.
\]

Mais :

\[
\dot q=2r\dot r,
\]

donc :

\[
\boxed{
L_{R,\rm time}
=
2\eta_Qr^2\dot r^2.
}
\]

Le momentum radial en \(r\) est :

\[
\boxed{
P_r=4\eta_Qr^2\dot r.
}
\]

Alors que :

\[
P_q=\eta_Q\dot q=2\eta_Qr\dot r,
\]

d'où :

\[
\boxed{
P_r=2rP_q.
}
\]

La transformation canonique devient singulière à :

\[
\boxed{r=0\Longleftrightarrow q=0}.
\]

Ainsi, le fait que la Hessienne soit non nulle dans le chart \(q\) ne permet **pas** de conclure qu'un mode radial régulier subsiste dans les variables fondamentales \(A\) exactement à la branche GR.

Le résultat `.28.19R` est donc strictement :

\[
\boxed{q>0}.
\]

```python
r,rdot,eta = sp.symbols("r rdot eta_Q", positive=True, real=True)
qdot_r = 2*r*rdot
Lr = sp.factor(eta*qdot_r**2/2)
Pr = sp.factor(sp.diff(Lr,rdot))
Pq_in_r = sp.factor(eta*qdot_r)
relation = sp.simplify(Pr-2*r*Pq_in_r)
Kr = sp.factor(sp.diff(Lr,rdot,2))

G2819R_Q_TO_R_CANONICAL_MAP_PASS = (relation==0)
G2819R_R_HESSIAN = Kr
G2819R_Q_CHART_SINGULAR_AT_GR = (sp.limit(Kr,r,0,dir="+")==0)

assert G2819R_Q_TO_R_CANONICAL_MAP_PASS
assert G2819R_Q_CHART_SINGULAR_AT_GR

print("L_R(r) =",Lr)
print("P_r =",Pr)
print("P_q =",Pq_in_r)
print("P_r - 2 r P_q =",relation)
print("K_r =",Kr)
```
