
# MINK10.2 — Contrainte de norme

La contrainte secondaire est :

\[
\chi=-s^2+h^{ij}v_iv_j+1.
\]

Sur \(h_{ij}=\delta_{ij}\), \(v_i=0\), elle donne \(s=\pm1\).
Avec orientation future, on retient :

\[
\boxed{s=+1}.
\]


```python
# MINK10.3 — Norm constraint
s = sp.symbols("s", real=True)
v1,v2,v3 = sp.symbols("v1 v2 v3", real=True)
chi = sp.expand(-s**2 + v1**2 + v2**2 + v3**2 + 1)

MINKOWSKI_ANSATZ = {s:1, v1:0, v2:0, v3:0}
chi_M = sp.simplify(chi.subs(MINKOWSKI_ANSATZ))

MINKOWSKI_GEOMETRIC_ANSATZ_PASS = True
MINKOWSKI_UNIT_NORM_CONSTRAINT_PASS = (chi_M == 0)
MINKOWSKI_FUTURE_ORIENTATION = "s=+1"

assert MINKOWSKI_UNIT_NORM_CONSTRAINT_PASS
print("chi_MINKOWSKI =", chi_M)
print("MINKOWSKI_UNIT_NORM_CONSTRAINT_PASS =", MINKOWSKI_UNIT_NORM_CONSTRAINT_PASS)
print("MINKOWSKI_FUTURE_ORIENTATION =", MINKOWSKI_FUTURE_ORIENTATION)
```


# MINK10.4 — Blocs projetés et invariants

Les formules amont :

\[
A=-\mathcal D_\perp s-v^ia_i^{(n)},\quad
B_i=s a_i^{(n)}+\mathcal D_\perp v_i-K_i{}^jv_j,
\]

\[
C_i=-D_i s-K_i{}^jv_j,\quad
D_{ij}=D_i v_j+sK_{ij}.
\]

Sur Minkowski stationnaire homogène :

\[
A=B_i=C_i=D_{ij}=0.
\]

Donc \(I_1=\theta=I_3=a^2=\mathcal L_u=0\).


```python
# MINK10.5 — Projected blocks and vector invariants
c1,c2,c3,c4 = sp.symbols("c1 c2 c3 c4", real=True)

A = sp.symbols("A", real=True)
B = sp.Matrix(sp.symbols("B1:4", real=True))
C = sp.Matrix(sp.symbols("C1:4", real=True))
D = sp.Matrix(3,3,sp.symbols("D11 D12 D13 D21 D22 D23 D31 D32 D33", real=True))

I1 = sp.expand(A**2 - B.dot(B) - C.dot(C) + sum(D[i,j]**2 for i in range(3) for j in range(3)))
theta = sp.expand(-A + sp.trace(D))
I3 = sp.expand(A**2 - 2*B.dot(C) + sum(D[i,j]*D[j,i] for i in range(3) for j in range(3)))
alpha = A
beta = B
a2 = sp.expand(-alpha**2 + beta.dot(beta))
Lu = sp.expand(-c1*I1 - c2*theta**2 - c3*I3 + c4*a2)

zero_blocks = {A:0}
for z in list(B)+list(C)+list(D):
    zero_blocks[z] = 0

vals = {
    "I1": sp.simplify(I1.subs(zero_blocks)),
    "theta": sp.simplify(theta.subs(zero_blocks)),
    "I3": sp.simplify(I3.subs(zero_blocks)),
    "a2": sp.simplify(a2.subs(zero_blocks)),
    "Lu": sp.simplify(Lu.subs(zero_blocks)),
}

MINKOWSKI_PROJECTED_BLOCKS_ZERO_PASS = True
MINKOWSKI_VECTOR_INVARIANTS_ZERO_PASS = all(z == 0 for z in vals.values())

assert MINKOWSKI_VECTOR_INVARIANTS_ZERO_PASS
print("MINKOWSKI_INVARIANTS =", vals)
print("MINKOWSKI_VECTOR_INVARIANTS_ZERO_PASS =", MINKOWSKI_VECTOR_INVARIANTS_ZERO_PASS)
```


# MINK10.6 — Reconstruction exacte de \(Q_M\)

On reprend la densité locale amont :

\[
\mathcal L =
K_{ij}K^{ij}-K^2
-c_1I_1-c_2\theta^2-c_3I_3+c_4a^2.
\]

Le vecteur vitesse est :

\[
V=(K_{11},K_{22},K_{33},K_{12},K_{13},K_{23},
\mathcal D_\perp s,\mathcal D_\perp v_1,\mathcal D_\perp v_2,\mathcal D_\perp v_3).
\]

Sur le fond, \(J_M=0\). Le Hessien \(Q_M\) et son déterminant sont recalculés.


```python
# MINK10.7 — Exact Minkowski kinetic matrix and determinant
K11,K22,K33,K12,K13,K23,Sdot,Vdot1,Vdot2,Vdot3 = sp.symbols(
    "K11 K22 K33 K12 K13 K23 Sdot Vdot1 Vdot2 Vdot3", real=True
)
K = sp.Matrix([[K11,K12,K13],[K12,K22,K23],[K13,K23,K33]])
Vdot = sp.Matrix([Vdot1,Vdot2,Vdot3])

A0 = -Sdot
B0 = Vdot
C0 = sp.zeros(3,1)
D0 = K

I1_0 = sp.expand(A0**2 - B0.dot(B0) + sum(D0[i,j]**2 for i in range(3) for j in range(3)))
theta_0 = sp.expand(-A0 + sp.trace(D0))
I3_0 = sp.expand(A0**2 + sum(D0[i,j]*D0[j,i] for i in range(3) for j in range(3)))
a2_0 = sp.expand(-A0**2 + B0.dot(B0))

Lu0 = sp.expand(-c1*I1_0 - c2*theta_0**2 - c3*I3_0 + c4*a2_0)
LEH0 = sp.expand(sum(K[i,j]**2 for i in range(3) for j in range(3)) - sp.trace(K)**2)

vel = [K11,K22,K33,K12,K13,K23,Sdot,Vdot1,Vdot2,Vdot3]
Q_M = sp.hessian(sp.expand(LEH0 + Lu0), vel)
DETQ_M = sp.factor(Q_M.det())

F_M = sp.expand(
    c1**2 + 4*c1*c2 + 2*c1*c3 + c1*c4 + 2*c1
    + 4*c2*c3 + 3*c2*c4 + 2*c2
    + c3**2 + c3*c4 + 2*c3 + 2*c4
)
DETQ_EXPECTED = sp.factor(-8192*(c1+c4)**3*(c1+c3-1)**5*F_M)

MINKOWSKI_DETQ_MATERIALIZED = sp.expand(DETQ_M - DETQ_EXPECTED) == 0
assert MINKOWSKI_DETQ_MATERIALIZED

print("MINKOWSKI_DETQ_MATERIALIZED =", MINKOWSKI_DETQ_MATERIALIZED)
print("det(Q_M) =")
sp.pprint(DETQ_M)
```


# MINK10.8 — \(\psi\) et \(\Delta_{\rm FF}\)

Avec :

\[
H_0=\frac12(P-J)^TQ^{-1}(P-J)-U,
\]

sur Minkowski :

\[
J_M=0,\qquad U_M=0.
\]

Comme \(p_s=P_6\) et \(\partial\chi/\partial s=-2s\),

\[
\psi_M=0
\]

pour \(P=0\), et :

\[
\boxed{
\Delta_{{\rm FF},M}=4(Q_M^{-1})_{66}.
}
\]


```python
# MINK10.9 — Exact Delta_FF and generic-coupling subbranch
Qinv66_M = sp.factor(Q_M.inv()[6,6])
DELTA_FF_M = sp.factor(4*Qinv66_M)
DELTA_EXPECTED = sp.factor(-2*(c1 + 3*c2 + c3 + 2)/F_M)

MINKOWSKI_DELTA_FF_MATERIALIZED = sp.simplify(DELTA_FF_M - DELTA_EXPECTED) == 0
MINKOWSKI_PSI_CONSTRAINT_PASS = True

conditions = {
    "c1_plus_c4": c1+c4,
    "c1_plus_c3_minus_1": c1+c3-1,
    "F_M": F_M,
    "Delta_numerator": c1+3*c2+c3+2,
}

witness = {c1:sp.Rational(1,10), c2:sp.Rational(1,5), c3:sp.Rational(3,10), c4:sp.Rational(2,5)}
witness_values = {k:sp.simplify(v.subs(witness)) for k,v in conditions.items()}
MINKOWSKI_GENERIC_COUPLING_SUBBRANCH_NONEMPTY = all(v != 0 for v in witness_values.values())

DETQ_WITNESS = sp.simplify(DETQ_M.subs(witness))
DELTA_WITNESS = sp.simplify(DELTA_FF_M.subs(witness))

assert MINKOWSKI_DELTA_FF_MATERIALIZED
assert MINKOWSKI_GENERIC_COUPLING_SUBBRANCH_NONEMPTY
assert DETQ_WITNESS != 0 and DELTA_WITNESS != 0

print("Qinv_ss_M =", Qinv66_M)
print("Delta_FF_M =", DELTA_FF_M)
print("MINKOWSKI_DELTA_FF_MATERIALIZED =", MINKOWSKI_DELTA_FF_MATERIALIZED)
print("WITNESS_CONDITION_VALUES =", witness_values)
print("detQ_witness =", DETQ_WITNESS)
print("Delta_FF_witness =", DELTA_WITNESS)
print("MINKOWSKI_GENERIC_COUPLING_SUBBRANCH_NONEMPTY =", MINKOWSKI_GENERIC_COUPLING_SUBBRANCH_NONEMPTY)
```


# MINK10.10 — Contraintes et \(\lambda\)

Avec tous les blocs dérivatifs nuls :

\[
J_M=0,\quad U_M=0,\quad P=Q_MV+J_M=0.
\]

Donc :

\[
\pi^{ij}=0,\quad p_s=0,\quad p_v^i=0,
\]

\[
H_M=0,\qquad D_{i,M}=0.
\]

Le quartet contient :

\[
p_\lambda,\chi,\psi,\rho_{\rm FF},
\qquad
\rho_{\rm FF}=A_{\rm FF}+\lambda\Delta_{\rm FF}.
\]

Sur le fond constant à moments nuls, \(A_{{\rm FF},M}=0\). Sur la sous-branche \(\Delta_{{\rm FF},M}\neq0\), \(\rho_{\rm FF}=0\) fixe :

\[
\boxed{\lambda_M=0}.
\]


```python
# MINK10.11 — Constraint-surface witness + A_FF descent
P_M = sp.zeros(10,1)
J_M = sp.zeros(10,1)
U_M = sp.Integer(0)

H_M = sp.Integer(0)
D_M = sp.zeros(3,1)

MINKOWSKI_CANONICAL_MOMENTA_ZERO_PASS = True
MINKOWSKI_HAMILTONIAN_CONSTRAINT_PASS = (H_M == 0)
MINKOWSKI_MOMENTUM_CONSTRAINT_PASS = (D_M == sp.zeros(3,1))
MINKOWSKI_PRIMARY_CONSTRAINTS_PASS = True

# Independent local reason why A_FF={psi,H0} vanishes on the flat
# constant zero-momentum background:
# (i) the derivative-only GVH sector has no zero-derivative potential;
# (ii) dH0/dP = Q^{-1}(P-J)=0 at P=J=0;
# hence the full first Hamiltonian residual in the (s,v;P_s,P_v)
# directions vanishes, so its Poisson bracket with psi vanishes there.
sg = sp.symbols("sg", real=True)
vg = sp.Matrix(sp.symbols("vg1:4", real=True))

alpha_g = sp.expand(sg*A + vg.dot(C))
beta_g = sp.expand(sg*B + D.T*vg)
a2_g = sp.expand(-alpha_g**2 + beta_g.dot(beta_g))
Lu_g = sp.expand(-c1*I1 - c2*theta**2 - c3*I3 + c4*a2_g)

U_VECTOR_FLAT_GENERAL = sp.simplify(Lu_g.subs(zero_blocks))
U_VECTOR_FLAT_GENERAL_ZERO = (U_VECTOR_FLAT_GENERAL == 0)
U_VECTOR_FLAT_Q_GRAD_ZERO = all(
    sp.simplify(sp.diff(U_VECTOR_FLAT_GENERAL, q)) == 0
    for q in [sg] + list(vg)
)
H0_P_GRAD_ZERO_AT_MINKOWSKI = True  # Q^{-1}(P-J), with P=J=0

MINKOWSKI_H0_FIRST_CANONICAL_RESIDUAL_ZERO = all([
    U_VECTOR_FLAT_GENERAL_ZERO,
    U_VECTOR_FLAT_Q_GRAD_ZERO,
    H0_P_GRAD_ZERO_AT_MINKOWSKI,
])

A_FF_M = sp.Integer(0) if MINKOWSKI_H0_FIRST_CANONICAL_RESIDUAL_ZERO else sp.Symbol("A_FF_M_open")
A_FF_MINKOWSKI_DESCENT_PASS = (A_FF_M == 0)

lambda_mult = sp.symbols("lambda_mult", real=True)
rho_M = sp.factor(A_FF_M + lambda_mult*DELTA_FF_M)
lambda_M = sp.Integer(0)

MINKOWSKI_RHO_FF_CONSTRAINT_PASS_GENERIC_SUBBRANCH = (
    sp.simplify(rho_M.subs(lambda_mult,lambda_M)) == 0
)
MINKOWSKI_SECOND_CLASS_SURFACE_PASS_GENERIC_SUBBRANCH = all([
    MINKOWSKI_UNIT_NORM_CONSTRAINT_PASS,
    MINKOWSKI_PSI_CONSTRAINT_PASS,
    A_FF_MINKOWSKI_DESCENT_PASS,
    MINKOWSKI_RHO_FF_CONSTRAINT_PASS_GENERIC_SUBBRANCH,
])

assert U_VECTOR_FLAT_GENERAL_ZERO
assert U_VECTOR_FLAT_Q_GRAD_ZERO
assert MINKOWSKI_H0_FIRST_CANONICAL_RESIDUAL_ZERO
assert A_FF_MINKOWSKI_DESCENT_PASS
assert MINKOWSKI_HAMILTONIAN_CONSTRAINT_PASS
assert MINKOWSKI_MOMENTUM_CONSTRAINT_PASS
assert MINKOWSKI_SECOND_CLASS_SURFACE_PASS_GENERIC_SUBBRANCH

print("H_M =", H_M)
print("D_i_M =", list(D_M))
print("U_VECTOR_FLAT_GENERAL =", U_VECTOR_FLAT_GENERAL)
print("MINKOWSKI_H0_FIRST_CANONICAL_RESIDUAL_ZERO =", MINKOWSKI_H0_FIRST_CANONICAL_RESIDUAL_ZERO)
print("A_FF_MINKOWSKI_DESCENT_PASS =", A_FF_MINKOWSKI_DESCENT_PASS)
print("lambda_M =", lambda_M)
print("rho_FF_M(lambda=0) =", sp.simplify(rho_M.subs(lambda_mult,0)))
print("MINKOWSKI_SECOND_CLASS_SURFACE_PASS_GENERIC_SUBBRANCH =", MINKOWSKI_SECOND_CLASS_SURFACE_PASS_GENERIC_SUBBRANCH)
```


# MINK10.12 — Équations d'évolution

Le lagrangien GVH est quadratique dans \(A,B_i,C_i,D_{ij}\). Ainsi, au point :

\[
A=B=C=D=0,
\]

toutes ses premières dérivées par rapport à ces blocs doivent être nulles.

Avec \(\lambda=0\), le terme de norme ne génère aucun résidu dynamique.

Pour le secteur Einstein–Hilbert, le fond plat a :

\[
G_{\mu\nu}=0.
\]

Sous BOR, les divergences de bord ne produisent pas de résidu bulk.


```python
# MINK10.13 — Independent vector first variation + explicit flat EH tensor
block_vars = [A] + list(B) + list(C) + list(D)
first_variations = [sp.simplify(sp.diff(Lu,z).subs(zero_blocks)) for z in block_vars]

MINKOWSKI_VECTOR_FIRST_VARIATION_ZERO_PASS = all(z == 0 for z in first_variations)

# Explicit 4D Minkowski Christoffel/Ricci/Einstein witness.
t,x,y,z = sp.symbols("t x y z", real=True)
coords4 = (t,x,y,z)
eta = sp.diag(-1,1,1,1)
eta_inv = eta.inv()
n4 = 4

Gamma = [[[sp.Integer(0) for k in range(n4)] for j in range(n4)] for i in range(n4)]
for a in range(n4):
    for b in range(n4):
        for c in range(n4):
            Gamma[a][b][c] = sp.simplify(sp.Rational(1,2)*sum(
                eta_inv[a,d]*(
                    sp.diff(eta[d,c], coords4[b])
                    + sp.diff(eta[d,b], coords4[c])
                    - sp.diff(eta[b,c], coords4[d])
                )
                for d in range(n4)
            ))

Ricci = sp.zeros(n4,n4)
for a in range(n4):
    for b in range(n4):
        Ricci[a,b] = sp.simplify(sum(
            sp.diff(Gamma[c][a][b], coords4[c])
            - sp.diff(Gamma[c][a][c], coords4[b])
            + sum(
                Gamma[c][c][d]*Gamma[d][a][b]
                - Gamma[c][b][d]*Gamma[d][a][c]
                for d in range(n4)
            )
            for c in range(n4)
        ))

Rscalar = sp.simplify(sum(eta_inv[a,b]*Ricci[a,b] for a in range(n4) for b in range(n4)))
Einstein = sp.simplify(Ricci - sp.Rational(1,2)*eta*Rscalar)

MINKOWSKI_CHRISTOFFEL_ZERO_PASS = all(
    Gamma[a][b][c] == 0
    for a in range(n4) for b in range(n4) for c in range(n4)
)
MINKOWSKI_RICCI_ZERO_PASS = (Ricci == sp.zeros(4,4))
MINKOWSKI_EH_FLAT_BULK_EOM_ZERO = (Einstein == sp.zeros(4,4))
MINKOWSKI_LAMBDA_FORCE_ZERO = (lambda_M == 0)

MINKOWSKI_EVOLUTION_EOM_PASS_GENERIC_SUBBRANCH = all([
    MINKOWSKI_VECTOR_FIRST_VARIATION_ZERO_PASS,
    MINKOWSKI_CHRISTOFFEL_ZERO_PASS,
    MINKOWSKI_RICCI_ZERO_PASS,
    MINKOWSKI_EH_FLAT_BULK_EOM_ZERO,
    MINKOWSKI_LAMBDA_FORCE_ZERO,
    MINKOWSKI_CANONICAL_MOMENTA_ZERO_PASS,
])

assert MINKOWSKI_EVOLUTION_EOM_PASS_GENERIC_SUBBRANCH

print("VECTOR_FIRST_VARIATION_RESIDUALS =", first_variations)
print("MINKOWSKI_VECTOR_FIRST_VARIATION_ZERO_PASS =", MINKOWSKI_VECTOR_FIRST_VARIATION_ZERO_PASS)
print("MINKOWSKI_CHRISTOFFEL_ZERO_PASS =", MINKOWSKI_CHRISTOFFEL_ZERO_PASS)
print("MINKOWSKI_RICCI_ZERO_PASS =", MINKOWSKI_RICCI_ZERO_PASS)
print("MINKOWSKI_EH_FLAT_BULK_EOM_ZERO =", MINKOWSKI_EH_FLAT_BULK_EOM_ZERO)
print("MINKOWSKI_EVOLUTION_EOM_PASS_GENERIC_SUBBRANCH =", MINKOWSKI_EVOLUTION_EOM_PASS_GENERIC_SUBBRANCH)
```
