
# 3. Free-\(C(r)\) static spherical geometry

Use

\[
ds^2
=
-A(r)\,dt^2
+
B(r)\,dr^2
+
C(r)\left(d\theta^2+\sin^2\theta\,d\phi^2\right).
\]

No areal-radius gauge is imposed.


````python

t, r, th, ph = sp.symbols("t r theta phi", real=True)
A = sp.Function("A")(r)
B = sp.Function("B")(r)
Cang = sp.Function("C")(r)

coords = [t, r, th, ph]
n = 4

g = sp.diag(-A, B, Cang, Cang*sp.sin(th)**2)
g_inv = sp.simplify(g.inv())

Gamma = [[[sp.Integer(0) for _ in range(n)] for _ in range(n)] for _ in range(n)]

for rho in range(n):
    for mu in range(n):
        for nu in range(n):
            Gamma[rho][mu][nu] = sp.simplify(
                sum(
                    g_inv[rho,sig] * (
                        sp.diff(g[sig,nu], coords[mu])
                        + sp.diff(g[sig,mu], coords[nu])
                        - sp.diff(g[mu,nu], coords[sig])
                    )
                    for sig in range(n)
                ) / 2
            )

connection_checks = {
    "Gamma_t_tr": sp.simplify(Gamma[0][0][1]),
    "Gamma_r_tt": sp.simplify(Gamma[1][0][0]),
    "Gamma_r_rr": sp.simplify(Gamma[1][1][1]),
    "Gamma_r_thth": sp.simplify(Gamma[1][2][2]),
    "Gamma_th_rth": sp.simplify(Gamma[2][1][2]),
    "Gamma_ph_rph": sp.simplify(Gamma[3][1][3]),
    "Gamma_th_phph": sp.simplify(Gamma[2][3][3]),
    "Gamma_ph_thph": sp.simplify(Gamma[3][2][3]),
}

for key, value in connection_checks.items():
    print(key, "=", value)
````


The expected nonzero connections include

\[
\Gamma^t{}_{tr}=\frac{A'}{2A},\qquad
\Gamma^r{}_{tt}=\frac{A'}{2B},\qquad
\Gamma^r{}_{rr}=\frac{B'}{2B},
\]

\[
\Gamma^r{}_{\theta\theta}=-\frac{C'}{2B},
\qquad
\Gamma^\theta{}_{r\theta}
=
\Gamma^\phi{}_{r\phi}
=
\frac{C'}{2C}.
\]


````python

assert sp.simplify(Gamma[0][0][1] - sp.diff(A,r)/(2*A)) == 0
assert sp.simplify(Gamma[1][0][0] - sp.diff(A,r)/(2*B)) == 0
assert sp.simplify(Gamma[1][1][1] - sp.diff(B,r)/(2*B)) == 0
assert sp.simplify(Gamma[1][2][2] + sp.diff(Cang,r)/(2*B)) == 0
assert sp.simplify(Gamma[2][1][2] - sp.diff(Cang,r)/(2*Cang)) == 0
assert sp.simplify(Gamma[3][1][3] - sp.diff(Cang,r)/(2*Cang)) == 0

print("PASS — free-C(r) SSS connection audit.")
````


# 4. Aligned timelike branch — direct rederivation

Take

\[
u^\mu=(A^{-1/2},0,0,0).
\]

The unit constraint is

\[
u^\mu u_\mu=-1.
\]

We now recompute \(a_\mu\) instead of assuming it vanishes.


````python

u_con = sp.Matrix([1/sp.sqrt(A), 0, 0, 0])
u_cov = sp.simplify(g*u_con)

norm = sp.simplify((u_con.T*u_cov)[0])
assert norm == -1

nabla_u_cov = [[sp.Integer(0) for _ in range(n)] for _ in range(n)]
nabla_u_con = [[sp.Integer(0) for _ in range(n)] for _ in range(n)]

for mu in range(n):
    for nu in range(n):
        nabla_u_cov[mu][nu] = sp.simplify(
            sp.diff(u_cov[nu], coords[mu])
            - sum(Gamma[rho][mu][nu]*u_cov[rho] for rho in range(n))
        )
        nabla_u_con[mu][nu] = sp.simplify(
            sp.diff(u_con[nu], coords[mu])
            + sum(Gamma[nu][mu][rho]*u_con[rho] for rho in range(n))
        )

a_cov = sp.Matrix([
    sp.simplify(sum(u_con[nu]*nabla_u_cov[nu][mu] for nu in range(n)))
    for mu in range(n)
])

a_sq = sp.simplify((a_cov.T*g_inv*a_cov)[0])

print("u.u =", norm)
print("a_mu =", a_cov.T)
print("a_mu a^mu =", a_sq)
````


## Correction established by direct computation

For a nontrivial static potential \(A'(r)\neq0\),

\[
\boxed{
a_\mu
=
\left(
0,\frac{A'}{2A},0,0
\right)
}

\[
\boxed{

Therefore the statement

is **not valid for the aligned static branch in general**.

This agrees with the primary 0.3.2.5 calculation and is why \(c_4\) contributes to the effective static combination \(c_{14}=c_1+c_4\).


````python

expected_a_cov = sp.Matrix([0, sp.diff(A,r)/(2*A), 0, 0])
expected_a_sq = sp.diff(A,r)**2/(4*A**2*B)

assert all(sp.simplify(a_cov[i]-expected_a_cov[i]) == 0 for i in range(4))
assert sp.simplify(a_sq-expected_a_sq) == 0

ACCELERATION_ZERO_CLAIM = all(sp.simplify(x)==0 for x in a_cov)

print("Claim a_mu = 0 ?", ACCELERATION_ZERO_CLAIM)
assert ACCELERATION_ZERO_CLAIM is False
````


# 5. Invariant reduction and \(c_{14}\)

\[

\[
I_3=(\nabla_\mu u_\nu)(\nabla^\nu u^\mu),

and \(a^2\).


````python

I1 = sp.Integer(0)
for mu in range(n):
    for nu in range(n):
        for aa in range(n):
            for bb in range(n):
                I1 += (
                    g_inv[mu,aa]*g_inv[nu,bb]
                    * nabla_u_cov[mu][nu]
                    * nabla_u_cov[aa][bb]
                )
I1 = sp.simplify(I1)

theta_exp = sp.simplify(sum(nabla_u_con[mu][mu] for mu in range(n)))

I3 = sp.simplify(sum(
    nabla_u_cov[mu][nu]*nabla_u_con[nu][mu]
    for mu in range(n) for nu in range(n)
))

L_u = sp.simplify(-c1*I1 - c2*theta_exp**2 - c3*I3 + c4*a_sq)

print("I1 =", I1)
print("theta =", theta_exp)
print("I3 =", I3)
print("a^2 =", a_sq)
print("L_u =", sp.factor(L_u))
````


# 6. \(J^\alpha{}_\mu\) and the vector equation — explicit SSS substitution

Construct

\[
J^\alpha{}_\mu
=
K^{\alpha\beta}{}_{\mu\nu}\nabla_\beta u^\nu
\]

directly from the primary \(K\)-tensor convention.


````python

J = [[sp.Integer(0) for _ in range(n)] for _ in range(n)]

for aa in range(n):
    for mu in range(n):
        expr = 0
        for bb in range(n):
            for nu in range(n):
                K = (
                    c1*g_inv[aa,bb]*g[mu,nu]
                    + c2*(1 if aa==mu else 0)*(1 if bb==nu else 0)
                    + c3*(1 if aa==nu else 0)*(1 if bb==mu else 0)
                    - c4*u_con[aa]*u_con[bb]*g[mu,nu]
                )
                expr += K*nabla_u_con[bb][nu]
        J[aa][mu] = sp.simplify(expr)

nonzero_J = [
    (aa,mu,sp.factor(J[aa][mu]))
    for aa in range(n)
    for mu in range(n)
    if J[aa][mu] != 0
]

nonzero_J
````

````python

# Mixed-tensor divergence nabla_alpha J^alpha_mu
divJ = []

for mu in range(n):
    expr = 0
    for aa in range(n):
        expr += sp.diff(J[aa][mu], coords[aa])
        for rho in range(n):
            expr += Gamma[aa][aa][rho]*J[rho][mu]
            expr -= Gamma[rho][aa][mu]*J[aa][rho]
    divJ.append(sp.simplify(expr))

E_without_lambda = []
for mu in range(n):
    extra = c4*sum(a_cov[aa]*nabla_u_con[mu][aa] for aa in range(n))
    E_without_lambda.append(sp.simplify(divJ[mu] + extra))

vector_nonzero = [
    (mu, sp.factor(E_without_lambda[mu]))
    for mu in range(n)
    if E_without_lambda[mu] != 0
]

vector_nonzero
````


On the aligned SSS branch, only the timelike vector equation is nontrivial before the multiplier term is added. Since

\[
u_t=-\sqrt A,
\]

the vector equation determines \(\lambda(r)\) rather than generating an additional independent radial equation for \(A,B,C\):

\[
E_t^{(u)}+\lambda u_t=0.
\]


````python

lambda_sss = sp.simplify(E_without_lambda[0]/sp.sqrt(A))

vector_full = [
    sp.simplify(E_without_lambda[mu] + lambda_sss*u_cov[mu])
    for mu in range(n)
]

print("lambda(r) =")
sp.pprint(sp.factor(lambda_sss))

print("\nVector EOM residuals:")
for mu, value in enumerate(vector_full):
    print(mu, sp.factor(value))

assert all(sp.simplify(v)==0 for v in vector_full)
print("PASS — aligned SSS vector equation closes by determining lambda(r).")
````
