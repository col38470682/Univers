# Niveau 1 — Variables polynomialement définies

On définit le tenseur mixte :

\[
M^\mu{}_\nu=g^{\mu\alpha}D_{\alpha\nu}.
\]

Puis sa trace :

\[
T_D=\operatorname{Tr}M.
\]

Le tenseur traceless 4D est :

\[
\boxed{
A^\mu{}_\nu
=
M^\mu{}_\nu
-
\frac14T_D\,\delta^\mu{}_\nu.
}
\]

Et :

\[
\boxed{
Q_D=\operatorname{Tr}(A^2).
}
\]

Tout ceci est polynomial en \(M\), donc en \(g^{-1}D\), sans projecteur spectral.

Sur la branche \(1+3\) :

\[
M_0=
\operatorname{diag}
(\lambda_t,\lambda_s,\lambda_s,\lambda_s),
\qquad
\Delta=\lambda_t-\lambda_s,
\]

on obtient :

\[
\boxed{
A_0=
\operatorname{diag}
\left(
\frac{3\Delta}{4},
-\frac{\Delta}{4},
-\frac{\Delta}{4},
-\frac{\Delta}{4}
\right),
}
\]

et :

\[
\boxed{
Q_D=\frac34\Delta^2.
}
\]

```python
lt,ls = sp.symbols("lambda_t lambda_s", real=True)
Delta = sp.factor(lt-ls)
I4 = sp.eye(4)

M0 = sp.diag(lt,ls,ls,ls)
T0 = sp.trace(M0)
A0 = sp.simplify(M0-T0*I4/4)
Q0 = sp.factor(sp.trace(A0*A0))

A0_expected = sp.diag(
    sp.Rational(3,4)*Delta,
    -sp.Rational(1,4)*Delta,
    -sp.Rational(1,4)*Delta,
    -sp.Rational(1,4)*Delta,
)

G2817_CENTERED_TENSOR_IDENTITY_PASS = (
    sp.simplify(A0-A0_expected) == sp.zeros(4)
)
G2817_Q_GAP_IDENTITY_PASS = (
    sp.simplify(Q0-sp.Rational(3,4)*Delta**2) == 0
)

assert G2817_CENTERED_TENSOR_IDENTITY_PASS
assert G2817_Q_GAP_IDENTITY_PASS

print("A0 =")
sp.pprint(A0)
print("Q_D =",Q0)
```

# Niveau 2.4 — Domaine de ghost-freedom sur un fond diagonal général

\[

\[
\boxed{

Pour une perturbation spatial-spatial \(ij\), la condition est :

\[
\boxed{
2K_SQ
>
\kappa_D(a_i-a_j)^2.
}
\]

Sur la branche \(1+3\), la première devient :

\[

\[

\]

Donc le candidat est sain dans un voisinage spectral du fond \(1+3\), mais la ghost-freedom globale sur tout l'espace anisotrope n'est pas démontrée.

```python
a0,a1,a2,a3,x = sp.symbols("a0 a1 a2 a3 x", real=True)
Qgen = sp.factor(a0**2+a1**2+a2**2+a3**2)

At = sp.diag(a0,a1,a2,a3)

# timelike-spatial pair 0-1
dAti = sp.zeros(4)
dAti[0,1] = -x
dAti[1,0] = x

# spatial-spatial pair 1-2
dAss = sp.zeros(4)
dAss[1,2] = x
dAss[2,1] = x

def pair_coeff(dA_pair):
    tr2 = sp.factor(sp.trace(dA_pair*dA_pair))
    commp = sp.simplify(At*dA_pair-dA_pair*At)
    trc2 = sp.factor(sp.trace(commp*commp))
    # coefficient of x^2 in time L
    L = sp.factor(KS*Qgen*tr2/2 + kappa*trc2/4)
    return sp.factor(L/x**2), tr2, trc2

coeff_ti,tr2_ti,trc2_ti = pair_coeff(dAti)
coeff_ss,tr2_ss,trc2_ss = pair_coeff(dAss)

expected_ti = sp.factor(-KS*Qgen+kappa*(a0-a1)**2/2)
expected_ss = sp.factor(KS*Qgen-kappa*(a1-a2)**2/2)

G2817_GENERAL_EIGENPAIR_KINETIC_COEFFICIENTS_PASS = all([
    sp.simplify(coeff_ti-expected_ti)==0,
    sp.simplify(coeff_ss-expected_ss)==0,
])

assert G2817_GENERAL_EIGENPAIR_KINETIC_COEFFICIENTS_PASS

print("timelike-spatial coefficient =",coeff_ti)
print("spatial-spatial coefficient =",coeff_ss)
```
