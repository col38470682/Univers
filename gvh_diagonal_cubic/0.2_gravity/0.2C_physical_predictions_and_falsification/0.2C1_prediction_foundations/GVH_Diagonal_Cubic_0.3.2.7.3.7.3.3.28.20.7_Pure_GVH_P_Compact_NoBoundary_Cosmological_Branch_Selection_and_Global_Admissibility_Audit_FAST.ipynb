# 4 — Équation cosmologique algébrique du gap

Le potentiel reste
\[
U(q)=\mu q-aq^2+bq^3,\qquad \mu=\frac{a^2}{4b}.
\]
À \(H\) fixé, le secteur radial voit
\[
\boxed{W_H(q)=U(q)+\frac32\alpha H^2q^2}.
\]
Comme \(q=r^2\), l'équation fondamentale est
\[
\boxed{2r\,W_H'(r^2)=0}.
\]
Ainsi \(r=0\) reste toujours la solution GR. Pour \(r\neq0\),
\[
\boxed{U'(q)+3\alpha H^2q=0}.
\]
Le gap demeure auxiliaire : l'expansion déforme ses extrema sans créer de \(\dot q^2\).

```python
aa,bb,h2=sp.symbols("a b h2", positive=True, real=True)
mu=aa**2/(4*bb); qp=aa/(2*bb)
Uq=sp.factor(mu*q-aa*q**2+bb*q**3)
Wp=sp.expand(sp.diff(Uq,q)+3*alpha*h2*q)
assert sp.simplify(Wp-(3*bb*q**2+(-2*aa+3*alpha*h2)*q+aa**2/(4*bb)))==0
print("U(q) =",Uq); print("W_H'(q) =",sp.factor(Wp))
```
