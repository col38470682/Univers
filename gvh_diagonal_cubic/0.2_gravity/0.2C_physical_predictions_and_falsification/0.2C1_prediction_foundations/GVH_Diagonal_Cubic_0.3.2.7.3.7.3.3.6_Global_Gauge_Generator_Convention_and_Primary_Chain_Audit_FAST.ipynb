# GGA.2 — Convention canonique minimale

On fixe le crochet canonique standard :

\[
\{q,p\}=+1.
\]

Pour une variation cinématique externe \(X(q)\), le momentum map usuel est :

\[
J_X=p\,X(q).
\]

Deux conventions d'action sont alors possibles :

\[
\delta_XF=\{F,J_X\},
\]

ou :

\[
\delta_XF=\{D_X,F\}.
\]

Elles ne peuvent pas utiliser le même signe du générateur.

Le témoin suivant décide ce point directement à partir du crochet canonique.

```python
# GGA.3 — Exact one-pair sign witness

q, p, Xq = sp.symbols("q p Xq")

def pb(F, G, qs, ps):
    return sp.expand(sum(
        sp.diff(F, qk)*sp.diff(G, pk)
        - sp.diff(F, pk)*sp.diff(G, qk)
        for qk, pk in zip(qs, ps)
    ))

J_plus = p*Xq
D_left = -J_plus

RIGHT_ACTION_PLUS = sp.expand(pb(q, J_plus, [q], [p]) - Xq)
LEFT_ACTION_PLUS = sp.expand(pb(J_plus, q, [q], [p]) - Xq)
LEFT_ACTION_MINUS = sp.expand(pb(D_left, q, [q], [p]) - Xq)

RIGHT_ACTION_PLUS_PASS = (RIGHT_ACTION_PLUS == 0)
LEFT_ACTION_PLUS_PASS = (LEFT_ACTION_PLUS == 0)
LEFT_ACTION_MINUS_PASS = (LEFT_ACTION_MINUS == 0)

assert RIGHT_ACTION_PLUS_PASS
assert not LEFT_ACTION_PLUS_PASS
assert LEFT_ACTION_MINUS_PASS

print("RIGHT_ACTION_PLUS_PASS =", RIGHT_ACTION_PLUS_PASS)
print("LEFT_ACTION_PLUS_PASS =", LEFT_ACTION_PLUS_PASS)
print("LEFT_ACTION_MINUS_PASS =", LEFT_ACTION_MINUS_PASS)
print("{+pX,q} =", pb(J_plus,q,[q],[p]))
print("{-pX,q} =", pb(D_left,q,[q],[p]))
```

Le résultat exact est :

\[
\{+pX,q\}=-X,
\qquad
\{-pX,q\}=+X.
\]

Donc, avec \(\{q,p\}=+1\) :

\[
\boxed{
\{D_X,F\}=+\delta_XF
\quad\Longrightarrow\quad
D_X=-J_X.
}
\]

À l'inverse :

\[
\boxed{
\{F,J_X\}=+\delta_XF
}
\]

utilise naturellement \(J_X=+\int p\,\delta_Xq\).

Le couple hérité

\[
D[\xi]=+\int p\,\mathcal L_\xi q,
\qquad
\{D[\xi],F\}=+\mathcal L_\xi F
\]

doit donc être réconcilié avant le gel d'un générateur global.

# GGA.4 — Témoin d'algèbre du momentum map

Le problème n'est pas limité à une paire canonique.

Pour :

\[
J_X=p_aX^a(q),
\]

le crochet canonique réalise un **anti-homomorphisme** lorsque l'action est écrite à droite :

\[
\{J_X,J_Y\}=-J_{[X,Y]}.
\]

En revanche :

\[
D_X=-J_X
\]

réalise :

\[
\{D_X,D_Y\}=D_{[X,Y]}
\]

et :

\[
\{D_X,q^a\}=X^a(q).
\]

Le témoin polynomial suivant vérifie ces deux relations exactement.

```python
# GGA.5 — Exact polynomial momentum-map Lie witness

q1,q2,p1,p2 = sp.symbols("q1 q2 p1 p2")
qs = [q1,q2]
ps = [p1,p2]

X = sp.Matrix([
    q1 + q1*q2,
    q2 + q1**2,
])
Y = sp.Matrix([
    q2 + q1*q2,
    q1 - q2**2,
])

def vec_bracket(Xv, Yv):
    return sp.Matrix([
        sp.expand(sum(
            Xv[b]*sp.diff(Yv[a], qs[b])
            - Yv[b]*sp.diff(Xv[a], qs[b])
            for b in range(2)
        ))
        for a in range(2)
    ])

br = vec_bracket(X,Y)

JX = sp.expand(p1*X[0] + p2*X[1])
JY = sp.expand(p1*Y[0] + p2*Y[1])
Jbr = sp.expand(p1*br[0] + p2*br[1])

DX = -JX
DY = -JY
Dbr = -Jbr

RIGHT_ANTI_HOM_RESID = sp.expand(pb(JX,JY,qs,ps) + Jbr)
LEFT_HOM_RESID = sp.expand(pb(DX,DY,qs,ps) - Dbr)

LEFT_ACTION_Q_RESID = [
    sp.expand(pb(DX, qs[a], qs, ps) - X[a])
    for a in range(2)
]

MOMENTUM_MAP_RIGHT_ANTI_HOM_PASS = (RIGHT_ANTI_HOM_RESID == 0)
MOMENTUM_MAP_LEFT_HOM_PASS = (LEFT_HOM_RESID == 0)
MOMENTUM_MAP_LEFT_ACTION_PASS = all(z == 0 for z in LEFT_ACTION_Q_RESID)

assert MOMENTUM_MAP_RIGHT_ANTI_HOM_PASS
assert MOMENTUM_MAP_LEFT_HOM_PASS
assert MOMENTUM_MAP_LEFT_ACTION_PASS

print("MOMENTUM_MAP_RIGHT_ANTI_HOM_PASS =", MOMENTUM_MAP_RIGHT_ANTI_HOM_PASS)
print("MOMENTUM_MAP_LEFT_HOM_PASS =", MOMENTUM_MAP_LEFT_HOM_PASS)
print("MOMENTUM_MAP_LEFT_ACTION_PASS =", MOMENTUM_MAP_LEFT_ACTION_PASS)
```
