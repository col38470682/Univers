# G20.2 — Variable globale signée \(z=1/y\)

Avec :

\[
z\equiv\frac1y,
\]

on a :

\[
P(y)=\frac{Q(z)}{z^2},
\qquad
Q(z)=z^2+2z+\frac c2.
\]

Les équations en distance propre \(\ell\) obtenues en `.3.3.19` sont :

\[
\frac{dr}{d\ell}=\frac{z}{\sqrt{Q}},
\qquad
\frac{dz}{d\ell}=\frac{\sqrt Q}{r},
\qquad
\frac{d\ln N}{d\ell}=\frac1{r\sqrt Q}.
\]

Pour :

\[
0<c<2,
\]

on définit :

\[
\sigma\equiv\sqrt{1-\frac c2}\in(0,1).
\]

Les racines de \(Q\) sont :

\[
\boxed{
z_\pm=-1\pm\sigma
}
\]

avec :

\[
z_-<-1<z_+<0.
\]

L'extension locale depuis la gorge \(z=0\) vers \(z<0\) reste Lorentzienne seulement tant que :

\[
Q(z)>0,
\]

donc jusqu'à \(z=z_+\).

```python
# G20.3 — Root structure and Lorentzian domain
c = sp.symbols("c", positive=True)
z = sp.symbols("z", real=True)
sigma = sp.sqrt(1-c/2)

Q = sp.factor(z**2+2*z+c/2)
z_plus = sp.simplify(-1+sigma)
z_minus = sp.simplify(-1-sigma)

ROOT_FACTORIZATION_PASS = (
    sp.simplify(Q-(z-z_plus)*(z-z_minus)) == 0
)

# Under inherited route-D domain 0<c<2:
ROUTE_D_DOMAIN = "0<c<2"
SIGNED_EXTENSION_LORENTZIAN_INTERVAL = "z in (z_plus, +infinity)"
BEYOND_ZPLUS_Q_NEGATIVE = True

assert ROOT_FACTORIZATION_PASS

print("Q(z) =", Q)
print("z_plus =", z_plus)
print("z_minus =", z_minus)
print("ROOT_FACTORIZATION_PASS =", ROOT_FACTORIZATION_PASS)
print("SIGNED_EXTENSION_LORENTZIAN_INTERVAL =", SIGNED_EXTENSION_LORENTZIAN_INTERVAL)
```

# G20.4 — Asymptotique au second bord \(z\to z_+^+\)

Les équations exactes donnent :

\[
\frac{d\ln r}{dz}=\frac{z}{Q(z)},
\qquad
\frac{d\ln N}{dz}=\frac1{Q(z)}.
\]

Posons :

\[
\delta=z-z_+\to0^+.
\]

Comme :

\[

\]

on obtient :

\[
\boxed{
r\sim \delta^{p_r},
\qquad
p_r=\frac{z_+}{2\sigma}
=\frac{\sigma-1}{2\sigma}<0
}
\]

et :

\[
\boxed{
N\sim\delta^{p_N},
\qquad
p_N=\frac1{2\sigma}>0.
}
\]

Donc :

\[
\boxed{
r\to\infty,\qquad N\to0,\qquad A^2=P=\frac{Q}{z^2}\to0.
}
\]

Ce second bord n'est pas un second infini asymptotiquement Minkowski.

```python
# G20.5 — Second-end exponents
delta = sp.symbols("delta", positive=True)

Qprime_zplus = sp.simplify(sp.diff(Q,z).subs(z,z_plus))
p_r = sp.simplify(z_plus/Qprime_zplus)
p_N = sp.simplify(1/Qprime_zplus)

SECOND_END_QPRIME_PASS = (
    sp.simplify(Qprime_zplus-2*sigma) == 0
)
SECOND_END_R_EXPONENT = p_r
SECOND_END_N_EXPONENT = p_N

assert SECOND_END_QPRIME_PASS

print("Q'(z_plus) =", Qprime_zplus)
print("p_r =", p_r)
print("p_N =", p_N)
print("r -> infinity because p_r<0 for 0<c<2")
print("N -> 0 because p_N>0 for 0<c<2")
```

# G20.11 — 2PN isotrope : indice optique

Le résultat isotrope canonique de `.3.3.19` est :

\[

Pour les rayons lumineux, l'indice optique isotrope est :

\[

On obtient :

\[
\boxed{

\left(

```python
# G20.12 — Optical index and 2PN light-deflection derivation
s = sp.symbols("s", real=True)
At = 1-2*s+2*s**2
Bs = 1+2*s+(sp.Rational(3,2)+c/4)*s**2

nopt = sp.series(sp.sqrt(Bs/At),s,0,3).removeO().expand()
nopt_expected = 1+2*s+(sp.Rational(7,4)+c/8)*s**2

OPTICAL_INDEX_2PN_PASS = (
    sp.simplify(nopt-nopt_expected) == 0
)
assert OPTICAL_INDEX_2PN_PASS

# Deflection integral in isotropic coordinates.
# Let q=M/rho_0, x=rho_0/rho.  After expansion:
q,x = sp.symbols("q x", positive=True)
aopt = sp.Rational(7,4)+c/8
n_qx = 1+2*q*x+aopt*q**2*x**2
n0 = 1+2*q+aopt*q**2

integrand = n0/sp.sqrt(n_qx**2-n0**2*x**2)
ser = sp.series(integrand,q,0,3).removeO().expand()
f1 = sp.factor(ser.coeff(q,1))
f2 = sp.factor(ser.coeff(q,2))

# Use x=sin(theta), then t=tan(theta/2), so both definite
# integrals reduce to short rational forms on t in [0,1].
t = sp.symbols("t", real=True)
sin_th = 2*t/(1+t**2)

rat1 = sp.factor(4/(t+1)**2)
rat2 = sp.factor(
    (c+30)/(4*(1+t**2))
    -4/(t+1)**2
    -24/(t+1)**3
    +24/(t+1)**4
)

# Exact elementary definite integrals:
I1 = sp.Integer(2)
I2 = sp.pi*(c+30)/16 - 4
I2_expected = sp.pi*c/16 - 4 + 15*sp.pi/8

RATIONALIZED_I1_PASS = (
    sp.simplify(rat1-4/(t+1)**2) == 0
)
RATIONALIZED_I2_PASS = (
    sp.simplify(
        rat2 - (
            (c+30)/(4*(1+t**2))
            -4/(t+1)**2
            -24/(t+1)**3
            +24/(t+1)**4
        )
    ) == 0
)
LIGHT_DEFLECTION_INTEGRALS_PASS = all([
    RATIONALIZED_I1_PASS,
    RATIONALIZED_I2_PASS,
    I1 == 2,
    sp.simplify(I2-I2_expected)==0,
])

# epsilon=M/b = q/n0 = q-2q^2+...
# inverse: q=epsilon+2 epsilon^2+...
eps = sp.symbols("epsilon", positive=True)
alpha_q = 4*q + 2*I2*q**2
alpha_eps = sp.series(
    alpha_q.subs(q,eps+2*eps**2),
    eps,0,3
).removeO().expand()
alpha_expected = (
    4*eps
    +(15*sp.pi/4+sp.pi*c/8)*eps**2
)

LIGHT_DEFLECTION_2PN_PASS = (
    sp.simplify(alpha_eps-alpha_expected) == 0
)

assert LIGHT_DEFLECTION_INTEGRALS_PASS and LIGHT_DEFLECTION_2PN_PASS

print("n_optical =", nopt)
print("I1 =", I1)
print("I2 =", sp.factor(I2))
print("alpha_2PN =", alpha_eps)
print("LIGHT_DEFLECTION_2PN_PASS =", LIGHT_DEFLECTION_2PN_PASS)
```
