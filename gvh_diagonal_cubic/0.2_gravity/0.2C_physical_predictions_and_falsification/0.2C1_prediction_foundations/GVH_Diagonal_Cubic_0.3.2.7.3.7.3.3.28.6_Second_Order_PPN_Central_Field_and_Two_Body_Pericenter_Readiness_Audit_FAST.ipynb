# Niveau 1 — Réduction radiale exacte de la branche statique alignée

Écrivons :

\[
ds^2
=
-N(r)^2dt^2
+
A(r)^2
\left[
dr^2+r^2d\Omega^2
\right].
\]

Pour :

\[
u^\mu=N^{-1}(1,0,0,0),
\]

la branche statique a :

\[
K_{ij}=0.
\]

Après retrait du terme de bord Einstein-Hilbert, l'action radiale vide se réduit, à un facteur global \(4\pi/(16\pi G_0)\) près, à :

\[
\boxed{
L_r
=
r^2
\left[
4N'A'
+
2N\frac{A'^2}{A}
+
c_{14}A\frac{N'^2}{N}
\right].
}
\]

Le troisième terme est le secteur accélération æther statique.

Cette expression doit reproduire le secteur faible de `.3.3.28.5` lorsqu'on pose :

\[
N=1+\Phi+\cdots,
\qquad
A=1-\Psi+\cdots.
\]

L'objectif est maintenant de ne plus tronquer au premier ordre, mais de résoudre les équations radiales jusqu'à \(O(U^2)\).

```python
r,c14 = sp.symbols("r c14", positive=True, real=True)
Nf = sp.Function("N")(r)
Af = sp.Function("A")(r)

Lrad = sp.simplify(
    r**2 * (
        4*sp.diff(Nf,r)*sp.diff(Af,r)
        + 2*Nf*sp.diff(Af,r)**2/Af
        + c14*Af*sp.diff(Nf,r)**2/Nf
    )
)

E_N = sp.simplify(
    sp.diff(sp.diff(Lrad,sp.diff(Nf,r)),r)
    - sp.diff(Lrad,Nf)
)
E_A = sp.simplify(
    sp.diff(sp.diff(Lrad,sp.diff(Af,r)),r)
    - sp.diff(Lrad,Af)
)

G286_EXACT_STATIC_RADIAL_REDUCED_ACTION_MATERIALIZED = True
G286_EXACT_STATIC_RADIAL_EULER_SYSTEM_MATERIALIZED = True

assert G286_EXACT_STATIC_RADIAL_REDUCED_ACTION_MATERIALIZED
assert G286_EXACT_STATIC_RADIAL_EULER_SYSTEM_MATERIALIZED

print("L_rad =")
sp.pprint(Lrad)
print("\nE_N =")
sp.pprint(E_N)
print("\nE_A =")
sp.pprint(E_A)
```
