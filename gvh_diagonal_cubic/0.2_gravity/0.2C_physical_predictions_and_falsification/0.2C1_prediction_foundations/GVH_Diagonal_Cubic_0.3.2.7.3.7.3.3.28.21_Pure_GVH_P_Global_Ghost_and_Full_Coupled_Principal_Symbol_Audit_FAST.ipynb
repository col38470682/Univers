# 1 — Coefficients spectraux du Hessien \(D\)

Sur une base propre locale

\[
A=\operatorname{diag}(a_0,a_1,a_2,a_3),\qquad \sum_Aa_A=0,\qquad q=\sum_Aa_A^2,
\]

les coefficients principaux déjà dérivés s'écrivent

\[
\boxed{K_{ij}^{\rm shape}=K_Sq-\frac{\kappa_{\rm eff}}2(a_i-a_j)^2}
\]

et

\[
\boxed{K_{0i}^{\rm tilt}=\frac{\kappa_{\rm eff}}2(a_0-a_i)^2-K_Sq}.
\]

La santé locale demande leur positivité.

```python
KS,kappa,q=sp.symbols('K_S kappa_eff q',positive=True,real=True)
a0,a1,a2,a3=sp.symbols('a0 a1 a2 a3',real=True)
Kshape12=sp.simplify(KS*q-kappa*(a1-a2)**2/2)
Ktilt01=sp.simplify(kappa*(a0-a1)**2/2-KS*q)
print('K_shape12 =',Kshape12); print('K_tilt01 =',Ktilt01)
```

\[
