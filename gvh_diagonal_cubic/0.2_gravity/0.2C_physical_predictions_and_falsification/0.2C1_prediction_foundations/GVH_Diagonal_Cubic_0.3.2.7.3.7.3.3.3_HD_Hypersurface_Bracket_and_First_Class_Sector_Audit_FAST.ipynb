# 2 — Générateur spatial canonique

On utilise le générateur spatial étendu :

\[
D_{\rm ext}[\xi]
=\int d^3x\,[
\pi^{ij}\mathcal L_\xi h_{ij}
+p_s\mathcal L_\xi s
+p_v^i\mathcal L_\xi v_i
+p_\lambda\mathcal L_\xi\lambda].
\]

Sa densité locale est :

\[
\boxed{\mathcal C_i^{\rm ext}
=-2D_j\pi^j{}_i+p_sD_is+p_v^jD_iv_j-D_j(p_v^jv_i)+p_\lambda D_i\lambda.}
\]

Le terme final assure l’action sur \(\lambda\) et s’annule faiblement puisque \(p_\lambda\approx0\).


```python
CANONICAL_PAIRS = (("h_ij","pi^ij"),("s","p_s"),("v_i","p_v^i"),("lambda","p_lambda"))
CI_UPSTREAM = ("-2 D_j pi^j_i","p_s D_i s","p_v^j D_i v_j","-D_j(p_v^j v_i)")
CI_EXTENDED = (*CI_UPSTREAM,"p_lambda D_i lambda")
SPATIAL_GENERATOR_RECONSTRUCTED = all([
    len(CANONICAL_PAIRS)==4,
    len(CI_EXTENDED)==5,
    CI_EXTENDED[-1]=="p_lambda D_i lambda",
    "p_lambda" in UPSTREAM["quartet"],
])
D_EXT_WEAKLY_EQ_UPSTREAM_D = SPATIAL_GENERATOR_RECONSTRUCTED and UPSTREAM["quartet_classified"]
assert SPATIAL_GENERATOR_RECONSTRUCTED
assert D_EXT_WEAKLY_EQ_UPSTREAM_D
print("SPATIAL_GENERATOR_RECONSTRUCTED =", SPATIAL_GENERATOR_RECONSTRUCTED)
print("D_EXT_WEAKLY_EQ_UPSTREAM_D =", D_EXT_WEAKLY_EQ_UPSTREAM_D)
```

# 3 — Convention et identité canonique HD

Convention :

\[
\boxed{\{D[\xi],F\}_{\rm can}=\mathcal L_\xi F.}
\]

Pour une densité hamiltonienne scalaire de poids +1 et un lapse scalaire, on doit avoir localement :

\[
N\mathcal L_\xi\mathcal H+\mathcal H\mathcal L_\xi N-D_i(N\xi^i\mathcal H)=0.
\]

Sous BOR, cela implique :

\[
\{H[N],D[\xi]\}_{\rm can}=H[\mathcal L_\xi N].
\]


```python
# Lightweight inherited covariance certificate
COVARIANCE_CERTIFICATE = {
    "Hamiltonian_density": "scalar_density_weight_1",
    "lapse": "scalar_weight_0",
    "Dg": "INHERITED_CLOSED",
    "Dq": "INHERITED_CLOSED",
    "DX": "INHERITED_CLOSED",
}
HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED = all([
    COVARIANCE_CERTIFICATE["Hamiltonian_density"]=="scalar_density_weight_1",
    COVARIANCE_CERTIFICATE["lapse"]=="scalar_weight_0",
    COVARIANCE_CERTIFICATE["Dg"]=="INHERITED_CLOSED",
    COVARIANCE_CERTIFICATE["Dq"]=="INHERITED_CLOSED",
    COVARIANCE_CERTIFICATE["DX"]=="INHERITED_CLOSED",
])
assert HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED
print("HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED =", HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED)
```

```python
Hdens,N = sp.symbols("Hdens N", commutative=True)
xi = sp.symbols("xi0:3", commutative=True)
dH = sp.symbols("dH0:3", commutative=True)
dN = sp.symbols("dN0:3", commutative=True)
dxi = sp.symbols("dxi00 dxi11 dxi22", commutative=True)
div_xi = sum(dxi)
Lie_H = sum(xi[i]*dH[i] for i in range(3)) + Hdens*div_xi
Lie_N = sum(xi[i]*dN[i] for i in range(3))
div_NxiH = sum(dN[i]*xi[i]*Hdens + N*dxi[i]*Hdens + N*xi[i]*dH[i] for i in range(3))
HD_LEIBNIZ_RESIDUAL = sp.expand(N*Lie_H + Hdens*Lie_N - div_NxiH)
HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED = (HD_LEIBNIZ_RESIDUAL == 0)
BOR_INHERITED = (UPSTREAM["boundary_domain"] == "compact_without_boundary OR sufficient_decay OR boundary_conditions_killing_flux")
HD_CANONICAL_RESIDUAL_BOR_ZERO = all([
    HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED,
    HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED,
    BOR_INHERITED,
])
HD_CANONICAL_CLASSIFICATION = "STRONG_ZERO_MOD_BOR" if HD_CANONICAL_RESIDUAL_BOR_ZERO else "UNCLASSIFIED"
assert HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED
assert HD_CANONICAL_RESIDUAL_BOR_ZERO
print("HD_LEIBNIZ_RESIDUAL =", HD_LEIBNIZ_RESIDUAL)
print("HD_CANONICAL_CLASSIFICATION =", HD_CANONICAL_CLASSIFICATION)
```

# 4 — Préservation du quartet par le générateur spatial

Pour un objet scalaire/densité scalaire de poids \(w_A\) :

\[
\mathcal L_\xi\Phi_A=\xi^iD_i\Phi_A+w_A\Phi_A D_i\xi^i.
\]

Cela reste dans l’idéal différentiel engendré par \(\Phi_A\) et \(D_i\Phi_A\). On teste ce support sans imposer les poids précis de \(\psi\) et \(\rho_{\rm FF}\).


```python
phi = sp.symbols("phi0:4", commutative=True)
w = sp.symbols("w0:4", commutative=True)
dphi = [[sp.Symbol(f"dphi{A}_{i}") for i in range(3)] for A in range(4)]
Lie_phi = [sp.expand(sum(xi[i]*dphi[A][i] for i in range(3)) + w[A]*phi[A]*div_xi) for A in range(4)]
zero_jet = {**{z:0 for z in phi}, **{dphi[A][i]:0 for A in range(4) for i in range(3)}}
D_PRESERVES_SECOND_CLASS_SURFACE = all(sp.expand(e.subs(zero_jet))==0 for e in Lie_phi)
assert D_PRESERVES_SECOND_CLASS_SURFACE
print("D_PRESERVES_SECOND_CLASS_SURFACE =", D_PRESERVES_SECOND_CLASS_SURFACE)
```
