# Niveau 1 — Énergie réduite du mode vectoriel

On reprend :

\[
c_{13}=c_1+c_3,
\qquad
c_{14}=c_1+c_4,
\qquad
c_-=c_1-c_3.
\]

Pour une orientation du trièdre définie par :

\[
\chi=\sin^2(2\phi),
\]

on pose :

\[
C_\sigma(\chi)
=
c_{13}-\frac{\zeta_C}{2}\chi.
\]

Après élimination du shift métrique, l'action réduite est :

\[
\mathcal L_{\rm red}
=
c_{14}W_t^2
-
\frac12\,G(\chi)\,W_x^2,
\]

avec :

\[
\boxed{
G(\chi)
=
\frac{
C_\sigma+c_--C_\sigma c_-
}{
1-C_\sigma
}.
}
\]

La densité d'énergie quadratique correspondante est :

\[
\boxed{
\mathcal H_{\rm red}
=
c_{14}W_t^2
+
\frac12\,G(\chi)\,W_x^2.
}
\]

L'orientation du trièdre n'affecte donc que le coefficient spatial.

```python
c1,c2,c3,c4,zeta,chi,phi = sp.symbols(
    "c1 c2 c3 c4 zeta_C chi phi", real=True
)

c13 = sp.factor(c1+c3)
c14 = sp.factor(c1+c4)
cminus = sp.factor(c1-c3)

Cs = sp.factor(c13-zeta*chi/2)
Gchi = sp.factor(
    (Cs+cminus-Cs*cminus)/(1-Cs)
)

Wt,Wx = sp.symbols("W_t W_x", real=True)

Lred = sp.factor(
    c14*Wt**2-Gchi*Wx**2/2
)
Hred = sp.factor(
    c14*Wt**2+Gchi*Wx**2/2
)

G2812_REDUCED_ENERGY_EXACT_MATERIALIZED = True

assert G2812_REDUCED_ENERGY_EXACT_MATERIALIZED

print("C_sigma(chi) =", Cs)
print("G(chi) =")
sp.pprint(Gchi)
print("H_red =")
sp.pprint(Hred)
```

\[

Hors surface singulière, le dénominateur est positif.

Donc :

\[
\boxed{
