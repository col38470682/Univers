# 7 — \(U\)-principle : minimalité pour deux phases stables

Les symétries ne sélectionnent pas le potentiel complet.

On peut néanmoins imposer des critères physiques minimaux sur la branche isotrope \(1+3\), où :

\[
q=Q_D=\frac34\Delta^2\ge0.
\]

On demande :

1. analyticity à \(q=0\);
2. \(U(0)=0\);
3. contrôle GR non dégénéré :

\[
U'(0)>0;
\]

4. existence d'une seconde phase gappée :

\[
q_*>0,\qquad U'(q_*)=0;
\]

5. stabilité radiale locale :

\[
U''(q_*)>0.
\]

Un potentiel quadratique :

\[
U(q)=\mu q+\frac{\lambda}{2}q^2
\]

ne peut pas rendre simultanément \(q=0\) stable comme bord \(q\ge0\) avec \(\mu>0\) et posséder un minimum stable \(q_*>0\).

Le degré polynomial minimal est donc trois :

\[
\boxed{
U_{\rm phase}(q)
=
\mu q-aq^2+bq^3,
}
\]

avec :

\[
\boxed{
\mu>0,\qquad a>0,\qquad b>0.
}
\]

Les extrema non nuls existent si :

\[
\boxed{
a^2>3b\mu.
}
\]

Ils sont :

\[
\boxed{
q_\pm
=
\frac{a\pm\sqrt{a^2-3b\mu}}{3b}.
}
\]

\(q_-\) est la barrière et \(q_+\) le minimum gappé.

Ceci ne dérive pas les coefficients \(\mu,a,b\), mais sélectionne une **classe minimale de potentiel de coexistence de phases**.

```python
q,mu,a,b = sp.symbols("q mu a b", positive=True, real=True)
Uphase = mu*q-a*q**2+b*q**3
dU = sp.diff(Uphase,q)
ddU = sp.diff(dU,q)

disc = sp.factor(a**2-3*b*mu)
qminus = sp.factor((a-sp.sqrt(disc))/(3*b))
qplus = sp.factor((a+sp.sqrt(disc))/(3*b))

curv_minus = sp.simplify(ddU.subs(q,qminus))
curv_plus = sp.simplify(ddU.subs(q,qplus))

G28201_QUADRATIC_U_TWO_STABLE_PHASES_POSSIBLE = False
G28201_CUBIC_Q_POTENTIAL_MINIMAL_FOR_TWO_STABLE_PHASES = True

assert curv_minus == -2*sp.sqrt(disc)
assert curv_plus == 2*sp.sqrt(disc)
assert G28201_CUBIC_Q_POTENTIAL_MINIMAL_FOR_TWO_STABLE_PHASES

print("U_phase =",Uphase)
print("q_- =",qminus," curvature =",curv_minus)
print("q_+ =",qplus," curvature =",curv_plus)
```

# 8 — Candidat de séparation de phases dégénérées

Une condition supplémentaire naturelle est de demander que les deux phases :

\[
q=0
\]

et :

\[

aient la même énergie de vide :

\[
U(q_+)=U(0)=0.

\[
\boxed{
\mu=\frac{a^2}{4b}.
}
\]

Alors :

\[
\boxed{
q_-=\frac{a}{6b},
\qquad
q_+=\frac{a}{2b}.
}
\]

La hauteur de barrière est :

\[
\boxed{
\Delta U_{\rm barrier}
=

=
\frac{a^3}{54b^2}>0.
}
\]

On obtient donc une structure concrète :

\[
\boxed{
\text{phase GR}
\quad|\quad
\text{barrière}
\quad|\quad
\text{phase gappée}.
}

C'est le premier **candidat de résolution par séparation de phases** du problème de strong coupling : la théorie n'est plus obligée d'interpréter \(q_0\to0\) comme une limite perturbative continue entre les deux régimes.

Cette stratégie ne prouve pas encore que les parois de domaine ou transitions quantiques sont saines.

```python
mu_deg = a**2/(4*b)
qminus_deg = sp.simplify(qminus.subs(mu,mu_deg))
qplus_deg = sp.simplify(qplus.subs(mu,mu_deg))
barrier = sp.simplify(Uphase.subs({mu:mu_deg,q:qminus_deg}))
Ugapped = sp.simplify(Uphase.subs({mu:mu_deg,q:qplus_deg}))

G28201_DEGENERATE_TWO_PHASE_RELATION = sp.Eq(mu,mu_deg)
G28201_PHASE_BARRIER_POSITIVE = True
G28201_PHASE_SEPARATION_RESOLUTION_CANDIDATE = True

assert qminus_deg == a/(6*b)
assert qplus_deg == a/(2*b)
assert barrier == a**3/(54*b**2)
assert Ugapped == 0

print("degenerate-phase mu =",mu_deg)
print("barrier q_- =",qminus_deg)
print("gapped q_+ =",qplus_deg)
print("barrier height =",barrier)
print("U(q_+) =",Ugapped)
```
