# Niveau 1 — Rang d'identifiabilité des couplages Einstein-æther

Les combinaisons déjà utilisées dans la chaîne sont :

\[
\boxed{
c_{13}=c_1+c_3,
\qquad
c_{14}=c_1+c_4,
\qquad
c_2=c_2.
}
\]

Définissons le vecteur de paramètres fondamentaux :

\[
\mathbf c=
(c_1,c_2,c_3,c_4)^T
\]

et la carte observable actuelle :

\[
\mathbf q=
(c_{13},c_2,c_{14})^T.
\]

Son Jacobien est :

\[
J
=
\frac{\partial\mathbf q}{\partial\mathbf c}.
\]

Si :

\[
\operatorname{rank}J<4,
\]

alors les quatre couplages fondamentaux ne sont pas identifiables séparément à partir du ledger actuel.

```python
c1,c2,c3,c4 = sp.symbols("c1 c2 c3 c4", real=True)
c13,c14 = sp.symbols("c13 c14", real=True)

cvec = sp.Matrix([c1,c2,c3,c4])
qvec = sp.Matrix([c1+c3, c2, c1+c4])
Jq = qvec.jacobian(cvec)

rank_q = Jq.rank()
null_q = Jq.nullspace()

G288_CURRENT_COUPLING_MAP_RANK = rank_q
G288_CURRENT_COUPLING_NULLITY = 4-rank_q
G288_COUPLING_IDENTIFIABILITY_FULL = (rank_q == 4)

assert rank_q == 3
assert len(null_q) == 1
assert not G288_COUPLING_IDENTIFIABILITY_FULL

print("J =")
sp.pprint(Jq)
print("rank(J) =", rank_q)
print("nullity =", G288_CURRENT_COUPLING_NULLITY)
print("null direction =")
sp.pprint(null_q[0])
print("G288_COUPLING_IDENTIFIABILITY_FULL =",
      G288_COUPLING_IDENTIFIABILITY_FULL)
```

## 1.1 Direction non identifiable

Le noyau du Jacobien donne une direction libre :

\[
\boxed{
\delta\mathbf c
\propto
(1,0,-1,-1).
}
\]

Ainsi, sous :

\[
c_1\mapsto c_1+\lambda,
\]

\[
c_3\mapsto c_3-\lambda,
\qquad
c_4\mapsto c_4-\lambda,
\]

les combinaisons :

\[
c_{13},\quad c_{14},\quad c_2
\]

restent inchangées.

Donc :

\[
\boxed{
(c_{13},c_{14},c_2)
\not\Rightarrow
(c_1,c_2,c_3,c_4)_{\rm unique}.
}
\]

La question suivante est de savoir si \(\alpha_1,\alpha_2\) sont invariants le long de cette direction libre. S'ils varient avec \(\lambda\), le secteur preferred-frame n'est pas identifiable.

```python
den1 = 2*c1-c1**2+c3**2
alpha1 = sp.simplify(
    -8*(c3**2+c1*c4)/den1
)
alpha2 = sp.simplify(
    alpha1/2
    - (c1+2*c3-c4)*(2*c1+3*c2+c3+c4)
      /((c1+c2+c3)*(2-c1-c4))
)

# Rewrite using c13, c14 and the unresolved c1.
subs_combo = {
    c3:c13-c1,
    c4:c14-c1,
}
a1_combo = sp.factor(sp.simplify(alpha1.subs(subs_combo)))
a2_combo = sp.factor(sp.simplify(alpha2.subs(subs_combo)))

da1_dc1 = sp.factor(sp.simplify(sp.diff(a1_combo,c1)))
da2_dc1 = sp.factor(sp.simplify(sp.diff(a2_combo,c1)))

G288_ALPHA1_VARIES_ALONG_UNIDENTIFIED_DIRECTION = (da1_dc1 != 0)
G288_ALPHA2_VARIES_ALONG_UNIDENTIFIED_DIRECTION = (da2_dc1 != 0)

G288_PREFERRED_FRAME_COUPLING_IDENTIFIABILITY_CLOSED = not (
    G288_ALPHA1_VARIES_ALONG_UNIDENTIFIED_DIRECTION
    or G288_ALPHA2_VARIES_ALONG_UNIDENTIFIED_DIRECTION
)

assert G288_ALPHA1_VARIES_ALONG_UNIDENTIFIED_DIRECTION
assert not G288_PREFERRED_FRAME_COUPLING_IDENTIFIABILITY_CLOSED

print("alpha1(c13,c14,c1) =")
sp.pprint(a1_combo)
print("\nd alpha1 / d c1 =")
sp.pprint(da1_dc1)
print("\nd alpha2 / d c1 =")
sp.pprint(da2_dc1)
print("G288_PREFERRED_FRAME_COUPLING_IDENTIFIABILITY_CLOSED =",
      G288_PREFERRED_FRAME_COUPLING_IDENTIFIABILITY_CLOSED)
```
