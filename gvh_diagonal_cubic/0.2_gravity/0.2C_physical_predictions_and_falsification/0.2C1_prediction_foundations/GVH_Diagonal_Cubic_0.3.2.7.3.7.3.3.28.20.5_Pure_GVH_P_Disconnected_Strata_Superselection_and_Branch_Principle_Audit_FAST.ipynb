# 1 — Vacua individuels admissibles

On conserve le potentiel dégénéré minimal
\[
U(q)=\mu q-aq^2+bq^3,\qquad \mu=\frac{a^2}{4b},\qquad q_+=\frac{a}{2b},
\]
soit
\[
\boxed{U(q)=bq(q_+-q)^2}.
\]
Le point GR \(A=0\) reste stationnaire parce que \(q=O(A^2)\) et \(\partial F/\partial A|_{A=0}=0\). Le vacuum gappé vérifie
\[
\boxed{U'(q_+)=0,\qquad U''(q_+)=a>0}.
\]
Les deux strates survivent donc individuellement dans leurs scopes déjà audités.

```python
q,a,b=sp.symbols("q a b",positive=True,real=True)
mu=a**2/(4*b); qp=a/(2*b); qm=a/(6*b)
U=sp.factor(mu*q-a*q**2+b*q**3); Up=sp.diff(U,q); Upp=sp.diff(U,q,2)
G28205_GAPPED_STATIONARY=sp.simplify(Up.subs(q,qp))==0
G28205_GAPPED_LOCAL_MINIMUM=sp.simplify(Upp.subs(q,qp))==a
G28205_GR_FUNDAMENTAL_A_STATIONARITY_RETAINED=True
assert sp.factor(U-b*q*(qp-q)**2)==0 and G28205_GAPPED_STATIONARY and G28205_GAPPED_LOCAL_MINIMUM
print("U(q) =",U); print("U'(q+) =",sp.simplify(Up.subs(q,qp))); print("U''(q+) =",sp.simplify(Upp.subs(q,qp)))
```

# 2 — Pas de séparation topologique

Prenons le représentant \(1+3\)
\[
A_+=\operatorname{diag}\!\left(\frac{3\Delta_+}{4},-\frac{\Delta_+}{4},-\frac{\Delta_+}{4},-\frac{\Delta_+}{4}\right),\qquad \operatorname{Tr}(A_+^2)=q_+.
\]
La famille
\[
\boxed{A(\lambda)=\lambda A_+,\quad 0\le\lambda\le1}
\]
relie continûment \(A=0\) à \(A=A_+\), avec
\[
\boxed{q(\lambda)=\lambda^2q_+}.
\]
Aucune charge topologique discrète n'est donc matérialisée par cette structure.

```python
lam=sp.symbols("lambda",nonnegative=True,real=True); Delta=sp.sqrt(4*qp/3)
Aplus=sp.diag(3*Delta/4,-Delta/4,-Delta/4,-Delta/4); Alam=lam*Aplus
qplus_check=sp.simplify(sp.trace(Aplus*Aplus)); qlam=sp.simplify(sp.trace(Alam*Alam))
G28205_CONFIGURATION_CONTINUOUS_PATH_EXISTS=(sp.simplify(qplus_check-qp)==0 and sp.simplify(qlam-lam**2*qp)==0)
G28205_TOPOLOGICAL_SUPERSELECTION_DERIVED=False
assert G28205_CONFIGURATION_CONTINUOUS_PATH_EXISTS and not G28205_TOPOLOGICAL_SUPERSELECTION_DERIVED
print("Tr(A_plus^2) =",qplus_check); print("q(lambda) =",qlam)
```
