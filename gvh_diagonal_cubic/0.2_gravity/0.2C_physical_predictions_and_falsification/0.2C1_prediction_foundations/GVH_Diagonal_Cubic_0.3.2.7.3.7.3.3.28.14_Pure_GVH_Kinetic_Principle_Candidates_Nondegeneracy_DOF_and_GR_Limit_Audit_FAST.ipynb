
# Niveau 2 — C2 : tenseur spatial STF

Un tenseur symétrique spatial \(3\times3\) possède 6 composantes. La trace nulle en retire une :

\[
\boxed{N_{\rm comp}^{\rm STF}=5}.
\]

Dans une base STF orthonormée,

\[
\bar D_{ij}=\sum_{A=1}^{5}q_AE^{(A)}_{ij},
\qquad
\mathrm{Tr}(E^{(A)}E^{(B)})=\delta^{AB},
\]

donc

\[
\bar D_{ij}\bar D^{ij}=\sum_Aq_A^2.
\]

La cinétique réduite est alors

\[
\boxed{
\mathcal L_D^{(2)}
=
\frac{K_D}{2}\sum_A\dot q_A^2
-\frac{G_D}{2}\sum_A(\nabla q_A)^2
-\frac{m_D^2}{2}\sum_Aq_A^2.
}
\]


```python

s2=sp.sqrt(2); s6=sp.sqrt(6)
E=[
 sp.Matrix([[1,0,0],[0,-1,0],[0,0,0]])/s2,
 sp.Matrix([[1,0,0],[0,1,0],[0,0,-2]])/s6,
 sp.Matrix([[0,1,0],[1,0,0],[0,0,0]])/s2,
 sp.Matrix([[0,0,1],[0,0,0],[1,0,0]])/s2,
 sp.Matrix([[0,0,0],[0,0,1],[0,1,0]])/s2,
]
Gram=sp.Matrix([[sp.simplify(sp.trace(a.T*b)) for b in E] for a in E])
q=sp.symbols("q1:6", real=True)
Dbar=sp.zeros(3)
for i in range(5): Dbar += q[i]*E[i]
norm=sp.simplify(sp.trace(Dbar.T*Dbar))
G2814_STF_BASIS_PASS=(Gram==sp.eye(5))
G2814_STF_TRACE_PASS=(sp.trace(Dbar)==0)
G2814_STF_NORM_PASS=(sp.simplify(norm-sum(x*x for x in q))==0)
assert G2814_STF_BASIS_PASS and G2814_STF_TRACE_PASS and G2814_STF_NORM_PASS
print("STF basis Gram rank =",Gram.rank())
print("trace(Dbar) =",sp.trace(Dbar))
print("Dbar norm =",norm)
```
