# Niveau 1 — GVH

## 1.1 Nondimensionnalisation minimale

Nous introduisons deux échelles fondamentales indépendantes :

\[
\boxed{L_* > 0}
\qquad\text{et}\qquad
\boxed{T_* > 0}.
\]

Les variables sans dimension de `.3.3.27.3` sont notées avec un chapeau :

\[
\hat t,\qquad
\hat x,\qquad
\hat R_c,\qquad
\hat \omega,\qquad
\hat u.
\]

Le pont SI est défini par :

\[
t_{\rm SI}=T_*\hat t,
\]

\[
x_{\rm SI}=L_*\hat x,
\]

\[
R_{c,\rm SI}=L_*\hat R_c,
\]

\[
\omega_{\rm SI}=\frac{\hat\omega}{T_*},
\]

\[
u_{\rm SI}=\frac{L_*}{T_*}\hat u,
\]

\[
v_{\rm SI}=\frac{L_*}{T_*}\hat v,
\]

\[
a_{\rm SI}=\frac{L_*}{T_*^2}\hat a.
\]

Le pont est mathématiquement suffisant pour les variables cinématiques
présentes, mais **il ne détermine pas \(L_*\) ni \(T_*\)**.


```python
# Symbolic dimensional bridge

Lstar, Tstar = sp.symbols("L_star T_star", positive=True, finite=True)
that = sp.symbols("t_hat", real=True)
Rhat, what, uhat = sp.symbols("R_hat omega_hat u_hat", real=True)

tSI = Tstar*that
RSI = Lstar*Rhat
omegaSI = what/Tstar
uSI = Lstar*uhat/Tstar
vperpSI = sp.simplify(RSI*omegaSI)
aperpSI = sp.simplify(RSI*omegaSI**2)

# Exact inverse normalization
Rhat_back = sp.simplify(RSI/Lstar)
what_back = sp.simplify(omegaSI*Tstar)
uhat_back = sp.simplify(uSI*Tstar/Lstar)

G274_DIMENSIONAL_MAP_MATERIALIZED = all([
    sp.simplify(Rhat_back-Rhat) == 0,
    sp.simplify(what_back-what) == 0,
    sp.simplify(uhat_back-uhat) == 0,
])

# Two free scales remain
G274_FREE_SCALE_COUNT = 2
G274_SI_SCALE_THEORETICALLY_DERIVED = False

assert G274_DIMENSIONAL_MAP_MATERIALIZED
assert G274_FREE_SCALE_COUNT == 2
assert not G274_SI_SCALE_THEORETICALLY_DERIVED

print("R_SI =", RSI)
print("omega_SI =", omegaSI)
print("u_SI =", uSI)
print("v_perp_SI =", vperpSI)
print("a_perp_SI =", aperpSI)
print("G274_DIMENSIONAL_MAP_MATERIALIZED =", G274_DIMENSIONAL_MAP_MATERIALIZED)
print("G274_FREE_SCALE_COUNT =", G274_FREE_SCALE_COUNT)
```

## 1.2 Quantité indépendante des échelles

Toutes les quantités physiques n'exigent pas nécessairement \(L_*\) et
\(T_*\) séparément.

Pour l'hélice :

\[
\chi_{\rm helix}
=
\frac{u_{\rm SI}}
{R_{c,\rm SI}\omega_{\rm SI}}
\]

donne :

\[
\boxed{
\chi_{\rm helix}
=
\frac{\hat u}
{\hat R_c\hat\omega}
}
\]

et dépend donc seulement des paramètres sans dimension.

De même, le pas par tour normalisé :

\[
\frac{p}{R_c}
=
\frac{2\pi u}
{R_c\omega}
=
2\pi\chi_{\rm helix}.
\]

Cela fournit un **observable scale-free candidat**.

Mais dans `.3.3.27.3`, \(\hat u,\hat R_c,\hat\omega\) sont encore
prescrits. Cette quantité est donc calculable, mais pas encore prédite
par les EOM GVH.


```python
chi_helix = sp.simplify(uSI/(RSI*omegaSI))
chi_hat = sp.simplify(uhat/(Rhat*what))
pitch_over_radius = sp.simplify(2*sp.pi*chi_helix)

G274_SCALE_FREE_HELIX_RATIO_IDENTITY = (
    sp.simplify(chi_helix-chi_hat) == 0
)
G274_SCALE_FREE_OBSERVABLE_MATERIALIZED = (
    G274_SCALE_FREE_HELIX_RATIO_IDENTITY
)
G274_SCALE_FREE_OBSERVABLE_EOM_DERIVED = False

# Canonical .3.3.27.3 values
Rc_hat_num = 7.0
omega_total_hat_num = 0.45 + 0.20
u2_hat_num = 0.35

chi_num = u2_hat_num/(Rc_hat_num*omega_total_hat_num)
pitch_ratio_num = 2*np.pi*chi_num

assert G274_SCALE_FREE_OBSERVABLE_MATERIALIZED
assert not G274_SCALE_FREE_OBSERVABLE_EOM_DERIVED

print("chi_helix symbolic =", chi_helix)
print("canonical chi_helix =", chi_num)
print("canonical pitch/R =", pitch_ratio_num)
print(
    "G274_SCALE_FREE_OBSERVABLE_MATERIALIZED =",
    G274_SCALE_FREE_OBSERVABLE_MATERIALIZED
)
print(
    "G274_SCALE_FREE_OBSERVABLE_EOM_DERIVED =",
    G274_SCALE_FREE_OBSERVABLE_EOM_DERIVED
)
```

## 2.1 Règle standard de nondimensionnalisation

En physique mathématique, une variable dimensionnée est reconstruite à
partir d'une variable sans dimension et d'une échelle :

\[
q_{\rm SI}=Q_*\,\hat q.
\]

Une telle opération est une **reconstruction dimensionnelle**.

Elle ne devient prédictive que si \(Q_*\) est :

1. dérivé de la théorie, ou
2. calibré indépendamment d'un observable ensuite testé.

### Classification à conserver

\[
\boxed{
\texttt{THEORY\_DERIVED}
}
\]

si l'échelle provient d'une EOM / action / constante dérivée ;

\[
\boxed{
\texttt{CALIBRATED}
}
\]

si elle provient d'une donnée mesurée ;

\[
\boxed{
\texttt{HELD\_OUT\_PREDICTION}
}
\]

uniquement pour une quantité qui n'a pas servi à la calibration.

Cette séparation est la protection principale contre la circularité.

## 2.2 Exemple contrôlé avec constantes terrestres

Nous utilisons les constantes déjà présentes dans le benchmark
`.3.3.26` uniquement comme **références établies** :

\[
R_\oplus=6\,378\,137\ {\rm m},
\]

\[
\Omega_\oplus=7.2921159\times10^{-5}\ {\rm s}^{-1},
\]

\[
g=9.80665\ {\rm m\,s^{-2}}.
\]

Si nous imposons :

\[
L_*=\frac{R_\oplus}{\hat R_c},
\]

et :

\[
T_*=\frac{\hat\omega_{\rm total}}{\Omega_\oplus},
\]

alors :

\[
R_{c,\rm SI}=R_\oplus
\]

et :

\[
\omega_{\rm SI}=\Omega_\oplus
\]

**par construction**.

Ces deux égalités ne sont donc pas des prédictions GVH.

En revanche, une troisième quantité non utilisée dans la calibration
peut servir de témoin indépendant.


```python
R_earth = 6_378_137.0          # m
Omega_earth = 7.2921159e-5     # s^-1
g_standard = 9.80665           # m/s^2

Lstar_earth_cal = R_earth/Rc_hat_num
Tstar_earth_cal = omega_total_hat_num/Omega_earth

Rc_SI_cal = Lstar_earth_cal*Rc_hat_num
omega_SI_cal = omega_total_hat_num/Tstar_earth_cal
u2_SI_cal = (Lstar_earth_cal/Tstar_earth_cal)*u2_hat_num
a_perp_cal = Rc_SI_cal*omega_SI_cal**2

G274_EARTH_RADIUS_REPRODUCED_BY_CALIBRATION = np.isclose(
    Rc_SI_cal, R_earth, rtol=0, atol=1e-9
)
G274_EARTH_OMEGA_REPRODUCED_BY_CALIBRATION = np.isclose(
    omega_SI_cal, Omega_earth, rtol=0, atol=1e-15
)

# This is a held-out comparison, not a claimed target of .3.3.27.2.
heldout_ratio_to_g = a_perp_cal/g_standard
heldout_relative_gap_to_g = abs(a_perp_cal-g_standard)/g_standard

G274_DOUBLE_CALIBRATION_IDENTIFIED = all([
    G274_EARTH_RADIUS_REPRODUCED_BY_CALIBRATION,
    G274_EARTH_OMEGA_REPRODUCED_BY_CALIBRATION,
])

assert G274_DOUBLE_CALIBRATION_IDENTIFIED

print("L_star from Earth-radius calibration =", Lstar_earth_cal, "m")
print("T_star from Earth-rotation calibration =", Tstar_earth_cal, "s")
print("R_c_SI (calibrated) =", Rc_SI_cal, "m")
print("omega_SI (calibrated) =", omega_SI_cal, "s^-1")
print("u2_SI (derived from both calibrations) =", u2_SI_cal, "m/s")
print("held-out centripetal acceleration =", a_perp_cal, "m/s^2")
print("held-out a_perp / g =", heldout_ratio_to_g)
print("held-out relative gap to g =", heldout_relative_gap_to_g)
print("G274_DOUBLE_CALIBRATION_IDENTIFIED =", G274_DOUBLE_CALIBRATION_IDENTIFIED)
```

# Niveau 3 — Test numérique minimal

## 3.1 Covariance du pont sous plusieurs choix d'échelles

Nous testons plusieurs couples arbitraires :

\[
(L_*,T_*).
\]

Pour chacun, nous convertissons la même hélice sans dimension en SI,
puis nous renormalisons vers les variables sans dimension.

Le pont est correct si la géométrie normalisée est récupérée à précision
numérique.

Ce test vérifie la **cohérence dimensionnelle du pont**, pas la physique
de la valeur choisie pour \(L_*,T_*\).


```python
def bridge_test(Ls, Ts, N=1201, tmax_hat=10.0):
    th = np.linspace(0.0, tmax_hat, N)
    Rc_hat = Rc_hat_num
    omega_hat = omega_total_hat_num
    u_hat = u2_hat_num

    xhat = np.column_stack([
        Rc_hat*np.cos(omega_hat*th),
        Rc_hat*np.sin(omega_hat*th),
        u_hat*th,
    ])

    t_si = Ts*th
    x_si = Ls*xhat

    # Recover dimensionless representation
    th_back = t_si/Ts
    xhat_back = x_si/Ls

    radius_hat_back = np.sqrt(
        xhat_back[:,0]**2 + xhat_back[:,1]**2
    )
    phi_back = np.unwrap(
        np.arctan2(xhat_back[:,1], xhat_back[:,0])
    )

    omega_fit_hat = np.polyfit(th_back, phi_back, 1)[0]
    u_fit_hat = np.polyfit(th_back, xhat_back[:,2], 1)[0]

    return {
        "L_star_m": Ls,
        "T_star_s": Ts,
        "max_state_recovery_error":
            float(np.max(np.abs(xhat_back-xhat))),
        "max_radius_hat_error":
            float(np.max(np.abs(radius_hat_back-Rc_hat))),
        "omega_hat_recovery_error":
            float(abs(omega_fit_hat-omega_hat)),
        "u_hat_recovery_error":
            float(abs(u_fit_hat-u_hat)),
    }

scale_sets = [
    (1.0, 1.0),
    (100.0, 10.0),
    (1e-3, 0.1),
    (1e6, 1e3),
    (Lstar_earth_cal, Tstar_earth_cal),
]

rows = []
for Ls,Ts in scale_sets:
    r = bridge_test(Ls,Ts)
    r["pass"] = max(
        r["max_state_recovery_error"],
        r["max_radius_hat_error"],
        r["omega_hat_recovery_error"],
        r["u_hat_recovery_error"],
    ) < 1e-10
    rows.append(r)

scale_sweep = pd.DataFrame(rows)
G274_NUMERIC_SCALE_COVARIANCE_PASS = bool(scale_sweep["pass"].all())
G274_SCALE_SET_COUNT = len(scale_sweep)

assert G274_NUMERIC_SCALE_COVARIANCE_PASS

print(scale_sweep.to_string(index=False))
print(
    "G274_NUMERIC_SCALE_COVARIANCE_PASS =",
    G274_NUMERIC_SCALE_COVARIANCE_PASS
)
print("G274_SCALE_SET_COUNT =", G274_SCALE_SET_COUNT)
```
