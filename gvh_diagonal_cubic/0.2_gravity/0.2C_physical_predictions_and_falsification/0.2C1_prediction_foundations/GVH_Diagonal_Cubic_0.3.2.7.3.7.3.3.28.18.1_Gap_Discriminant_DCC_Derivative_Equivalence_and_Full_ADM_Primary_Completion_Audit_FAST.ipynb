## Route R

\[

\[

## Route P

\[

\[

```python
r,rdot,eta = sp.symbols("r rdot eta_Q", positive=True, real=True)
Q = r**2
Qdot = sp.diff(Q,r)*rdot
LR_time = sp.factor(eta*Qdot**2/2)
KR = sp.factor(sp.diff(LR_time,rdot,2))

q0,U1,U2,dr = sp.symbols("Q0 U1 U2 dr", positive=True, real=True)
# radial second variation of U(r^2) at r0=sqrt(q0), with U'(q0)=0
r0 = sp.sqrt(q0)
U_radial_second_at_stationary = sp.factor(4*q0*U2)

G28181_ROUTE_R_RADIAL_HESSIAN_EXACT = (
    sp.simplify(KR-4*eta*r**2)==0
)
G28181_ROUTE_R_RADIAL_KINETIC_VANISHES_AT_GR = (
    sp.limit(KR,r,0,dir="+")==0
)
G28181_ROUTE_P_GAPPED_STATIONARITY_REQUIRES_UPRIME_ZERO = True
G28181_ROUTE_P_RADIAL_CURVATURE_AT_GAPPED_STATIONARY = (
    U_radial_second_at_stationary
)

assert G28181_ROUTE_R_RADIAL_HESSIAN_EXACT
assert G28181_ROUTE_R_RADIAL_KINETIC_VANISHES_AT_GR

print("L_R,time =",LR_time)
print("K_r^(R) =",KR)
print("Route P radial curvature at U'(Q0)=0 =",U_radial_second_at_stationary)
```

\[

\begin{pmatrix}
b&0\\
0&-b

\[

```python
phi,a,b = sp.symbols("phi a b", real=True, nonzero=True)

R = sp.Matrix([
    [sp.cos(phi),-sp.sin(phi)],
    [sp.sin(phi), sp.cos(phi)],
])
A0 = sp.diag(a,-a)
B0 = sp.diag(b,-b)
Aphi = sp.simplify(R*A0*R.T)

commAB = sp.simplify(Aphi*B0-B0*Aphi)
tr_comm2 = sp.trigsimp(sp.factor(sp.trace(commAB*commAB)))
QA = sp.factor(sp.trace(Aphi*Aphi))
QB = sp.factor(sp.trace(B0*B0))

chi_D = sp.trigsimp(sp.factor(-tr_comm2/(2*QA*QB)))
chi_target = sp.sin(2*phi)**2

G28181_DCC_ORIENTATION_FUNCTIONAL_MATCH_EXACT = (
    sp.trigsimp(chi_D-chi_target)==0
)

assert G28181_DCC_ORIENTATION_FUNCTIONAL_MATCH_EXACT

print("Tr([A,B]^2) =")
sp.pprint(tr_comm2)
print("Q_A =",QA)
print("Q_B =",QB)
print("chi_D =")
sp.pprint(chi_D)
```
