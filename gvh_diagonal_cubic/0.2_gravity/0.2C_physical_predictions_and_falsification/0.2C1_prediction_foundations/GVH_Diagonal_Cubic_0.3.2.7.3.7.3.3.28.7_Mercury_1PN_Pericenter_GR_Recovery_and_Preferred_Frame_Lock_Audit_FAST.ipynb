# Niveau 2 — Preferred-frame : formules exactes de référence

Dans Einstein-æther, les PPN qui diffèrent génériquement de GR sont les paramètres preferred-frame :

\[
\boxed{
\alpha_1
=
-\frac{
8(c_3^2+c_1c_4)
}{
2c_1-c_1^2+c_3^2
}.
}
\]

et :

\[
\boxed{
\alpha_2
=
\frac{\alpha_1}{2}
-
\frac{
(c_1+2c_3-c_4)
(2c_1+3c_2+c_3+c_4)
}{
(c_1+c_2+c_3)(2-c_1-c_4)
}.
}
\]

Référence de formule : littérature Einstein-æther / revue de Jacobson.

Les dénominateurs imposent notamment, sur la branche générique utilisée par ces expressions :

\[
2c_1-c_1^2+c_3^2\neq0,
\]

\[
c_{123}:=c_1+c_2+c_3\neq0,
\]

\[
c_{14}:=c_1+c_4\neq2.
\]

Le but de `.28.7` n'est pas d'ajuster \(\alpha_1,\alpha_2\) aux observations.

Il doit d'abord répondre à la question plus fondamentale :

\[
\boxed{
\text{les paramètres GVH actuellement dérivés suffisent-ils à calculer }
\alpha_1,\alpha_2
\text{ de façon unique ?}
}
\]

```python
c1,c2,c3,c4 = sp.symbols("c1 c2 c3 c4", real=True)
c13,c14 = sp.symbols("c13 c14", real=True)

den1 = 2*c1-c1**2+c3**2
c123 = c1+c2+c3

alpha1 = sp.simplify(
    -8*(c3**2+c1*c4)/den1
)
alpha2 = sp.simplify(
    alpha1/2
    - (c1+2*c3-c4)*(2*c1+3*c2+c3+c4)
      /(c123*(2-c1-c4))
)

# Rewrite in combinations already common in the GVH/EA chain:
subs_combo = {
    c3:c13-c1,
    c4:c14-c1,
}
alpha1_combo = sp.factor(sp.simplify(alpha1.subs(subs_combo)))
alpha2_combo = sp.factor(sp.simplify(alpha2.subs(subs_combo)))

alpha1_combo_expected = sp.factor(
    -8*(c13**2+c1*(c14-2*c13))
    /(c13**2+2*c1*(1-c13))
)
alpha2_combo_expected = sp.factor(
    alpha1_combo_expected/2
    - (2*c13-c14)*(c13+c14+3*c2)
      /((c13+c2)*(2-c14))
)

G287_ALPHA1_COMBINATION_REWRITE_EXACT_PASS = (
    sp.simplify(alpha1_combo-alpha1_combo_expected) == 0
)
G287_ALPHA2_COMBINATION_REWRITE_EXACT_PASS = (
    sp.simplify(alpha2_combo-alpha2_combo_expected) == 0
)

# alpha1 still depends generically on c1 even if c13,c14 are known.
alpha1_c1_dependence = sp.simplify(sp.diff(alpha1_combo,c1))
G287_ALPHA1_REQUIRES_ADDITIONAL_COUPLING_INFORMATION = (
    alpha1_c1_dependence != 0
)

assert G287_ALPHA1_COMBINATION_REWRITE_EXACT_PASS
assert G287_ALPHA2_COMBINATION_REWRITE_EXACT_PASS
assert G287_ALPHA1_REQUIRES_ADDITIONAL_COUPLING_INFORMATION

print("alpha1 =")
sp.pprint(alpha1)
print("\nalpha2 =")
sp.pprint(alpha2)
print("\nalpha1(c13,c14,c1) =")
sp.pprint(alpha1_combo)
print("\nalpha2(c13,c14,c2,c1) =")
sp.pprint(alpha2_combo)
print("\nd(alpha1)/dc1 =")
sp.pprint(alpha1_c1_dependence)
print("G287_ALPHA1_REQUIRES_ADDITIONAL_COUPLING_INFORMATION =",
      G287_ALPHA1_REQUIRES_ADDITIONAL_COUPLING_INFORMATION)
```
