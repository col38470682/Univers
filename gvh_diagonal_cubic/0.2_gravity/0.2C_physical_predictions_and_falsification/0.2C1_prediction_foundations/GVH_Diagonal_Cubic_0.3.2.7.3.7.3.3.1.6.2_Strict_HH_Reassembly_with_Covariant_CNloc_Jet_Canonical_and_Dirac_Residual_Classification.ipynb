
# 2 — Reconstruction exacte du secteur local \(Q,J_0,U_0\)

On reconstruit le même secteur déjà utilisé dans `.1.6` et `.1.6.1` :

\[
A=-S-v^ia_i,
\]

\[
B_i=s\,a_i+W_i-K_i{}^jv_j,
\]

\[
C_i=-D_is-K_i{}^jv_j,
\]

\[
D_{ij}=D_iv_j+sK_{ij}.
\]

Les invariants directionnels sont combinés avec le bloc cinétique EH :

\[
K_{ij}K^{ij}-K^2.
\]

Aucun terme physique nouveau n'est ajouté.


```python

c1,c2,c3,c4,s = sp.symbols(
    "c1 c2 c3 c4 s",
    real=True
)

v = sp.Matrix(
    sp.symbols("v1:4", real=True)
)

avec = sp.Matrix(
    sp.symbols("a1:4", real=True)
)

g = sp.Matrix(
    sp.symbols("g1:4", real=True)
)  # g_i = D_i s

qsyms = sp.symbols(
    "q11 q12 q13 q21 q22 q23 q31 q32 q33",
    real=True
)
q = sp.Matrix(3,3,qsyms)  # q_ij = D_i v_j

K11,K22,K33,K12,K13,K23,S,W1,W2,W3 = sp.symbols(
    "K11 K22 K33 K12 K13 K23 S W1 W2 W3",
    real=True
)

vel = sp.Matrix([
    K11,K22,K33,K12,K13,K23,
    S,W1,W2,W3,
])

K = sp.Matrix([
    [K11,K12,K13],
    [K12,K22,K23],
    [K13,K23,K33],
])

W = sp.Matrix([W1,W2,W3])

A = sp.expand(
    -S-v.dot(avec)
)

Bvec = sp.expand(
    s*avec+W-K*v
)

Cvec = sp.expand(
    -g-K*v
)

Dmat = sp.expand(
    q+s*K
)

I1 = sp.expand(
    A**2
    -Bvec.dot(Bvec)
    -Cvec.dot(Cvec)
    +sum(
        Dmat[i,j]**2
        for i in range(3)
        for j in range(3)
    )
)

theta = sp.expand(
    -A+sp.trace(Dmat)
)

I3 = sp.expand(
    A**2
    -2*Bvec.dot(Cvec)
    +sum(
        Dmat[i,j]*Dmat[j,i]
        for i in range(3)
        for j in range(3)
    )
)

alpha = sp.expand(
    s*A+v.dot(Cvec)
)

beta_vec = sp.expand(
    s*Bvec+Dmat.T*v
)

acc2 = sp.expand(
    -alpha**2+beta_vec.dot(beta_vec)
)

Lu = sp.expand(
    -c1*I1
    -c2*theta**2
    -c3*I3
    +c4*acc2
)

LEH = sp.expand(
    sum(
        K[i,j]**2
        for i in range(3)
        for j in range(3)
    )
    -sp.trace(K)**2
)

Ltot = sp.expand(
    Lu+LEH
)

zero_vel = {
    x:0
    for x in vel
}

zero_a = {
    x:0
    for x in avec
}

Q = sp.hessian(
    Ltot,
    list(vel)
)

J = sp.Matrix([
    sp.diff(Ltot,x).subs(zero_vel)
    for x in vel
])

J0 = sp.Matrix([
    sp.expand(x)
    for x in J.subs(zero_a)
])

U0 = sp.expand(
    Ltot
    .subs(zero_vel)
    .subs(zero_a)
)

assert Q == Q.T
assert all(not Q.has(a) for a in avec)

QJU_RECONSTRUCTED = True

print("Q shape =",Q.shape)
print("J0 shape =",J0.shape)
print("QJU_RECONSTRUCTED =",QJU_RECONSTRUCTED)
```


# 3 — Variable de Legendre compacte et dérivées locales

On garde :

\[
\boxed{
QX=P-J_0
}
\]

sans développer \(Q^{-1}\).

Pour :

\[
C_N^{\rm loc}
=
-\frac12(P-J_0)^TX+U_0,
\]

l'identité de `.1.6` donne, pour toute variable locale \(y\) :

\[
\boxed{
C_{,y}
=
J_{0,y}^{T}X
+
\frac12X^TQ_{,y}X
+
U_{0,y}.
}
\]

On construit en particulier :

\[
C_{g_i}
=
\frac{\partial C_N^{\rm loc}}{\partial(D_is)},
\]

\[
\mathcal A^{ij}
=
\frac{\partial C_N^{\rm loc}}{\partial(D_iv_j)}.
\]


```python

P = sp.Matrix(
    sp.symbols("P0:10", real=True)
)

X = sp.Matrix(
    sp.symbols("X0:10", real=True)
)

def Cpartial(y):
    return sp.expand(
        (J0.diff(y).T*X)[0]
        +
        sp.Rational(1,2)*
        (X.T*Q.diff(y)*X)[0]
        +
        sp.diff(U0,y)
    )

Cg = sp.Matrix([
    Cpartial(g[i])
    for i in range(3)
])

Aq = sp.Matrix(
    3,3,
    [
        Cpartial(qij)
        for qij in list(q)
    ]
)

Cs = Cpartial(s)

Cv = sp.Matrix([
    Cpartial(v[j])
    for j in range(3)
])

assert Aq.shape == (3,3)

LOCAL_FUNCTIONAL_PARTIALS_EXPLICIT = True

print(
    "C_g lengths =",
    [len(str(x)) for x in Cg]
)

print(
    "A_q lengths =",
    [
        len(str(Aq[i,j]))
        for i in range(3)
        for j in range(3)
    ]
)

print(
    "LOCAL_FUNCTIONAL_PARTIALS_EXPLICIT =",
    LOCAL_FUNCTIONAL_PARTIALS_EXPLICIT
)
```


# 4 — Opérateur de dérivée spatiale formelle

Pour ne pas perdre les termes provenant des dérivées des coefficients,
on introduit explicitement les jets :

\[
D_i g_k,\qquad
D_i q_{kj},\qquad
D_iX_A,\qquad
D_iP_A.
\]

Puis on définit un opérateur de chaîne exact \(D_i\).

Les constantes \(c_1,\ldots,c_4\) sont ici des couplages constants.


```python

Dg = [
    [
        sp.Symbol(f"D{r}g{k}", real=True)
        for k in range(3)
    ]
    for r in range(3)
]

Dq = [
    [
        [
            sp.Symbol(
                f"D{r}q{k}{j}",
                real=True
            )
            for j in range(3)
        ]
        for k in range(3)
    ]
    for r in range(3)
]

DX = [
    [
        sp.Symbol(
            f"D{r}X{A}",
            real=True
        )
        for A in range(10)
    ]
    for r in range(3)
]

DP = [
    [
        sp.Symbol(
            f"D{r}P{A}",
            real=True
        )
        for A in range(10)
    ]
    for r in range(3)
]

def Dop(expr,r):
    out = sp.diff(expr,s)*g[r]

    for j in range(3):
        out += (
            sp.diff(expr,v[j])
            *q[r,j]
        )

    for k in range(3):
        out += (
            sp.diff(expr,g[k])
            *Dg[r][k]
        )

    for k in range(3):
        for j in range(3):
            out += (
                sp.diff(expr,q[k,j])
                *Dq[r][k][j]
            )

    for A in range(10):
        out += (
            sp.diff(expr,X[A])
            *DX[r][A]
        )

        out += (
            sp.diff(expr,P[A])
            *DP[r][A]
        )

    return sp.expand(out)

SPATIAL_CHAIN_RULE_OPERATOR_READY = True

print(
    "SPATIAL_CHAIN_RULE_OPERATOR_READY =",
    SPATIAL_CHAIN_RULE_OPERATOR_READY
)
```


# 7 — Dérivées fonctionnelles utilisées dans le crochet

À \(h_{ij}=\delta_{ij}\), \(\sqrt h=1\), la dérivée impulsionnelle du

\frac{\delta H[N]}{\delta\Pi_A}

\[

\[
\frac{\delta H[N]}{\delta p_v^j}

Les dérivées configurationnelles incluent :

- les termes de connexion de `.1.6.1` ;

- les termes d'Euler de \(C_N^{\rm loc}\).

La courbure \(R^{(3)}\) est traitée séparément dans la section suivante.


```python

N0,M0 = sp.symbols(
    "N M",
    real=True
)

Ng = sp.Matrix(
    sp.symbols("N1:4", real=True)
)

Mg = sp.Matrix(
    sp.symbols("M1:4", real=True)
)

def metric_qgrad(smgrad,A):
    m,n = metric_pairs[A]
    w = metric_weights[A]

    connection = sum(
        kappa[i][m][n]*smgrad[i]
        for i in range(3)
    )

    divb = (
        -sp.Rational(1,2)
        *P[6]
        *(
            v[m]*smgrad[n]
            +
            v[n]*smgrad[m]
        )
    )

    return sp.expand(
        w*(connection+divb)
    )

def metric_p(sm,A):
    return sp.expand(
        -2*sm*X[A]
    )

def scalar_q(sm,smgrad):
    return sp.expand(
        sm*Es
        -
        sum(
            smgrad[i]*Cg[i]
            for i in range(3)
        )
        +
        sum(
            P[7+i]*smgrad[i]
            for i in range(3)
        )
    )

def scalar_p(sm,smgrad):
    return sp.expand(
        -sm*X[6]
        +
        sum(
            v[i]*smgrad[i]
            for i in range(3)
        )
    )

def vector_q(j,sm,smgrad):
    return sp.expand(
        sm*Ev[j]
        -
        sum(
            smgrad[i]*Aq[i,j]
            for i in range(3)
        )
        +
        P[6]*smgrad[j]
    )

def vector_p(j,sm,smgrad):
    return sp.expand(
        -sm*X[7+j]
        +
        s*smgrad[j]
    )

FUNCTIONAL_DERIVATIVE_BLOCKS_ASSEMBLED = True

print(
    "FUNCTIONAL_DERIVATIVE_BLOCKS_ASSEMBLED =",
    FUNCTIONAL_DERIVATIVE_BLOCKS_ASSEMBLED
)
```


# 8 — Crochet canonique non-courbure

On calcule directement :

\[
\{H[N],H[M]\}_{\rm can}
=
\sum_A
\left(
\frac{\delta H[N]}{\delta q_A}
\frac{\delta H[M]}{\delta p_A}
-
(N\leftrightarrow M)
\right)
\]

sur les six coordonnées métriques indépendantes, puis sur :

\[
(s,p_s),
\qquad
(v_j,p_v^j).
\]

À ce stade, aucun \(D[\beta]\) n'est soustrait.


```python

Bnon = sp.Integer(0)

for A in range(6):
    qN = metric_qgrad(Ng,A)
    pN = metric_p(N0,A)

    qM = metric_qgrad(Mg,A)
    pM = metric_p(M0,A)

    Bnon += (
        qN*pM
        -
        pN*qM
    )

qN = scalar_q(N0,Ng)
pN = scalar_p(N0,Ng)

qM = scalar_q(M0,Mg)
pM = scalar_p(M0,Mg)

Bnon += (
    qN*pM
    -
    pN*qM
)

for j in range(3):
    qN = vector_q(j,N0,Ng)
    pN = vector_p(j,N0,Ng)

    qM = vector_q(j,M0,Mg)
    pM = vector_p(j,M0,Mg)

    Bnon += (
        qN*pM
        -
        pN*qM
    )

Bnon = sp.expand(Bnon)

# No N*M ultralocal term may survive antisymmetrization.
assert sp.diff(
    sp.diff(Bnon,N0),
    M0
) == 0

NONCURVATURE_HH_BRACKET_MATERIALIZED = True

print(
    "NONCURVATURE_HH_BRACKET_MATERIALIZED =",
    NONCURVATURE_HH_BRACKET_MATERIALIZED
)

print(
    "non-curvature bracket expression length =",
    len(str(Bnon))
)
```


# 9 — Décomposition en \(W_i\) et wedge \(D_iN\,D_jM\)

Posons :

\[
W_i
=
ND_iM-MD_iN.
\]

Le crochet non-courbure peut contenir :

\[
F_i^{\rm direct}W_i
\]

et des termes :

\[
C_{ij}D_iN D_jM.
\]

L'antisymétrie du crochet impose :

\[
C_{ij}=-C_{ji}.
\]

Pour une matrice antisymétrique \(C_{ij}\), l'intégration par parties donne,
modulo terme de bord :

\[
\boxed{
C_{ij}D_iN D_jM
\equiv
-\frac12(D_iC_{ij})W_j.
}
\]

Le facteur \(1/2\) vient de la moyenne des deux intégrations par parties
reliées par l'antisymétrie \(N\leftrightarrow M\).


```python

Fdirect = []

for i in range(3):
    coeff = sp.expand(
        Bnon
        .coeff(N0,1)
        .coeff(Mg[i],1)
    )

    reverse = sp.expand(
        Bnon
        .coeff(M0,1)
        .coeff(Ng[i],1)
    )

    assert sp.expand(
        reverse+coeff
    ) == 0

    Fdirect.append(coeff)

Cmat = sp.zeros(3,3)

for i in range(3):
    for j in range(3):
        Cmat[i,j] = sp.expand(
            Bnon
            .coeff(Ng[i],1)
            .coeff(Mg[j],1)
        )

assert sp.simplify(
    Cmat+Cmat.T
) == sp.zeros(3)

Fwedge = [
    sp.expand(
        -sp.Rational(1,2)
        *
        sum(
            Dop(Cmat[i,j],i)
            for i in range(3)
        )
    )
    for j in range(3)
]

WEDGE_ANTISYMMETRY_EXACT = True
WEDGE_IBP_REDUCED = True

print(
    "WEDGE_ANTISYMMETRY_EXACT =",
    WEDGE_ANTISYMMETRY_EXACT
)

print(
    "nonzero off-diagonal wedge entries =",
    sum(
        1
        for i in range(3)
        for j in range(i+1,3)
        if Cmat[i,j] != 0
    )
)

print(
    "WEDGE_IBP_REDUCED =",
    WEDGE_IBP_REDUCED
)
```


# 10 — Contribution de courbure \(+{}^{(3)}R\)

Le bloc fermé dans `.1.3` est :

\[
\frac{\delta}{\delta h_{mn}}
\int d^3x\,
N\sqrt h\,{}^{(3)}R

\left[

+

\right].
\]

La partie \(-NG^{mn}\) est ultralocale et s'annule dans le crochet HH
par antisymétrie.

Définissons le tenseur symétrique associé aux six composantes cinétiques :

\[
X^{mn}
=
\begin{pmatrix}
X_0&X_3&X_4\\
X_3&X_1&X_5\\
X_4&X_5&X_2
\end{pmatrix},

\[
X_{\rm tr}=X_0+X_1+X_2.

Après intégration par parties, le secteur de courbure contribue au
coefficient de \(W_n\) par :

\[
\boxed{
F_n^{(R)}
=
-2D_m
\left(
X^{mn}-h^{mn}X_{\rm tr}
\right).
}
\]


```python

Xmetric = sp.Matrix([
    [X[0],X[3],X[4]],
    [X[3],X[1],X[5]],
    [X[4],X[5],X[2]],
])

Xtrace = (
    X[0]+X[1]+X[2]
)

Fcurv = []

for n in range(3):
    value = sp.Integer(0)

    for m in range(3):
        Ymn = (
            Xmetric[m,n]
            -
            (1 if m == n else 0)
            *Xtrace
        )

        value += Dop(Ymn,m)

    Fcurv.append(
        sp.expand(-2*value)
    )

CURVATURE_HESSIAN_IBP_REDUCED = True

print(
    "CURVATURE_HESSIAN_IBP_REDUCED =",
    CURVATURE_HESSIAN_IBP_REDUCED
)
```


# 12 — Cible indépendante : contrainte de momentum

La contrainte de momentum full-field amont est :

\[
\boxed{
\mathcal C_i
=
-2D_j\pi^j{}_i
+
p_sD_is
+
p_v^jD_iv_j
-
D_j(p_v^jv_i).
}
\]

En coordonnées normales :

\[
p_s=P_6,
\qquad
p_v^j=P_{7+j},
\]

et :

\[
\pi^{11}=\frac{P_0}{2},
\quad
\pi^{22}=\frac{P_1}{2},
\quad
\pi^{33}=\frac{P_2}{2},
\]

\[
\pi^{12}=\frac{P_3}{4},
\quad
\pi^{13}=\frac{P_4}{4},
\quad
\pi^{23}=\frac{P_5}{4}.
\]

Cette cible est construite **indépendamment** de \(F_i^{HH}\).


```python

piP = sp.Matrix([
    [P[0]/2, P[3]/4, P[4]/4],
    [P[3]/4, P[1]/2, P[5]/4],
    [P[4]/4, P[5]/4, P[2]/2],
])

Ci = []

for i in range(3):
    value = (
        -2
        *
        sum(
            Dop(piP[j,i],j)
            for j in range(3)
        )
    )

    value += (
        P[6]*g[i]
    )

    value += sum(
        P[7+j]*q[i,j]
        for j in range(3)
    )

    value -= sum(
        DP[j][7+j]*v[i]
        +
        P[7+j]*q[j,i]
        for j in range(3)
    )

    Ci.append(
        sp.expand(value)
    )

EXACT_MOMENTUM_CONSTRAINT_TARGET_MATERIALIZED = True

print(
    "C_i component lengths =",
    [len(str(x)) for x in Ci]
)

print(
    "EXACT_MOMENTUM_CONSTRAINT_TARGET_MATERIALIZED =",
    EXACT_MOMENTUM_CONSTRAINT_TARGET_MATERIALIZED
)
```


# 14 — Réduction exacte par la relation de Legendre

On ne réduit pas le résidu par une identité ADM supposée.

On substitue seulement :

\[
P_A=(QX+J_0)_A,
\]

puis :

\[
D_iP_A
=
D_i(QX+J_0)_A.
\]

Si les trois composantes deviennent exactement nulles dans SymPy,
le résultat est une fermeture algébrique réelle du secteur assemblé.


```python

Prel = sp.Matrix([
    sp.expand(x)
    for x in (Q*X+J0)
])

DPrel = [
    [
        Dop(Prel[A],r)
        for A in range(10)
    ]
    for r in range(3)
]

sub_P = {
    P[A]:Prel[A]
    for A in range(10)
}

sub_DP = {
    DP[r][A]:DPrel[r][A]
    for r in range(3)
    for A in range(10)
}

Rreduced = []

for i in range(3):
    rr = sp.expand(
        Rraw[i]
        .subs(sub_DP)
        .subs(sub_P)
    )

    Rreduced.append(rr)

for i,rr in enumerate(Rreduced):
    print(
        f"R_HH_can coefficient {i} =",
        rr
    )

assert all(
    rr == 0
    for rr in Rreduced
)

LEGENDRE_AND_DERIVATIVE_RELATION_APPLIED = True
CANONICAL_RHH_STRONG_ZERO = True

print(
    "LEGENDRE_AND_DERIVATIVE_RELATION_APPLIED =",
    LEGENDRE_AND_DERIVATIVE_RELATION_APPLIED
)

print(
    "CANONICAL_RHH_STRONG_ZERO =",
    CANONICAL_RHH_STRONG_ZERO
)
```


# 16 — Contrôle du multiplicateur de norme

Le secteur auxiliaire amont utilise aussi le terme local :

\[
-\lambda_{\rm mult}\chi,
\qquad
\chi=-s^2+v^iv_i+1.
\]

Ce terme est ultralocal.

Ses dérivées configurationnelles pertinentes sont :

\[
\frac{\partial(-\lambda\chi)}{\partial s}
=
2\lambda s,
\]

\[
\frac{\partial(-\lambda\chi)}{\partial v_j}
=
-2\lambda v^j.
\]

Les seuls morceaux à gradient de smearing dans les dérivées
impulsionnelles déjà établies sont :

\[
v^iD_iN,
\qquad
sD_jN.
\]

La contribution croisée HH est donc proportionnelle à :

\[
2\lambda s\,v^i
-
2\lambda v^i s
=
0.
\]

Les morceaux ultralocaux s'annulent par \(NM-MN\).

Ainsi, le multiplicateur de norme ne change pas le résidu canonique
calculé ci-dessus.


```python

lam = sp.Symbol(
    "lambda_mult",
    real=True
)

norm_cross_gradient = sp.Matrix([
    sp.expand(
        2*lam*s*v[i]
        -
        2*lam*v[i]*s
    )
    for i in range(3)
])

assert norm_cross_gradient == sp.zeros(3,1)

NORM_MULTIPLIER_HH_CROSS_CANCELS = True

print(
    "NORM_MULTIPLIER_HH_CROSS_CANCELS =",
    NORM_MULTIPLIER_HH_CROSS_CANCELS
)
```


# 18 — Reconstruction du noyau auxiliaire local de `.1.5`

Avant toute tentative Dirac, on reconstruit seulement ce qui est réellement
établi en amont :

\[
\Phi_A
=
(p_\lambda,\chi,\psi,\rho)
\]

sur la branche locale générique :

\[
c_{14}c_{\rm time}\Delta\neq0.
\]

Le noyau local vérifie :

\[
\det C_4=\Delta^4,
\qquad
\operatorname{rank}C_4=4.
\]

Cette reconstruction ne constitue **pas encore** une preuve que ces
quatre fonctions sont le noyau fonctionnel full-field complet du
Hamiltonien \(Q,J,U\) utilisé ci-dessus.


```python

c14,ct = sp.symbols(
    "c14 c_time",
    nonzero=True,
    real=True
)

sa,v1a,v2a,v3a = sp.symbols(
    "s_aux v1_aux v2_aux v3_aux",
    real=True
)

psa,p1a,p2a,p3a = sp.symbols(
    "ps_aux p1_aux p2_aux p3_aux",
    real=True
)

lama,plam = sp.symbols(
    "lambda_aux p_lambda",
    real=True
)

vnorm2 = (
    v1a**2+v2a**2+v3a**2
)

pnorm2 = (
    p1a**2+p2a**2+p3a**2
)

vp = (
    v1a*p1a
    +v2a*p2a
    +v3a*p3a
)

chi = sp.expand(
    -sa**2+vnorm2+1
)

psi = sp.factor(
    sa*psa/ct
    +
    vp/c14
)

Delta = sp.factor(
    2*(
        vnorm2/c14
        -
        sa**2/ct
    )
)

Akin = sp.factor(
    -psa**2/(2*ct**2)
    +
    pnorm2/(2*c14**2)
)

rho = sp.factor(
    Akin+lama*Delta
)

xaux,yaux = sp.symbols(
    "x_aux y_aux"
)

C4 = sp.Matrix([
    [0,0,0,-Delta],
    [0,0,Delta,xaux],
    [0,-Delta,0,yaux],
    [Delta,-xaux,-yaux,0],
])

C4inv = sp.Matrix([
    [0,yaux/Delta**2,-xaux/Delta**2,1/Delta],
    [-yaux/Delta**2,0,-1/Delta,0],
    [xaux/Delta**2,1/Delta,0,0],
    [-1/Delta,0,0,0],
])

assert sp.factor(
    C4.det()-Delta**4
) == 0

assert sp.simplify(
    C4*C4inv-sp.eye(4)
) == sp.zeros(4)

LOCAL_AUXILIARY_C4_READY = True
LOCAL_AUXILIARY_DIRAC_INVERSE_READY = True

print(
    "LOCAL_AUXILIARY_C4_READY =",
    LOCAL_AUXILIARY_C4_READY
)

print(
    "det(C4)=Delta^4: PASS"
)

print(
    "LOCAL_AUXILIARY_DIRAC_INVERSE_READY =",
    LOCAL_AUXILIARY_DIRAC_INVERSE_READY
)
```
