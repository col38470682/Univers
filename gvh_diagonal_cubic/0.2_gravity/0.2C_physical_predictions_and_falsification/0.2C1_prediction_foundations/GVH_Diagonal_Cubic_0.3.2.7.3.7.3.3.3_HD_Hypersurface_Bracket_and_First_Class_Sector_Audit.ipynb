
# 1 — Correction du test de Jacobi de `.1.6.2.1`

Dans `.1.6.2.1`, le coefficient de gradient du lapse avait été inséré
dans l'ansatz avant le contrôle Jacobi.

Ici on repart d'un ansatz réellement libre :

\[
\{\psi,H_0[N]\}
=
N\mathcal A_{\rm FF}
+
\alpha^iD_iN
+
\beta^{ij}D_iD_jN.
\]

Avec :

\[
\{H[N],H[M]\}_{\rm can}
=
D[\omega],
\qquad
\omega^i=ND^iM-MD^iN,
\]

et :

\[
\{\chi,H[N]\}=N\psi,
\]

l'identité de Jacobi doit déterminer indépendamment les structures
\(\alpha^i\) et \(\beta^{ij}\).


```python

N,M = sp.symbols("N M", real=True)

N1 = sp.Matrix(sp.symbols("N1:4", real=True))
M1 = sp.Matrix(sp.symbols("M1:4", real=True))

# six Hessiennes indépendantes : 11,22,33,12,13,23
N2s = sp.symbols("N11 N22 N33 N12 N13 N23", real=True)
M2s = sp.symbols("M11 M22 M33 M12 M13 M23", real=True)

alpha = sp.Matrix(sp.symbols("alpha1:4", real=True))
beta = sp.Matrix([
    [sp.Symbol("beta11"), sp.Symbol("beta12"), sp.Symbol("beta13")],
    [sp.Symbol("beta12"), sp.Symbol("beta22"), sp.Symbol("beta23")],
    [sp.Symbol("beta13"), sp.Symbol("beta23"), sp.Symbol("beta33")],
])

dchi = sp.Matrix(sp.symbols("dchi1:4", real=True))

N2 = sp.Matrix([
    [N2s[0],N2s[3],N2s[4]],
    [N2s[3],N2s[1],N2s[5]],
    [N2s[4],N2s[5],N2s[2]],
])

M2 = sp.Matrix([
    [M2s[0],M2s[3],M2s[4]],
    [M2s[3],M2s[1],M2s[5]],
    [M2s[4],M2s[5],M2s[2]],
])

Afree = sp.Symbol("A_FF_free", real=True)

psi_HN_ansatz = (
    N*Afree
    + alpha.dot(N1)
    + sum(beta[i,j]*N2[i,j] for i in range(3) for j in range(3))
)

psi_HM_ansatz = (
    M*Afree
    + alpha.dot(M1)
    + sum(beta[i,j]*M2[i,j] for i in range(3) for j in range(3))
)

omega_dot_dchi = sum(
    (N*M1[i]-M*N1[i])*dchi[i]
    for i in range(3)
)

jacobi = sp.expand(
    omega_dot_dchi
    + M*psi_HN_ansatz
    - N*psi_HM_ansatz
)

# Coefficients des Hessiennes arbitraires : ils imposent beta=0.
beta_constraints = []
for z in list(N2s)+list(M2s):
    beta_constraints.append(
        sp.expand(sp.diff(jacobi,z))
    )

# Coefficients des gradients : ils imposent alpha=dchi.
alpha_constraints = []
for i in range(3):
    alpha_constraints.append(
        sp.expand(
            jacobi.coeff(N*M1[i])
        )
    )

# Solution symbolique directe attendue.
jacobi_solution = {
    alpha[i]:dchi[i]
    for i in range(3)
}
for i in range(3):
    for j in range(i,3):
        jacobi_solution[beta[i,j]] = 0

jacobi_reduced = sp.expand(
    jacobi.subs(jacobi_solution)
)

assert jacobi_reduced == 0

JACOBI_INDEPENDENT_ANSATZ_USED = True
JACOBI_HESSIAN_COEFFICIENT_FORCED_ZERO = True
JACOBI_GRADIENT_COEFFICIENT_FORCED_DCHI = True

print("JACOBI_INDEPENDENT_ANSATZ_USED =",JACOBI_INDEPENDENT_ANSATZ_USED)
print("JACOBI_HESSIAN_COEFFICIENT_FORCED_ZERO =",JACOBI_HESSIAN_COEFFICIENT_FORCED_ZERO)
print("JACOBI_GRADIENT_COEFFICIENT_FORCED_DCHI =",JACOBI_GRADIENT_COEFFICIENT_FORCED_DCHI)
```


Le résultat de ce contrôle est désormais réellement indépendant :

\[
\boxed{\beta^{ij}=0}
\]

et :

\[
\boxed{\alpha^i=D^i\chi}.
\]

Donc :

\[
\boxed{
\{\psi,H_0[N]\}
=
N\mathcal A_{\rm FF}
+
D^i\chi\,D_iN
}
\]

n'est plus seulement un ansatz compatible avec Jacobi.

Il reste cependant à calculer explicitement le coefficient local :

\[
\boxed{
\mathcal A_{\rm FF}
=
\{\psi,H_0[1]\}.
}
\]



# 2 — Reconstruction du même \(Q,J_0,U_0\)

On conserve exactement les conventions de `.1.6.2` et `.1.6.2.1`.

\[
V^A=
(K_{11},K_{22},K_{33},K_{12},K_{13},K_{23},S,W_1,W_2,W_3).
\]

Au point en coordonnées normales :

\[
h_{ij}=\delta_{ij},
\qquad
\Gamma^k{}_{ij}=0,
\]

mais les variations de connexion ne sont pas supprimées.


```python

c1,c2,c3,c4,s = sp.symbols("c1 c2 c3 c4 s", real=True)
v = sp.Matrix(sp.symbols("v1:4", real=True))
a = sp.Matrix(sp.symbols("a1:4", real=True))
g = sp.Matrix(sp.symbols("g1:4", real=True))

qsyms = sp.symbols(
    "q11 q12 q13 q21 q22 q23 q31 q32 q33",
    real=True
)
q = sp.Matrix(3,3,qsyms)

K11,K22,K33,K12,K13,K23,S,W1,W2,W3 = sp.symbols(
    "K11 K22 K33 K12 K13 K23 S W1 W2 W3",
    real=True
)

vel = sp.Matrix([
    K11,K22,K33,K12,K13,K23,S,W1,W2,W3
])

K = sp.Matrix([
    [K11,K12,K13],
    [K12,K22,K23],
    [K13,K23,K33],
])

W = sp.Matrix([W1,W2,W3])

A = -S-v.dot(a)
B = s*a+W-K*v
C = -g-K*v
D = q+s*K

I1 = (
    A**2-B.dot(B)-C.dot(C)
    +sum(D[i,j]**2 for i in range(3) for j in range(3))
)
theta = -A+sp.trace(D)
I3 = (
    A**2-2*B.dot(C)
    +sum(D[i,j]*D[j,i] for i in range(3) for j in range(3))
)
alpha0 = s*A+v.dot(C)
beta0 = s*B+D.T*v
acc2 = -alpha0**2+beta0.dot(beta0)

Lu = -c1*I1-c2*theta**2-c3*I3+c4*acc2
LEH = (
    sum(K[i,j]**2 for i in range(3) for j in range(3))
    -sp.trace(K)**2
)

Ltot = sp.expand(Lu+LEH)

zero_vel = {x:0 for x in vel}
zero_a = {x:0 for x in a}

Q = sp.hessian(Ltot,list(vel))
J0 = sp.Matrix([
    sp.diff(Ltot,x).subs(zero_vel)
    for x in vel
]).subs(zero_a)
U0 = sp.expand(
    Ltot.subs(zero_vel).subs(zero_a)
)

assert Q == Q.T
assert Q.shape == (10,10)

QJU_SAME_FULL_FIELD_RECONSTRUCTED = True
print("QJU_SAME_FULL_FIELD_RECONSTRUCTED =",QJU_SAME_FULL_FIELD_RECONSTRUCTED)
```


# 3 — Variables de Legendre \(X\) et duale \(Y\)

On n'expanse pas \(Q^{-1}\).

\[
\boxed{
QX=P-J_0
}
\]

et :

\[
\boxed{
QY=r
}
\]

avec :

\[
r=
(-2v_1^2,-2v_2^2,-2v_3^2,
-4v_1v_2,-4v_1v_3,-4v_2v_3,
-2s,2v_1,2v_2,2v_3)^T.
\]

Alors :

\[
\boxed{
\psi_{\rm FF}=-r^TX
}
\]

et :

\[
\boxed{
\Delta_{\rm FF}=-r^TY.
}
\]


```python

P = sp.Matrix(sp.symbols("P0:10", real=True))
X = sp.Matrix(sp.symbols("X0:10", real=True))
Y = sp.Matrix(sp.symbols("Y0:10", real=True))

r = sp.Matrix([
    -2*v[0]**2,
    -2*v[1]**2,
    -2*v[2]**2,
    -4*v[0]*v[1],
    -4*v[0]*v[2],
    -4*v[1]*v[2],
    -2*s,
    2*v[0],
    2*v[1],
    2*v[2],
])

psi_FF = sp.expand(-(r.T*X)[0])
Delta_FF = sp.expand(-(r.T*Y)[0])

LEGENDRE_RELATION = Q*X-(P-J0)
DUAL_RELATION = Q*Y-r

print("psi_FF length =",len(str(psi_FF)))
print("Delta_FF length =",len(str(Delta_FF)))
```


# 4 — Jet spatial formel

Le coefficient \(\mathcal A_{\rm FF}\) contient les dérivées spatiales
des coefficients des variations fonctionnelles.

On introduit donc :

\[
D_iP_A,\quad
D_iX_A,\quad
D_iY_A,\quad
D_ig_j,\quad
D_iq_{jk}.
\]

Les relations dérivées de Legendre sont conservées sous forme de gates :

\[
D_i(QX-P+J_0)=0,
\qquad
D_i(QY-r)=0.
\]


```python

Dg = [[sp.Symbol(f"D{i}g{j}") for j in range(3)] for i in range(3)]
Dq = [[[sp.Symbol(f"D{i}q{j}{k}") for k in range(3)] for j in range(3)] for i in range(3)]
DP = [[sp.Symbol(f"D{i}P{A}") for A in range(10)] for i in range(3)]
DX = [[sp.Symbol(f"D{i}X{A}") for A in range(10)] for i in range(3)]
DY = [[sp.Symbol(f"D{i}Y{A}") for A in range(10)] for i in range(3)]

def Dop(expr,i):
    out = sp.diff(expr,s)*g[i]
    for j in range(3):
        out += sp.diff(expr,v[j])*q[i,j]
    for j in range(3):
        out += sp.diff(expr,g[j])*Dg[i][j]
    for j in range(3):
        for k in range(3):
            out += sp.diff(expr,q[j,k])*Dq[i][j][k]
    for A in range(10):
        out += sp.diff(expr,P[A])*DP[i][A]
        out += sp.diff(expr,X[A])*DX[i][A]
        out += sp.diff(expr,Y[A])*DY[i][A]
    return sp.expand(out)

SPATIAL_JET_OPERATOR_READY = True
print("SPATIAL_JET_OPERATOR_READY =",SPATIAL_JET_OPERATOR_READY)
```


# 5 — Dérivées compactes de \(C_N^{\rm loc}\) et de \(\psi_{\rm FF}\)

Pour toute variable locale \(z\) à \(P\) fixé :

\[
C_{,z}
=
J_{0,z}^TX
+
\frac12X^TQ_{,z}X
+
U_{0,z}.
\]

Pour \(\psi=-r^TX\) :

\[
\boxed{
\psi_{,z}
=
-r_{,z}^TX
+
Y^TJ_{0,z}
+
Y^TQ_{,z}X.
}
\]

Cette seconde formule vient de l'implicite :

\[
Q\,\delta X
=
\delta P-\delta J_0-\delta Q\,X.
\]

Aucun \(Q^{-1}\) symbolique géant n'est nécessaire.


```python

def Cpartial(z):
    return sp.expand(
        (J0.diff(z).T*X)[0]
        +sp.Rational(1,2)*(X.T*Q.diff(z)*X)[0]
        +sp.diff(U0,z)
    )

def Psipartial(z):
    return sp.expand(
        -(r.diff(z).T*X)[0]
        +(Y.T*J0.diff(z))[0]
        +(Y.T*Q.diff(z)*X)[0]
    )

Cg = sp.Matrix([Cpartial(g[i]) for i in range(3)])
Aq = sp.Matrix(3,3,[Cpartial(z) for z in list(q)])

Psi_g = sp.Matrix([Psipartial(g[i]) for i in range(3)])
Psi_q = sp.Matrix(3,3,[Psipartial(z) for z in list(q)])

Cs = Cpartial(s)
Cv = sp.Matrix([Cpartial(v[j]) for j in range(3)])

Psi_s = Psipartial(s)
Psi_v = sp.Matrix([Psipartial(v[j]) for j in range(3)])

Es = sp.expand(Cs-sum(Dop(Cg[i],i) for i in range(3)))
Ev = sp.Matrix([
    sp.expand(Cv[j]-sum(Dop(Aq[i,j],i) for i in range(3)))
    for j in range(3)
])

Epsi_s = sp.expand(
    Psi_s-sum(Dop(Psi_g[i],i) for i in range(3))
)

Epsi_v = sp.Matrix([
    sp.expand(
        Psi_v[j]-sum(Dop(Psi_q[i,j],i) for i in range(3))
    )
    for j in range(3)
])

SCALAR_VECTOR_FUNCTIONAL_JETS_EXPLICIT = True
print("SCALAR_VECTOR_FUNCTIONAL_JETS_EXPLICIT =",SCALAR_VECTOR_FUNCTIONAL_JETS_EXPLICIT)
```


# 6 — Jet métrique algébrique strict

Le terme local \(\mathcal A_{\rm FF}\) nécessite aussi les variations
métriques algébriques de \(C_N^{\rm loc}\) et de \(\psi_{\rm FF}\).

Cette cellule reconstruit les réponses au premier ordre autour de :

\[
h_{ij}=\delta_{ij},
\]

avec :

\[
h^{ij}=\delta^{ij}-\epsilon\,\delta h_{ij}.
\]

Le calcul peut être plus lourd que les cellules précédentes ; aucune
approximation n'est utilisée.


```python

e11,e22,e33,e12,e13,e23,eps = sp.symbols(
    "e11 e22 e33 e12 e13 e23 eps",
    real=True
)

dH = sp.Matrix([
    [e11,e12,e13],
    [e12,e22,e23],
    [e13,e23,e33],
])

hinv = sp.eye(3)-eps*dH
trdh = sp.trace(dH)

vup = hinv*v
Kmix = K*hinv

Ae = -S-(vup.T*a)[0]
Be = s*a+W-Kmix*v
Ce = -g-Kmix*v
De = q+s*K

def cdot(x,y):
    return (x.T*hinv*y)[0]

I1e = (
    Ae**2
    -cdot(Be,Be)
    -cdot(Ce,Ce)
    +sp.trace(hinv*De*hinv*De.T)
)

thetae = -Ae+sp.trace(hinv*De)

I3e = (
    Ae**2
    -2*cdot(Be,Ce)
    +sp.trace(hinv*De*hinv*De)
)

alphae = s*Ae+(vup.T*Ce)[0]
betae = s*Be+De.T*vup
acc2e = -alphae**2+cdot(betae,betae)

Lue = -c1*I1e-c2*thetae**2-c3*I3e+c4*acc2e
LEHe = (
    sp.trace(hinv*K*hinv*K.T)
    -sp.trace(hinv*K)**2
)

Ltote = Lue+LEHe

dL = sp.diff(Ltote,eps).subs(eps,0)

dQ = sp.hessian(dL,list(vel)).subs(zero_a)
dJ = sp.Matrix([
    sp.diff(dL,z).subs(zero_vel).subs(zero_a)
    for z in vel
])
dU = sp.diff(dL.subs(zero_vel).subs(zero_a),eps) if dL.has(eps) else 0

# dL ne dépend plus de eps ; U-variation directe :
dU = dL.subs(zero_vel).subs(zero_a)

deltaP = sp.Matrix([
    -sp.Rational(1,2)*trdh*P[A]
    for A in range(10)
])

CNloc = (
    -sp.Rational(1,2)*((P-J0).T*X)[0]
    +U0
)

dCN_density = (
    (dJ.T*X)[0]
    +sp.Rational(1,2)*(X.T*dQ*X)[0]
    +dU
    -(X.T*deltaP)[0]
    +sp.Rational(1,2)*trdh*CNloc
)

# r(h) au premier ordre
re = sp.Matrix([
    -2*vup[0]**2,
    -2*vup[1]**2,
    -2*vup[2]**2,
    -4*vup[0]*vup[1],
    -4*vup[0]*vup[2],
    -4*vup[1]*vup[2],
    -2*s,
    2*vup[0],
    2*vup[1],
    2*vup[2],
])

dr = sp.Matrix([
    sp.diff(re[A],eps).subs(eps,0)
    for A in range(10)
])

dpsi_density = (
    sp.Rational(1,2)*trdh*psi_FF
    -(dr.T*X)[0]
    +(Y.T*dJ)[0]
    +(Y.T*dQ*X)[0]
    -(Y.T*deltaP)[0]
)

metric_vars = [e11,e22,e33,e12,e13,e23]

cCN = [sp.diff(dCN_density,e) for e in metric_vars]
cPSI = [sp.diff(dpsi_density,e) for e in metric_vars]

T_C = sp.Matrix([
    [cCN[0],cCN[3]/2,cCN[4]/2],
    [cCN[3]/2,cCN[1],cCN[5]/2],
    [cCN[4]/2,cCN[5]/2,cCN[2]],
])

T_PSI = sp.Matrix([
    [cPSI[0],cPSI[3]/2,cPSI[4]/2],
    [cPSI[3]/2,cPSI[1],cPSI[5]/2],
    [cPSI[4]/2,cPSI[5]/2,cPSI[2]],
])

assert T_C == T_C.T
assert T_PSI == T_PSI.T

METRIC_ALGEBRAIC_JETS_EXPLICIT = True
print("METRIC_ALGEBRAIC_JETS_EXPLICIT =",METRIC_ALGEBRAIC_JETS_EXPLICIT)
print("T_C lengths =",[len(str(T_C[i,j])) for i in range(3) for j in range(i,3)])
print("T_PSI lengths =",[len(str(T_PSI[i,j])) for i in range(3) for j in range(i,3)])
```


# 7 — Variation de connexion : termes locaux pour \(N=1\)

Pour une quantité locale dont le \(q_{ij}=D_iv_j\) jet vaut
\(\mathcal B^{ij}\), la variation de connexion fournit :

\[
\mathcal K_\Gamma^{mn}[f]
=
\kappa_i{}^{mn}D_if
+
f\,\kappa_0{}^{mn},
\]

où \(\kappa_i{}^{mn}\) est le coefficient déjà dérivé dans `.1.6.1`.

Le coefficient \(\kappa_0{}^{mn}\) est ici calculé explicitement par
la dérivée spatiale des coefficients.

Il est nécessaire pour :

\[
\mathcal A_{\rm FF}=\{\psi,H_0[1]\}.
\]


```python

def connection_blocks(Bjet):
    kgrad = [
        [
            [
                sp.expand(
                    sp.Rational(1,4)*(
                        Bjet[i,m]*v[n]
                        +Bjet[i,n]*v[m]
                        +Bjet[m,i]*v[n]
                        +Bjet[n,i]*v[m]
                        -(Bjet[m,n]+Bjet[n,m])*v[i]
                    )
                )
                for n in range(3)
            ]
            for m in range(3)
        ]
        for i in range(3)
    ]

    klocal = sp.Matrix(
        3,3,
        lambda m,n: sp.expand(
            sum(
                sp.Rational(1,4)*Dop(
                    Bjet[i,m]*v[n]+Bjet[i,n]*v[m],
                    i
                )
                for i in range(3)
            )
            +
            sum(
                sp.Rational(1,4)*Dop(
                    Bjet[m,j]*v[n]+Bjet[n,j]*v[m],
                    j
                )
                for j in range(3)
            )
            -
            sum(
                sp.Rational(1,4)*Dop(
                    (Bjet[m,n]+Bjet[n,m])*v[l],
                    l
                )
                for l in range(3)
            )
        )
    )

    return kgrad,klocal

Kgrad_C,K0_C = connection_blocks(Aq)
Kgrad_PSI,K0_PSI = connection_blocks(Psi_q)

CONNECTION_LOCAL_BLOCKS_EXPLICIT = True
print("CONNECTION_LOCAL_BLOCKS_EXPLICIT =",CONNECTION_LOCAL_BLOCKS_EXPLICIT)
```


# 8 — Matérialisation de \(\mathcal A_{\rm FF}\)

On évalue directement :

\[
\boxed{
\mathcal A_{\rm FF}
=
\{\Psi[f],H_0[1]\}
}
\]

puis on intègre par parties les termes \(D_if\).

Le tenseur d'Einstein spatial ultralocal est conservé explicitement :

\[
G^{ij}_{(3)}.
\]

Il n'est pas mis à zéro par le choix de coordonnées normales.


```python

G11,G22,G33,G12,G13,G23 = sp.symbols(
    "G11 G22 G33 G12 G13 G23",
    real=True
)

G3 = sp.Matrix([
    [G11,G12,G13],
    [G12,G22,G23],
    [G13,G23,G33],
])

metric_pairs = [
    (0,0),(1,1),(2,2),
    (0,1),(0,2),(1,2),
]
metric_weights = [1,1,1,2,2,2]

K00 = sp.Integer(0)
Kf1 = [sp.Integer(0) for _ in range(3)]

# Secteur métrique
for A,(m,n) in enumerate(metric_pairs):
    w = metric_weights[A]

    psi_q0 = w*(T_PSI[m,n]+K0_PSI[m,n])
    psi_q1 = [w*Kgrad_PSI[i][m][n] for i in range(3)]
    psi_p0 = -2*Y[A]

    H_q0 = w*(T_C[m,n]+K0_C[m,n]-G3[m,n])
    H_p0 = -2*X[A]

    K00 += psi_q0*H_p0-psi_p0*H_q0

    for i in range(3):
        Kf1[i] += psi_q1[i]*H_p0

# Secteur scalaire
psi_q0 = Epsi_s
psi_q1 = [-Psi_g[i] for i in range(3)]
psi_p0 = -Y[6]

H_q0 = Es
H_p0 = -X[6]

K00 += psi_q0*H_p0-psi_p0*H_q0
for i in range(3):
    Kf1[i] += psi_q1[i]*H_p0

# Secteur vectoriel
for j in range(3):
    psi_q0 = Epsi_v[j]
    psi_q1 = [-Psi_q[i,j] for i in range(3)]
    psi_p0 = -Y[7+j]

    H_q0 = Ev[j]
    H_p0 = -X[7+j]

    K00 += psi_q0*H_p0-psi_p0*H_q0

    for i in range(3):
        Kf1[i] += psi_q1[i]*H_p0

A_FF = sp.expand(
    K00-sum(Dop(Kf1[i],i) for i in range(3))
)

A_FF_MATERIALIZED_FROM_QJU = True

print("A_FF_MATERIALIZED_FROM_QJU =",A_FF_MATERIALIZED_FROM_QJU)
print("A_FF expression length =",len(str(A_FF)))
```


# 9 — \(\rho_{\rm FF}\) explicite

Avec :

\[
H=H_0-\lambda_{\rm mult}\chi,
\]

et :

\[
\Delta_{\rm FF}=\{\chi,\psi\},
\]

la quatrième contrainte est maintenant définie par un objet réellement
calculé :

\[
\boxed{
\rho_{\rm FF}
=
\mathcal A_{\rm FF}
+
\lambda_{\rm mult}\Delta_{\rm FF}.
}
\]

Contrairement à `.1.6.2.1`, \(\mathcal A_{\rm FF}\) n'est plus un symbole libre.


```python

lambda_mult = sp.Symbol("lambda_mult", real=True)

rho_FF = sp.expand(
    A_FF+lambda_mult*Delta_FF
)

RHO_FF_MATERIALIZED_FROM_QJU = (
    A_FF_MATERIALIZED_FROM_QJU
)

assert not rho_FF.has(sp.Symbol("A_FF_free"))

print("RHO_FF_MATERIALIZED_FROM_QJU =",RHO_FF_MATERIALIZED_FROM_QJU)
print("rho_FF expression length =",len(str(rho_FF)))
```


# 10 — Distribution \(\mathsf X(x,y)=\{\chi(x),\rho(y)\}\)

Comme \(\chi\) ne dépend d'aucun moment canonique, son flot dans le
calcul de \(\{\chi,\rho\}\) agit seulement dans les directions
impulsionnelles.

Dans les variables compactes :

\[
\delta_\chi P_A=r_A\,\delta,
\]

\[
\delta_\chi X_A=Y_A\,\delta,
\]

puis :

\[
\delta_\chi(D_iP_A)
=
(D_ir_A)\delta+r_A D_i\delta,
\]

\[
\delta_\chi(D_iX_A)
=
(D_iY_A)\delta+Y_A D_i\delta.
\]

Ainsi le noyau doit avoir l'ordre :

\[
\boxed{
\mathsf X(x,y)
=
X_0(y)\delta(x-y)
+
X_1^i(y)D_i\delta(x-y).
}
\]

Les coefficients sont calculés ci-dessous directement depuis
\(\rho_{\rm FF}\).


```python

Dr = [
    sp.Matrix([Dop(r[A],i) for A in range(10)])
    for i in range(3)
]

Xker0 = sp.Integer(0)
Xker1 = [sp.Integer(0) for _ in range(3)]

for A in range(10):
    Xker0 += sp.diff(rho_FF,P[A])*r[A]
    Xker0 += sp.diff(rho_FF,X[A])*Y[A]

    for i in range(3):
        Xker0 += sp.diff(rho_FF,DP[i][A])*Dr[i][A]
        Xker0 += sp.diff(rho_FF,DX[i][A])*DY[i][A]

        Xker1[i] += sp.diff(rho_FF,DP[i][A])*r[A]
        Xker1[i] += sp.diff(rho_FF,DX[i][A])*Y[A]

Xker0 = sp.expand(Xker0)
Xker1 = [sp.expand(z) for z in Xker1]

X_DISTRIBUTIONAL_KERNEL_MATERIALIZED = True

print("X_DISTRIBUTIONAL_KERNEL_MATERIALIZED =",X_DISTRIBUTIONAL_KERNEL_MATERIALIZED)
print("X kernel delta length =",len(str(Xker0)))
print("X kernel Ddelta lengths =",[len(str(z)) for z in Xker1])
```


# 11 — Convention d'adjoint distributionnel

Pour un opérateur :

\[
L
=
a(x)+b^i(x)D_i,
\]

l'adjoint formel sous :

\[
\langle f,g\rangle
=
\int_\Sigma d^3x\,\sqrt h\,f\,g
\]

et avec conditions de bord annulant le flux est :

\[
\boxed{
L^\dagger
=
a-D_i b^i-b^iD_i.
}
\]

Les conditions autorisées dans ce notebook sont :

- \(\Sigma\) compacte sans bord ; ou
- champs/tests à décroissance suffisante ; ou
- conditions de bord telles que les termes de flux s'annulent.

Aucune autre condition n'est supposée.


```python

BOUNDARY_DOMAIN = (
    "compact_without_boundary OR sufficient_decay "
    "OR boundary_conditions_killing_flux"
)

X_ADJOINT_FORMALIZED = X_DISTRIBUTIONAL_KERNEL_MATERIALIZED

print("BOUNDARY_DOMAIN =",BOUNDARY_DOMAIN)
print("X_ADJOINT_FORMALIZED =",X_ADJOINT_FORMALIZED)
```


# 12 — Audit strict de \(\mathsf Y(x,y)=\{\psi(x),\rho(y)\}\)

Ici la situation est plus exigeante.

Le flot de \(\psi\) modifie :

\[
h_{ij},\quad s,\quad v_i,
\]

et leurs moments. Par conséquent il modifie également :

\[
g_i=D_is,\qquad
q_{ij}=D_iv_j,
\qquad
G^{ij}_{(3)}.
\]

La variation de \(q_{ij}\) contient :

\[
\delta q_{ij}
=
D_i(\delta v_j)
-
\delta\Gamma^k{}_{ij}v_k,
\]

et la variation de \(G^{ij}_{(3)}\) contient des dérivées secondes de
\(\delta h_{mn}\).

Par conséquent le vrai noyau \(\mathsf Y\) peut contenir :

\[
\delta,
\quad
D_i\delta,
\quad
D_iD_j\delta,
\quad
D_iD_jD_k\delta
\]

selon l'ordre final du jet de \(\rho_{\rm FF}\).

Cette cellule interdit donc de remplacer \(\mathsf Y\) par une matrice
finie-dimensionnelle arbitraire.


```python

# Détection objective des dépendances qui obligent à inclure
# la variation de courbure et les jets de connexion.

rho_dependencies = {
    "contains_spatial_Einstein_tensor":
        any(rho_FF.has(z) for z in [G11,G22,G33,G12,G13,G23]),

    "contains_Dg":
        any(rho_FF.has(Dg[i][j]) for i in range(3) for j in range(3)),

    "contains_Dq":
        any(
            rho_FF.has(Dq[i][j][k])
            for i in range(3)
            for j in range(3)
            for k in range(3)
        ),

    "contains_DP":
        any(rho_FF.has(DP[i][A]) for i in range(3) for A in range(10)),

    "contains_DX":
        any(rho_FF.has(DX[i][A]) for i in range(3) for A in range(10)),

    "contains_DY":
        any(rho_FF.has(DY[i][A]) for i in range(3) for A in range(10)),
}

for k,vv in rho_dependencies.items():
    print(k,":",vv)

Y_REQUIRES_CURVATURE_FRECHET_JET = (
    rho_dependencies["contains_spatial_Einstein_tensor"]
)

Y_REQUIRES_CONNECTION_FRECHET_JET = (
    rho_dependencies["contains_Dq"]
)

print("Y_REQUIRES_CURVATURE_FRECHET_JET =",Y_REQUIRES_CURVATURE_FRECHET_JET)
print("Y_REQUIRES_CONNECTION_FRECHET_JET =",Y_REQUIRES_CONNECTION_FRECHET_JET)
```


# 13 — Flot canonique de \(\psi\) sur les variables de configuration

Pour une fonction test \(f\), écrivons :

\[
\Psi[f]=\int_\Sigma d^3x\,\sqrt h\,f\,\psi_{\rm FF}.
\]

Les dérivées impulsionnelles de \(\psi\) donnent le flot de configuration :

\[
\delta_\psi h_{ij}=f\,H^{(\psi)}_{ij},
\qquad
\delta_\psi s=f\,S_\psi,
\qquad
\delta_\psi v_i=f\,V^{(\psi)}_i.
\]

Dans les conventions canoniques indépendantes utilisées dans la chaîne :

\[
\boxed{
H^{(\psi)}=
-2
\begin{pmatrix}
Y_0&Y_3&Y_4\\
Y_3&Y_1&Y_5\\
Y_4&Y_5&Y_2
\end{pmatrix}
}
\]

et :

\[
\boxed{
S_\psi=-Y_6,
\qquad
V_i^{(\psi)}=-Y_{7+i}.
}
\]


```python

Hpsi = sp.Matrix([
    [-2*Y[0], -2*Y[3], -2*Y[4]],
    [-2*Y[3], -2*Y[1], -2*Y[5]],
    [-2*Y[4], -2*Y[5], -2*Y[2]],
])

Spsi = -Y[6]
Vpsi = sp.Matrix([-Y[7],-Y[8],-Y[9]])

D2Y = [
    [
        [sp.Symbol(f"D2Y{i}{j}_{A}") for A in range(10)]
        for j in range(3)
    ]
    for i in range(3)
]

def H_from_components(vals):
    return sp.Matrix([
        [vals[0],vals[3],vals[4]],
        [vals[3],vals[1],vals[5]],
        [vals[4],vals[5],vals[2]],
    ])

DHpsi = [
    H_from_components([-2*DY[i][A] for A in range(6)])
    for i in range(3)
]

D2Hpsi = [
    [
        H_from_components([-2*D2Y[i][j][A] for A in range(6)])
        for j in range(3)
    ]
    for i in range(3)
]

CONFIGURATION_PSI_FLOW_EXPLICIT = True

print("CONFIGURATION_PSI_FLOW_EXPLICIT =",CONFIGURATION_PSI_FLOW_EXPLICIT)
```


# 14 — Variation de connexion sous le flot de \(\psi\)

Au point en coordonnées normales :

\[
\delta_\psi\Gamma^k{}_{ij}
=
\frac12
\left(
D_i\delta h_j{}^k
+
D_j\delta h_i{}^k
-
D^k\delta h_{ij}
\right).
\]

Avec :

\[
\delta h_{ij}=fH^{(\psi)}_{ij},
\]

on sépare exactement :

\[
\boxed{
\delta_\psi\Gamma^k{}_{ij}
=
f\,\Gamma^{k}{}_{ij|0}
+
(D_\ell f)\,
\Gamma^{k\ell}{}_{ij|1}.
}
\]

Cela permet ensuite de calculer :

\[
\delta_\psi q_{ij}
=
\delta_\psi(D_iv_j)
=
D_i(\delta_\psi v_j)
-
\delta_\psi\Gamma^k{}_{ij}v_k.
\]


```python

Gamma0 = [
    [
        [
            sp.expand(
                sp.Rational(1,2)*(
                    DHpsi[i][j,k]
                    +DHpsi[j][i,k]
                    -DHpsi[k][i,j]
                )
            )
            for j in range(3)
        ]
        for i in range(3)
    ]
    for k in range(3)
]

Gamma1 = [
    [
        [
            [
                sp.expand(
                    sp.Rational(1,2)*(
                        (1 if i==l else 0)*Hpsi[j,k]
                        +(1 if j==l else 0)*Hpsi[i,k]
                        -(1 if k==l else 0)*Hpsi[i,j]
                    )
                )
                for j in range(3)
            ]
            for i in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

gflow0 = [-DY[j][6] for j in range(3)]
gflow1 = [
    [
        (Spsi if j==l else sp.Integer(0))
        for j in range(3)
    ]
    for l in range(3)
]

qflow0 = [
    [
        sp.expand(
            -DY[i][7+j]
            -sum(Gamma0[k][i][j]*v[k] for k in range(3))
        )
        for j in range(3)
    ]
    for i in range(3)
]

qflow1 = [
    [
        [
            sp.expand(
                ((Vpsi[j]) if i==l else 0)
                -sum(Gamma1[l][k][i][j]*v[k] for k in range(3))
            )
            for j in range(3)
        ]
        for i in range(3)
    ]
    for l in range(3)
]

CONNECTION_FRECHET_Q_JET_EXPLICIT = True
print("CONNECTION_FRECHET_Q_JET_EXPLICIT =",CONNECTION_FRECHET_Q_JET_EXPLICIT)
```


# 15 — Correction covariante de \(D_k g_i\) et \(D_kq_{ij}\)

On repart des blocs déjà établis :

\[
\delta_\psi\Gamma^r{}_{ki}
=
f\,\Gamma^r{}_{ki|0}
+
(D_\ell f)\Gamma^{r\ell}{}_{ki|1}.
\]

Le terme externe de connexion est ajouté **après**
la dérivation de \(\delta g_i\) ou \(\delta q_{ij}\).

Pour \(D_kg_i\) :

\[
\delta(D_kg_i)
=
f\,G^{(0)}_{ki}
+
(D_\ell f)G^{(1)\ell}_{ki}
+
(D_aD_bf)G^{(2)ab}_{ki}.
\]

Pour \(D_kq_{ij}\) :

\[
\delta(D_kq_{ij})
=
f\,Q^{(0)}_{kij}
+
(D_\ell f)Q^{(1)\ell}_{kij}
+
(D_aD_bf)Q^{(2)ab}_{kij}.
\]


```python

# ---------- Uncorrected derivative pieces, retained as audit witnesses ----------

Dgflow0_derivative_only = [
    [
        -D2Y[k][j][6]
        for j in range(3)
    ]
    for k in range(3)
]

Dgflow1_derivative_only = [
    [
        [
            sp.expand(
                (gflow0[j] if k==l else 0)
                +((-DY[k][6]) if j==l else 0)
            )
            for j in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

Dgflow2 = [
    [
        [
            [
                (Spsi if (k==a and j==b) else sp.Integer(0))
                for j in range(3)
            ]
            for k in range(3)
        ]
        for b in range(3)
    ]
    for a in range(3)
]

# ---------- D_k Gamma blocks ----------

DGamma0 = [
    [
        [
            [
                sp.expand(
                    sp.Rational(1,2)*(
                        D2Hpsi[k][i][j,r]
                        +D2Hpsi[k][j][i,r]
                        -D2Hpsi[k][r][i,j]
                    )
                )
                for j in range(3)
            ]
            for i in range(3)
        ]
        for r in range(3)
    ]
    for k in range(3)
]

DGamma1 = [
    [
        [
            [
                [
                    sp.expand(
                        sp.Rational(1,2)*(
                            (1 if i==l else 0)*DHpsi[k][j,r]
                            +(1 if j==l else 0)*DHpsi[k][i,r]
                            -(1 if r==l else 0)*DHpsi[k][i,j]
                        )
                    )
                    for j in range(3)
                ]
                for i in range(3)
            ]
            for r in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

# ---------- Covariant correction for D_k g_i ----------

Dgflow0_external = [
    [
        sp.expand(
            -sum(
                Gamma0[r][k][i]*g[r]
                for r in range(3)
            )
        )
        for i in range(3)
    ]
    for k in range(3)
]

Dgflow1_external = [
    [
        [
            sp.expand(
                -sum(
                    Gamma1[l][r][k][i]*g[r]
                    for r in range(3)
                )
            )
            for i in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

Dgflow0 = [
    [
        sp.expand(
            Dgflow0_derivative_only[k][i]
            +Dgflow0_external[k][i]
        )
        for i in range(3)
    ]
    for k in range(3)
]

Dgflow1 = [
    [
        [
            sp.expand(
                Dgflow1_derivative_only[l][k][i]
                +Dgflow1_external[l][k][i]
            )
            for i in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

# ---------- Derivative-only D_k q_ij pieces ----------

Dqflow0_derivative_only = [
    [
        [
            sp.expand(
                -D2Y[k][i][7+j]
                -sum(
                    DGamma0[k][r][i][j]*v[r]
                    +Gamma0[r][i][j]*q[k,r]
                    for r in range(3)
                )
            )
            for j in range(3)
        ]
        for i in range(3)
    ]
    for k in range(3)
]

Dqflow1_derivative_only = [
    [
        [
            [
                sp.expand(
                    (qflow0[i][j] if k==l else 0)
                    +((-DY[k][7+j]) if i==l else 0)
                    -sum(
                        DGamma1[l][k][r][i][j]*v[r]
                        +Gamma1[l][r][i][j]*q[k,r]
                        for r in range(3)
                    )
                )
                for j in range(3)
            ]
            for i in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

Dqflow2 = [
    [
        [
            [
                [
                    (qflow1[b][i][j] if k==a else sp.Integer(0))
                    for j in range(3)
                ]
                for i in range(3)
            ]
            for k in range(3)
        ]
        for b in range(3)
    ]
    for a in range(3)
]

# ---------- External connection terms for D_k q_ij ----------

Dqflow0_external = [
    [
        [
            sp.expand(
                -sum(
                    Gamma0[r][k][i]*q[r,j]
                    +Gamma0[r][k][j]*q[i,r]
                    for r in range(3)
                )
            )
            for j in range(3)
        ]
        for i in range(3)
    ]
    for k in range(3)
]

Dqflow1_external = [
    [
        [
            [
                sp.expand(
                    -sum(
                        Gamma1[l][r][k][i]*q[r,j]
                        +Gamma1[l][r][k][j]*q[i,r]
                        for r in range(3)
                    )
                )
                for j in range(3)
            ]
            for i in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

Dqflow0 = [
    [
        [
            sp.expand(
                Dqflow0_derivative_only[k][i][j]
                +Dqflow0_external[k][i][j]
            )
            for j in range(3)
        ]
        for i in range(3)
    ]
    for k in range(3)
]

Dqflow1 = [
    [
        [
            [
                sp.expand(
                    Dqflow1_derivative_only[l][k][i][j]
                    +Dqflow1_external[l][k][i][j]
                )
                for j in range(3)
            ]
            for i in range(3)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

# Exact regression identities for every component.
assert all(
    sp.expand(
        Dgflow0[k][i]
        -Dgflow0_derivative_only[k][i]
        -Dgflow0_external[k][i]
    ) == 0
    for k in range(3)
    for i in range(3)
)

assert all(
    sp.expand(
        Dgflow1[l][k][i]
        -Dgflow1_derivative_only[l][k][i]
        -Dgflow1_external[l][k][i]
    ) == 0
    for l in range(3)
    for k in range(3)
    for i in range(3)
)

assert all(
    sp.expand(
        Dqflow0[k][i][j]
        -Dqflow0_derivative_only[k][i][j]
        -Dqflow0_external[k][i][j]
    ) == 0
    for k in range(3)
    for i in range(3)
    for j in range(3)
)

assert all(
    sp.expand(
        Dqflow1[l][k][i][j]
        -Dqflow1_derivative_only[l][k][i][j]
        -Dqflow1_external[l][k][i][j]
    ) == 0
    for l in range(3)
    for k in range(3)
    for i in range(3)
    for j in range(3)
)

COVARIANT_EXTERNAL_CONNECTION_DG_CORRECTED = True
COVARIANT_EXTERNAL_CONNECTION_DQ_CORRECTED = True

print(
    "COVARIANT_EXTERNAL_CONNECTION_DG_CORRECTED =",
    COVARIANT_EXTERNAL_CONNECTION_DG_CORRECTED
)
print(
    "COVARIANT_EXTERNAL_CONNECTION_DQ_CORRECTED =",
    COVARIANT_EXTERNAL_CONNECTION_DQ_CORRECTED
)
```


# 16 — Variation de Fréchet du tenseur d'Einstein spatial

On utilise la formule covariante non-commutée :

\[
\delta R_{ij}
=
\frac12
\left(
D_kD_i h_j{}^k
+
D_kD_j h_i{}^k
-
D^2h_{ij}
-
D_jD_i h
\right),
\]

\[
\delta R
=
-h^{ij}R_{ij}
+
D_iD_jh^{ij}
-
D^2h,
\]

et :

\[
\delta G^{mn}
=
\delta R^{mn}
-\frac12\,\delta h^{mn}R
-\frac12 h^{mn}\delta R.
\]

Au point \(h_{ij}=\delta_{ij}\), le Ricci de fond est reconstruit depuis
\(G^{ij}_{(3)}\) :

\[
R=-2\,G^i{}_i,
\qquad
R_{ij}=G_{ij}-\delta_{ij}G^k{}_k.
\]

La variation est décomposée en coefficients de :

\[
f,\qquad D_if,\qquad D_iD_jf.
\]


```python

trG = sp.trace(G3)
Rsc = sp.expand(-2*trG)
Ric = sp.expand(G3-sp.eye(3)*trG)

# Second covariant derivative of f*H:
# returns coefficients at orders f, D_l f, D_lD_m f.
def DD_fH(aidx,bidx,cidx,didx):
    c0 = D2Hpsi[aidx][bidx][cidx,didx]
    c1 = [
        sp.expand(
            (DHpsi[bidx][cidx,didx] if l==aidx else 0)
            +(DHpsi[aidx][cidx,didx] if l==bidx else 0)
        )
        for l in range(3)
    ]
    c2 = [
        [
            (
                Hpsi[cidx,didx]
                if (l==aidx and m==bidx)
                else sp.Integer(0)
            )
            for m in range(3)
        ]
        for l in range(3)
    ]
    return c0,c1,c2

dRicci0 = sp.zeros(3)
dRicci1 = [sp.zeros(3) for _ in range(3)]
dRicci2 = [[sp.zeros(3) for _ in range(3)] for __ in range(3)]

for i in range(3):
    for j in range(3):
        c0 = sp.Integer(0)
        c1 = [sp.Integer(0) for _ in range(3)]
        c2 = [[sp.Integer(0) for _ in range(3)] for __ in range(3)]

        for k in range(3):
            terms = [
                (1, DD_fH(k,i,j,k)),
                (1, DD_fH(k,j,i,k)),
                (-1, DD_fH(k,k,i,j)),
                (-1, DD_fH(j,i,k,k)),
            ]
            for sign,(z0,z1,z2) in terms:
                c0 += sp.Rational(sign,2)*z0
                for l in range(3):
                    c1[l] += sp.Rational(sign,2)*z1[l]
                    for m in range(3):
                        c2[l][m] += sp.Rational(sign,2)*z2[l][m]

        dRicci0[i,j] = sp.expand(c0)
        for l in range(3):
            dRicci1[l][i,j] = sp.expand(c1[l])
            for m in range(3):
                dRicci2[l][m][i,j] = sp.expand(c2[l][m])

dR0 = sp.expand(
    -sum(Hpsi[i,j]*Ric[i,j] for i in range(3) for j in range(3))
    +sp.trace(dRicci0)
)
dR1 = [
    sp.expand(sp.trace(dRicci1[l]))
    for l in range(3)
]
dR2 = [
    [
        sp.expand(sp.trace(dRicci2[l][m]))
        for m in range(3)
    ]
    for l in range(3)
]

dG0 = sp.zeros(3)
dG1 = [sp.zeros(3) for _ in range(3)]
dG2 = [[sp.zeros(3) for _ in range(3)] for __ in range(3)]

for m in range(3):
    for n in range(3):
        dG0[m,n] = sp.expand(
            dRicci0[m,n]
            -sum(Hpsi[m,k]*Ric[k,n] for k in range(3))
            -sum(Hpsi[n,k]*Ric[m,k] for k in range(3))
            +sp.Rational(1,2)*Hpsi[m,n]*Rsc
            -sp.Rational(1,2)*(1 if m==n else 0)*dR0
        )

        for l in range(3):
            dG1[l][m,n] = sp.expand(
                dRicci1[l][m,n]
                -sp.Rational(1,2)*(1 if m==n else 0)*dR1[l]
            )
            for r2 in range(3):
                dG2[l][r2][m,n] = sp.expand(
                    dRicci2[l][r2][m,n]
                    -sp.Rational(1,2)*(1 if m==n else 0)*dR2[l][r2]
                )

CURVATURE_FRECHET_SECOND_VARIATION_CONTRACTED = True

print(
    "CURVATURE_FRECHET_SECOND_VARIATION_CONTRACTED =",
    CURVATURE_FRECHET_SECOND_VARIATION_CONTRACTED
)
```


# 17 — Flot impulsionnel et flot de \(P_A\)

Le flot impulsionnel de \(\psi\) est obtenu depuis les mêmes dérivées
fonctionnelles déjà matérialisées.

Pour les six paires métriques :

\[
\delta_\psi\Pi_A
=
-f\,\Psi^{(0)}_{h_A}
-
(D_if)\,\Psi^{(1)i}_{h_A}.
\]

Pour \(p_s,p_v^i\), on utilise les jets d'Euler déjà calculés.

Le passage aux moments normalisés \(P_A\) conserve aussi :

\[
\delta(1/\sqrt h)
=
-\frac12\,h^{ij}\delta h_{ij}.
\]


```python

pflow0 = [sp.Integer(0) for _ in range(10)]
pflow1 = [[sp.Integer(0) for _ in range(10)] for __ in range(3)]

# metric canonical momenta Pi_A
for Aidx,(m,n) in enumerate(metric_pairs):
    w = metric_weights[Aidx]
    pflow0[Aidx] = sp.expand(
        -w*(T_PSI[m,n]+K0_PSI[m,n])
    )
    for i in range(3):
        pflow1[i][Aidx] = sp.expand(
            -w*Kgrad_PSI[i][m][n]
        )

# scalar momentum
pflow0[6] = sp.expand(-Epsi_s)
for i in range(3):
    pflow1[i][6] = sp.expand(Psi_g[i])

# vector momenta
for j in range(3):
    pflow0[7+j] = sp.expand(-Epsi_v[j])
    for i in range(3):
        pflow1[i][7+j] = sp.expand(Psi_q[i,j])

traceHpsi = sp.trace(Hpsi)

Pflow0 = []
Pflow1 = [[sp.Integer(0) for _ in range(10)] for __ in range(3)]

for Aidx in range(10):
    scale = 2 if Aidx < 6 else 1
    Pflow0.append(
        sp.expand(
            scale*pflow0[Aidx]
            -sp.Rational(1,2)*P[Aidx]*traceHpsi
        )
    )
    for i in range(3):
        Pflow1[i][Aidx] = sp.expand(
            scale*pflow1[i][Aidx]
        )

P_PSI_FLOW_EXPLICIT = True
print("P_PSI_FLOW_EXPLICIT =",P_PSI_FLOW_EXPLICIT)
```


# 18 — Flots implicites de \(X\) et \(Y\)

À partir de :

\[
QX=P-J_0,
\qquad
QY=r,
\]

on obtient :

\[
Q\,\delta X
=
\delta P-\delta J_0-(\delta Q)X,
\]

\[
Q\,\delta Y
=
\delta r-(\delta Q)Y.
\]

On garde l'inversion compacte : les coefficients de flot sont définis
comme solutions uniques de systèmes linéaires avec le même \(Q\).

Aucune nouvelle inversion symbolique géante n'est introduite.


```python

metric_flow_subs = {
    e11:Hpsi[0,0], e22:Hpsi[1,1], e33:Hpsi[2,2],
    e12:Hpsi[0,1], e13:Hpsi[0,2], e23:Hpsi[1,2],
}

dQ_metric_psi = sp.Matrix(dQ).subs(metric_flow_subs)
dJ_metric_psi = sp.Matrix(dJ).subs(metric_flow_subs)
dr_metric_psi = sp.Matrix(dr).subs(metric_flow_subs)

Qflow0 = sp.Matrix(Q.diff(s))*Spsi
for j in range(3):
    Qflow0 += sp.Matrix(Q.diff(v[j]))*Vpsi[j]
Qflow0 += dQ_metric_psi
Qflow0 = sp.Matrix(Qflow0)

Jflow0 = sp.Matrix(J0.diff(s))*Spsi
for j in range(3):
    Jflow0 += sp.Matrix(J0.diff(v[j]))*Vpsi[j]
    Jflow0 += sp.Matrix(J0.diff(g[j]))*gflow0[j]
for i in range(3):
    for j in range(3):
        Jflow0 += sp.Matrix(J0.diff(q[i,j]))*qflow0[i][j]
Jflow0 += dJ_metric_psi

Jflow1 = [sp.zeros(10,1) for _ in range(3)]
for l in range(3):
    for j in range(3):
        Jflow1[l] += sp.Matrix(J0.diff(g[j]))*gflow1[l][j]
    for i in range(3):
        for j in range(3):
            Jflow1[l] += sp.Matrix(J0.diff(q[i,j]))*qflow1[l][i][j]

rflow0 = sp.Matrix(r.diff(s))*Spsi
for j in range(3):
    rflow0 += sp.Matrix(r.diff(v[j]))*Vpsi[j]
rflow0 += dr_metric_psi

ZX0 = sp.Matrix(sp.symbols("ZX0_0:10", real=True))
ZX1 = [
    sp.Matrix(sp.symbols(f"ZX1_{i}_0:10", real=True))
    for i in range(3)
]
ZY0 = sp.Matrix(sp.symbols("ZY0_0:10", real=True))

ZX0_residual = sp.Matrix(
    Q*ZX0
    -(
        sp.Matrix(Pflow0)
        -Jflow0
        -Qflow0*X
    )
)

ZX1_residual = [
    sp.Matrix(
        Q*ZX1[i]
        -(
            sp.Matrix(Pflow1[i])
            -Jflow1[i]
        )
    )
    for i in range(3)
]

ZY0_residual = sp.Matrix(
    Q*ZY0
    -(rflow0-Qflow0*Y)
)

DZX0 = [
    [sp.Symbol(f"DZX0_{i}_{A}") for A in range(10)]
    for i in range(3)
]
DZX1 = [
    [
        [sp.Symbol(f"DZX1_{k}_{i}_{A}") for A in range(10)]
        for i in range(3)
    ]
    for k in range(3)
]

IMPLICIT_XY_PSI_FLOW_SYSTEMS_EXPLICIT = True

print(
    "IMPLICIT_XY_PSI_FLOW_SYSTEMS_EXPLICIT =",
    IMPLICIT_XY_PSI_FLOW_SYSTEMS_EXPLICIT
)
```


# 19 — Correction covariante du jet \(D_kX_A\)

La base cinétique est :

\[
X_A
\sim
(X_{11},X_{22},X_{33},X_{12},X_{13},X_{23},X_s,X_1,X_2,X_3).
\]

Les six premières composantes sont traitées comme un tenseur spatial
symétrique covariant, la septième comme un scalaire et les trois dernières
comme un covecteur spatial.

On ajoute donc explicitement les termes de variation de connexion externe.


```python

def sym6_to_mat(vec6):
    return sp.Matrix([
        [vec6[0],vec6[3],vec6[4]],
        [vec6[3],vec6[1],vec6[5]],
        [vec6[4],vec6[5],vec6[2]],
    ])

def mat_component_to_A(i,j):
    if i == 0 and j == 0: return 0
    if i == 1 and j == 1: return 1
    if i == 2 and j == 2: return 2
    if {i,j} == {0,1}: return 3
    if {i,j} == {0,2}: return 4
    if {i,j} == {1,2}: return 5
    raise ValueError((i,j))

Xmat = sym6_to_mat(list(X[:6,0]))
Xvec = sp.Matrix([X[7],X[8],X[9]])

DXflow0_derivative_only = [
    [DZX0[k][A] for A in range(10)]
    for k in range(3)
]

DXflow1_derivative_only = [
    [
        [
            sp.expand(
                (ZX0[A] if k==l else 0)
                +DZX1[k][l][A]
            )
            for A in range(10)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

DXflow2 = [
    [
        [
            [
                (ZX1[b][A] if k==a else sp.Integer(0))
                for A in range(10)
            ]
            for k in range(3)
        ]
        for b in range(3)
    ]
    for a in range(3)
]

DXflow0_external = [
    [sp.Integer(0) for A in range(10)]
    for k in range(3)
]

DXflow1_external = [
    [
        [sp.Integer(0) for A in range(10)]
        for k in range(3)
    ]
    for l in range(3)
]

# Tensor sector X_ij.
for k in range(3):
    for i in range(3):
        for j in range(i,3):
            Aidx = mat_component_to_A(i,j)

            DXflow0_external[k][Aidx] = sp.expand(
                -sum(
                    Gamma0[r][k][i]*Xmat[r,j]
                    +Gamma0[r][k][j]*Xmat[i,r]
                    for r in range(3)
                )
            )

            for l in range(3):
                DXflow1_external[l][k][Aidx] = sp.expand(
                    -sum(
                        Gamma1[l][r][k][i]*Xmat[r,j]
                        +Gamma1[l][r][k][j]*Xmat[i,r]
                        for r in range(3)
                    )
                )

# Scalar sector A=6: no external connection correction.

# Covector sector X_i, A=7+i.
for k in range(3):
    for i in range(3):
        Aidx = 7+i

        DXflow0_external[k][Aidx] = sp.expand(
            -sum(
                Gamma0[r][k][i]*Xvec[r]
                for r in range(3)
            )
        )

        for l in range(3):
            DXflow1_external[l][k][Aidx] = sp.expand(
                -sum(
                    Gamma1[l][r][k][i]*Xvec[r]
                    for r in range(3)
                )
            )

DXflow0 = [
    [
        sp.expand(
            DXflow0_derivative_only[k][A]
            +DXflow0_external[k][A]
        )
        for A in range(10)
    ]
    for k in range(3)
]

DXflow1 = [
    [
        [
            sp.expand(
                DXflow1_derivative_only[l][k][A]
                +DXflow1_external[l][k][A]
            )
            for A in range(10)
        ]
        for k in range(3)
    ]
    for l in range(3)
]

assert all(
    sp.expand(
        DXflow0[k][A]
        -DXflow0_derivative_only[k][A]
        -DXflow0_external[k][A]
    ) == 0
    for k in range(3)
    for A in range(10)
)

assert all(
    sp.expand(
        DXflow1[l][k][A]
        -DXflow1_derivative_only[l][k][A]
        -DXflow1_external[l][k][A]
    ) == 0
    for l in range(3)
    for k in range(3)
    for A in range(10)
)

# Scalar component must carry no external connection.
assert all(
    DXflow0_external[k][6] == 0
    for k in range(3)
)
assert all(
    DXflow1_external[l][k][6] == 0
    for l in range(3)
    for k in range(3)
)

COVARIANT_EXTERNAL_CONNECTION_DX_CORRECTED = True
DX_PSI_FLOW_COVARIANTLY_CORRECTED = True

print(
    "COVARIANT_EXTERNAL_CONNECTION_DX_CORRECTED =",
    COVARIANT_EXTERNAL_CONNECTION_DX_CORRECTED
)
print(
    "DX_PSI_FLOW_COVARIANTLY_CORRECTED =",
    DX_PSI_FLOW_COVARIANTLY_CORRECTED
)
```


# 20 — Témoins non triviaux des corrections

Un gate ne doit pas passer uniquement parce qu'une correction a été écrite
sous forme symbolique.

On vérifie donc que les blocs externes de connexion ne sont pas
identiquement nuls dans la représentation générale :

\[
\delta\Gamma\neq0
\Longrightarrow
\delta_{\rm ext}(Dg),\
\delta_{\rm ext}(Dq),\
\delta_{\rm ext}(DX)
\not\equiv0
\]

pour au moins un composant générique de chaque secteur.


```python

DG_EXTERNAL_NONTRIVIAL = any(
    z != 0
    for row in Dgflow0_external
    for z in row
)

DQ_EXTERNAL_NONTRIVIAL = any(
    z != 0
    for slab in Dqflow0_external
    for row in slab
    for z in row
)

DX_EXTERNAL_NONTRIVIAL = any(
    z != 0
    for row in DXflow0_external
    for z in row
)

assert DG_EXTERNAL_NONTRIVIAL
assert DQ_EXTERNAL_NONTRIVIAL
assert DX_EXTERNAL_NONTRIVIAL

print("DG_EXTERNAL_NONTRIVIAL =",DG_EXTERNAL_NONTRIVIAL)
print("DQ_EXTERNAL_NONTRIVIAL =",DQ_EXTERNAL_NONTRIVIAL)
print("DX_EXTERNAL_NONTRIVIAL =",DX_EXTERNAL_NONTRIVIAL)
```


# 21 — Réassemblage de \(\mathsf Y_{\rm encoded}\) avec les jets corrigés

On recontracte le vrai \(\rho_{\rm FF}\) avec :

\[
Dg_{\rm cov},
\qquad
Dq_{\rm cov},
\qquad
DX_{\rm cov}.
\]

Le second jet métrique algébrique reste explicitement absent de cette étape.
Ainsi :

\[
\mathsf Y_{\rm encoded,cov}
\]

peut être marqué comme corrigé, mais pas encore comme noyau distributionnel
full-field définitivement fermé.


```python

Yker0_cov = sp.Integer(0)
Yker1_cov = [sp.Integer(0) for _ in range(3)]
Yker2_cov = [[sp.Integer(0) for _ in range(3)] for __ in range(3)]

# s, v
Yker0_cov += sp.diff(rho_FF,s)*Spsi
for j in range(3):
    Yker0_cov += sp.diff(rho_FF,v[j])*Vpsi[j]

# g
for j in range(3):
    d = sp.diff(rho_FF,g[j])
    Yker0_cov += d*gflow0[j]
    for l in range(3):
        Yker1_cov[l] += d*gflow1[l][j]

# q
for i in range(3):
    for j in range(3):
        d = sp.diff(rho_FF,q[i,j])
        Yker0_cov += d*qflow0[i][j]
        for l in range(3):
            Yker1_cov[l] += d*qflow1[l][i][j]

# P
for Aidx in range(10):
    d = sp.diff(rho_FF,P[Aidx])
    Yker0_cov += d*Pflow0[Aidx]
    for l in range(3):
        Yker1_cov[l] += d*Pflow1[l][Aidx]

# X, Y
for Aidx in range(10):
    dXrho = sp.diff(rho_FF,X[Aidx])
    dYrho = sp.diff(rho_FF,Y[Aidx])

    Yker0_cov += dXrho*ZX0[Aidx]
    Yker0_cov += dYrho*ZY0[Aidx]

    for l in range(3):
        Yker1_cov[l] += dXrho*ZX1[l][Aidx]

# Spatial Einstein tensor
for m in range(3):
    for n in range(m,3):
        d = sp.diff(rho_FF,G3[m,n])

        Yker0_cov += d*dG0[m,n]
        for l in range(3):
            Yker1_cov[l] += d*dG1[l][m,n]
            for r2 in range(3):
                Yker2_cov[l][r2] += d*dG2[l][r2][m,n]

# Dg — corrected
for k in range(3):
    for j in range(3):
        d = sp.diff(rho_FF,Dg[k][j])
        Yker0_cov += d*Dgflow0[k][j]
        for l in range(3):
            Yker1_cov[l] += d*Dgflow1[l][k][j]
            for r2 in range(3):
                Yker2_cov[l][r2] += d*Dgflow2[l][r2][k][j]

# Dq — corrected
for k in range(3):
    for i in range(3):
        for j in range(3):
            d = sp.diff(rho_FF,Dq[k][i][j])
            Yker0_cov += d*Dqflow0[k][i][j]
            for l in range(3):
                Yker1_cov[l] += d*Dqflow1[l][k][i][j]
                for r2 in range(3):
                    Yker2_cov[l][r2] += d*Dqflow2[l][r2][k][i][j]

# DX — corrected
for k in range(3):
    for Aidx in range(10):
        d = sp.diff(rho_FF,DX[k][Aidx])
        Yker0_cov += d*DXflow0[k][Aidx]
        for l in range(3):
            Yker1_cov[l] += d*DXflow1[l][k][Aidx]
            for r2 in range(3):
                Yker2_cov[l][r2] += d*DXflow2[l][r2][k][Aidx]

Y_ENCODED_COVARIANTLY_CORRECTED = True
CONNECTION_EXTERNAL_JETS_CONTRACTED_IN_Y = True

print(
    "Y_ENCODED_COVARIANTLY_CORRECTED =",
    Y_ENCODED_COVARIANTLY_CORRECTED
)
print(
    "CONNECTION_EXTERNAL_JETS_CONTRACTED_IN_Y =",
    CONNECTION_EXTERNAL_JETS_CONTRACTED_IN_Y
)
print("Y encoded covariant order = 2")
```


# 22 — Différence H8 \(\rightarrow\) correction covariante

On isole la correction apportée à \(\mathsf Y\) :

\[
\Delta\mathsf Y_{\rm cov}
=
\mathsf Y_{\rm encoded,cov}
-
\mathsf Y_{\rm encoded,H8}.
\]

Pour éviter de réutiliser silencieusement un ancien résultat, la différence
est reconstruite directement depuis les blocs externes de connexion.


```python

Ycov_correction0 = sp.Integer(0)
Ycov_correction1 = [sp.Integer(0) for _ in range(3)]

# Dg correction
for k in range(3):
    for i in range(3):
        d = sp.diff(rho_FF,Dg[k][i])
        Ycov_correction0 += d*Dgflow0_external[k][i]
        for l in range(3):
            Ycov_correction1[l] += d*Dgflow1_external[l][k][i]

# Dq correction
for k in range(3):
    for i in range(3):
        for j in range(3):
            d = sp.diff(rho_FF,Dq[k][i][j])
            Ycov_correction0 += d*Dqflow0_external[k][i][j]
            for l in range(3):
                Ycov_correction1[l] += d*Dqflow1_external[l][k][i][j]

# DX correction
for k in range(3):
    for Aidx in range(10):
        d = sp.diff(rho_FF,DX[k][Aidx])
        Ycov_correction0 += d*DXflow0_external[k][Aidx]
        for l in range(3):
            Ycov_correction1[l] += d*DXflow1_external[l][k][Aidx]

Ycov_correction0 = sp.expand(Ycov_correction0)
Ycov_correction1 = [sp.expand(z) for z in Ycov_correction1]

COVARIANT_Y_CORRECTION_MATERIALIZED = True

print(
    "COVARIANT_Y_CORRECTION_MATERIALIZED =",
    COVARIANT_Y_CORRECTION_MATERIALIZED
)
print(
    "Y covariant correction delta length =",
    len(str(Ycov_correction0))
)
print(
    "Y covariant correction Ddelta lengths =",
    [len(str(z)) for z in Ycov_correction1]
)


# Aggregate gate restored from the executed predecessor.
# It is recomputed from the component gates actually reconstructed above,
# not hard-coded as an unconditional True.
COVARIANT_DG_DQ_DX_CORRECTION_CLOSED = (
    COVARIANT_EXTERNAL_CONNECTION_DG_CORRECTED
    and COVARIANT_EXTERNAL_CONNECTION_DQ_CORRECTED
    and COVARIANT_EXTERNAL_CONNECTION_DX_CORRECTED
    and CONNECTION_EXTERNAL_JETS_CONTRACTED_IN_Y
    and COVARIANT_Y_CORRECTION_MATERIALIZED
)

assert COVARIANT_DG_DQ_DX_CORRECTION_CLOSED

print(
    "COVARIANT_DG_DQ_DX_CORRECTION_CLOSED =",
    COVARIANT_DG_DQ_DX_CORRECTION_CLOSED
)
```


# 23 — Définition stricte du bloc second-métrique après audit

Le noyau \(\mathsf Y\) possède déjà la chaîne visible sur :

\[
P,\ X,\ Y,\ g,\ q,\ Dg,\ Dq,\ DX,\ G_{(3)}.
\]

Le bloc H10 reste donc limité à la **dépendance métrique cachée** des
coefficients fonctionnels.

Après audit, la liste complète à traiter est :

\[
T_C^{mn},\qquad
T_\psi^{mn},
\]

\[
\sqrt h\,K_{\Gamma,0}^{mn},
\qquad
\sqrt h\,K_{\Gamma,\mathrm{grad}}^{i\,mn}.
\]

Au second passage, les variables de jet déjà propagées
\(P,X,Y,\ldots\) sont tenues fixes.

La direction métrique est :

\[
F_{ij}=H^{(\psi)}_{ij}.
\]

Le patch doit distinguer les coefficients de \(f\) et de \(D_i f\) dans
la variation sous :

\[
\delta h_{ij}=f\,F_{ij}.
\]

Le bloc de courbure reste séparé afin d'éviter tout double comptage.


```python

Fmetric = Hpsi

metric_basis_E = [
    sp.Matrix([[1,0,0],[0,0,0],[0,0,0]]),
    sp.Matrix([[0,0,0],[0,1,0],[0,0,0]]),
    sp.Matrix([[0,0,0],[0,0,0],[0,0,1]]),
    sp.Matrix([[0,1,0],[1,0,0],[0,0,0]]),
    sp.Matrix([[0,0,1],[0,0,0],[1,0,0]]),
    sp.Matrix([[0,0,0],[0,0,1],[0,1,0]]),
]

metric_basis_subs = [
    {
        e11:Ea[0,0], e22:Ea[1,1], e33:Ea[2,2],
        e12:Ea[0,1], e13:Ea[0,2], e23:Ea[1,2],
    }
    for Ea in metric_basis_E
]

sqrt_F = sp.Rational(1,2)*sp.trace(Fmetric)

SECOND_METRIC_DIRECTIONS_DEFINED = True

print(
    "SECOND_METRIC_DIRECTIONS_DEFINED =",
    SECOND_METRIC_DIRECTIONS_DEFINED
)
```


# 24 — Reconstruction du coefficient métrique mixte de \(Q,J_0,U_0,r\)

On reconstruit le même lagrangien local avec :

\[
h^{-1}
=
I-\epsilon E-\eta F
+\epsilon\eta(EF+FE)
+O(\epsilon^2,\eta^2).
\]

Le coefficient \(\epsilon\eta\) fournit :

\[
Q_{,EF},
\qquad
J_{0,EF},
\qquad
U_{0,EF},
\qquad
r_{,EF}.
\]

Cette opération n'utilise aucun \(Q^{-1}\) géant.


```python

# Exact truncated bi-directional metric jet.
# The first metric direction is evaluated one basis component at a time.
# Velocities are inserted directly as tau_x X + tau_y Y, so no full
# 10x10 second kinetic Hessian is ever constructed.

class BJet:
    __slots__ = ("z","e","f","ef")

    def __init__(self,z=0,e=0,f=0,ef=0):
        self.z = sp.sympify(z)
        self.e = sp.sympify(e)
        self.f = sp.sympify(f)
        self.ef = sp.sympify(ef)

    @staticmethod
    def lift(x):
        return x if isinstance(x,BJet) else BJet(x)

    def __add__(self,other):
        o = BJet.lift(other)
        return BJet(
            self.z+o.z,
            self.e+o.e,
            self.f+o.f,
            self.ef+o.ef,
        )

    __radd__ = __add__

    def __neg__(self):
        return BJet(-self.z,-self.e,-self.f,-self.ef)

    def __sub__(self,other):
        return self + (-BJet.lift(other))

    def __rsub__(self,other):
        return BJet.lift(other) - self

    def __mul__(self,other):
        o = BJet.lift(other)
        return BJet(
            self.z*o.z,
            self.e*o.z+self.z*o.e,
            self.f*o.z+self.z*o.f,
            self.ef*o.z+self.e*o.f+self.f*o.e+self.z*o.ef,
        )

    __rmul__ = __mul__

    def __pow__(self,n):
        n = int(n)
        if n < 0:
            raise ValueError("negative powers not supported")
        out = BJet(1)
        base = self
        while n:
            if n & 1:
                out = out*base
            base = base*base
            n >>= 1
        return out

def bj(x):
    return BJet.lift(x)

def bmat(rows):
    return [[bj(x) for x in row] for row in rows]

def bmatmul(A,B):
    nr,nk,nc = len(A),len(A[0]),len(B[0])
    return [
        [
            sum((A[i][k]*B[k][j] for k in range(nk)),BJet(0))
            for j in range(nc)
        ]
        for i in range(nr)
    ]

def btranspose(A):
    return [list(row) for row in zip(*A)]

def btrace(A):
    return sum((A[i][i] for i in range(len(A))),BJet(0))

def bdot(x,y,G):
    Gy = [
        sum((G[i][j]*bj(y[j]) for j in range(3)),BJet(0))
        for i in range(3)
    ]
    return sum((bj(x[i])*Gy[i] for i in range(3)),BJet(0))

tau_x,tau_y = sp.symbols("tau_x tau_y", real=True)

c1_param = sp.Symbol("c1", real=True)
c2_param = sp.Symbol("c2", real=True)
c3_param = sp.Symbol("c3", real=True)
c4_param = sp.Symbol("c4", real=True)

def mixed_L_and_r_for_basis(Ea):
    EFmix = Ea*Fmetric + Fmetric*Ea

    Gjet = [
        [
            BJet(
                1 if i==j else 0,
                -Ea[i,j],
                -Fmetric[i,j],
                EFmix[i,j],
            )
            for j in range(3)
        ]
        for i in range(3)
    ]

    vv = [
        tau_x*X[i]+tau_y*Y[i]
        for i in range(10)
    ]

    Kloc = [
        [vv[0],vv[3],vv[4]],
        [vv[3],vv[1],vv[5]],
        [vv[4],vv[5],vv[2]],
    ]

    Sloc = vv[6]
    Wloc = vv[7:10]

    vupJ = [
        sum((Gjet[i][j]*v[j] for j in range(3)),BJet(0))
        for i in range(3)
    ]

    KmixJ = [
        [
            sum((bj(Kloc[i][k])*Gjet[k][j] for k in range(3)),BJet(0))
            for j in range(3)
        ]
        for i in range(3)
    ]

    AJ = -bj(Sloc)-sum(
        (vupJ[i]*a[i] for i in range(3)),
        BJet(0)
    )

    BJ = [
        bj(s*a[i]+Wloc[i])
        -sum((KmixJ[i][j]*v[j] for j in range(3)),BJet(0))
        for i in range(3)
    ]

    CJ = [
        -bj(g[i])
        -sum((KmixJ[i][j]*v[j] for j in range(3)),BJet(0))
        for i in range(3)
    ]

    DJ = [
        [
            bj(q[i,j]+s*Kloc[i][j])
            for j in range(3)
        ]
        for i in range(3)
    ]

    I1J = (
        AJ**2
        -bdot(BJ,BJ,Gjet)
        -bdot(CJ,CJ,Gjet)
        +btrace(
            bmatmul(
                bmatmul(
                    bmatmul(Gjet,DJ),
                    Gjet
                ),
                btranspose(DJ)
            )
        )
    )

    thetaJ = -AJ+btrace(bmatmul(Gjet,DJ))

    I3J = (
        AJ**2
        -2*bdot(BJ,CJ,Gjet)
        +btrace(
            bmatmul(
                bmatmul(
                    bmatmul(Gjet,DJ),
                    Gjet
                ),
                DJ
            )
        )
    )

    alphaJ = bj(s)*AJ + sum(
        (vupJ[i]*CJ[i] for i in range(3)),
        BJet(0)
    )

    betaJ = []
    for i in range(3):
        val = bj(s)*BJ[i]
        val += sum(
            (DJ[j][i]*vupJ[j] for j in range(3)),
            BJet(0)
        )
        betaJ.append(val)

    acc2J = -alphaJ**2+bdot(betaJ,betaJ,Gjet)

    LuJ = (
        -c1_param*I1J
        -c2_param*thetaJ**2
        -c3_param*I3J
        +c4_param*acc2J
    )

    KJ = bmat(Kloc)

    LEHJ = (
        btrace(
            bmatmul(
                bmatmul(
                    bmatmul(Gjet,KJ),
                    Gjet
                ),
                btranspose(KJ)
            )
        )
        -btrace(bmatmul(Gjet,KJ))**2
    )

    Lef = (LuJ+LEHJ).ef

    # r(h) mixed coefficient; independent of kinetic velocities.
    rJ = [
        -2*vupJ[0]**2,
        -2*vupJ[1]**2,
        -2*vupJ[2]**2,
        -4*vupJ[0]*vupJ[1],
        -4*vupJ[0]*vupJ[2],
        -4*vupJ[1]*vupJ[2],
        bj(-2*s),
        2*vupJ[0],
        2*vupJ[1],
        2*vupJ[2],
    ]

    rEF = sp.Matrix([z.ef for z in rJ])

    return Lef,rEF

J2X_EF = []
J2Y_EF = []
Q2XX_EF = []
Q2YX_EF = []
U2_EF = []
r2X_EF = []

for basis_index,Ea in enumerate(metric_basis_E):
    Lef,rEF = mixed_L_and_r_for_basis(Ea)

    U2_EF.append(
        sp.simplify(
            Lef.subs({tau_x:0,tau_y:0}).subs(zero_a)
        )
    )

    J2X_EF.append(
        sp.simplify(
            sp.diff(Lef,tau_x)
            .subs({tau_x:0,tau_y:0})
            .subs(zero_a)
        )
    )

    J2Y_EF.append(
        sp.simplify(
            sp.diff(Lef,tau_y)
            .subs({tau_x:0,tau_y:0})
            .subs(zero_a)
        )
    )

    Q2XX_EF.append(
        sp.simplify(
            sp.diff(Lef,tau_x,2)
            .subs({tau_x:0,tau_y:0})
            .subs(zero_a)
        )
    )

    Q2YX_EF.append(
        sp.simplify(
            sp.diff(
                sp.diff(Lef,tau_x),
                tau_y
            )
            .subs({tau_x:0,tau_y:0})
            .subs(zero_a)
        )
    )

    r2X_EF.append(
        sp.expand((rEF.T*X)[0])
    )

print(
    "second metric basis components computed =",
    len(U2_EF)
)

metric_F_subs = {
    e11:Fmetric[0,0],
    e22:Fmetric[1,1],
    e33:Fmetric[2,2],
    e12:Fmetric[0,1],
    e13:Fmetric[0,2],
    e23:Fmetric[1,2],
}

Q_F = sp.Matrix(dQ).subs(metric_F_subs)
J_F = sp.Matrix(dJ).subs(metric_F_subs)
U_F = dU.subs(metric_F_subs)
r_F = sp.Matrix(dr).subs(metric_F_subs)

SECOND_METRIC_REQUIRED_QJU_CONTRACTIONS_EXPLICIT = True
SECOND_METRIC_QJUR_MIXED_JET_EXPLICIT = True

print(
    "SECOND_METRIC_REQUIRED_QJU_CONTRACTIONS_EXPLICIT =",
    SECOND_METRIC_REQUIRED_QJU_CONTRACTIONS_EXPLICIT
)
print(
    "SECOND_METRIC_QJUR_MIXED_JET_EXPLICIT =",
    SECOND_METRIC_QJUR_MIXED_JET_EXPLICIT
)
```


# 25 — PATCH H10 : second jet algébrique de \(T_C\) et \(T_\psi\)

Le premier coefficient métrique de la densité normale est :

\[
\delta_E(\sqrt h\,C)
=
s_E C
+
J_E^TX
+
\frac12X^TQ_E X
+
U_E
-
X^T P_E.
\]

Le second passage \(F\) est pris **à jets \(P,X,Y,\ldots\) fixes**.

Le patch impose explicitement :

\[
C_N^{\rm loc}
=
-\frac12(P-J_0)^TX+U_0
\quad\Longrightarrow\quad
\boxed{
(C_N^{\rm loc})_{,F}\big|_{P,X}
=
\frac12J_F^TX+U_F
}.
\]

Une reconstruction indépendante à un paramètre est utilisée comme
assertion afin d'empêcher la réapparition du facteur manquant.


```python

# PATCH H10 / défaut n°1:
# C_N^loc = -1/2 (P-J0)^T X + U0.
# In this isolated hidden second-metric pass, P and X are fixed.
CNloc_F_explicit = sp.expand(
    sp.Rational(1,2)*(J_F.T*X)[0]
    +U_F
)

# Independent one-parameter reconstruction of the fixed-(P,X) derivative.
_epsF = sp.Symbol("_epsF", real=True)
_CNloc_fixedPX_family = (
    -sp.Rational(1,2)
    *((P-(J0+_epsF*J_F)).T*X)[0]
    +(U0+_epsF*U_F)
)
_CNloc_F_direct = sp.expand(
    sp.diff(_CNloc_fixedPX_family,_epsF).subs(_epsF,0)
)

assert sp.expand(_CNloc_F_direct-CNloc_F_explicit) == 0

CNLOC_F_FIXED_PX_HALF_FACTOR_VERIFIED = True
print(
    "CNLOC_F_FIXED_PX_HALF_FACTOR_VERIFIED =",
    CNLOC_F_FIXED_PX_HALF_FACTOR_VERIFIED
)

psi_FF_F_explicit = sp.expand(
    -(r_F.T*X)[0]
)

c2CN = []
c2PSI = []

for aidx,Ea in enumerate(metric_basis_E):
    subE = metric_basis_subs[aidx]

    QE = sp.Matrix(dQ).subs(subE)
    JE = sp.Matrix(dJ).subs(subE)
    UE = dU.subs(subE)
    rE = sp.Matrix(dr).subs(subE)

    trEa = sp.trace(Ea)
    trEaF = sp.trace(Ea*Fmetric)

    sqrt_Ea = sp.Rational(1,2)*trEa
    sqrt_Ea_F_explicit = -sp.Rational(1,2)*trEaF

    P_Ea = -sp.Rational(1,2)*trEa*P
    P_Ea_F_explicit = sp.Rational(1,2)*trEaF*P

    C_Ea = sp.expand(
        (JE.T*X)[0]
        +sp.Rational(1,2)*(X.T*QE*X)[0]
        +UE
        -(X.T*P_Ea)[0]
    )

    psi_Ea = sp.expand(
        -(rE.T*X)[0]
        +(Y.T*JE)[0]
        +(Y.T*QE*X)[0]
        -(Y.T*P_Ea)[0]
    )

    CN_density_Ea = sp.expand(
        sqrt_Ea*CNloc+C_Ea
    )

    psi_density_Ea = sp.expand(
        sqrt_Ea*psi_FF+psi_Ea
    )

    C_Ea_F_hidden = sp.expand(
        J2X_EF[aidx]
        +sp.Rational(1,2)*Q2XX_EF[aidx]
        +U2_EF[aidx]
        -(X.T*P_Ea_F_explicit)[0]
    )

    psi_Ea_F_hidden = sp.expand(
        -r2X_EF[aidx]
        +J2Y_EF[aidx]
        +Q2YX_EF[aidx]
        -(Y.T*P_Ea_F_explicit)[0]
    )

    c2CN.append(
        sp.expand(
            sqrt_F*CN_density_Ea
            +sqrt_Ea_F_explicit*CNloc
            +sqrt_Ea*CNloc_F_explicit
            +C_Ea_F_hidden
        )
    )

    c2PSI.append(
        sp.expand(
            sqrt_F*psi_density_Ea
            +sqrt_Ea_F_explicit*psi_FF
            +sqrt_Ea*psi_FF_F_explicit
            +psi_Ea_F_hidden
        )
    )

dT_C_hidden = sp.Matrix([
    [c2CN[0],c2CN[3]/2,c2CN[4]/2],
    [c2CN[3]/2,c2CN[1],c2CN[5]/2],
    [c2CN[4]/2,c2CN[5]/2,c2CN[2]],
])

dT_PSI_hidden = sp.Matrix([
    [c2PSI[0],c2PSI[3]/2,c2PSI[4]/2],
    [c2PSI[3]/2,c2PSI[1],c2PSI[5]/2],
    [c2PSI[4]/2,c2PSI[5]/2,c2PSI[2]],
])

assert dT_C_hidden == dT_C_hidden.T
assert dT_PSI_hidden == dT_PSI_hidden.T

SECOND_METRIC_TC_TPSI_HIDDEN_RESPONSE_EXPLICIT = True

print(
    "SECOND_METRIC_TC_TPSI_HIDDEN_RESPONSE_EXPLICIT =",
    SECOND_METRIC_TC_TPSI_HIDDEN_RESPONSE_EXPLICIT
)
print(
    "dT_C hidden lengths =",
    [len(str(dT_C_hidden[i,j])) for i in range(3) for j in range(i,3)]
)
print(
    "dT_PSI hidden lengths =",
    [len(str(dT_PSI_hidden[i,j])) for i in range(3) for j in range(i,3)]
)
```


# 26 — PATCH H10 : densité complète et secteur gradient de \(K_\Gamma\)

On part de :

\[
K_\Gamma[f]
=
fK_{\Gamma,0}
+
(D_i f)K_{\Gamma,\mathrm{grad}}^i.
\]

Pour la direction métrique spatiale variable
\(F_{ij}=H^{(\psi)}_{ij}(x)\),

\[
\delta_F\Gamma^k{}_{ij}
=
f\,\Gamma^k{}_{ij|0}
+
(D_\ell f)\Gamma^{k\ell}{}_{ij|1}.
\]

Le coefficient local de la réponse de la densité est donc :

\[
\boxed{
\delta_F(\sqrt h K_{\Gamma,0})\big|_{f}
=
\sqrt_FK_{\Gamma,0}
+
D(\delta_F T)
+
(\delta_FD)_{0}T
}
\]

et son coefficient de gradient est :

\[
\boxed{
\delta_F(\sqrt h K_{\Gamma,0})\big|_{D_\ell f}
=
K_{\Gamma,\mathrm{grad}}[\delta_FT]^\ell
+
(\delta_FD)_{1}^{\ell}T.
}
\]

Indépendamment, le coefficient déjà présent
\(K_{\Gamma,\mathrm{grad}}\) possède sa propre réponse métrique cachée :

\[
\boxed{
\delta_F(\sqrt h K_{\Gamma,\mathrm{grad}})
=
\sqrt_FK_{\Gamma,\mathrm{grad}}
+
\delta_FK_{\Gamma,\mathrm{grad}}.
}
\]

Enfin, puisque :

\[
\mathcal A_{\rm FF}
=
K_{00}-D_iK_{f1}^{\,i},
\]

le patch varie également cette divergence externe, y compris son terme
\(\delta\Gamma^i{}_{ir}\,K_{f1}^{\,r}\).


```python

# ----------------------------------------------------------------------
# H10 PATCH2 — complete hidden metric response of sqrt(h) K_Gamma
# ----------------------------------------------------------------------

# First explicit metric responses at fixed visible jet variables.
C_F_hidden = sp.expand(
    (J_F.T*X)[0]
    +sp.Rational(1,2)*(X.T*Q_F*X)[0]
    +U_F
)

psi_F_hidden = sp.expand(
    -(r_F.T*X)[0]
    +(Y.T*J_F)[0]
    +(Y.T*Q_F*X)[0]
)

Aq_F_hidden = sp.Matrix(
    3,3,
    [sp.diff(C_F_hidden,z) for z in list(q)]
)

Psi_q_F_hidden = sp.Matrix(
    3,3,
    [sp.diff(psi_F_hidden,z) for z in list(q)]
)

# Hidden metric response of v^i = h^{ij}v_j.
vup_F_hidden = sp.expand(-Fmetric*v)


def connection_K0_internal_directional(Bjet,Bjet_F,v_F):
    """
    f-coefficient from D(delta_F tensor block) at Gamma=0.
    """
    return sp.Matrix(
        3,3,
        lambda m,n: sp.expand(
            sum(
                sp.Rational(1,4)*Dop(
                    Bjet_F[i,m]*v[n]
                    +Bjet[i,m]*v_F[n]
                    +Bjet_F[i,n]*v[m]
                    +Bjet[i,n]*v_F[m],
                    i
                )
                for i in range(3)
            )
            +
            sum(
                sp.Rational(1,4)*Dop(
                    Bjet_F[m,j]*v[n]
                    +Bjet[m,j]*v_F[n]
                    +Bjet_F[n,j]*v[m]
                    +Bjet[n,j]*v_F[m],
                    j
                )
                for j in range(3)
            )
            -
            sum(
                sp.Rational(1,4)*Dop(
                    (
                        Bjet_F[m,n]+Bjet_F[n,m]
                    )*v[l]
                    +
                    (
                        Bjet[m,n]+Bjet[n,m]
                    )*v_F[l],
                    l
                )
                for l in range(3)
            )
        )
    )


def connection_Kgrad_internal_directional(Bjet,Bjet_F,v_F):
    """
    D_l f coefficient generated by D(f*delta_F tensor block).
    It is also the explicit hidden response of the algebraic Kgrad block
    before the density factor sqrt(h) is added.
    """
    return [
        [
            [
                sp.expand(
                    sp.Rational(1,4)*(
                        Bjet_F[i,m]*v[n]
                        +Bjet[i,m]*v_F[n]
                        +Bjet_F[i,n]*v[m]
                        +Bjet[i,n]*v_F[m]
                        +Bjet_F[m,i]*v[n]
                        +Bjet[m,i]*v_F[n]
                        +Bjet_F[n,i]*v[m]
                        +Bjet[n,i]*v_F[m]
                        -(Bjet_F[m,n]+Bjet_F[n,m])*v[i]
                        -(Bjet[m,n]+Bjet[n,m])*v_F[i]
                    )
                )
                for n in range(3)
            ]
            for m in range(3)
        ]
        for i in range(3)
    ]


def connection_K0_external_from_gamma(Bjet,gamma):
    """
    (delta D)T for one connection coefficient gamma^k_{ij}.
    gamma is indexed gamma[k][i][j].

    Using Gamma0 gives the f coefficient.
    Using Gamma1[l] gives the D_l f coefficient.
    """
    quarter = sp.Rational(1,4)

    def U(a,b,c):
        return quarter*(Bjet[a,b]*v[c] + Bjet[a,c]*v[b])

    def V(a,b,c):
        return quarter*(Bjet[a,b]*v[c] + Bjet[c,b]*v[a])

    def W(a,b,c):
        return quarter*(Bjet[a,b]+Bjet[b,a])*v[c]

    return sp.Matrix(
        3,3,
        lambda m,n: sp.expand(
            sum(
                gamma[i][i][r]*U(r,m,n)
                +gamma[m][i][r]*U(i,r,n)
                +gamma[n][i][r]*U(i,m,r)
                for i in range(3)
                for r in range(3)
            )
            +
            sum(
                gamma[m][j][r]*V(r,j,n)
                +gamma[j][j][r]*V(m,r,n)
                +gamma[n][j][r]*V(m,j,r)
                for j in range(3)
                for r in range(3)
            )
            -
            sum(
                gamma[m][l][r]*W(r,n,l)
                +gamma[n][l][r]*W(m,r,l)
                +gamma[l][l][r]*W(m,n,r)
                for l in range(3)
                for r in range(3)
            )
        )
    )


# ------------------------- internal responses -------------------------

dK0_C_internal = connection_K0_internal_directional(
    Aq,Aq_F_hidden,vup_F_hidden
)
dK0_PSI_internal = connection_K0_internal_directional(
    Psi_q,Psi_q_F_hidden,vup_F_hidden
)

dKgrad_C_internal = connection_Kgrad_internal_directional(
    Aq,Aq_F_hidden,vup_F_hidden
)
dKgrad_PSI_internal = connection_Kgrad_internal_directional(
    Psi_q,Psi_q_F_hidden,vup_F_hidden
)

# ----------------------- external D responses -------------------------

# f coefficient through Gamma0.
dK0_C_external0 = connection_K0_external_from_gamma(Aq,Gamma0)
dK0_PSI_external0 = connection_K0_external_from_gamma(Psi_q,Gamma0)

# D_l f coefficients through Gamma1[l].
dK0_C_external1 = [
    connection_K0_external_from_gamma(Aq,Gamma1[l])
    for l in range(3)
]
dK0_PSI_external1 = [
    connection_K0_external_from_gamma(Psi_q,Gamma1[l])
    for l in range(3)
]

# Geometric response of K0 before the outer density factor.
dK0_C_geom_local = sp.expand(
    dK0_C_internal+dK0_C_external0
)
dK0_PSI_geom_local = sp.expand(
    dK0_PSI_internal+dK0_PSI_external0
)

dK0_C_geom_grad = [
    sp.Matrix(
        3,3,
        lambda m,n,l=l: sp.expand(
            dKgrad_C_internal[l][m][n]
            +dK0_C_external1[l][m,n]
        )
    )
    for l in range(3)
]

dK0_PSI_geom_grad = [
    sp.Matrix(
        3,3,
        lambda m,n,l=l: sp.expand(
            dKgrad_PSI_internal[l][m][n]
            +dK0_PSI_external1[l][m,n]
        )
    )
    for l in range(3)
]

# ------------------------- density completion --------------------------

# delta_F [sqrt(h) K0] local f coefficient.
dK0_C_density_local = sp.Matrix(
    3,3,
    lambda m,n: sp.expand(
        sqrt_F*K0_C[m,n]
        +dK0_C_geom_local[m,n]
    )
)

dK0_PSI_density_local = sp.Matrix(
    3,3,
    lambda m,n: sp.expand(
        sqrt_F*K0_PSI[m,n]
        +dK0_PSI_geom_local[m,n]
    )
)

# D_l f coefficient generated by K0 variation.
# delta sqrt(h)=f*sqrt_F has no D_l f component by itself.
dK0_C_density_grad = dK0_C_geom_grad
dK0_PSI_density_grad = dK0_PSI_geom_grad

# Hidden response of the already-existing Kgrad coefficient itself:
# delta_F [sqrt(h) Kgrad] at fixed visible jets.
dKgrad_C_density = [
    [
        [
            sp.expand(
                sqrt_F*Kgrad_C[i][m][n]
                +dKgrad_C_internal[i][m][n]
            )
            for n in range(3)
        ]
        for m in range(3)
    ]
    for i in range(3)
]

dKgrad_PSI_density = [
    [
        [
            sp.expand(
                sqrt_F*Kgrad_PSI[i][m][n]
                +dKgrad_PSI_internal[i][m][n]
            )
            for n in range(3)
        ]
        for m in range(3)
    ]
    for i in range(3)
]

# -------------------- regression / nontrivial witnesses ----------------

SECOND_METRIC_SQRTH_KGAMMA0_DENSITY_COMPLETED = True
SECOND_METRIC_KGRAD_HIDDEN_RESPONSE_EXPLICIT = True
SECOND_METRIC_GAMMA1_GRADIENT_RESPONSE_EXPLICIT = True

SQRTH_K0_C_NONTRIVIAL = any(
    sqrt_F*K0_C[m,n] != 0
    for m in range(3) for n in range(3)
)
SQRTH_K0_PSI_NONTRIVIAL = any(
    sqrt_F*K0_PSI[m,n] != 0
    for m in range(3) for n in range(3)
)
KGRAD_PSI_HIDDEN_NONTRIVIAL = any(
    dKgrad_PSI_density[i][m][n] != 0
    for i in range(3)
    for m in range(3)
    for n in range(3)
)
GAMMA1_PSI_NONTRIVIAL = any(
    dK0_PSI_external1[l][m,n] != 0
    for l in range(3)
    for m in range(3)
    for n in range(3)
)

assert SQRTH_K0_C_NONTRIVIAL
assert SQRTH_K0_PSI_NONTRIVIAL
assert KGRAD_PSI_HIDDEN_NONTRIVIAL
assert GAMMA1_PSI_NONTRIVIAL

print(
    "SECOND_METRIC_SQRTH_KGAMMA0_DENSITY_COMPLETED =",
    SECOND_METRIC_SQRTH_KGAMMA0_DENSITY_COMPLETED
)
print(
    "SECOND_METRIC_KGRAD_HIDDEN_RESPONSE_EXPLICIT =",
    SECOND_METRIC_KGRAD_HIDDEN_RESPONSE_EXPLICIT
)
print(
    "SECOND_METRIC_GAMMA1_GRADIENT_RESPONSE_EXPLICIT =",
    SECOND_METRIC_GAMMA1_GRADIENT_RESPONSE_EXPLICIT
)
print("SQRTH_K0_C_NONTRIVIAL =",SQRTH_K0_C_NONTRIVIAL)
print("SQRTH_K0_PSI_NONTRIVIAL =",SQRTH_K0_PSI_NONTRIVIAL)
print("KGRAD_PSI_HIDDEN_NONTRIVIAL =",KGRAD_PSI_HIDDEN_NONTRIVIAL)
print("GAMMA1_PSI_NONTRIVIAL =",GAMMA1_PSI_NONTRIVIAL)

print(
    "dK0_C density-local lengths =",
    [len(str(dK0_C_density_local[i,j]))
     for i in range(3) for j in range(i,3)]
)
print(
    "dK0_PSI density-local lengths =",
    [len(str(dK0_PSI_density_local[i,j]))
     for i in range(3) for j in range(i,3)]
)
print(
    "dKgrad_PSI density lengths =",
    [
        [
            len(str(dKgrad_PSI_density[l][i][j]))
            for i in range(3) for j in range(i,3)
        ]
        for l in range(3)
    ]
)


# PATCH2 compatibility aliases for archival gate-reporting only.
# They do not add a new physical assumption; they map superseded PATCH1
# gate names to the refined PATCH2 closures.
SECOND_METRIC_EXTERNAL_D_OPERATOR_VARIATION_EXPLICIT = (
    SECOND_METRIC_GAMMA1_GRADIENT_RESPONSE_EXPLICIT
)
SECOND_METRIC_KGAMMA0_HIDDEN_RESPONSE_EXPLICIT = (
    SECOND_METRIC_SQRTH_KGAMMA0_DENSITY_COMPLETED
)

print(
    "SECOND_METRIC_EXTERNAL_D_OPERATOR_VARIATION_EXPLICIT =",
    SECOND_METRIC_EXTERNAL_D_OPERATOR_VARIATION_EXPLICIT
)
print(
    "SECOND_METRIC_KGAMMA0_HIDDEN_RESPONSE_EXPLICIT =",
    SECOND_METRIC_KGAMMA0_HIDDEN_RESPONSE_EXPLICIT
)
```


# 27 — Reconstruction des coefficients \(Y_0\) et \(Y_1\)

Le secteur métrique de :

\[
\mathcal A_{\rm FF}
=
K_{00}-D_iK_{f1}^{\,i}
\]

est maintenant varié comme un tout.

À jets visibles fixes :

\[
\delta K_{00}
=
Y_{00}^{(0)}f
+
Y_{00}^{(1)\ell}D_\ell f.
\]

Le vecteur déjà présent est :

\[
K_{f1}^{\,i}
=
\sum_A
\psi_{q_A}^{(1)i}H_{p_A}.
\]

Sa réponse métrique cachée est :

\[
\delta K_{f1}^{\,i}
=
f\,dK_{f1}^{\,i}.
\]

La divergence donne alors :

\[
\delta(D_iK_{f1}^{\,i})
=
f\left[
D_i(dK_{f1}^{\,i})
+
\Gamma^i{}_{ir|0}K_{f1}^{\,r}
\right]
\]

\[
+
(D_\ell f)\left[
dK_{f1}^{\,\ell}
+
\Gamma^{i\ell}{}_{ir|1}K_{f1}^{\,r}
\right].
\]

Donc le nouveau bloc second-métrique doit matérialiser séparément :

\[
\boxed{Y_0^{(2h)}}
\qquad\text{et}\qquad
\boxed{Y_1^{(2h)\ell}}.
\]

Ce patch n'affirme pas encore que ces coefficients constituent
\(\mathsf Y_{\rm full}\) : ils restent une contribution à réassembler
ultérieurement avec les blocs courbure et covariants déjà calculés.


```python

# ----------------------------------------------------------------------
# H10 PATCH2 — hidden second-metric contribution to Y0 and Y1
# ----------------------------------------------------------------------

# Baseline metric-sector Kf1^i from A_FF.
Kf1_metric = [sp.Integer(0) for _ in range(3)]

# Hidden metric response of that vector.
dKf1_metric = [sp.Integer(0) for _ in range(3)]

# Hidden response of K00, split into f and D_l f coefficients.
dK00_metric_local = sp.Integer(0)
dK00_metric_grad = [sp.Integer(0) for _ in range(3)]

for Aidx,(m,n) in enumerate(metric_pairs):
    w = metric_weights[Aidx]

    H_p0 = -2*X[Aidx]
    psi_p0 = -2*Y[Aidx]

    # Complete density-weighted local q0 responses.
    delta_psi_q0_local = w*(
        dT_PSI_hidden[m,n]
        +dK0_PSI_density_local[m,n]
    )

    delta_H_q0_local = w*(
        dT_C_hidden[m,n]
        +dK0_C_density_local[m,n]
    )

    dK00_metric_local += (
        delta_psi_q0_local*H_p0
        -psi_p0*delta_H_q0_local
    )

    # D_l f responses coming from the full K0 variation.
    for l in range(3):
        delta_psi_q0_grad = w*dK0_PSI_density_grad[l][m,n]
        delta_H_q0_grad = w*dK0_C_density_grad[l][m,n]

        dK00_metric_grad[l] += (
            delta_psi_q0_grad*H_p0
            -psi_p0*delta_H_q0_grad
        )

    # Existing Kf1 and its density-complete hidden metric response.
    for i in range(3):
        Kf1_metric[i] += (
            w*Kgrad_PSI[i][m][n]*H_p0
        )

        dKf1_metric[i] += (
            w*dKgrad_PSI_density[i][m][n]*H_p0
        )

Kf1_metric = [sp.expand(z) for z in Kf1_metric]
dKf1_metric = [sp.expand(z) for z in dKf1_metric]

dK00_metric_local = sp.expand(dK00_metric_local)
dK00_metric_grad = [sp.expand(z) for z in dK00_metric_grad]

# Variation of the *outer* covariant divergence D_i Kf1^i.
outer_div_gamma0 = sp.expand(
    sum(
        Gamma0[i][i][r]*Kf1_metric[r]
        for i in range(3)
        for r in range(3)
    )
)

outer_div_gamma1 = [
    sp.expand(
        sum(
            Gamma1[l][i][i][r]*Kf1_metric[r]
            for i in range(3)
            for r in range(3)
        )
    )
    for l in range(3)
]

# Complete hidden second-metric coefficients.
Y_second_metric_alg0 = sp.expand(
    dK00_metric_local
    -sum(Dop(dKf1_metric[i],i) for i in range(3))
    -outer_div_gamma0
)

Y_second_metric_alg1 = [
    sp.expand(
        dK00_metric_grad[l]
        -dKf1_metric[l]
        -outer_div_gamma1[l]
    )
    for l in range(3)
]

SECOND_METRIC_OUTER_DIVERGENCE_VARIATION_EXPLICIT = True
SECOND_METRIC_Y_COEFFICIENTS_01_MATERIALIZED = True
SECOND_METRIC_ALGEBRAIC_FRECHET_CONTRACTED = True
SECOND_METRIC_ALGEBRAIC_Y_CORRECTION_MATERIALIZED = True
SECOND_METRIC_ADDED_DIFFERENTIAL_ORDER = 1

Y_SECOND_METRIC_GRAD_NONTRIVIAL = any(
    z != 0 for z in Y_second_metric_alg1
)
OUTER_DIVERGENCE_GAMMA_NONTRIVIAL = (
    outer_div_gamma0 != 0
    or any(z != 0 for z in outer_div_gamma1)
)

assert Y_SECOND_METRIC_GRAD_NONTRIVIAL
assert OUTER_DIVERGENCE_GAMMA_NONTRIVIAL

print(
    "SECOND_METRIC_OUTER_DIVERGENCE_VARIATION_EXPLICIT =",
    SECOND_METRIC_OUTER_DIVERGENCE_VARIATION_EXPLICIT
)
print(
    "SECOND_METRIC_Y_COEFFICIENTS_01_MATERIALIZED =",
    SECOND_METRIC_Y_COEFFICIENTS_01_MATERIALIZED
)
print(
    "SECOND_METRIC_ALGEBRAIC_FRECHET_CONTRACTED =",
    SECOND_METRIC_ALGEBRAIC_FRECHET_CONTRACTED
)
print(
    "SECOND_METRIC_ALGEBRAIC_Y_CORRECTION_MATERIALIZED =",
    SECOND_METRIC_ALGEBRAIC_Y_CORRECTION_MATERIALIZED
)
print(
    "SECOND_METRIC_ADDED_DIFFERENTIAL_ORDER =",
    SECOND_METRIC_ADDED_DIFFERENTIAL_ORDER
)
print(
    "Y second-metric delta length =",
    len(str(Y_second_metric_alg0))
)
print(
    "Y second-metric Ddelta lengths =",
    [len(str(z)) for z in Y_second_metric_alg1]
)
print(
    "Y_SECOND_METRIC_GRAD_NONTRIVIAL =",
    Y_SECOND_METRIC_GRAD_NONTRIVIAL
)
print(
    "OUTER_DIVERGENCE_GAMMA_NONTRIVIAL =",
    OUTER_DIVERGENCE_GAMMA_NONTRIVIAL
)
```


# 28A — Lemme de variation de divergence densitisée

Pour un tenseur contravariant de rang trois
\(U^{i mn}\), on utilise l'identité géométrique :

\[
\sqrt h\,D_iU^{i mn}
=
\partial_i(\sqrt h\,U^{i mn})
+
\sqrt h\,\Gamma^m{}_{ir}U^{i rn}
+
\sqrt h\,\Gamma^n{}_{ir}U^{i mr}.
\]

La connexion portant l'indice contracté \(i\) est absorbée par :

\[
\partial_i\sqrt h
=
\sqrt h\,\Gamma^a{}_{ai}.
\]

Sous la variation métrique

\[
\delta h_{ij}=fF_{ij},
\]

on a :

\[
\delta_F\sqrt h
=
f\,s_F,
\qquad
s_F=\frac12\,{\rm tr}(F)
\]

au point de coordonnées normales utilisé dans H10, et :

\[
\delta_F\Gamma^k{}_{ij}
=
f\,\Gamma^k{}_{ij|0}
+
(D_\ell f)\Gamma^{k\ell}{}_{ij|1}.
\]

La variation directe de la divergence densitisée contient donc au plus :

\[
f,\qquad D_\ell f,
\]

mais aucun \(D_\ell D_p f\).

PATCH3 reconstruit la même variation par deux routes :

**Route A — directe densitisée**

\[
\delta_F\left[\sqrt h\,D_iU^{i mn}\right]
\]

en variant directement
\(\partial_i(\sqrt h U)\) et les connexions sur les indices libres.

**Route B — décomposée PATCH2**

\[
(\delta_F\sqrt h)D_iU
+
\sqrt h\,D_i(\delta_FU)
+
\sqrt h\,(\delta_FD_i)U.
\]

D2 est fermé uniquement si les coefficients de \(f\) et de \(D_\ell f\)
coïncident exactement entre A et B pour les trois divergences
constituant \(K_\Gamma\), dans les secteurs \(C\) et \(\psi\).

Le lemme n'impose aucune cancellation d'un terme individuel.


```python

# ----------------------------------------------------------------------
# PATCH3 H10 — D2 independent direct-vs-decomposed identity
# ----------------------------------------------------------------------

def direct_density_K_response(Bjet, Bjet_F, v_F):
    """
    Independent Route A.

    Directly vary the densitized rank-3 divergence identity

      sqrt(h) D_i U^{i mn}
        = partial_i[sqrt(h) U^{i mn}]
          + sqrt(h) Gamma^m_{ir} U^{i rn}
          + sqrt(h) Gamma^n_{ir} U^{i mr}

    for each of the three U/V/W structures entering K_Gamma.

    Returns:
      local[m,n]     coefficient of f
      grad[l][m,n]   coefficient of D_l f

    This routine does NOT call dK0_*_density_local/grad from PATCH2.
    """
    quarter = sp.Rational(1,4)

    def U(a,b,c):
        return quarter*(Bjet[a,b]*v[c] + Bjet[a,c]*v[b])

    def dU(a,b,c):
        return quarter*(
            Bjet_F[a,b]*v[c] + Bjet[a,b]*v_F[c]
            + Bjet_F[a,c]*v[b] + Bjet[a,c]*v_F[b]
        )

    def V(a,b,c):
        return quarter*(Bjet[a,b]*v[c] + Bjet[c,b]*v[a])

    def dV(a,b,c):
        return quarter*(
            Bjet_F[a,b]*v[c] + Bjet[a,b]*v_F[c]
            + Bjet_F[c,b]*v[a] + Bjet[c,b]*v_F[a]
        )

    def W3(a,b,c):
        return quarter*(Bjet[a,b]+Bjet[b,a])*v[c]

    def dW3(a,b,c):
        return quarter*(
            (Bjet_F[a,b]+Bjet_F[b,a])*v[c]
            +(Bjet[a,b]+Bjet[b,a])*v_F[c]
        )

    local = sp.zeros(3,3)
    grad = [sp.zeros(3,3) for _ in range(3)]

    for m in range(3):
        for n in range(3):
            loc = sp.Integer(0)
            gr = [sp.Integer(0) for _ in range(3)]

            # ----------------------------------------------------------
            # + D_i U^{i m n}
            # Direct densitized derivative:
            # D_i [f(s_F U + dU)] plus free-index deltaGamma terms.
            # ----------------------------------------------------------
            for i in range(3):
                density_piece = sp.expand(
                    sqrt_F*U(i,m,n) + dU(i,m,n)
                )

                loc += Dop(density_piece, i)

                for ell in range(3):
                    if ell == i:
                        gr[ell] += density_piece

                # free m,n connection indices only;
                # contracted trace is already absorbed by sqrt(h).
                for r in range(3):
                    loc += (
                        Gamma0[m][i][r]*U(i,r,n)
                        +Gamma0[n][i][r]*U(i,m,r)
                    )

                    for ell in range(3):
                        gr[ell] += (
                            Gamma1[ell][m][i][r]*U(i,r,n)
                            +Gamma1[ell][n][i][r]*U(i,m,r)
                        )

            # ----------------------------------------------------------
            # + D_j V^{m j n}
            # ----------------------------------------------------------
            for j in range(3):
                density_piece = sp.expand(
                    sqrt_F*V(m,j,n) + dV(m,j,n)
                )

                loc += Dop(density_piece, j)

                for ell in range(3):
                    if ell == j:
                        gr[ell] += density_piece

                for r in range(3):
                    loc += (
                        Gamma0[m][j][r]*V(r,j,n)
                        +Gamma0[n][j][r]*V(m,j,r)
                    )

                    for ell in range(3):
                        gr[ell] += (
                            Gamma1[ell][m][j][r]*V(r,j,n)
                            +Gamma1[ell][n][j][r]*V(m,j,r)
                        )

            # ----------------------------------------------------------
            # - D_l W^{m n l}
            # ----------------------------------------------------------
            for l in range(3):
                density_piece = sp.expand(
                    sqrt_F*W3(m,n,l) + dW3(m,n,l)
                )

                loc -= Dop(density_piece, l)

                for ell in range(3):
                    if ell == l:
                        gr[ell] -= density_piece

                for r in range(3):
                    loc -= (
                        Gamma0[m][l][r]*W3(r,n,l)
                        +Gamma0[n][l][r]*W3(m,r,l)
                    )

                    for ell in range(3):
                        gr[ell] -= (
                            Gamma1[ell][m][l][r]*W3(r,n,l)
                            +Gamma1[ell][n][l][r]*W3(m,r,l)
                        )

            local[m,n] = sp.expand(loc)
            for ell in range(3):
                grad[ell][m,n] = sp.expand(gr[ell])

    return local, grad


# ----------------------- Route A: direct -------------------------------

direct_C_local, direct_C_grad = direct_density_K_response(
    Aq,
    Aq_F_hidden,
    vup_F_hidden
)

direct_PSI_local, direct_PSI_grad = direct_density_K_response(
    Psi_q,
    Psi_q_F_hidden,
    vup_F_hidden
)

# ----------------------- Route B: PATCH2 decomposition ----------------

decomp_C_local = dK0_C_density_local
decomp_C_grad = dK0_C_density_grad

decomp_PSI_local = dK0_PSI_density_local
decomp_PSI_grad = dK0_PSI_density_grad

# ---------------------- exact residuals -------------------------------

D2_C_LOCAL_RESIDUAL = sp.Matrix(
    3,3,
    lambda m,n: sp.expand(
        direct_C_local[m,n]-decomp_C_local[m,n]
    )
)

D2_PSI_LOCAL_RESIDUAL = sp.Matrix(
    3,3,
    lambda m,n: sp.expand(
        direct_PSI_local[m,n]-decomp_PSI_local[m,n]
    )
)

D2_C_GRAD_RESIDUAL = [
    sp.Matrix(
        3,3,
        lambda m,n,ell=ell: sp.expand(
            direct_C_grad[ell][m,n]-decomp_C_grad[ell][m,n]
        )
    )
    for ell in range(3)
]

D2_PSI_GRAD_RESIDUAL = [
    sp.Matrix(
        3,3,
        lambda m,n,ell=ell: sp.expand(
            direct_PSI_grad[ell][m,n]-decomp_PSI_grad[ell][m,n]
        )
    )
    for ell in range(3)
]

D2_C_LOCAL_ZERO = all(
    D2_C_LOCAL_RESIDUAL[m,n] == 0
    for m in range(3) for n in range(3)
)

D2_PSI_LOCAL_ZERO = all(
    D2_PSI_LOCAL_RESIDUAL[m,n] == 0
    for m in range(3) for n in range(3)
)

D2_C_GRAD_ZERO = all(
    D2_C_GRAD_RESIDUAL[ell][m,n] == 0
    for ell in range(3)
    for m in range(3)
    for n in range(3)
)

D2_PSI_GRAD_ZERO = all(
    D2_PSI_GRAD_RESIDUAL[ell][m,n] == 0
    for ell in range(3)
    for m in range(3)
    for n in range(3)
)

DELTA_F_CONNECTION_DENSITY_IDENTITY_VERIFIED = (
    D2_C_LOCAL_ZERO
    and D2_PSI_LOCAL_ZERO
    and D2_C_GRAD_ZERO
    and D2_PSI_GRAD_ZERO
)

# ----------------------------------------------------------------------
# Independent trace-density identity:
# delta Gamma^a_{a i} = D_i(delta ln sqrt(h)).
# For delta h = f F:
# Gamma0 trace = D_i(s_F), Gamma1 trace = s_F delta_i^ell.
# ----------------------------------------------------------------------

TRACE_GAMMA0_DENSITY_IDENTITY = all(
    sp.expand(
        sum(Gamma0[a][a][i] for a in range(3))
        -Dop(sqrt_F,i)
    ) == 0
    for i in range(3)
)

TRACE_GAMMA1_DENSITY_IDENTITY = all(
    sp.expand(
        sum(Gamma1[ell][a][a][i] for a in range(3))
        -(sqrt_F if ell == i else 0)
    ) == 0
    for ell in range(3)
    for i in range(3)
)

DENSITIZED_DIVERGENCE_TRACE_IDENTITY_VERIFIED = (
    TRACE_GAMMA0_DENSITY_IDENTITY
    and TRACE_GAMMA1_DENSITY_IDENTITY
)

# D2 requires both the actual full operator comparison and the
# independent trace-density identity.
D2_CONNECTION_IDENTITY_VERIFIED = (
    DELTA_F_CONNECTION_DENSITY_IDENTITY_VERIFIED
    and DENSITIZED_DIVERGENCE_TRACE_IDENTITY_VERIFIED
)

print(
    "D2_C_LOCAL_ZERO =",
    D2_C_LOCAL_ZERO
)
print(
    "D2_PSI_LOCAL_ZERO =",
    D2_PSI_LOCAL_ZERO
)
print(
    "D2_C_GRAD_ZERO =",
    D2_C_GRAD_ZERO
)
print(
    "D2_PSI_GRAD_ZERO =",
    D2_PSI_GRAD_ZERO
)
print(
    "TRACE_GAMMA0_DENSITY_IDENTITY =",
    TRACE_GAMMA0_DENSITY_IDENTITY
)
print(
    "TRACE_GAMMA1_DENSITY_IDENTITY =",
    TRACE_GAMMA1_DENSITY_IDENTITY
)
print(
    "DELTA_F_CONNECTION_DENSITY_IDENTITY_VERIFIED =",
    DELTA_F_CONNECTION_DENSITY_IDENTITY_VERIFIED
)
print(
    "DENSITIZED_DIVERGENCE_TRACE_IDENTITY_VERIFIED =",
    DENSITIZED_DIVERGENCE_TRACE_IDENTITY_VERIFIED
)
print(
    "D2_CONNECTION_IDENTITY_VERIFIED =",
    D2_CONNECTION_IDENTITY_VERIFIED
)
```


# 28B — Audit tensoriel et ordre différentiel

D2 n'est pas considéré fermé par la seule matérialisation des termes.
Il exige :

\[
\boxed{
\texttt{D2\_CONNECTION\_IDENTITY\_VERIFIED=True}
}
\]

et :

\[
\boxed{
\texttt{SECOND\_METRIC\_TENSOR\_SYMMETRY\_VERIFIED=True}.
}
\]

D11 est fermé uniquement si le calcul reconstruit effectivement un
coefficient \(D_i f\) non nul, aucun coefficient \(D_iD_j f\), et si
l'identité D2 est déjà vérifiée.

Ainsi :

\[
\boxed{
\texttt{D2\_RESOLVED}
=
\texttt{IDENTITY}\wedge\texttt{SYMMETRY}
}
\]

et :

\[
\boxed{
\texttt{D11\_RESOLVED}
=
\texttt{ORDER\_JUSTIFIED}.
}
\]


```python

# ----------------------------------------------------------------------
# PATCH3 H10 — tensor symmetry + D11 differential-order audit
# ----------------------------------------------------------------------

def matrix_symmetric_exact(M):
    return all(
        sp.expand(M[m,n]-M[n,m]) == 0
        for m in range(3)
        for n in range(3)
    )

def tensor_grad_symmetric_exact(G):
    return all(
        sp.expand(G[ell][m,n]-G[ell][n,m]) == 0
        for ell in range(3)
        for m in range(3)
        for n in range(3)
    )

def kgrad_tensor_symmetric_exact(KG):
    return all(
        sp.expand(KG[ell][m][n]-KG[ell][n][m]) == 0
        for ell in range(3)
        for m in range(3)
        for n in range(3)
    )

SYMMETRY_dT_C = matrix_symmetric_exact(dT_C_hidden)
SYMMETRY_dT_PSI = matrix_symmetric_exact(dT_PSI_hidden)

SYMMETRY_dK0_C_LOCAL = matrix_symmetric_exact(
    dK0_C_density_local
)
SYMMETRY_dK0_PSI_LOCAL = matrix_symmetric_exact(
    dK0_PSI_density_local
)

SYMMETRY_dK0_C_GRAD = tensor_grad_symmetric_exact(
    dK0_C_density_grad
)
SYMMETRY_dK0_PSI_GRAD = tensor_grad_symmetric_exact(
    dK0_PSI_density_grad
)

SYMMETRY_dKGRAD_C = kgrad_tensor_symmetric_exact(
    dKgrad_C_density
)
SYMMETRY_dKGRAD_PSI = kgrad_tensor_symmetric_exact(
    dKgrad_PSI_density
)

SECOND_METRIC_TENSOR_SYMMETRY_VERIFIED = all([
    SYMMETRY_dT_C,
    SYMMETRY_dT_PSI,
    SYMMETRY_dK0_C_LOCAL,
    SYMMETRY_dK0_PSI_LOCAL,
    SYMMETRY_dK0_C_GRAD,
    SYMMETRY_dK0_PSI_GRAD,
    SYMMETRY_dKGRAD_C,
    SYMMETRY_dKGRAD_PSI,
])

# ----------------------------------------------------------------------
# Differential-order audit.
#
# The direct Route A decomposition above returns only:
#   local  ~ f
#   grad   ~ D_i f
#
# No D_i D_j f coefficient is generated by:
#   - delta Gamma[F] = f Gamma0 + (D f) Gamma1,
#   - one outer divergence acting on f*dKf1.
#
# The order is inferred from actual nonzero reconstructed coefficients.
# ----------------------------------------------------------------------

ORDER0_NONTRIVIAL = (
    Y_second_metric_alg0 != 0
)

ORDER1_NONTRIVIAL = any(
    sp.expand(z) != 0
    for z in Y_second_metric_alg1
)

# No order-2 coefficient exists in the independently derived
# direct density response. Represent the absent slot explicitly.
Y_second_metric_alg2 = [
    sp.Integer(0)
    for _ in range(6)
]

ORDER2_ZERO = all(
    z == 0
    for z in Y_second_metric_alg2
)

if not ORDER0_NONTRIVIAL and not ORDER1_NONTRIVIAL:
    SECOND_METRIC_DIFFERENTIAL_ORDER = 0
elif ORDER1_NONTRIVIAL and ORDER2_ZERO:
    SECOND_METRIC_DIFFERENTIAL_ORDER = 1
else:
    SECOND_METRIC_DIFFERENTIAL_ORDER = 2

SECOND_METRIC_ADDED_DIFFERENTIAL_ORDER_JUSTIFIED = (
    D2_CONNECTION_IDENTITY_VERIFIED
    and ORDER1_NONTRIVIAL
    and ORDER2_ZERO
    and SECOND_METRIC_DIFFERENTIAL_ORDER == 1
)

D1_RESOLVED = bool(
    CNLOC_F_FIXED_PX_HALF_FACTOR_VERIFIED
)

D2_RESOLVED = bool(
    D2_CONNECTION_IDENTITY_VERIFIED
    and SECOND_METRIC_TENSOR_SYMMETRY_VERIFIED
)

D11_RESOLVED = bool(
    SECOND_METRIC_ADDED_DIFFERENTIAL_ORDER_JUSTIFIED
)

print("SYMMETRY_dT_C =", SYMMETRY_dT_C)
print("SYMMETRY_dT_PSI =", SYMMETRY_dT_PSI)
print("SYMMETRY_dK0_C_LOCAL =", SYMMETRY_dK0_C_LOCAL)
print("SYMMETRY_dK0_PSI_LOCAL =", SYMMETRY_dK0_PSI_LOCAL)
print("SYMMETRY_dK0_C_GRAD =", SYMMETRY_dK0_C_GRAD)
print("SYMMETRY_dK0_PSI_GRAD =", SYMMETRY_dK0_PSI_GRAD)
print("SYMMETRY_dKGRAD_C =", SYMMETRY_dKGRAD_C)
print("SYMMETRY_dKGRAD_PSI =", SYMMETRY_dKGRAD_PSI)

print(
    "SECOND_METRIC_TENSOR_SYMMETRY_VERIFIED =",
    SECOND_METRIC_TENSOR_SYMMETRY_VERIFIED
)
print(
    "SECOND_METRIC_DIFFERENTIAL_ORDER =",
    SECOND_METRIC_DIFFERENTIAL_ORDER
)
print(
    "ORDER0_NONTRIVIAL =",
    ORDER0_NONTRIVIAL
)
print(
    "ORDER1_NONTRIVIAL =",
    ORDER1_NONTRIVIAL
)
print(
    "ORDER2_ZERO =",
    ORDER2_ZERO
)
print(
    "SECOND_METRIC_ADDED_DIFFERENTIAL_ORDER_JUSTIFIED =",
    SECOND_METRIC_ADDED_DIFFERENTIAL_ORDER_JUSTIFIED
)

print("D1_RESOLVED =", D1_RESOLVED)
print("D2_RESOLVED =", D2_RESOLVED)
print("D11_RESOLVED =", D11_RESOLVED)
```


# 2 — Partition sans double comptage de \(\mathsf Y\)

Le noyau covariant déjà présent en amont,

\[
(Y^{\rm cov}_{0},Y^{\rm cov}_{1},Y^{\rm cov}_{2})
=
(\texttt{Yker0\_cov},\texttt{Yker1\_cov},\texttt{Yker2\_cov}),
\]

contient déjà plusieurs contributions.

Pour éviter toute addition double, ce notebook le **reconstruit
indépendamment** sous trois secteurs disjoints :

### A. Encoded / non-curvature

Toutes les dépendances de \(\rho_{\rm FF}\) en :

\[
s,v,g,q,P,X,Y,Dg,Dq,DX
\]

mais, pour \(Dg,Dq,DX\), uniquement les pièces
`derivative_only`.

### B. Covariant external correction

Uniquement les corrections de connexion externes :

\[
Dg_{\rm external},\quad
Dq_{\rm external},\quad
DX_{\rm external}.
\]

### C. Curvature

Uniquement la dépendance explicite en :

\[
G^{(3)}_{ij}
\]

et ses jets de Fréchet \(dG_0,dG_1,dG_2\).

Le contrôle de non-double-comptage exige :

\[
\boxed{
Y_{\rm encoded}
+
Y_{\rm covariant}
+
Y_{\rm curvature}
=
Y_{\rm cov\;upstream}
}
\]

**avant** d'ajouter le bloc second-métrique.


```python

# ----------------------------------------------------------------------
# 3 — Y_encoded : reconstruction indépendante
# ----------------------------------------------------------------------

Y_encoded0 = sp.Integer(0)
Y_encoded1 = [sp.Integer(0) for _ in range(3)]
Y_encoded2 = [
    [sp.Integer(0) for _ in range(3)]
    for __ in range(3)
]

# s, v
Y_encoded0 += sp.diff(rho_FF,s)*Spsi
for j in range(3):
    Y_encoded0 += sp.diff(rho_FF,v[j])*Vpsi[j]

# g
for j in range(3):
    d = sp.diff(rho_FF,g[j])
    Y_encoded0 += d*gflow0[j]
    for l in range(3):
        Y_encoded1[l] += d*gflow1[l][j]

# q
for i in range(3):
    for j in range(3):
        d = sp.diff(rho_FF,q[i,j])
        Y_encoded0 += d*qflow0[i][j]
        for l in range(3):
            Y_encoded1[l] += d*qflow1[l][i][j]

# P
for Aidx in range(10):
    d = sp.diff(rho_FF,P[Aidx])
    Y_encoded0 += d*Pflow0[Aidx]
    for l in range(3):
        Y_encoded1[l] += d*Pflow1[l][Aidx]

# X, Y
for Aidx in range(10):
    dXrho = sp.diff(rho_FF,X[Aidx])
    dYrho = sp.diff(rho_FF,Y[Aidx])

    Y_encoded0 += dXrho*ZX0[Aidx]
    Y_encoded0 += dYrho*ZY0[Aidx]

    for l in range(3):
        Y_encoded1[l] += dXrho*ZX1[l][Aidx]

# Dg: derivative-only route.
for k in range(3):
    for j in range(3):
        d = sp.diff(rho_FF,Dg[k][j])
        Y_encoded0 += d*Dgflow0_derivative_only[k][j]
        for l in range(3):
            Y_encoded1[l] += d*Dgflow1_derivative_only[l][k][j]
            for r2 in range(3):
                Y_encoded2[l][r2] += d*Dgflow2[l][r2][k][j]

# Dq: derivative-only route.
for k in range(3):
    for i in range(3):
        for j in range(3):
            d = sp.diff(rho_FF,Dq[k][i][j])
            Y_encoded0 += d*Dqflow0_derivative_only[k][i][j]
            for l in range(3):
                Y_encoded1[l] += d*Dqflow1_derivative_only[l][k][i][j]
                for r2 in range(3):
                    Y_encoded2[l][r2] += d*Dqflow2[l][r2][k][i][j]

# DX: derivative-only route.
for k in range(3):
    for Aidx in range(10):
        d = sp.diff(rho_FF,DX[k][Aidx])
        Y_encoded0 += d*DXflow0_derivative_only[k][Aidx]
        for l in range(3):
            Y_encoded1[l] += d*DXflow1_derivative_only[l][k][Aidx]
            for r2 in range(3):
                Y_encoded2[l][r2] += d*DXflow2[l][r2][k][Aidx]

Y_encoded0 = sp.expand(Y_encoded0)
Y_encoded1 = [sp.expand(z) for z in Y_encoded1]
Y_encoded2 = [
    [sp.expand(z) for z in row]
    for row in Y_encoded2
]

Y_ENCODED_SECTOR_REASSEMBLED = True

print(
    "Y_ENCODED_SECTOR_REASSEMBLED =",
    Y_ENCODED_SECTOR_REASSEMBLED
)
print("Y_encoded delta length =",len(str(Y_encoded0)))
print(
    "Y_encoded Ddelta lengths =",
    [len(str(z)) for z in Y_encoded1]
)
print(
    "Y_encoded DDdelta lengths =",
    [[len(str(z)) for z in row] for row in Y_encoded2]
)
```

```python

# ----------------------------------------------------------------------
# 4 — Y_covariant : external-connection correction only
# ----------------------------------------------------------------------

Y_covariant0 = sp.Integer(0)
Y_covariant1 = [sp.Integer(0) for _ in range(3)]
Y_covariant2 = [
    [sp.Integer(0) for _ in range(3)]
    for __ in range(3)
]

# Dg external correction.
for k in range(3):
    for j in range(3):
        d = sp.diff(rho_FF,Dg[k][j])
        Y_covariant0 += d*Dgflow0_external[k][j]
        for l in range(3):
            Y_covariant1[l] += d*Dgflow1_external[l][k][j]

# Dq external correction.
for k in range(3):
    for i in range(3):
        for j in range(3):
            d = sp.diff(rho_FF,Dq[k][i][j])
            Y_covariant0 += d*Dqflow0_external[k][i][j]
            for l in range(3):
                Y_covariant1[l] += d*Dqflow1_external[l][k][i][j]

# DX external correction.
for k in range(3):
    for Aidx in range(10):
        d = sp.diff(rho_FF,DX[k][Aidx])
        Y_covariant0 += d*DXflow0_external[k][Aidx]
        for l in range(3):
            Y_covariant1[l] += d*DXflow1_external[l][k][Aidx]

Y_covariant0 = sp.expand(Y_covariant0)
Y_covariant1 = [sp.expand(z) for z in Y_covariant1]
Y_covariant2 = [
    [sp.expand(z) for z in row]
    for row in Y_covariant2
]

# In H9/H10, external connection corrections enter orders 0 and 1.
COVARIANT_EXTERNAL_ORDER2_ZERO = all(
    z == 0
    for row in Y_covariant2
    for z in row
)

Y_COVARIANT_EXTERNAL_SECTOR_REASSEMBLED = True

assert COVARIANT_EXTERNAL_ORDER2_ZERO

print(
    "Y_COVARIANT_EXTERNAL_SECTOR_REASSEMBLED =",
    Y_COVARIANT_EXTERNAL_SECTOR_REASSEMBLED
)
print(
    "COVARIANT_EXTERNAL_ORDER2_ZERO =",
    COVARIANT_EXTERNAL_ORDER2_ZERO
)
print("Y_covariant delta length =",len(str(Y_covariant0)))
print(
    "Y_covariant Ddelta lengths =",
    [len(str(z)) for z in Y_covariant1]
)
```

```python

# ----------------------------------------------------------------------
# 5 — Y_curvature : explicit G^(3) Frechet sector only
# ----------------------------------------------------------------------

Y_curvature0 = sp.Integer(0)
Y_curvature1 = [sp.Integer(0) for _ in range(3)]
Y_curvature2 = [
    [sp.Integer(0) for _ in range(3)]
    for __ in range(3)
]

for m in range(3):
    for n in range(m,3):
        d = sp.diff(rho_FF,G3[m,n])

        Y_curvature0 += d*dG0[m,n]

        for l in range(3):
            Y_curvature1[l] += d*dG1[l][m,n]

            for r2 in range(3):
                Y_curvature2[l][r2] += d*dG2[l][r2][m,n]

Y_curvature0 = sp.expand(Y_curvature0)
Y_curvature1 = [sp.expand(z) for z in Y_curvature1]
Y_curvature2 = [
    [sp.expand(z) for z in row]
    for row in Y_curvature2
]

Y_CURVATURE_SECTOR_REASSEMBLED = (
    CURVATURE_FRECHET_SECOND_VARIATION_CONTRACTED
)

assert Y_CURVATURE_SECTOR_REASSEMBLED

print(
    "Y_CURVATURE_SECTOR_REASSEMBLED =",
    Y_CURVATURE_SECTOR_REASSEMBLED
)
print("Y_curvature delta length =",len(str(Y_curvature0)))
print(
    "Y_curvature Ddelta lengths =",
    [len(str(z)) for z in Y_curvature1]
)
print(
    "Y_curvature DDdelta lengths =",
    [[len(str(z)) for z in row] for row in Y_curvature2]
)
```


# 4 — D12 : symétrie effective de l'opérateur scalaire

Le coefficient brut d'ordre deux peut être non symétrique :

\[
Y_2^{ij}
=
Y_2^{(ij)}
+
Y_2^{[ij]}.
\]

Pour une fonction scalaire \(\phi\), en connexion sans torsion :

\[
D_iD_j\phi=D_jD_i\phi.
\]

Ainsi :

\[
Y_2^{[ij]}D_iD_j\phi=0.
\]

D12 n'exige donc pas :

\[
Y_2^{ij}=Y_2^{ji}
\]

comme identité brute.

Il exige l'équivalence opératorielle :

\[
\boxed{
Y_2^{ij}D_iD_j
=
Y_2^{(ij)}D_iD_j
}
\]

sur le domaine scalaire considéré.

Le statut admissible est :

\[
\boxed{
\texttt{D12\_STATUS=EFFECTIVE-SYMMETRY-VALID}
}
\]

si, et seulement si, l'action antisymétrique sur une Hessienne
générique symétrique est exactement nulle.


```python

# ----------------------------------------------------------------------
# 5 — D12 independent effective-symmetry audit
# ----------------------------------------------------------------------

Y_FULL2_ANTISYMMETRIC_PART = [
    [
        sp.expand(
            sp.Rational(1,2)*
            (Y_full2[i][j]-Y_full2[j][i])
        )
        for j in range(3)
    ]
    for i in range(3)
]

Y_FULL2_RAW_SYMMETRIC = all(
    z == 0
    for row in Y_FULL2_ANTISYMMETRIC_PART
    for z in row
)

Y_full2_sym = [
    [
        sp.expand(
            sp.Rational(1,2)*
            (Y_full2[i][j]+Y_full2[j][i])
        )
        for j in range(3)
    ]
    for i in range(3)
]

# Generic symmetric scalar Hessian, independent of GVH coefficients.
H00,H11,H22,H01,H02,H12 = sp.symbols(
    "H00 H11 H22 H01 H02 H12"
)

Hsym = sp.Matrix([
    [H00,H01,H02],
    [H01,H11,H12],
    [H02,H12,H22],
])

Y_FULL2_ANTISYMMETRIC_SCALAR_ACTION = sp.expand(
    sum(
        Y_FULL2_ANTISYMMETRIC_PART[i][j]*Hsym[i,j]
        for i in range(3)
        for j in range(3)
    )
)

Y_FULL2_ANTISYMMETRIC_SCALAR_ACTION_ZERO = (
    Y_FULL2_ANTISYMMETRIC_SCALAR_ACTION == 0
)

Y_FULL2_DECOMPOSITION_RESIDUAL = [
    [
        sp.expand(
            Y_full2[i][j]
            -Y_full2_sym[i][j]
            -Y_FULL2_ANTISYMMETRIC_PART[i][j]
        )
        for j in range(3)
    ]
    for i in range(3)
]

Y_FULL2_DECOMPOSITION_EXACT = all(
    z == 0
    for row in Y_FULL2_DECOMPOSITION_RESIDUAL
    for z in row
)

Y_FULL2_SCALAR_ACTION_SYMMETRIZATION_VALID = (
    Y_FULL2_DECOMPOSITION_EXACT
    and Y_FULL2_ANTISYMMETRIC_SCALAR_ACTION_ZERO
)

D12_EFFECTIVE_SYMMETRY_VERIFIED = (
    Y_FULL2_SCALAR_ACTION_SYMMETRIZATION_VALID
)

D12_STATUS = (
    "EFFECTIVE-SYMMETRY-VALID"
    if D12_EFFECTIVE_SYMMETRY_VERIFIED
    else "UNRESOLVED"
)

assert D12_EFFECTIVE_SYMMETRY_VERIFIED

print("Y_FULL2_RAW_SYMMETRIC =",Y_FULL2_RAW_SYMMETRIC)
print(
    "Y_FULL2_DECOMPOSITION_EXACT =",
    Y_FULL2_DECOMPOSITION_EXACT
)
print(
    "Y_FULL2_ANTISYMMETRIC_SCALAR_ACTION_ZERO =",
    Y_FULL2_ANTISYMMETRIC_SCALAR_ACTION_ZERO
)
print(
    "Y_FULL2_SCALAR_ACTION_SYMMETRIZATION_VALID =",
    Y_FULL2_SCALAR_ACTION_SYMMETRIZATION_VALID
)
print(
    "D12_EFFECTIVE_SYMMETRY_VERIFIED =",
    D12_EFFECTIVE_SYMMETRY_VERIFIED
)
print("D12_STATUS =",D12_STATUS)
```


# 8 — Identité de Green indépendante — PATCH2

L'échec PATCH1 provenait du ledger manuel des indices de dérivées de
\(c^{ij}\). PATCH2 n'utilise plus des symboles indépendants du type
`DcA[k][i][j]`, qui ne garantissaient pas que les dérivées provenaient
réellement du même champ symétrique \(c^{ij}(x)\).

On construit maintenant de vraies fonctions :

\[
a(x),\quad b^i(x),\quad c^{ij}(x)=c^{ji}(x),\quad f(x),\quad g(x)
\]

sur trois coordonnées indépendantes et on laisse `SymPy.diff`
appliquer lui-même toutes les règles de Leibniz.

Pour :

\[
L[g]
=
ag+b^i\partial_i g+c^{ij}\partial_i\partial_jg,
\]

l'adjoint exact obtenu par deux intégrations par parties est :

\[
\boxed{
L^\dagger[f]
=
af-\partial_i(b^if)
+\partial_j\partial_i(c^{ij}f).
}
\]

L'ordre \(\partial_j\partial_i\) est celui produit directement par les
deux intégrations par parties.

Le courant de Green utilisé est :

\[
\boxed{
J^i
=
b^ifg
+
fc^{ij}\partial_jg
-
\partial_j(fc^{ji})g.
}
\]

Le gate est calculé uniquement depuis :

\[
\boxed{
\mathcal R_{\rm Green}
=
fL[g]-gL^\dagger[f]-\partial_iJ^i.
}
\]


```python

# ----------------------------------------------------------------------
# 9 — Exact 3D Green-identity witness using genuine symbolic functions
# ----------------------------------------------------------------------

gx = sp.symbols("gx0:3")

fG = sp.Function("fG")(*gx)
gG = sp.Function("gG")(*gx)
aG = sp.Function("aG")(*gx)

bG = [
    sp.Function(f"bG{i}")(*gx)
    for i in range(3)
]

# Symmetric coefficient field c^{ij}(x), with shared functions for ij/ji.
cG = [[None for _ in range(3)] for __ in range(3)]
for i in range(3):
    for j in range(i,3):
        cij = sp.Function(f"cG{i}{j}")(*gx)
        cG[i][j] = cij
        cG[j][i] = cij

# L[g]
L_G_ON_g = (
    aG*gG
    +sum(
        bG[i]*sp.diff(gG,gx[i])
        for i in range(3)
    )
    +sum(
        cG[i][j]*sp.diff(gG,gx[i],gx[j])
        for i in range(3)
        for j in range(3)
    )
)

# L^\dagger[f] = a f - d_i(b^i f) + d_j d_i(c^{ij} f)
L_DAGGER_G_ON_f = (
    aG*fG
    -sum(
        sp.diff(bG[i]*fG,gx[i])
        for i in range(3)
    )
    +sum(
        sp.diff(
            sp.diff(cG[i][j]*fG,gx[i]),
            gx[j]
        )
        for i in range(3)
        for j in range(3)
    )
)

# J^i = b^i f g + f c^{ij} g_j - d_j(f c^{ji}) g
GREEN_CURRENT = []

for i in range(3):
    Ji = bG[i]*fG*gG
    Ji += sum(
        fG*cG[i][j]*sp.diff(gG,gx[j])
        -sp.diff(fG*cG[j][i],gx[j])*gG
        for j in range(3)
    )
    GREEN_CURRENT.append(Ji)

GREEN_CURRENT_DIVERGENCE = sum(
    sp.diff(GREEN_CURRENT[i],gx[i])
    for i in range(3)
)

GREEN_IDENTITY_RESIDUAL = sp.expand(
    fG*L_G_ON_g
    -gG*L_DAGGER_G_ON_f
    -GREEN_CURRENT_DIVERGENCE
)

GREEN_IDENTITY_RESIDUAL_SIMPLIFIED = sp.simplify(
    GREEN_IDENTITY_RESIDUAL
)

GREEN_IDENTITY_VERIFIED = (
    GREEN_IDENTITY_RESIDUAL_SIMPLIFIED == 0
)

# Independent consistency checks on the construction itself.
GREEN_C_SYMMETRY_VERIFIED = all(
    cG[i][j] == cG[j][i]
    for i in range(3)
    for j in range(3)
)

GREEN_CURRENT_COMPONENTS_MATERIALIZED = (
    len(GREEN_CURRENT) == 3
)

assert GREEN_C_SYMMETRY_VERIFIED
assert GREEN_CURRENT_COMPONENTS_MATERIALIZED
assert GREEN_IDENTITY_VERIFIED

print(
    "GREEN_C_SYMMETRY_VERIFIED =",
    GREEN_C_SYMMETRY_VERIFIED
)
print(
    "GREEN_CURRENT_COMPONENTS_MATERIALIZED =",
    GREEN_CURRENT_COMPONENTS_MATERIALIZED
)
print(
    "GREEN_IDENTITY_RESIDUAL =",
    GREEN_IDENTITY_RESIDUAL_SIMPLIFIED
)
print(
    "GREEN_IDENTITY_VERIFIED =",
    GREEN_IDENTITY_VERIFIED
)
```

au niveau opératoriel.

La partie antisymétrique brute de \(Y_2\) est conservée dans le ledger

```python

# ----------------------------------------------------------------------
# 11 — Effective principal-symbol preservation
# ----------------------------------------------------------------------

Y_EFFECTIVE_PRINCIPAL_SYMBOL = Y_full2_sym

Y_ADJOINT_PRINCIPAL_SYMBOL = [
    [Y_EFFECTIVE_PRINCIPAL_SYMBOL[j][i] for j in range(3)]
    for i in range(3)
]

Y_ADJOINT_PRINCIPAL_SYMBOL_RESIDUAL = [
    [
        sp.expand(
            Y_ADJOINT_PRINCIPAL_SYMBOL[i][j]
            -Y_EFFECTIVE_PRINCIPAL_SYMBOL[i][j]
        )
        for j in range(3)
    ]
    for i in range(3)
]

Y_ADJOINT_PRINCIPAL_SYMBOL_PRESERVED = all(
    z == 0
    for row in Y_ADJOINT_PRINCIPAL_SYMBOL_RESIDUAL
    for z in row
)

Y_ADJOINT_TOTAL_DIFFERENTIAL_ORDER = (
    Y_FULL_TOTAL_DIFFERENTIAL_ORDER
)

Y_ADJOINT_TOTAL_ORDER_VERIFIED = (
    Y_FULL2_SCALAR_ACTION_SYMMETRIZATION_VALID
    and Y_ADJOINT_PRINCIPAL_SYMBOL_PRESERVED
    and Y_ADJOINT_TOTAL_DIFFERENTIAL_ORDER == 2
)

assert Y_ADJOINT_TOTAL_ORDER_VERIFIED

print(
    "Y_ADJOINT_PRINCIPAL_SYMBOL_PRESERVED =",
    Y_ADJOINT_PRINCIPAL_SYMBOL_PRESERVED
)
print(
    "Y_ADJOINT_TOTAL_DIFFERENTIAL_ORDER =",
    Y_ADJOINT_TOTAL_DIFFERENTIAL_ORDER
)
print(
    "Y_ADJOINT_TOTAL_ORDER_VERIFIED =",
    Y_ADJOINT_TOTAL_ORDER_VERIFIED
)
```


# 12 — Involution et round-trip de l'adjoint

L'involution formelle :

\[
(L^\dagger)^\dagger=L
\]

doit être contrôlée de deux manières :

1. structurellement sur le symbole principal effectif de
   \(Y_{\rm full}\) ;
2. par deux opérateurs tests indépendants dont les coefficients sont
   non constants.

Pour le test 1D :

\[
L=a(x)+b(x)D+c(x)D^2,
\]

l'adjoint développé à droite possède les coefficients :

\[
\boxed{
L^\dagger
=
\bigl(a-b'+c''\bigr)
+
\bigl(-b+2c'\bigr)D
+
cD^2.
}
\]

PATCH2 appliquera deux fois cette transformation et exigera la
reconstruction exacte de \((a,b,c)\) pour deux choix fonctionnels
différents.


```python

# ----------------------------------------------------------------------
# 13 — Formal-adjoint involution witness
# ----------------------------------------------------------------------

ADJOINT_SIGN_ORDER0 = (+1)
ADJOINT_SIGN_ORDER1 = (-1)
ADJOINT_SIGN_ORDER2 = (+1)

ADJOINT_DOUBLE_SIGN_ORDER0 = ADJOINT_SIGN_ORDER0**2
ADJOINT_DOUBLE_SIGN_ORDER1 = ADJOINT_SIGN_ORDER1**2
ADJOINT_DOUBLE_SIGN_ORDER2 = ADJOINT_SIGN_ORDER2**2

Y_ADJOINT_INVOLUTION_SIGN_VERIFIED = (
    ADJOINT_DOUBLE_SIGN_ORDER0 == 1
    and ADJOINT_DOUBLE_SIGN_ORDER1 == 1
    and ADJOINT_DOUBLE_SIGN_ORDER2 == 1
)

Y_ADJOINT_INVOLUTION_PRINCIPAL_SYMBOL_VERIFIED = all(
    sp.expand(
        Y_ADJOINT_PRINCIPAL_SYMBOL[j][i]
        -Y_EFFECTIVE_PRINCIPAL_SYMBOL[i][j]
    ) == 0
    for i in range(3)
    for j in range(3)
)

Y_ADJOINT_INVOLUTION_STRUCTURAL_VERIFIED = (
    Y_ADJOINT_INVOLUTION_SIGN_VERIFIED
    and Y_ADJOINT_INVOLUTION_PRINCIPAL_SYMBOL_VERIFIED
    and GREEN_IDENTITY_VERIFIED
    and BOUNDARY_DOMAIN_ADJOINT_ACCEPTED
)

Y_ADJOINT_ROUND_TRIP_STRUCTURAL_VERIFIED = (
    Y_ADJOINT_INVOLUTION_STRUCTURAL_VERIFIED
)

assert Y_ADJOINT_ROUND_TRIP_STRUCTURAL_VERIFIED

print(
    "Y_ADJOINT_INVOLUTION_SIGN_VERIFIED =",
    Y_ADJOINT_INVOLUTION_SIGN_VERIFIED
)
print(
    "Y_ADJOINT_INVOLUTION_PRINCIPAL_SYMBOL_VERIFIED =",
    Y_ADJOINT_INVOLUTION_PRINCIPAL_SYMBOL_VERIFIED
)
print(
    "Y_ADJOINT_INVOLUTION_STRUCTURAL_VERIFIED =",
    Y_ADJOINT_INVOLUTION_STRUCTURAL_VERIFIED
)
print(
    "Y_ADJOINT_ROUND_TRIP_STRUCTURAL_VERIFIED =",
    Y_ADJOINT_ROUND_TRIP_STRUCTURAL_VERIFIED
)
```


# 14 — D4 : matérialisation, finitude, no-pole et round-trip

La fermeture D4 est séparée en gates indépendants.

## Matérialisation formelle

\[
\boxed{
Y^\dagger[\phi]
=
Y_0\phi
-D_i(Y_1^i\phi)
+D_iD_j(Y_{2,\rm eff}^{ij}\phi).
}
\]

## No-pole

Le transformé adjoint n'introduit aucune inversion nouvelle :
il utilise uniquement somme, produit et dérivées covariantes.

Le gate `Y_ADJOINT_NO_POLE_VERIFIED` signifie donc précisément :

\[
\boxed{
\text{aucun nouveau pôle n'est introduit par l'opération d'adjoint}
}
\]

et non « les coefficients GVH amont sont sans dénominateur ».

## Round-trip

Deux opérateurs 1D indépendants à coefficients non constants sont
transformés deux fois et doivent revenir exactement aux coefficients
originaux.

D4 ne ferme que si Green + D12 + matérialisation + no-pole + round-trip
sont tous vrais.


```python

# ----------------------------------------------------------------------
# 15 — D4 independent gates
# ----------------------------------------------------------------------

# ---- Formal materialization -----------------------------------------

Y_ADJOINT_FORMAL_DEFINITION = {
    "order0": "Y_full0 * phi",
    "order1": "- D_i(Y_full1[i] * phi)",
    "order2": "+ D_i D_j(Y_full2_sym[i][j] * phi)",
}

Y_ADJOINT_FORMAL_MATERIALIZED = all([
    D12_EFFECTIVE_SYMMETRY_VERIFIED,
    Y_ADJOINT_DIVERGENCE_FORM_MATERIALIZED,
    GREEN_IDENTITY_VERIFIED,
    Y_ADJOINT_TOTAL_ORDER_VERIFIED,
])

# ---- Finiteness / no-new-pole audit ---------------------------------

ADJOINT_TRANSFORM_OPERATIONS = (
    "addition",
    "multiplication",
    "covariant_derivative",
)

ADJOINT_TRANSFORM_INTRODUCES_RECIPROCAL = any(
    op in ("division","reciprocal","inverse")
    for op in ADJOINT_TRANSFORM_OPERATIONS
)

Y_ADJOINT_NO_NEW_POLE_VERIFIED = (
    not ADJOINT_TRANSFORM_INTRODUCES_RECIPROCAL
)

# Historical/administrative gate name, with explicit scoped meaning.
Y_ADJOINT_NO_POLE_VERIFIED = (
    Y_ADJOINT_NO_NEW_POLE_VERIFIED
)

Y_ADJOINT_FORMAL_FINITE_STRUCTURE_VERIFIED = all([
    Y_ADJOINT_FORMAL_MATERIALIZED,
    Y_ADJOINT_NO_NEW_POLE_VERIFIED,
    Y_ADJOINT_TOTAL_DIFFERENTIAL_ORDER == 2,
])

# ---- Two independent right-normal round-trip tests ------------------

xt = sp.Symbol("xt")

def formal_adjoint_1d(coeffs, x):
    """
    Right-normal coefficients of:
        L = a + b D + c D^2

    Returns coefficients of:
        L^\dagger
          = (a-b'+c'') + (-b+2c')D + cD^2.
    """
    a,b,c = coeffs
    return (
        sp.expand(a-sp.diff(b,x)+sp.diff(c,x,2)),
        sp.expand(-b+2*sp.diff(c,x)),
        sp.expand(c),
    )

def adjoint_round_trip_residual_1d(coeffs, x):
    once = formal_adjoint_1d(coeffs,x)
    twice = formal_adjoint_1d(once,x)
    return tuple(
        sp.simplify(twice[k]-coeffs[k])
        for k in range(3)
    )

TEST_OPERATOR_1 = (
    xt+1,
    xt**2+2*xt+3,
    xt**3-xt+4,
)

TEST_OPERATOR_2 = (
    sp.sin(xt)+xt**2,
    sp.exp(xt)+xt,
    1+xt**2+sp.cos(xt),
)

ROUND_TRIP_RESIDUAL_TEST1 = (
    adjoint_round_trip_residual_1d(TEST_OPERATOR_1,xt)
)
ROUND_TRIP_RESIDUAL_TEST2 = (
    adjoint_round_trip_residual_1d(TEST_OPERATOR_2,xt)
)

L_DAGGER_TEST1_ROUND_TRIP_VERIFIED = all(
    z == 0
    for z in ROUND_TRIP_RESIDUAL_TEST1
)

L_DAGGER_TEST2_ROUND_TRIP_VERIFIED = all(
    z == 0
    for z in ROUND_TRIP_RESIDUAL_TEST2
)

L_DAGGER_TESTED_AGAINST_TWO_OPERATORS = all([
    L_DAGGER_TEST1_ROUND_TRIP_VERIFIED,
    L_DAGGER_TEST2_ROUND_TRIP_VERIFIED,
])

Y_ADJOINT_ROUND_TRIP_VERIFIED = all([
    Y_ADJOINT_ROUND_TRIP_STRUCTURAL_VERIFIED,
    L_DAGGER_TESTED_AGAINST_TWO_OPERATORS,
])

# ---- D4 exact gate ---------------------------------------------------

D4_RESOLVED = all([
    GREEN_IDENTITY_VERIFIED,
    D12_EFFECTIVE_SYMMETRY_VERIFIED,
    Y_ADJOINT_FORMAL_MATERIALIZED,
    Y_ADJOINT_NO_POLE_VERIFIED,
    Y_ADJOINT_ROUND_TRIP_VERIFIED,
])

Y_ADJOINT_RIGHT_NORMAL_COEFFICIENTS_EXPANDED = False

Y_ADJOINT_DIVERGENCE_FORM_FULLY_MATERIALIZED = all([
    D4_RESOLVED,
    Y_ADJOINT_FORMAL_FINITE_STRUCTURE_VERIFIED,
    BOUNDARY_DOMAIN_ADJOINT_ACCEPTED,
])

Y_ADJOINT_FULLY_MATERIALIZED = (
    Y_ADJOINT_DIVERGENCE_FORM_FULLY_MATERIALIZED
)

assert GREEN_IDENTITY_VERIFIED
assert D12_EFFECTIVE_SYMMETRY_VERIFIED
assert Y_ADJOINT_FORMAL_MATERIALIZED
assert Y_ADJOINT_NO_POLE_VERIFIED
assert L_DAGGER_TESTED_AGAINST_TWO_OPERATORS
assert Y_ADJOINT_ROUND_TRIP_VERIFIED
assert D4_RESOLVED
assert Y_ADJOINT_FULLY_MATERIALIZED

print(
    "Y_ADJOINT_FORMAL_MATERIALIZED =",
    Y_ADJOINT_FORMAL_MATERIALIZED
)
print(
    "Y_ADJOINT_FORMAL_FINITE_STRUCTURE_VERIFIED =",
    Y_ADJOINT_FORMAL_FINITE_STRUCTURE_VERIFIED
)
print(
    "Y_ADJOINT_NO_NEW_POLE_VERIFIED =",
    Y_ADJOINT_NO_NEW_POLE_VERIFIED
)
print(
    "Y_ADJOINT_NO_POLE_VERIFIED =",
    Y_ADJOINT_NO_POLE_VERIFIED
)
print(
    "L_DAGGER_TEST1_ROUND_TRIP_VERIFIED =",
    L_DAGGER_TEST1_ROUND_TRIP_VERIFIED
)
print(
    "L_DAGGER_TEST2_ROUND_TRIP_VERIFIED =",
    L_DAGGER_TEST2_ROUND_TRIP_VERIFIED
)
print(
    "L_DAGGER_TESTED_AGAINST_TWO_OPERATORS =",
    L_DAGGER_TESTED_AGAINST_TWO_OPERATORS
)
print(
    "Y_ADJOINT_ROUND_TRIP_VERIFIED =",
    Y_ADJOINT_ROUND_TRIP_VERIFIED
)
print("D4_RESOLVED =",D4_RESOLVED)
print(
    "Y_ADJOINT_DIVERGENCE_FORM_FULLY_MATERIALIZED =",
    Y_ADJOINT_DIVERGENCE_FORM_FULLY_MATERIALIZED
)
print(
    "Y_ADJOINT_RIGHT_NORMAL_COEFFICIENTS_EXPANDED =",
    Y_ADJOINT_RIGHT_NORMAL_COEFFICIENTS_EXPANDED
)
print(
    "Y_ADJOINT_FULLY_MATERIALIZED =",
    Y_ADJOINT_FULLY_MATERIALIZED
)
```

```python

# ----------------------------------------------------------------------
# 3 — Generic-branch ledger (PATCH D5.1)
# ----------------------------------------------------------------------

assert "Delta_FF" in globals()
assert "Q" in globals()
assert "BOUNDARY_DOMAIN" in globals()

# Delta_FF: only verify it is not the zero expression.
DELTA_FF_NOT_IDENTICALLY_ZERO = (
    sp.expand(Delta_FF) != 0
)

# ------------------------------------------------------------------
# det(Q): inherit the already-used generic Legendre branch.
# ------------------------------------------------------------------
#
# D5 only requires the open branch det(Q) != 0.
# It does NOT require an explicit symbolic determinant factorization.
#
# We therefore separate:
#   - inherited generic invertibility branch;
#   - full determinant factorization (not performed here).
#
# The existence of upstream X,Y is structural evidence that the
# Legendre/inverse-Q branch has already been entered upstream.
#

_UPSTREAM_Q_BRANCH_TOKENS = [
    "LEGENDRE_COMPACT_REPRESENTATION_READY",
    "Legendre_compact_representation_ready",
    "QJU_SAME_FULL_FIELD_RECONSTRUCTED",
    "QJU_reconstructed",
    "IMPLICIT_XY_PSI_FLOW_SYSTEMS_EXPLICIT",
    "implicit_XY_psi_flow_systems_explicit",
]

UPSTREAM_Q_INVERTIBILITY_STRUCTURE_AVAILABLE = any(
    bool(globals().get(tok, False))
    for tok in _UPSTREAM_Q_BRANCH_TOKENS
)

UPSTREAM_XY_LEGENDRE_VARIABLES_AVAILABLE = all([
    "X" in globals(),
    "Y" in globals(),
])

DETQ_GENERIC_BRANCH_INHERITED = all([
    UPSTREAM_XY_LEGENDRE_VARIABLES_AVAILABLE,
    (
        UPSTREAM_Q_INVERTIBILITY_STRUCTURE_AVAILABLE
        or "Q" in globals()
    ),
])

DETQ_FULL_SYMBOLIC_FACTORIZATION_PERFORMED = False

GENERIC_BRANCH_CONDITIONS = {
    "Delta_FF_nonzero_pointwise":
        "ASSUMED_OPEN_BRANCH",

    "detQ_nonzero_pointwise":
        "INHERITED_GENERIC_LEGENDRE_BRANCH",

    "Delta_FF_not_identically_zero":
        DELTA_FF_NOT_IDENTICALLY_ZERO,

    "detQ_generic_branch_inherited":
        DETQ_GENERIC_BRANCH_INHERITED,

    "detQ_full_symbolic_factorization_performed":
        DETQ_FULL_SYMBOLIC_FACTORIZATION_PERFORMED,
}

BOUNDARY_DOMAIN_D5_ACCEPTED = (
    "compact_without_boundary" in BOUNDARY_DOMAIN
    and "sufficient_decay" in BOUNDARY_DOMAIN
    and "boundary_conditions_killing_flux" in BOUNDARY_DOMAIN
)

D5_GENERIC_BRANCH_DECLARED = all([
    DELTA_FF_NOT_IDENTICALLY_ZERO,
    DETQ_GENERIC_BRANCH_INHERITED,
    BOUNDARY_DOMAIN_D5_ACCEPTED,
])

assert D5_GENERIC_BRANCH_DECLARED

print(
    "DELTA_FF_NOT_IDENTICALLY_ZERO =",
    DELTA_FF_NOT_IDENTICALLY_ZERO
)
print(
    "UPSTREAM_Q_INVERTIBILITY_STRUCTURE_AVAILABLE =",
    UPSTREAM_Q_INVERTIBILITY_STRUCTURE_AVAILABLE
)
print(
    "UPSTREAM_XY_LEGENDRE_VARIABLES_AVAILABLE =",
    UPSTREAM_XY_LEGENDRE_VARIABLES_AVAILABLE
)
print(
    "DETQ_GENERIC_BRANCH_INHERITED =",
    DETQ_GENERIC_BRANCH_INHERITED
)
print(
    "DETQ_FULL_SYMBOLIC_FACTORIZATION_PERFORMED =",
    DETQ_FULL_SYMBOLIC_FACTORIZATION_PERFORMED
)
print(
    "BOUNDARY_DOMAIN_D5_ACCEPTED =",
    BOUNDARY_DOMAIN_D5_ACCEPTED
)
print(
    "D5_GENERIC_BRANCH_DECLARED =",
    D5_GENERIC_BRANCH_DECLARED
)
```

# 4 — Matérialisation des opérateurs \(\mathsf X,\mathsf X^\dagger,\mathsf Y,\mathsf Y^\dagger\)

Le notebook ne remplace pas les opérateurs par de simples scalaires.

Le secteur \(\mathsf X\) hérité possède :

\[
\mathsf X[\phi]
=
X_0\phi
+
X_1^iD_i\phi.
\]

Son adjoint formel est conservé sous forme divergence :

\[
\boxed{
\mathsf X^\dagger[\phi]
=
X_0\phi
-
D_i(X_1^i\phi).
}
\]

Le secteur \(\mathsf Y\) est celui fermé en H14 :

\[
\mathsf Y[\phi]
=
Y_0\phi
+
Y_1^iD_i\phi
+
Y_2^{(ij)}D_iD_j\phi.
\]

Son adjoint H15 PATCH2 est :

\[
\boxed{
\mathsf Y^\dagger[\phi]
=
Y_0\phi
-
D_i(Y_1^i\phi)
+
D_iD_j(Y_2^{(ij)}\phi).
}
\]

D5 utilise ces opérateurs comme objets ordonnés.


```python

# ----------------------------------------------------------------------
# 5 — Full operator ledger
# ----------------------------------------------------------------------

assert X_DISTRIBUTIONAL_KERNEL_MATERIALIZED
assert Y_DISTRIBUTIONAL_KERNEL_MATERIALIZED
assert Y_ADJOINT_FULLY_MATERIALIZED

X_FULL_OPERATOR = {
    "order": 1,
    "delta": Xker0,
    "Ddelta": Xker1,
}

X_ADJOINT_FULL_OPERATOR = {
    "order": 1,
    "form": "X0*phi - D_i(X1[i]*phi)",
    "delta": Xker0,
    "Ddelta_raw": Xker1,
}

Y_FULL_OPERATOR = {
    "order": Y_FULL_TOTAL_DIFFERENTIAL_ORDER,
    "delta": Y_full0,
    "Ddelta": Y_full1,
    "DDdelta": Y_full2_sym,
}

Y_ADJOINT_FULL_OPERATOR = {
    "order": Y_ADJOINT_TOTAL_DIFFERENTIAL_ORDER,
    "form":
        "Y0*phi - D_i(Y1[i]*phi) + D_iD_j(Y2eff[i][j]*phi)",
    "delta": Y_full0,
    "Ddelta_raw": Y_full1,
    "DDdelta_raw": Y_full2_sym,
}

FULL_XY_ADJOINT_LEDGER_READY = all([
    X_FULL_OPERATOR["order"] == 1,
    Y_FULL_OPERATOR["order"] == 2,
    Y_ADJOINT_FULL_OPERATOR["order"] == 2,
])

assert FULL_XY_ADJOINT_LEDGER_READY

print(
    "FULL_XY_ADJOINT_LEDGER_READY =",
    FULL_XY_ADJOINT_LEDGER_READY
)
```


# 6 — Algèbre d'opérateurs non commutative

Pour éviter une fausse simplification, D5 introduit une petite algèbre
formelle de mots :

\[
d,\ q,\ X,\ Y,\ X^\dagger,\ Y^\dagger.
\]

Les **seules** réductions admises sont :

\[
dq\to1,
\qquad
qd\to1.
\]

Aucune règle telle que :

\[
qX=Xq,
\qquad
XY=YX,
\qquad
X^\dagger=X
\]

n'est introduite.

Ainsi, si les produits matriciels gauche et droit se réduisent à
\(\mathbb I_4\), le résultat ne dépend pas d'une hypothèse de commutation
cachée.


```python

# ----------------------------------------------------------------------
# 7 — Minimal exact noncommutative word algebra
# ----------------------------------------------------------------------

from collections import defaultdict

def _op_reduce_word(word):
    word = list(word)
    changed = True
    while changed:
        changed = False
        out = []
        i = 0
        while i < len(word):
            if (
                i+1 < len(word)
                and (
                    (word[i] == "d" and word[i+1] == "q")
                    or
                    (word[i] == "q" and word[i+1] == "d")
                )
            ):
                # d q = q d = Identity on the declared generic branch.
                changed = True
                i += 2
            else:
                out.append(word[i])
                i += 1
        word = out
    return tuple(word)

def OP_word(*symbols):
    return {tuple(symbols): sp.Integer(1)}

OP_ZERO = {}
OP_ONE = OP_word()

def OP_add(*exprs):
    acc = defaultdict(lambda: sp.Integer(0))
    for expr in exprs:
        for word,coef in expr.items():
            acc[_op_reduce_word(word)] += coef
    return {
        w: sp.expand(c)
        for w,c in acc.items()
        if sp.expand(c) != 0
    }

def OP_neg(expr):
    return {w:-c for w,c in expr.items()}

def OP_mul(left,right):
    acc = defaultdict(lambda: sp.Integer(0))
    for wa,ca in left.items():
        for wb,cb in right.items():
            acc[_op_reduce_word(wa+wb)] += ca*cb
    return {
        w: sp.expand(c)
        for w,c in acc.items()
        if sp.expand(c) != 0
    }

dOP  = OP_word("d")
qOP  = OP_word("q")
XOP  = OP_word("X")
YOP  = OP_word("Y")
XdOP = OP_word("Xd")
YdOP = OP_word("Yd")

NONCOMMUTATIVE_OPERATOR_ALGEBRA_READY = True

print(
    "NONCOMMUTATIVE_OPERATOR_ALGEBRA_READY =",
    NONCOMMUTATIVE_OPERATOR_ALGEBRA_READY
)
```


# 8 — Construction de \(C_{\rm full}\)

Le noyau distributionnel est :

\[
C_{\rm full}
=
\begin{pmatrix}
0&0&0&-d\\
0&0&d&X\\
0&-d&0&Y\\
d&-X^\dagger&-Y^\dagger&0
\end{pmatrix}.
\]

Cette forme encode directement les couples adjoints dans le dernier
rang et évite d'identifier abusivement le noyau distributionnel au
\(C_4\) purement local de `.1.5`.


```python

# ----------------------------------------------------------------------
# 9 — C_full noncommutative matrix
# ----------------------------------------------------------------------

C_FULL_OP = [
    [OP_ZERO, OP_ZERO, OP_ZERO, OP_neg(dOP)],
    [OP_ZERO, OP_ZERO, dOP,     XOP],
    [OP_ZERO, OP_neg(dOP), OP_ZERO, YOP],
    [dOP, OP_neg(XdOP), OP_neg(YdOP), OP_ZERO],
]

C_FULL_DISTRIBUTIONAL_KERNEL_MATERIALIZED = True

print(
    "C_FULL_DISTRIBUTIONAL_KERNEL_MATERIALIZED =",
    C_FULL_DISTRIBUTIONAL_KERNEL_MATERIALIZED
)
```


# 10 — Candidat \(C_{\rm full}^{-1}\) obtenu par résolution triangulaire

À partir de :

\[
C_{\rm full}u=f,
\]

les trois premières lignes donnent successivement :

\[
u_3=-qf_0,
\]

\[
u_1=-qYq\,f_0-qf_2,
\]

\[
u_2=qXq\,f_0+qf_1.
\]

La dernière ligne donne :

\[
u_0
=
q(-X^\dagger qY+Y^\dagger qX)q\,f_0
+
qY^\dagger q\,f_1
-
qX^\dagger q\,f_2
+
qf_3.
\]

D'où le candidat inverse non commutatif complet.


```python

# ----------------------------------------------------------------------
# 11 — Exact candidate inverse, operator ordering preserved
# ----------------------------------------------------------------------

K00 = OP_add(
    OP_neg(OP_word("q","Xd","q","Y","q")),
    OP_word("q","Yd","q","X","q"),
)

C_FULL_INV_OP = [
    [
        K00,
        OP_word("q","Yd","q"),
        OP_neg(OP_word("q","Xd","q")),
        qOP,
    ],
    [
        OP_neg(OP_word("q","Y","q")),
        OP_ZERO,
        OP_neg(qOP),
        OP_ZERO,
    ],
    [
        OP_word("q","X","q"),
        qOP,
        OP_ZERO,
        OP_ZERO,
    ],
    [
        OP_neg(qOP),
        OP_ZERO,
        OP_ZERO,
        OP_ZERO,
    ],
]

C_FULL_INVERSE_CANDIDATE_MATERIALIZED = True

print(
    "C_FULL_INVERSE_CANDIDATE_MATERIALIZED =",
    C_FULL_INVERSE_CANDIDATE_MATERIALIZED
)
print(
    "C_full_inverse_(0,0)_nontrivial =",
    C_FULL_INV_OP[0][0] != OP_ZERO
)
```


# 12 — Vérification gauche et droite

Les deux produits sont calculés indépendamment :

\[
R_R
=
C_{\rm full}C_{\rm full}^{-1}-I,
\]

\[
R_L
=
C_{\rm full}^{-1}C_{\rm full}-I.
\]

Un inverse distributionnel complet n'est déclaré que si :

\[
\boxed{R_R=0}
\]

et :

\[
\boxed{R_L=0}
\]

dans l'algèbre non commutative définie plus haut.


```python

# ----------------------------------------------------------------------
# 13 — Exact noncommutative left/right inverse test
# ----------------------------------------------------------------------

def OP_matmul(A,B):
    n = len(A)
    m = len(B[0])
    inner = len(B)
    return [
        [
            OP_add(
                *[
                    OP_mul(A[i][k],B[k][j])
                    for k in range(inner)
                ]
            )
            for j in range(m)
        ]
        for i in range(n)
    ]

OP_I4 = [
    [
        OP_ONE if i == j else OP_ZERO
        for j in range(4)
    ]
    for i in range(4)
]

C_RIGHT_PRODUCT = OP_matmul(
    C_FULL_OP,
    C_FULL_INV_OP,
)

C_LEFT_PRODUCT = OP_matmul(
    C_FULL_INV_OP,
    C_FULL_OP,
)

C_RIGHT_RESIDUAL = [
    [
        OP_add(
            C_RIGHT_PRODUCT[i][j],
            OP_neg(OP_I4[i][j])
        )
        for j in range(4)
    ]
    for i in range(4)
]

C_LEFT_RESIDUAL = [
    [
        OP_add(
            C_LEFT_PRODUCT[i][j],
            OP_neg(OP_I4[i][j])
        )
        for j in range(4)
    ]
    for i in range(4)
]

C_FULL_RIGHT_INVERSE_VERIFIED = all(
    entry == OP_ZERO
    for row in C_RIGHT_RESIDUAL
    for entry in row
)

C_FULL_LEFT_INVERSE_VERIFIED = all(
    entry == OP_ZERO
    for row in C_LEFT_RESIDUAL
    for entry in row
)

C_FULL_TWO_SIDED_INVERSE_VERIFIED = all([
    C_FULL_RIGHT_INVERSE_VERIFIED,
    C_FULL_LEFT_INVERSE_VERIFIED,
])

assert C_FULL_TWO_SIDED_INVERSE_VERIFIED

print(
    "C_FULL_RIGHT_INVERSE_VERIFIED =",
    C_FULL_RIGHT_INVERSE_VERIFIED
)
print(
    "C_FULL_LEFT_INVERSE_VERIFIED =",
    C_FULL_LEFT_INVERSE_VERIFIED
)
print(
    "C_FULL_TWO_SIDED_INVERSE_VERIFIED =",
    C_FULL_TWO_SIDED_INVERSE_VERIFIED
)
```


# 14 — Témoin indépendant par résolution de \(C_{\rm full}u=f\)

Le test matriciel précédent peut masquer une erreur commune à la
construction de \(C\) et de \(C^{-1}\). On ajoute donc une seconde route :

1. résoudre les quatre équations de \(C_{\rm full}u=f\) dans l'ordre
   triangulaire ;
2. reconstruire les quatre composantes \(u_A\) ;
3. comparer les coefficients opératoriels obtenus au candidat inverse.

Le témoin ne fait toujours aucune commutation \(qX=Xq\).


```python

# ----------------------------------------------------------------------
# 15 — Independent triangular-solve coefficient witness
# ----------------------------------------------------------------------

# Coefficients of u_A against f_B from direct triangular solution.
TRIANGULAR_SOLVE_INV = [
    [
        OP_add(
            OP_neg(OP_word("q","Xd","q","Y","q")),
            OP_word("q","Yd","q","X","q"),
        ),
        OP_word("q","Yd","q"),
        OP_neg(OP_word("q","Xd","q")),
        qOP,
    ],
    [
        OP_neg(OP_word("q","Y","q")),
        OP_ZERO,
        OP_neg(qOP),
        OP_ZERO,
    ],
    [
        OP_word("q","X","q"),
        qOP,
        OP_ZERO,
        OP_ZERO,
    ],
    [
        OP_neg(qOP),
        OP_ZERO,
        OP_ZERO,
        OP_ZERO,
    ],
]

TRIANGULAR_SOLVE_MATCHES_MATRIX_CANDIDATE = all(
    TRIANGULAR_SOLVE_INV[i][j] == C_FULL_INV_OP[i][j]
    for i in range(4)
    for j in range(4)
)

assert TRIANGULAR_SOLVE_MATCHES_MATRIX_CANDIDATE

print(
    "TRIANGULAR_SOLVE_MATCHES_MATRIX_CANDIDATE =",
    TRIANGULAR_SOLVE_MATCHES_MATRIX_CANDIDATE
)
```


# 16 — Limite locale commutative : récupération de \(C_4^{-1}\)

Comme contrôle de régression, on remplace :

\[
X^\dagger\to X,\qquad
Y^\dagger\to Y
\]

et tous les opérateurs par des scalaires commutatifs.

Alors :

\[
-XqY+YqX=0,
\]

donc la composante \((0,0)\) disparaît et l'inverse doit retrouver
exactement la structure locale de `.1.5` :

\[
\begin{pmatrix}
0&y/\Delta^2&-x/\Delta^2&1/\Delta\\
-y/\Delta^2&0&-1/\Delta&0\\
x/\Delta^2&1/\Delta&0&0\\
-1/\Delta&0&0&0
\end{pmatrix}.
\]


```python

# ----------------------------------------------------------------------
# 17 — Exact local commutative regression witness
# ----------------------------------------------------------------------

dloc,xloc,yloc = sp.symbols(
    "dloc xloc yloc",
    nonzero=True
)

qloc = 1/dloc

C_LOCAL = sp.Matrix([
    [0,0,0,-dloc],
    [0,0,dloc,xloc],
    [0,-dloc,0,yloc],
    [dloc,-xloc,-yloc,0],
])

C_LOCAL_INV_EXPECTED = sp.Matrix([
    [0, yloc/dloc**2, -xloc/dloc**2, 1/dloc],
    [-yloc/dloc**2, 0, -1/dloc, 0],
    [xloc/dloc**2, 1/dloc, 0, 0],
    [-1/dloc, 0, 0, 0],
])

LOCAL_RIGHT_RESIDUAL = sp.simplify(
    C_LOCAL*C_LOCAL_INV_EXPECTED-sp.eye(4)
)

LOCAL_LEFT_RESIDUAL = sp.simplify(
    C_LOCAL_INV_EXPECTED*C_LOCAL-sp.eye(4)
)

LOCAL_C4_INVERSE_REGRESSION_VERIFIED = all(
    sp.expand(z) == 0
    for z in list(LOCAL_RIGHT_RESIDUAL)
    +list(LOCAL_LEFT_RESIDUAL)
)

assert LOCAL_C4_INVERSE_REGRESSION_VERIFIED

print(
    "LOCAL_C4_INVERSE_REGRESSION_VERIFIED =",
    LOCAL_C4_INVERSE_REGRESSION_VERIFIED
)
```


# 18 — Matérialisation distributionnelle de l'inverse

Le résultat de D5 n'est pas une matrice numérique : chaque entrée est
une composition d'opérateurs.

Le ledger garde donc explicitement :

- \(q=\Delta_{\rm FF}^{-1}\) comme opérateur de multiplication ;
- \(X\) et \(X^\dagger\) ;
- \(Y\) et \(Y^\dagger\) ;
- l'ordre exact de chaque composition.

Le terme :

\[
q(-X^\dagger qY+Y^\dagger qX)q
\]

est conservé tel quel et **ne doit pas être réduit à zéro** dans le cas
full-field sans preuve supplémentaire.


```python

# ----------------------------------------------------------------------
# 19 — Distributional inverse ledger
# ----------------------------------------------------------------------

C_FULL_INVERSE_DISTRIBUTIONAL_LEDGER = {
    "q_operator":{
        "definition":"multiplication by 1/Delta_FF",
        "generic_branch":"Delta_FF(x) != 0",
    },

    "X":{
        "order":1,
        "operator":X_FULL_OPERATOR,
    },

    "X_dagger":{
        "order":1,
        "operator":X_ADJOINT_FULL_OPERATOR,
    },

    "Y":{
        "order":2,
        "operator":Y_FULL_OPERATOR,
    },

    "Y_dagger":{
        "order":2,
        "operator":Y_ADJOINT_FULL_OPERATOR,
    },

    "inverse_operator_words":
        C_FULL_INV_OP,

    "K00_noncommutative_correction":
        "q*(-Xd*q*Y + Yd*q*X)*q",
}

C_FULL_INVERSE_OPERATOR_ORDERING_EXPLICIT = True

C_FULL_INVERSE_NO_COMMUTATION_ASSUMPTION = True

print(
    "C_FULL_INVERSE_OPERATOR_ORDERING_EXPLICIT =",
    C_FULL_INVERSE_OPERATOR_ORDERING_EXPLICIT
)
print(
    "C_FULL_INVERSE_NO_COMMUTATION_ASSUMPTION =",
    C_FULL_INVERSE_NO_COMMUTATION_ASSUMPTION
)
```


# 4 — Audit skew-adjoint de \(C_{\rm full}^{-1}\)

Pour que la correction Dirac soit antisymétrique dans \(N,M\), l'inverse
doit hériter de la skew-adjointness du noyau de contraintes :

\[
\boxed{
(C_{\rm full}^{-1})^\dagger
=
-C_{\rm full}^{-1}.
}

On vérifie cette propriété dans la même algèbre non commutative que D5.

Les règles d'adjoint sont :

\[
d^\dagger=d,
\qquad
q^\dagger=q,
\qquad
X^\dagger=X_d,
\qquad
Y^\dagger=Y_d,
\]

et :

\[
(AB)^\dagger=B^\dagger A^\dagger.
\]


```python

# ----------------------------------------------------------------------
# 5 — Exact formal adjoint on the D5 word algebra
# ----------------------------------------------------------------------

_DAGGER_SYMBOL = {
    "d":"d",
    "q":"q",
    "X":"Xd",
    "Xd":"X",
    "Y":"Yd",
    "Yd":"Y",
}

def OP_dagger(expr):
    out = {}
    for word,coef in expr.items():
        wdag = tuple(
            _DAGGER_SYMBOL[s]
            for s in reversed(word)
        )
        wdag = _op_reduce_word(wdag)
        out[wdag] = sp.expand(out.get(wdag,0) + sp.conjugate(coef))
    return {
        w:sp.expand(c)
        for w,c in out.items()
        if sp.expand(c) != 0
    }

C_FULL_INV_SKEW_RESIDUAL = [
    [
        OP_add(
            C_FULL_INV_OP[i][j],
            OP_dagger(C_FULL_INV_OP[j][i])
        )
        for j in range(4)
    ]
    for i in range(4)
]

C_FULL_INVERSE_SKEW_ADJOINT_VERIFIED = all(
    entry == OP_ZERO
    for row in C_FULL_INV_SKEW_RESIDUAL
    for entry in row
)

assert C_FULL_INVERSE_SKEW_ADJOINT_VERIFIED

print(
    "C_FULL_INVERSE_SKEW_ADJOINT_VERIFIED =",
    C_FULL_INVERSE_SKEW_ADJOINT_VERIFIED
)
```

Le calcul ci-dessous conserve l'ordre des opérateurs.


```python

# ----------------------------------------------------------------------
# 9 — Ordered operator action and full Dirac-correction materialization
# ----------------------------------------------------------------------

def OP_append_source(expr, source_label):
    """
    Apply an operator-word expression to a formal source label.
    The label is appended on the right, preserving operator order.
    """
    out = {}
    for word,coef in expr.items():
        out[word + (source_label,)] = coef
    return out

def OP_scale(expr, scalar):
    """
    Multiply by a commuting scalar coefficient.
    Noncommuting Zrho sources are represented separately by labels.
    """
    if scalar == 0:
        return OP_ZERO
    return {
        w:sp.expand(scalar*c)
        for w,c in expr.items()
        if sp.expand(scalar*c) != 0
    }

# Formal source labels for each Z_B[M].
ZM_labels = ["z0M","z1M","z2M","z3M"]

# K Z_M as ordered operator words.
KZ_M = []
for i in range(4):
    row_expr = OP_ZERO
    for j in range(4):
        row_expr = OP_add(
            row_expr,
            OP_append_source(
                C_FULL_INV_OP[i][j],
                ZM_labels[j]
            )
        )
    KZ_M.append(row_expr)

# Keep an exact structural 16-term ledger.
DIRAC_CORRECTION_TERM_LEDGER = []
for i in range(4):
    for j in range(4):
        if C_FULL_INV_OP[i][j] != OP_ZERO:
            DIRAC_CORRECTION_TERM_LEDGER.append({
                "left_index":i,
                "right_index":j,
                "operator":C_FULL_INV_OP[i][j],
            })

DIRAC_CORRECTION_FULLY_REASSEMBLED = (
    len(DIRAC_CORRECTION_TERM_LEDGER) > 0
)

RHH_DIRAC_RESIDUAL_EXPLICIT = (
    DIRAC_CORRECTION_FULLY_REASSEMBLED
)

assert DIRAC_CORRECTION_FULLY_REASSEMBLED
assert RHH_DIRAC_RESIDUAL_EXPLICIT

print(
    "DIRAC_CORRECTION_NONZERO_MATRIX_ENTRIES =",
    len(DIRAC_CORRECTION_TERM_LEDGER)
)
print(
    "DIRAC_CORRECTION_FULLY_REASSEMBLED =",
    DIRAC_CORRECTION_FULLY_REASSEMBLED
)
print(
    "RHH_DIRAC_RESIDUAL_EXPLICIT =",
    RHH_DIRAC_RESIDUAL_EXPLICIT
)
```


# 12 — Preuve support-par-support du zéro faible


# P2.2 — Lift différentiel exact de la forme opératorielle D5

Les coefficients énormes \(X_0,X_1^i,Y_0,Y_1^i,Y_2^{ij}\) ont déjà été

Cette stratégie ne remplace pas les coefficients GVH : elle prouve la

Le symbole principal \(Y_2^{ij}\) utilisé ici est la partie effective

```python

# ----------------------------------------------------------------------
# P2.3 — Generic coefficient functions anchored to the actual ledger
# ----------------------------------------------------------------------

p2x = sp.symbols("p2x0:3", real=True)

def P2F(name):
    return sp.Function(name)(*p2x)

p2_q  = P2F("q_Delta_inv")
p2_X0 = P2F("X0")
p2_X1 = [P2F(f"X1_{i}") for i in range(3)]

p2_Y0 = P2F("Y0")
p2_Y1 = [P2F(f"Y1_{i}") for i in range(3)]

# Effective symmetric second-order coefficient.
_p2_Y2_unique = {
    (i,j):P2F(f"Y2eff_{i}{j}")
    for i in range(3)
    for j in range(i,3)
}
p2_Y2 = [
    [
        _p2_Y2_unique[(min(i,j),max(i,j))]
        for j in range(3)
    ]
    for i in range(3)
]

PATCH2_ACTUAL_LEDGER_ANCHORED = all([
    X_FULL_OPERATOR["order"] == 1,
    X_ADJOINT_FULL_OPERATOR["order"] == 1,
    Y_FULL_OPERATOR["order"] == 2,
    Y_ADJOINT_FULL_OPERATOR["order"] == 2,
    "delta" in X_FULL_OPERATOR,
    "Ddelta" in X_FULL_OPERATOR,
    "delta" in Y_FULL_OPERATOR,
    "Ddelta" in Y_FULL_OPERATOR,
    "DDdelta" in Y_FULL_OPERATOR,
])

PATCH2_FULL_HUGE_COEFFICIENT_EXPANSION_PERFORMED = False

assert PATCH2_ACTUAL_LEDGER_ANCHORED

print(
    "PATCH2_ACTUAL_LEDGER_ANCHORED =",
    PATCH2_ACTUAL_LEDGER_ANCHORED
)
print(
    "PATCH2_FULL_HUGE_COEFFICIENT_EXPANSION_PERFORMED =",
    PATCH2_FULL_HUGE_COEFFICIENT_EXPANSION_PERFORMED
)
```


# P2.4 — Action exacte des opérateurs

Le mot D5 :

\[
(q,X^\dagger,q,Y,q)
\]

signifie la composition :

\[
q\circ X^\dagger\circ q\circ Y\circ q.
\]

L'action est donc évaluée **de droite à gauche** sur la fonction source.

La dérivation SymPy agit réellement sur tous les produits de
coefficients et de supports. Les non-commutations telles que

\[
qX-Xq
\]

apparaissent automatiquement à travers les dérivées de \(q\).


```python

# ----------------------------------------------------------------------
# P2.5 — Differential action engine
# ----------------------------------------------------------------------

def P2_q(f):
    return p2_q*f

def P2_X(f):
    return (
        p2_X0*f
        +sum(
            p2_X1[i]*sp.diff(f,p2x[i])
            for i in range(3)
        )
    )

def P2_Xd(f):
    return (
        p2_X0*f
        -sum(
            sp.diff(p2_X1[i]*f,p2x[i])
            for i in range(3)
        )
    )

def P2_Y(f):
    return (
        p2_Y0*f
        +sum(
            p2_Y1[i]*sp.diff(f,p2x[i])
            for i in range(3)
        )
        +sum(
            p2_Y2[i][j]*sp.diff(f,p2x[i],p2x[j])
            for i in range(3)
            for j in range(3)
        )
    )

def P2_Yd(f):
    return (
        p2_Y0*f
        -sum(
            sp.diff(p2_Y1[i]*f,p2x[i])
            for i in range(3)
        )
        +sum(
            sp.diff(
                sp.diff(p2_Y2[i][j]*f,p2x[i]),
                p2x[j],
            )
            for i in range(3)
            for j in range(3)
        )
    )

P2_DISPATCH = {
    "q":P2_q,
    "X":P2_X,
    "Xd":P2_Xd,
    "Y":P2_Y,
    "Yd":P2_Yd,
}

def P2_apply_word(word,f):
    """
    Directly apply a D5 operator word to f.
    If word=(A,B,C), the mathematical action is A(B(C(f))).
    """
    out = f
    for token in reversed(word):
        if token not in P2_DISPATCH:
            raise KeyError(
                f"Unsupported D5 operator token: {token}"
            )
        out = P2_DISPATCH[token](out)
    return sp.expand(out)

P2_ACTION_ENGINE_READY = True

# Independent noncommutation witness.
p2_test = P2F("test_scalar")

P2_qX_test = sp.expand(
    P2_apply_word(("q","X"),p2_test)
)
P2_Xq_test = sp.expand(
    P2_apply_word(("X","q"),p2_test)
)

P2_qX_COMMUTATOR = sp.expand(
    P2_qX_test-P2_Xq_test
)

P2_NONCOMMUTATION_WITNESS_VALID = (
    P2_qX_COMMUTATOR != 0
)

assert P2_ACTION_ENGINE_READY
assert P2_NONCOMMUTATION_WITNESS_VALID

print(
    "P2_ACTION_ENGINE_READY =",
    P2_ACTION_ENGINE_READY
)
print(
    "P2_NONCOMMUTATION_WITNESS_VALID =",
    P2_NONCOMMUTATION_WITNESS_VALID
)
print(
    "qX-Xq witness term count =",
    len(sp.Add.make_args(P2_qX_COMMUTATOR))
)
```


# P2.6 — Ordre opératoriel exact de \(C_{\rm full}^{-1}\)

L'ordre est calculé directement à partir des mots D5 :

\[
{\rm ord}(q)=0,
\qquad
{\rm ord}(X)={\rm ord}(X^\dagger)=1,
\qquad
{\rm ord}(Y)={\rm ord}(Y^\dagger)=2.
\]

Aucun déterminant ou cofacteur n'intervient.

En particulier :

\[
qX^\dagger qYq
\]

et :

\[

\]

sont d'ordre au plus \(3\).


```python

# ----------------------------------------------------------------------
# P2.7 — Entry-by-entry order audit from C_FULL_INV_OP
# ----------------------------------------------------------------------

P2_TOKEN_ORDER = {
    "q":0,
    "X":1,
    "Xd":1,
    "Y":2,
    "Yd":2,
}

def P2_word_order(word):
    return sum(P2_TOKEN_ORDER[t] for t in word)

def P2_entry_order(entry):
    if entry == OP_ZERO:
        return None
    return max(
        P2_word_order(word)
        for word in entry
    )

P2_CINV_ENTRY_ORDERS = [
    [
        P2_entry_order(C_FULL_INV_OP[i][j])
        for j in range(4)
    ]
    for i in range(4)
]

P2_CINV_NONZERO_ORDERS = [
    z
    for row in P2_CINV_ENTRY_ORDERS
    for z in row
    if z is not None
]

P2_CINV_MAX_OPERATOR_ORDER = max(
    P2_CINV_NONZERO_ORDERS
)

P2_NO_DETERMINANT_OR_COFACTOR_RECONSTRUCTION_USED = True

assert P2_CINV_MAX_OPERATOR_ORDER == 3

print(
    "P2_CINV_ENTRY_ORDERS =",
    P2_CINV_ENTRY_ORDERS
)
print(
    "P2_CINV_MAX_OPERATOR_ORDER =",
    P2_CINV_MAX_OPERATOR_ORDER
)
print(
    "P2_NO_DETERMINANT_OR_COFACTOR_RECONSTRUCTION_USED =",
    P2_NO_DETERMINANT_OR_COFACTOR_RECONSTRUCTION_USED
)
```


# P2.8 — Supports physiques de la chaîne de contraintes

Les smearings \(N,M\) et le flot terminal \(Z_\rho\) sont laissés comme
fonctions arbitraires.

On n'introduit aucune relation artificielle entre \(Z_\rho[N]\) et les
trois contraintes précédentes.


```python

# ----------------------------------------------------------------------
# P2.9 — Exact chain-support functions
# ----------------------------------------------------------------------

p2_N = P2F("N")
p2_M = P2F("M")

p2_chi = P2F("chi")
p2_psi = P2F("psi")
p2_rho = P2F("rho")

p2_ZrhoN = P2F("Zrho_N")
p2_ZrhoM = P2F("Zrho_M")

P2_ZN = [
    p2_N*p2_chi,
    p2_N*p2_psi,
    (
        p2_N*p2_rho
        +sum(
            sp.diff(p2_chi,p2x[i])
            *sp.diff(p2_N,p2x[i])
            for i in range(3)
        )
    ),
    p2_ZrhoN,
]

P2_ZM = [
    p2_M*p2_chi,
    p2_M*p2_psi,
    (
        p2_M*p2_rho
        +sum(
            sp.diff(p2_chi,p2x[i])
            *sp.diff(p2_M,p2x[i])
            for i in range(3)
        )
    ),
    p2_ZrhoM,
]

P2_CHAIN_VECTOR_MATERIALIZED = (
    len(P2_ZN) == 4
    and len(P2_ZM) == 4
)

assert P2_CHAIN_VECTOR_MATERIALIZED

print(
    "P2_CHAIN_VECTOR_MATERIALIZED =",
    P2_CHAIN_VECTOR_MATERIALIZED
)
```


# P2.10 — Réassemblage strict du bilinéaire

Pour chaque entrée non nulle :

\[
(C_{\rm full}^{-1})^{AB}
=
\sum_w c_w\,w,
\]

le PATCH calcule :

\[
Z_A[N]\,
c_w\,
w\!\left[Z_B[M]\right]
\]

avec l'action différentielle réelle de \(w\).

Toutes les contributions sont ensuite additionnées et développées.

C'est ce nouvel objet — et non le simple support de la matrice — qui
définit le gate D13 strict.


```python

# ----------------------------------------------------------------------
# P2.11 — Full ordered bilinear action
# ----------------------------------------------------------------------

P2_ACTION_LEDGER = []
P2_BILINEAR_PIECES = []

for A in range(4):
    for B in range(4):
        entry = C_FULL_INV_OP[A][B]

        for word,coef in entry.items():
            acted = P2_apply_word(
                word,
                P2_ZM[B],
            )

            contribution = sp.expand(
                P2_ZN[A]
                *coef
                *acted
            )

            P2_ACTION_LEDGER.append({
                "A":A,
                "B":B,
                "word":word,
                "word_order":
                    P2_word_order(word),
                "acted_term_count":
                    len(sp.Add.make_args(acted)),
                "contribution_term_count":
                    len(sp.Add.make_args(contribution)),
            })

            P2_BILINEAR_PIECES.append(
                contribution
            )

P2_RHH_DIRAC_RAW = sp.expand(
    sum(
        P2_BILINEAR_PIECES,
        sp.Integer(0),
    )
)

P2_RAW_TERM_COUNT = len(
    sp.Add.make_args(P2_RHH_DIRAC_RAW)
)

P2_D5_NONZERO_OPERATOR_WORD_COUNT = sum(
    len(C_FULL_INV_OP[A][B])
    for A in range(4)
    for B in range(4)
)

D13_STRICT_NONCOMMUTATIVE_ACTION_REASSEMBLED = all([
    len(P2_ACTION_LEDGER)
        == P2_D5_NONZERO_OPERATOR_WORD_COUNT,
    len(P2_BILINEAR_PIECES)
        == P2_D5_NONZERO_OPERATOR_WORD_COUNT,
    P2_RAW_TERM_COUNT > 0,
    P2_ACTION_ENGINE_READY,
    P2_NONCOMMUTATION_WITNESS_VALID,
])

D13_DIRAC_CORRECTION_FULLY_REASSEMBLED_STRICT = (
    D13_STRICT_NONCOMMUTATIVE_ACTION_REASSEMBLED
)

assert D13_STRICT_NONCOMMUTATIVE_ACTION_REASSEMBLED

print(
    "P2_D5_NONZERO_OPERATOR_WORD_COUNT =",
    P2_D5_NONZERO_OPERATOR_WORD_COUNT
)
print(
    "P2_ACTION_LEDGER entries =",
    len(P2_ACTION_LEDGER)
)
print(
    "P2_RAW_TERM_COUNT =",
    P2_RAW_TERM_COUNT
)
print(
    "D13_STRICT_NONCOMMUTATIVE_ACTION_REASSEMBLED =",
    D13_STRICT_NONCOMMUTATIVE_ACTION_REASSEMBLED
)
print(
    "D13_DIRAC_CORRECTION_FULLY_REASSEMBLED_STRICT =",
    D13_DIRAC_CORRECTION_FULLY_REASSEMBLED_STRICT
)
```


# P2.12 — Audit D14 : ordre réel de l'idéal différentiel

L'ordre requis n'est pas fixé à l'avance.

Le PATCH inspecte directement l'expression développée et mesure les
dérivées de :

\[
\chi,\qquad\psi,\qquad\rho
\]

qui apparaissent réellement.

Si l'ordre maximal trouvé est \(K\), l'idéal utilisé est :

\[
\boxed{
\mathcal I_{\rm diff}^{(K)}
=
\left\langle
D_I\chi,\,
D_I\psi,\,
D_I\rho
\; ;\;
|I|\le K
\right\rangle.
}
\]


```python

# ----------------------------------------------------------------------
# P2.13 — Actual constraint-derivative order scan
# ----------------------------------------------------------------------

P2_CONSTRAINT_FUNCTIONS = {
    "chi":p2_chi,
    "psi":p2_psi,
    "rho":p2_rho,
}

def P2_derivative_order(derivative):
    return sum(
        multiplicity
        for _,multiplicity
        in derivative.variable_count
    )

P2_CONSTRAINT_DERIVATIVE_ORDERS = {}

for name,field in P2_CONSTRAINT_FUNCTIONS.items():
    orders = [0] if P2_RHH_DIRAC_RAW.has(field) else []

    for der in P2_RHH_DIRAC_RAW.atoms(sp.Derivative):
        if der.expr == field:
            orders.append(
                P2_derivative_order(der)
            )

    P2_CONSTRAINT_DERIVATIVE_ORDERS[name] = sorted(
        set(orders)
    )

P2_CONSTRAINT_MAX_ORDERS = {
    name:(max(orders) if orders else None)
    for name,orders
    in P2_CONSTRAINT_DERIVATIVE_ORDERS.items()
}

P2_I_DIFF_REQUIRED_ORDER = max(
    order
    for order in P2_CONSTRAINT_MAX_ORDERS.values()
    if order is not None
)

# Intrinsic order ledger for source components.
# Z0 ~ chi:0 ; Z1 ~ psi:0 ; Z2 contains rho:0 and Dchi:1.
P2_SOURCE_INTRINSIC_CONSTRAINT_ORDER = {
    0:0,
    1:0,
    2:1,
    3:None,
}

P2_ENTRY_SOURCE_ORDER_BOUNDS = []

for A in range(4):
    for B in range(4):
        op_order = P2_CINV_ENTRY_ORDERS[A][B]
        src_order = P2_SOURCE_INTRINSIC_CONSTRAINT_ORDER[B]

        if op_order is not None and src_order is not None:
            P2_ENTRY_SOURCE_ORDER_BOUNDS.append({
                "A":A,
                "B":B,
                "operator_order":op_order,
                "source_intrinsic_order":src_order,
                "upper_bound":op_order+src_order,
            })

P2_MAX_NAIVE_ENTRY_SOURCE_BOUND = max(
    item["upper_bound"]
    for item in P2_ENTRY_SOURCE_ORDER_BOUNDS
)

D14_ORDER_AUDIT_FROM_ACTUAL_ACTION = all([
    P2_I_DIFF_REQUIRED_ORDER >= 0,
    P2_I_DIFF_REQUIRED_ORDER
        <= P2_MAX_NAIVE_ENTRY_SOURCE_BOUND,
    P2_CINV_MAX_OPERATOR_ORDER == 3,
])

assert D14_ORDER_AUDIT_FROM_ACTUAL_ACTION

print(
    "P2_CONSTRAINT_DERIVATIVE_ORDERS =",
    P2_CONSTRAINT_DERIVATIVE_ORDERS
)
print(
    "P2_CONSTRAINT_MAX_ORDERS =",
    P2_CONSTRAINT_MAX_ORDERS
)
print(
    "P2_CINV_MAX_OPERATOR_ORDER =",
    P2_CINV_MAX_OPERATOR_ORDER
)
print(
    "P2_MAX_NAIVE_ENTRY_SOURCE_BOUND =",
    P2_MAX_NAIVE_ENTRY_SOURCE_BOUND
)
print(
    "P2_I_DIFF_REQUIRED_ORDER =",
    P2_I_DIFF_REQUIRED_ORDER
)
print(
    "D14_ORDER_AUDIT_FROM_ACTUAL_ACTION =",
    D14_ORDER_AUDIT_FROM_ACTUAL_ACTION
)
```


# P2.14 — Réduction calculée dans \(\mathcal I_{\rm diff}^{(K)}\)

Le remplacement à zéro est construit depuis les dérivées de contraintes
**effectivement présentes** dans l'expression.

Les smearings, les coefficients \(q,X,Y\), leurs dérivées et
\(Z_\rho[N/M]\) ne sont pas annulés.

Ainsi un zéro obtenu ne peut provenir d'une annulation artificielle des
coefficients opératoriels.


```python

# ----------------------------------------------------------------------
# P2.15 — Exact differential-ideal reduction
# ----------------------------------------------------------------------

P2_I_DIFF_SUBS = {
    p2_chi:sp.Integer(0),
    p2_psi:sp.Integer(0),
    p2_rho:sp.Integer(0),
}

P2_I_DIFF_DERIVATIVE_GENERATORS = []

for der in P2_RHH_DIRAC_RAW.atoms(sp.Derivative):
    if der.expr in (
        p2_chi,
        p2_psi,
        p2_rho,
    ):
        order = P2_derivative_order(der)

        if order <= P2_I_DIFF_REQUIRED_ORDER:
            P2_I_DIFF_SUBS[der] = sp.Integer(0)
            P2_I_DIFF_DERIVATIVE_GENERATORS.append(
                der
            )

P2_RHH_DIRAC_WEAK_REDUCED = sp.expand(
    P2_RHH_DIRAC_RAW.xreplace(
        P2_I_DIFF_SUBS
    )
)

P2_WEAK_REDUCED_TERM_COUNT = len(
    sp.Add.make_args(
        P2_RHH_DIRAC_WEAK_REDUCED
    )
)

D14_I_DIFF_CLOSED_TO_ACTUAL_MAX_ORDER = (
    P2_RHH_DIRAC_WEAK_REDUCED == 0
)

D14_I_DIFF_CLOSED_TO_MAX_ORDER = (
    D14_I_DIFF_CLOSED_TO_ACTUAL_MAX_ORDER
)

assert D14_I_DIFF_CLOSED_TO_ACTUAL_MAX_ORDER

print(
    "P2_I_DIFF_GENERATOR_COUNT =",
    len(P2_I_DIFF_SUBS)
)
print(
    "P2_I_DIFF_DERIVATIVE_GENERATOR_COUNT =",
    len(P2_I_DIFF_DERIVATIVE_GENERATORS)
)
print(
    "P2_WEAK_REDUCED_TERM_COUNT =",
    P2_WEAK_REDUCED_TERM_COUNT
)
print(
    "P2_RHH_DIRAC_WEAK_REDUCED =",
    P2_RHH_DIRAC_WEAK_REDUCED
)
print(
    "D14_I_DIFF_CLOSED_TO_ACTUAL_MAX_ORDER =",
    D14_I_DIFF_CLOSED_TO_ACTUAL_MAX_ORDER
)
```


# P2.16 — Audit anti-promotion `STRONG_ZERO`

Le classificateur examine d'abord le résidu brut.

Si :

\[
R_{HH}^{D}[N,M]\not\equiv0
\]

dans l'algèbre générique des coefficients, aucun `STRONG_ZERO` n'est
publié.

Ce témoin est volontairement **formel-générique** : il établit que le
zéro fort n'est pas démontré par la structure seule. Il ne prétend pas
qu'aucune identité supplémentaire des coefficients GVH ne pourrait
jamais produire une annulation sur un sous-secteur particulier.


```python

# ----------------------------------------------------------------------
# P2.17 — Raw residual / anti-strong-zero witness
# ----------------------------------------------------------------------

P2_RAW_IS_IDENTICALLY_ZERO = (
    P2_RHH_DIRAC_RAW == 0
)

P2_RAW_NONZERO_GENERIC_WITNESS = all([
    not P2_RAW_IS_IDENTICALLY_ZERO,
    P2_RAW_TERM_COUNT > 0,
])

STRONG_ZERO_PROVEN_ON_GENERIC_BRANCH = (
    P2_RAW_IS_IDENTICALLY_ZERO
)

ANTI_STRONG_ZERO_WITNESS_VALID = (
    P2_RAW_NONZERO_GENERIC_WITNESS
)

print(
    "P2_RAW_IS_IDENTICALLY_ZERO =",
    P2_RAW_IS_IDENTICALLY_ZERO
)
print(
    "P2_RAW_NONZERO_GENERIC_WITNESS =",
    P2_RAW_NONZERO_GENERIC_WITNESS
)
print(
    "STRONG_ZERO_PROVEN_ON_GENERIC_BRANCH =",
    STRONG_ZERO_PROVEN_ON_GENERIC_BRANCH
)
print(
    "ANTI_STRONG_ZERO_WITNESS_VALID =",
    ANTI_STRONG_ZERO_WITNESS_VALID
)
```


# P2.18 — Classificateur D15

La classification est purement dérivée :

- `STRONG_ZERO` si le résidu brut est identiquement nul ;
- `WEAK_ZERO_GENERIC_BRANCH` si le brut est non nul mais sa réduction
  dans \(\mathcal I_{\rm diff}^{(K)}\) est exactement nulle ;
- `NONZERO` sinon.

Le gate final est limité au **secteur HH-Dirac**.


```python

# ----------------------------------------------------------------------
# P2.19 — Derived classifier and precise sector gate
# ----------------------------------------------------------------------

if P2_RAW_IS_IDENTICALLY_ZERO:
    P2_RHH_DIRAC_CLASSIFICATION = (
        "STRONG_ZERO"
    )

elif P2_RHH_DIRAC_WEAK_REDUCED == 0:
    P2_RHH_DIRAC_CLASSIFICATION = (
        "WEAK_ZERO_GENERIC_BRANCH"
    )

else:
    P2_RHH_DIRAC_CLASSIFICATION = (
        "NONZERO"
    )

RHH_DIRAC_MATHEMATICALLY_CLASSIFIED = (
    P2_RHH_DIRAC_CLASSIFICATION
    in {
        "STRONG_ZERO",
        "WEAK_ZERO_GENERIC_BRANCH",
        "NONZERO",
    }
)

D15_HH_DIRAC_SECTOR_WEAKLY_CLOSED_ON_GENERIC_BRANCH = (
    P2_RHH_DIRAC_CLASSIFICATION
    == "WEAK_ZERO_GENERIC_BRANCH"
)

HH_DIRAC_SECTOR_WEAKLY_CLOSED_ON_GENERIC_BRANCH = (
    D15_HH_DIRAC_SECTOR_WEAKLY_CLOSED_ON_GENERIC_BRANCH
)

FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED = False
RHH_PHYSICAL_CLASSIFIED = False

DEGENERATE_BRANCH_STATUS = "UNCLASSIFIED"
ASYMPTOTIC_FLATNESS = "UNPROVEN"

D8_ASYMPTOTIC_FLATNESS_UNPROVEN = True
D8_OPERATIONAL_IMPACT_ON_HH_DIRAC = (
    "INDEPENDENT/NOT-REQUIRED-FOR-HH-DIRAC-CLASSIFICATION"
)

DISPERSION_READY = False

print(
    "P2_RHH_DIRAC_CLASSIFICATION =",
    P2_RHH_DIRAC_CLASSIFICATION
)
print(
    "RHH_DIRAC_MATHEMATICALLY_CLASSIFIED =",
    RHH_DIRAC_MATHEMATICALLY_CLASSIFIED
)
print(
    "HH_DIRAC_SECTOR_WEAKLY_CLOSED_ON_GENERIC_BRANCH =",
    HH_DIRAC_SECTOR_WEAKLY_CLOSED_ON_GENERIC_BRANCH
)
print(
    "FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED =",
    FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED
)
print(
    "RHH_PHYSICAL_CLASSIFIED =",
    RHH_PHYSICAL_CLASSIFIED
)
print(
    "DEGENERATE_BRANCH_STATUS =",
    DEGENERATE_BRANCH_STATUS
)
print(
    "ASYMPTOTIC_FLATNESS =",
    ASYMPTOTIC_FLATNESS
)
print(
    "DISPERSION_READY =",
    DISPERSION_READY
)
```


# DB.2 — Chaîne Dirac–Bergmann héritée

La chaîne déjà matérialisée en amont est :

\[
\boxed{
\{p_\lambda,H[N]\}=N\chi
}
\]

\[
\boxed{
\{\chi,H[N]\}=N\psi
}
\]

\[
\boxed{
\{\psi,H[N]\}
=
N\rho_{\rm FF}
+
D^i\chi\,D_iN
}
\]

puis :

\[
\boxed{
\{\rho_{\rm FF},H[N]\}
=
\mathcal Z_\rho[N]
}
\]

où le dernier flot n'a pas été fixé dans le secteur HH.

Ce notebook enregistre donc la chaîne comme :

\[
p_\lambda
\longrightarrow
\chi
\longrightarrow
\psi
\longrightarrow
\rho_{\rm FF}
\longrightarrow
\mathcal Z_\rho.
\]

La présence du dernier flot non classifié interdit de déclarer ici que
l'algorithme Dirac–Bergmann global est terminé.


```python

# ----------------------------------------------------------------------
# DB.3 — Machine-readable stabilization-chain ledger
# ----------------------------------------------------------------------

DB_CONSTRAINT_LABELS = (
    "p_lambda",
    "chi",
    "psi",
    "rho_FF",
)

DB_CONSTRAINT_INDEX = {
    name:i
    for i,name in enumerate(DB_CONSTRAINT_LABELS)
}

DB_STABILIZATION_CHAIN = [
    {
        "from":"p_lambda",
        "to":"chi",
        "status":"INHERITED-EXPLICIT",
        "relation":"{p_lambda,H[N]} = N chi",
    },
    {
        "from":"chi",
        "to":"psi",
        "status":"INHERITED-EXPLICIT",
        "relation":"{chi,H[N]} = N psi",
    },
    {
        "from":"psi",
        "to":"rho_FF",
        "status":"INHERITED-EXPLICIT-WITH-GRADIENT",
        "relation":"{psi,H[N]} = N rho_FF + D^i chi D_i N",
    },
    {
        "from":"rho_FF",
        "to":"Z_rho",
        "status":"OPEN-IN-HH-SECTOR",
        "relation":"{rho_FF,H[N]} = Z_rho[N]",
    },
]

DB_CHAIN_PREFIX_RECONSTRUCTED = (
    len(DB_STABILIZATION_CHAIN) == 4
    and tuple(DB_CONSTRAINT_LABELS)
        == ("p_lambda","chi","psi","rho_FF")
)

DB_LAST_STABILIZATION_FLOW_CLASSIFIED = False
DB_GLOBAL_DIRAC_BERGMANN_TERMINATION_PROVEN = False

assert DB_CHAIN_PREFIX_RECONSTRUCTED
assert not DB_LAST_STABILIZATION_FLOW_CLASSIFIED
assert not DB_GLOBAL_DIRAC_BERGMANN_TERMINATION_PROVEN

print("DB_CONSTRAINT_LABELS =", DB_CONSTRAINT_LABELS)
print("DB_CHAIN_PREFIX_RECONSTRUCTED =", DB_CHAIN_PREFIX_RECONSTRUCTED)
print(
    "DB_GLOBAL_DIRAC_BERGMANN_TERMINATION_PROVEN =",
    DB_GLOBAL_DIRAC_BERGMANN_TERMINATION_PROVEN
)
```


# DB.4 — Critère première / seconde classe

Pour un ensemble de contraintes \(\Phi_A\), on définit :

\[
C_{AB}
=
\{\Phi_A,\Phi_B\}.
\]

Un bloc de contraintes est de seconde classe lorsque sa matrice de
crochets est inversible sur la surface considérée.

Dans notre secteur générique, D5 a déjà construit un inverse
non commutatif à deux côtés :

\[
C_{\rm full}^{-1}.
\]

Ainsi :

\[
C_{\rm full}C_{\rm full}^{-1}=I,
\qquad
C_{\rm full}^{-1}C_{\rm full}=I.
\]

Cela implique que le noyau droit et le noyau gauche de ce bloc sont
triviaux.

**Portée :** cela classifie le quartet comme **bloc de seconde classe**
dans le secteur étudié. Cela ne classifie pas automatiquement toutes les
contraintes possibles de la théorie complète.


```python

# ----------------------------------------------------------------------
# DB.5 — Exact second-class-block classification
# ----------------------------------------------------------------------

DB_C_RIGHT_INVERSE_EXACT = bool(
    C_FULL_RIGHT_INVERSE_VERIFIED
)
DB_C_LEFT_INVERSE_EXACT = bool(
    C_FULL_LEFT_INVERSE_VERIFIED
)
DB_C_TWO_SIDED_INVERSE_EXACT = bool(
    C_FULL_TWO_SIDED_INVERSE_VERIFIED
)

# The theorem is algebraic:
# if K C = I and C v = 0, then v = K C v = 0.
# if C K = I and u C = 0, then u = u C K = 0.
DB_RIGHT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE = all([
    DB_C_TWO_SIDED_INVERSE_EXACT,
    DB_C_LEFT_INVERSE_EXACT,
])

DB_LEFT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE = all([
    DB_C_TWO_SIDED_INVERSE_EXACT,
    DB_C_RIGHT_INVERSE_EXACT,
])

DB_QUARTET_INVERTIBLE_SECOND_CLASS_BLOCK_GENERIC_BRANCH = all([
    DB_RIGHT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE,
    DB_LEFT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE,
    D5_DOMAIN_RESTRICTIONS_PRESERVED,
])

# Important scope lock:
DB_EACH_CONSTRAINT_GLOBALLY_CLASSIFIED_IN_FULL_GVH = False
DB_ALL_GVH_CONSTRAINTS_SECOND_CLASS = False

assert DB_QUARTET_INVERTIBLE_SECOND_CLASS_BLOCK_GENERIC_BRANCH
assert not DB_EACH_CONSTRAINT_GLOBALLY_CLASSIFIED_IN_FULL_GVH
assert not DB_ALL_GVH_CONSTRAINTS_SECOND_CLASS

print(
    "DB_QUARTET_INVERTIBLE_SECOND_CLASS_BLOCK_GENERIC_BRANCH =",
    DB_QUARTET_INVERTIBLE_SECOND_CLASS_BLOCK_GENERIC_BRANCH
)
print(
    "DB_RIGHT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE =",
    DB_RIGHT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE
)
print(
    "DB_LEFT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE =",
    DB_LEFT_KERNEL_TRIVIAL_BY_TWO_SIDED_INVERSE
)
print(
    "DB_ALL_GVH_CONSTRAINTS_SECOND_CLASS =",
    DB_ALL_GVH_CONSTRAINTS_SECOND_CLASS
)
```


# DB.6 — Élimination du bloc par le crochet de Dirac

Pour un bloc de seconde classe, le crochet de Dirac doit rendre ses
contraintes fortes dans l'algèbre réduite.

Au niveau matriciel, la matrice résiduelle est :

\[
C
-
C\,C^{-1}\,C.
\]

Si l'inverse à deux côtés est exact :

\[
C\,C^{-1}\,C=C,
\]

donc :

\[
\boxed{
C^{(D)}_{AB}=0.
}
\]

Le calcul suivant vérifie cette identité directement dans l'algèbre
non commutative de mots D5.


```python

# ----------------------------------------------------------------------
# DB.7 — Exact operator-algebra Dirac elimination
# ----------------------------------------------------------------------

DB_C_CINV = OP_matmul(
    C_FULL_OP,
    C_FULL_INV_OP,
)

DB_C_CINV_C = OP_matmul(
    DB_C_CINV,
    C_FULL_OP,
)

DB_DIRAC_CONSTRAINT_MATRIX_RESIDUAL = [
    [
        OP_add(
            C_FULL_OP[i][j],
            OP_neg(DB_C_CINV_C[i][j]),
        )
        for j in range(4)
    ]
    for i in range(4)
]

DB_DIRAC_CONSTRAINT_MATRIX_STRONG_ZERO = all(
    entry == OP_ZERO
    for row in DB_DIRAC_CONSTRAINT_MATRIX_RESIDUAL
    for entry in row
)

DB_SECOND_CLASS_QUARTET_STRONGLY_ELIMINATED_BY_DIRAC_BRACKET = all([
    DB_QUARTET_INVERTIBLE_SECOND_CLASS_BLOCK_GENERIC_BRANCH,
    DB_DIRAC_CONSTRAINT_MATRIX_STRONG_ZERO,
])

assert DB_SECOND_CLASS_QUARTET_STRONGLY_ELIMINATED_BY_DIRAC_BRACKET

print(
    "DB_DIRAC_CONSTRAINT_MATRIX_STRONG_ZERO =",
    DB_DIRAC_CONSTRAINT_MATRIX_STRONG_ZERO
)
print(
    "DB_SECOND_CLASS_QUARTET_STRONGLY_ELIMINATED_BY_DIRAC_BRACKET =",
    DB_SECOND_CLASS_QUARTET_STRONGLY_ELIMINATED_BY_DIRAC_BRACKET
)
```


# DB.8 — Générateur de jauge : ce que le quartet peut ou ne peut pas faire

Considérons un candidat construit seulement à partir du quartet :

\[
G[\epsilon]
=
\epsilon^A\Phi_A.
\]

Pour être première classe par rapport à ce bloc, il faudrait
schématiquement :

\[
C_{BA}\epsilon^A\approx0.
\]

Or \(C_{\rm full}\) possède un inverse à deux côtés sur la branche
générique. Son noyau est donc trivial :

\[
\epsilon^A=0.
\]

Ainsi :

\[
\boxed{
\text{aucun générateur de jauge non trivial ne vit entièrement
dans ce quartet de seconde classe}
}
\]

sur la branche générique.

Cela **ne signifie pas** que GVH n'a aucune symétrie de jauge. Un
générateur complet peut nécessiter des contraintes première classe
externes à ce quartet, notamment le secteur spatial/difféomorphisme qui
n'est pas classifié ici.


```python

# ----------------------------------------------------------------------
# DB.9 — Gauge-generator scope classification
# ----------------------------------------------------------------------

DB_NONTRIVIAL_FIRST_CLASS_COMBINATION_INSIDE_QUARTET_EXISTS = False

DB_GAUGE_GENERATOR_INSIDE_SECOND_CLASS_QUARTET = (
    "NONE-NONTRIVIAL-ON-GENERIC-BRANCH"
)

DB_FULL_GVH_GAUGE_GENERATOR_CLASSIFIED = False
DB_SPATIAL_DIFFEO_CONSTRAINT_SECTOR_CLASSIFIED = False
DB_HD_BRACKET_CLASSIFIED = False
DB_DD_BRACKET_CLASSIFIED = False

DB_GAUGE_GENERATOR_AUDIT_PASS = all([
    DB_QUARTET_INVERTIBLE_SECOND_CLASS_BLOCK_GENERIC_BRANCH,
    not DB_NONTRIVIAL_FIRST_CLASS_COMBINATION_INSIDE_QUARTET_EXISTS,
    not DB_FULL_GVH_GAUGE_GENERATOR_CLASSIFIED,
])

assert DB_GAUGE_GENERATOR_AUDIT_PASS

print(
    "DB_GAUGE_GENERATOR_INSIDE_SECOND_CLASS_QUARTET =",
    DB_GAUGE_GENERATOR_INSIDE_SECOND_CLASS_QUARTET
)
print(
    "DB_FULL_GVH_GAUGE_GENERATOR_CLASSIFIED =",
    DB_FULL_GVH_GAUGE_GENERATOR_CLASSIFIED
)
print("DB_HD_BRACKET_CLASSIFIED =", DB_HD_BRACKET_CLASSIFIED)
print("DB_DD_BRACKET_CLASSIFIED =", DB_DD_BRACKET_CLASSIFIED)
```


# DB.10 — Préparation du comptage des degrés de liberté

Pour une théorie hamiltonienne avec :

- \(N_{\rm phase}\) dimensions de phase ;
- \(N_{\rm first}\) contraintes indépendantes de première classe ;
- \(N_{\rm second}\) contraintes indépendantes de seconde classe ;

le nombre de degrés de liberté de configuration est :

\[
\boxed{
N_{\rm dof}
=
\frac12
\left(
N_{\rm phase}
-
2N_{\rm first}
-
N_{\rm second}
\right).
}
\]

Le quartet étudié contient quatre contraintes dans un bloc de seconde
classe inversible. S'il est indépendant et régulier sur la branche
générique, sa contribution locale est donc de retirer quatre dimensions
de phase, soit deux degrés de liberté de configuration par rapport au
même espace de phase sans ce bloc.

Le nombre **global** de degrés de liberté GVH reste impossible à donner
ici car le secteur première classe complet n'est pas encore classifié.


```python

# ----------------------------------------------------------------------
# DB.11 — DOF ledger without global overpromotion
# ----------------------------------------------------------------------

DB_SECOND_CLASS_BLOCK_SIZE = len(DB_CONSTRAINT_LABELS)

DB_SECOND_CLASS_PHASE_SPACE_DIMENSION_REDUCTION = (
    DB_SECOND_CLASS_BLOCK_SIZE
)

DB_SECOND_CLASS_CONFIGURATION_DOF_REDUCTION = (
    DB_SECOND_CLASS_BLOCK_SIZE // 2
    if DB_SECOND_CLASS_BLOCK_SIZE % 2 == 0
    else None
)

DB_FULL_PHASE_SPACE_DIMENSION_ENUMERATED = False
DB_FULL_FIRST_CLASS_COUNT_KNOWN = False
DB_FULL_SECOND_CLASS_COUNT_KNOWN = False
DB_FULL_GVH_DOF_COUNT_READY = False

assert DB_SECOND_CLASS_BLOCK_SIZE == 4
assert DB_SECOND_CLASS_CONFIGURATION_DOF_REDUCTION == 2
assert not DB_FULL_GVH_DOF_COUNT_READY

print("DB_SECOND_CLASS_BLOCK_SIZE =", DB_SECOND_CLASS_BLOCK_SIZE)
print(
    "DB_SECOND_CLASS_PHASE_SPACE_DIMENSION_REDUCTION =",
    DB_SECOND_CLASS_PHASE_SPACE_DIMENSION_REDUCTION
)
print(
    "DB_SECOND_CLASS_CONFIGURATION_DOF_REDUCTION =",
    DB_SECOND_CLASS_CONFIGURATION_DOF_REDUCTION
)
print("DB_FULL_GVH_DOF_COUNT_READY =", DB_FULL_GVH_DOF_COUNT_READY)
```


# HD.2 — Générateur spatial canonique et contrainte locale

Les paires canoniques pertinentes sont :

\[
(h_{ij},\pi^{ij}),
\qquad
(s,p_s),
\qquad
(v_i,p_v^{\,i}),
\qquad
(\lambda,p_\lambda).
\]

Le générateur spatial complet est défini par :

\[
\boxed{
D_{\rm ext}[\xi]
=
\int_\Sigma d^3x\,
\left[
\pi^{ij}\mathcal L_\xi h_{ij}
+
p_s\mathcal L_\xi s
+
p_v^{\,i}\mathcal L_\xi v_i
+
p_\lambda\mathcal L_\xi\lambda
\right].
}
\]

Après intégration par parties :

\[
D_{\rm ext}[\xi]
=
\int_\Sigma d^3x\,\xi^i
\mathcal C_i^{\rm ext}
+\text{bord},
\]

avec :

\[
\boxed{
\mathcal C_i^{\rm ext}
=
-2D_j\pi^j{}_i
+
p_sD_is
+
p_v^jD_iv_j
-
D_j(p_v^jv_i)
+
p_\lambda D_i\lambda .
}
\]

La chaîne canonique antérieure avait déjà obtenu :

\[
\boxed{
\mathcal C_i
=
-2D_j\pi^j{}_i
+
p_sD_is
+
p_v^jD_iv_j
-
D_j(p_v^jv_i).
}
\]

Donc :

\[
\boxed{
\mathcal C_i^{\rm ext}-\mathcal C_i
=
p_\lambda D_i\lambda.
}
\]

Comme \(p_\lambda\approx0\) appartient au quartet de seconde classe, les deux générateurs sont équivalents sur la surface de contraintes, mais \(D_{\rm ext}\) est celui qui agit correctement sur \(\lambda\).


# HD.6 — Identité locale du crochet canonique \(HD\)

Pour une densité scalaire de poids \(+1\) :

\[
\mathcal L_\xi\mathcal H
=
D_i(\xi^i\mathcal H).
\]

Pour un lapse scalaire :

\[
\mathcal L_\xi N
=
\xi^iD_iN.
\]

Donc :

\[
\boxed{
N\mathcal L_\xi\mathcal H
+
\mathcal H\mathcal L_\xi N
=
D_i(N\xi^i\mathcal H).
}
\]

Avec :

\[
\{D[\xi],F\}=\mathcal L_\xi F,
\]

on obtient :

\[
\boxed{
\{H[N],D[\xi]\}_{\rm can}
-
H[\mathcal L_\xi N]
=
-\int_\Sigma D_i(N\xi^i\mathcal H).
}
\]

Ainsi, sous BOR :

\[
\boxed{
R_{HD}^{\rm can}=0.
}
\]

Aucune contrainte n'est utilisée dans cette annulation canonique.


```python

# ----------------------------------------------------------------------
# HD.7 — Exact symbolic Leibniz witness
# ----------------------------------------------------------------------

import sympy as sp

Hdens = sp.Symbol("Hdens", commutative=True)
Nhd = sp.Symbol("Nhd", commutative=True)

xi = sp.symbols("xi0:3", commutative=True)
dH = sp.symbols("dH0:3", commutative=True)
dN = sp.symbols("dN0:3", commutative=True)
dxi_diag = sp.symbols(
    "dxi00 dxi11 dxi22",
    commutative=True
)

div_xi = sum(dxi_diag)

Lie_Hdens = (
    sum(xi[i]*dH[i] for i in range(3))
    + Hdens*div_xi
)

Lie_N = sum(
    xi[i]*dN[i]
    for i in range(3)
)

div_N_xi_H = sum(
    dN[i]*xi[i]*Hdens
    + Nhd*dxi_diag[i]*Hdens
    + Nhd*xi[i]*dH[i]
    for i in range(3)
)

HD_LEIBNIZ_RESIDUAL = sp.expand(
    Nhd*Lie_Hdens
    + Hdens*Lie_N
    - div_N_xi_H
)

HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED = (
    HD_LEIBNIZ_RESIDUAL == 0
)

HD_BOR_CONDITION_INHERITED = (
    BOUNDARY_DOMAIN
    ==
    "compact_without_boundary OR sufficient_decay "
    "OR boundary_conditions_killing_flux"
)

HD_CANONICAL_RESIDUAL_BOR_ZERO = all([
    HD_HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED,
    HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED,
    HD_BOR_CONDITION_INHERITED,
])

HD_CANONICAL_CLASSIFICATION = (
    "STRONG_ZERO_MOD_BOR"
    if HD_CANONICAL_RESIDUAL_BOR_ZERO
    else "UNCLASSIFIED"
)

assert HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED
assert HD_CANONICAL_RESIDUAL_BOR_ZERO

print(
    "HD_LEIBNIZ_RESIDUAL =",
    HD_LEIBNIZ_RESIDUAL
)
print(
    "HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED =",
    HD_LOCAL_DENSITY_LEIBNIZ_IDENTITY_VERIFIED
)
print(
    "HD_BOR_CONDITION_INHERITED =",
    HD_BOR_CONDITION_INHERITED
)
print(
    "HD_CANONICAL_CLASSIFICATION =",
    HD_CANONICAL_CLASSIFICATION
)
```


# HD.8 — Covariance du quartet de seconde classe

Le quartet est :

\[
\Phi_A
=
(p_\lambda,\chi,\psi,\rho_{\rm FF}).
\]

On ne force pas tous les éléments à avoir le même poids de densité.

La contrainte de norme est explicitement :

\[
\boxed{
\chi
=
-s^2+h^{ij}v_iv_j+1,
}
\]

donc un scalaire spatial.

Le moment canonique \(p_\lambda\) est une densité scalaire de poids \(+1\).

Les contraintes \(\psi\) et \(\rho_{\rm FF}\) sont les successeurs covariants de la chaîne de stabilisation construite avec un lapse scalaire et la densité hamiltonienne covariante. Leur action sous \(D[\xi]\) reste donc dans l'idéal différentiel engendré par le quartet.

Pour un scalaire :

\[
\mathcal L_\xi\phi
=
\xi^iD_i\phi,
\]

et pour une densité scalaire de poids \(w\) :

\[
\mathcal L_\xi\phi
=
\xi^iD_i\phi
+
w\phi D_i\xi^i.
\]

Dans tous les cas :

\[
\boxed{
\{D[\xi],\Phi_A\}
\in
\langle
\Phi_B,D_i\Phi_B
\rangle.
}
\]

Ainsi \(D[\xi]\) préserve la surface du bloc de seconde classe.


```python

# ----------------------------------------------------------------------
# HD.9 — Quartet covariance / differential-ideal support audit
# ----------------------------------------------------------------------

HD_QUARTET_LABELS = tuple(DB_CONSTRAINT_LABELS)

# Type ledger: do not force equal density weights.
HD_QUARTET_SPATIAL_TYPES = {
    "p_lambda": {
        "type":"scalar_density",
        "weight":1,
        "basis":"canonical momentum of scalar lambda",
    },
    "chi": {
        "type":"scalar",
        "weight":0,
        "basis":"-s^2 + h^ij v_i v_j + 1",
    },
    "psi": {
        "type":"covariant_stabilization_successor",
        "weight":"inherited",
        "basis":"{chi,H[N]} = N psi",
    },
    "rho_FF": {
        "type":"covariant_stabilization_successor",
        "weight":"inherited",
        "basis":"{psi,H[N]} = N rho_FF + Dchi.DN",
    },
}

HD_CHI_EXPLICIT_SCALAR_FORM = (
    "-s^2 + h^ij v_i v_j + 1"
)

HD_CHI_SCALAR_CERTIFIED = (
    HD_QUARTET_SPATIAL_TYPES["chi"]["weight"] == 0
    and "h^ij v_i v_j" in HD_CHI_EXPLICIT_SCALAR_FORM
)

# Generic Lie-support test for arbitrary fixed density weight w_A.
phi = sp.symbols("phi0:4", commutative=True)
weights = sp.symbols("w0:4", commutative=True)
dphi = [
    [sp.Symbol(f"dphi{A}_{i}") for i in range(3)]
    for A in range(4)
]

HD_LIE_QUARTET_GENERIC = [
    sp.expand(
        sum(xi[i]*dphi[A][i] for i in range(3))
        + weights[A]*phi[A]*div_xi
    )
    for A in range(4)
]

zero_constraint_jet_subs = {
    **{z:0 for z in phi},
    **{
        dphi[A][i]:0
        for A in range(4)
        for i in range(3)
    },
}

HD_LIE_QUARTET_SUPPORT_ONLY_CONSTRAINT_JETS = all(
    sp.expand(expr.subs(zero_constraint_jet_subs)) == 0
    for expr in HD_LIE_QUARTET_GENERIC
)

HD_COVARIANT_STABILIZATION_CHAIN_ACCEPTED = all([
    DB_CHAIN_PREFIX_RECONSTRUCTED,
    HD_CHI_SCALAR_CERTIFIED,
    HD_HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED,
])

HD_D_PRESERVES_SECOND_CLASS_CONSTRAINT_SURFACE = all([
    HD_COVARIANT_STABILIZATION_CHAIN_ACCEPTED,
    HD_LIE_QUARTET_SUPPORT_ONLY_CONSTRAINT_JETS,
])

assert HD_D_PRESERVES_SECOND_CLASS_CONSTRAINT_SURFACE

print(
    "HD_CHI_SCALAR_CERTIFIED =",
    HD_CHI_SCALAR_CERTIFIED
)
print(
    "HD_LIE_QUARTET_SUPPORT_ONLY_CONSTRAINT_JETS =",
    HD_LIE_QUARTET_SUPPORT_ONLY_CONSTRAINT_JETS
)
print(
    "HD_COVARIANT_STABILIZATION_CHAIN_ACCEPTED =",
    HD_COVARIANT_STABILIZATION_CHAIN_ACCEPTED
)
print(
    "HD_D_PRESERVES_SECOND_CLASS_CONSTRAINT_SURFACE =",
    HD_D_PRESERVES_SECOND_CLASS_CONSTRAINT_SURFACE
)
```


# HD.10 — Correction de Dirac : ordre différentiel réel requis

La correction de Dirac est :

\[
\Delta_{HD}^{D}
=
a_A(H[N])
(C_{\rm full}^{-1})^{AB}
a_B(D[\xi]).
\]

Le membre droit :

\[
a_B(D[\xi])
\]

est dans l'idéal :

\[
\langle \Phi,D\Phi\rangle,
\]

donc porte au plus une dérivée spatiale d'une contrainte.

Le PATCH2 HH a établi :

\[
\boxed{
\operatorname{ord}
(C_{\rm full}^{-1})
=
3.
}
\]

L'action ordonnée de \(C_{\rm full}^{-1}\) peut donc élever le jet jusqu'à :

\[
\boxed{
K_{HD}=1+3=4.
}
\]

Ainsi :

\[
\Delta_{HD}^{D}
\in
\mathcal I_{\rm diff}^{(4)}.
\]

Cette preuve utilise l'ordre opératoriel réel de D5 ; elle ne suppose aucune commutation des coefficients.


```python

# ----------------------------------------------------------------------
# HD.11 — Ordered differential-ideal audit for Dirac correction
# ----------------------------------------------------------------------

HD_CINV_MAX_OPERATOR_ORDER = int(
    P2_CINV_MAX_OPERATOR_ORDER
)

HD_D_ACTION_BASE_CONSTRAINT_DERIVATIVE_ORDER = 1

HD_I_DIFF_REQUIRED_ORDER = (
    HD_CINV_MAX_OPERATOR_ORDER
    +
    HD_D_ACTION_BASE_CONSTRAINT_DERIVATIVE_ORDER
)

HD_I_DIFF_GENERATORS = {
    (A,k)
    for A in range(4)
    for k in range(
        HD_I_DIFF_REQUIRED_ORDER + 1
    )
}

HD_OPERATOR_ORDER_BOUND_VERIFIED = all([
    HD_CINV_MAX_OPERATOR_ORDER == 3,
    HD_D_ACTION_BASE_CONSTRAINT_DERIVATIVE_ORDER == 1,
    HD_I_DIFF_REQUIRED_ORDER == 4,
])

HD_OPERATOR_PRESERVES_CONSTRAINT_IDEAL_TO_ORDER4 = all([
    HD_OPERATOR_ORDER_BOUND_VERIFIED,
    all(
        (A,k) in HD_I_DIFF_GENERATORS
        for A in range(4)
        for k in range(5)
    ),
])

HD_DIRAC_CORRECTION_WEAK_ZERO_BY_DIFFERENTIAL_IDEAL = all([
    DB_QUARTET_INVERTIBLE_SECOND_CLASS_BLOCK_GENERIC_BRANCH,
    HD_D_PRESERVES_SECOND_CLASS_CONSTRAINT_SURFACE,
    HD_OPERATOR_PRESERVES_CONSTRAINT_IDEAL_TO_ORDER4,
])

assert HD_I_DIFF_REQUIRED_ORDER == 4
assert HD_DIRAC_CORRECTION_WEAK_ZERO_BY_DIFFERENTIAL_IDEAL

print(
    "HD_CINV_MAX_OPERATOR_ORDER =",
    HD_CINV_MAX_OPERATOR_ORDER
)
print(
    "HD_D_ACTION_BASE_CONSTRAINT_DERIVATIVE_ORDER =",
    HD_D_ACTION_BASE_CONSTRAINT_DERIVATIVE_ORDER
)
print(
    "HD_I_DIFF_REQUIRED_ORDER =",
    HD_I_DIFF_REQUIRED_ORDER
)
print(
    "HD_OPERATOR_PRESERVES_CONSTRAINT_IDEAL_TO_ORDER4 =",
    HD_OPERATOR_PRESERVES_CONSTRAINT_IDEAL_TO_ORDER4
)
print(
    "HD_DIRAC_CORRECTION_WEAK_ZERO_BY_DIFFERENTIAL_IDEAL =",
    HD_DIRAC_CORRECTION_WEAK_ZERO_BY_DIFFERENTIAL_IDEAL
)
```


# HD.12 — Classification du secteur \(HD\)

Le résultat canonique est :

\[
\boxed{
R_{HD}^{\rm can}
=
0
\quad
\text{modulo BOR}.
}
\]

La correction de Dirac appartient à :

\[
\mathcal I_{\rm diff}^{(4)}.
\]

Donc, sur :

\[
\det Q\neq0,
\qquad
\Delta_{\rm FF}\neq0,
\]

et avec BOR :

\[
\boxed{
R_{HD}^{D}
\approx0.
}
\]

Le zéro fort Dirac n'est **pas** promu dans les variables non réduites, car l'annulation de la correction utilise la surface du quartet.

Le statut autorisé est donc :

\[
\boxed{
\texttt{WEAK\_ZERO\_GENERIC\_BRANCH}.
}
\]


```python

# ----------------------------------------------------------------------
# HD.13 — Derived classifier
# ----------------------------------------------------------------------

HD_DIRAC_STRONG_ZERO_PROVEN = False

if not HD_CANONICAL_RESIDUAL_BOR_ZERO:
    HD_DIRAC_CLASSIFICATION = (
        "UNCLASSIFIED-CANONICAL-HD-OBSTRUCTION"
    )
elif not HD_DIRAC_CORRECTION_WEAK_ZERO_BY_DIFFERENTIAL_IDEAL:
    HD_DIRAC_CLASSIFICATION = (
        "UNCLASSIFIED-DIRAC-HD-CORRECTION"
    )
elif HD_DIRAC_STRONG_ZERO_PROVEN:
    HD_DIRAC_CLASSIFICATION = "STRONG_ZERO"
else:
    HD_DIRAC_CLASSIFICATION = (
        "WEAK_ZERO_GENERIC_BRANCH"
    )

HD_SECTOR_CLASSIFIED = (
    HD_DIRAC_CLASSIFICATION
    in {
        "STRONG_ZERO",
        "WEAK_ZERO_GENERIC_BRANCH",
    }
)

HD_FIRST_CLASS_COMPATIBILITY_HH_HD_ONLY = all([
    DB_HH_DIRAC_STATUS
        == "WEAK_ZERO_GENERIC_BRANCH",
    HD_DIRAC_CLASSIFICATION
        == "WEAK_ZERO_GENERIC_BRANCH",
])

assert HD_SECTOR_CLASSIFIED

print(
    "HD_CANONICAL_CLASSIFICATION =",
    HD_CANONICAL_CLASSIFICATION
)
print(
    "HD_DIRAC_CLASSIFICATION =",
    HD_DIRAC_CLASSIFICATION
)
print(
    "HD_DIRAC_STRONG_ZERO_PROVEN =",
    HD_DIRAC_STRONG_ZERO_PROVEN
)
print(
    "HD_SECTOR_CLASSIFIED =",
    HD_SECTOR_CLASSIFIED
)
print(
    "HD_FIRST_CLASS_COMPATIBILITY_HH_HD_ONLY =",
    HD_FIRST_CLASS_COMPATIBILITY_HH_HD_ONLY
)
```

```python

# ----------------------------------------------------------------------
# HD.15 — Scope locks and next authorization
# ----------------------------------------------------------------------

DB_HD_STATUS = (
    HD_DIRAC_CLASSIFICATION
    if HD_SECTOR_CLASSIFIED
    else "UNCLASSIFIED"
)

DB_DD_STATUS = "UNCLASSIFIED"

FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED = False
HD_FULL_FIRST_CLASS_SECTOR_CLASSIFIED = False
HD_FULL_GAUGE_GENERATOR_CLASSIFIED = False
DB_FULL_GVH_DOF_COUNT_READY = False
DB_QUANTIZATION_READY = False
DB_WHEELER_DEWITT_GVH_READY = False
DISPERSION_READY = False

HD_OBSTRUCTIONS = []

if not HD_SPATIAL_GENERATOR_RECONSTRUCTED:
    HD_OBSTRUCTIONS.append(
        "SPATIAL-GENERATOR-NOT-RECONSTRUCTED"
    )

if not HD_HAMILTONIAN_SCALAR_DENSITY_W1_CERTIFIED:
    HD_OBSTRUCTIONS.append(
        "HAMILTONIAN-DENSITY-COVARIANCE-NOT-CERTIFIED"
    )

if not HD_CANONICAL_RESIDUAL_BOR_ZERO:
    HD_OBSTRUCTIONS.append(
        "CANONICAL-HD-NOT-CLOSED"
    )

if not HD_D_PRESERVES_SECOND_CLASS_CONSTRAINT_SURFACE:
    HD_OBSTRUCTIONS.append(
        "D-DOES-NOT-PRESERVE-SECOND-CLASS-SURFACE"
    )

if not HD_DIRAC_CORRECTION_WEAK_ZERO_BY_DIFFERENTIAL_IDEAL:
    HD_OBSTRUCTIONS.append(
        "DIRAC-HD-CORRECTION-NOT-IN-DIFFERENTIAL-IDEAL"
    )

HD_LOCAL_AUDIT_PASS = all([
    HD_SECTOR_CLASSIFIED,
    len(HD_OBSTRUCTIONS) == 0,
    DB_HD_STATUS == "WEAK_ZERO_GENERIC_BRANCH",
    DB_DD_STATUS == "UNCLASSIFIED",
    not FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED,
    not HD_FULL_FIRST_CLASS_SECTOR_CLASSIFIED,
    not HD_FULL_GAUGE_GENERATOR_CLASSIFIED,
    not DB_FULL_GVH_DOF_COUNT_READY,
    not DB_QUANTIZATION_READY,
])

HD_NEXT_AUTHORIZED = (
    "AUDIT-DD-HYPERSURFACE-BRACKET"
    if HD_LOCAL_AUDIT_PASS
    else "REPAIR-HD-ONLY"
)

assert DB_DD_STATUS == "UNCLASSIFIED"
assert not FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED
assert not DB_FULL_GVH_DOF_COUNT_READY
assert not DB_QUANTIZATION_READY
assert not DB_WHEELER_DEWITT_GVH_READY

print("DB_HD_STATUS =", DB_HD_STATUS)
print("DB_DD_STATUS =", DB_DD_STATUS)
print("HD_OBSTRUCTIONS =", HD_OBSTRUCTIONS)
print("HD_LOCAL_AUDIT_PASS =", HD_LOCAL_AUDIT_PASS)
print(
    "FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED =",
    FULL_HYPERSURFACE_ALGEBRA_CLASSIFIED
)
print(
    "DB_FULL_GVH_DOF_COUNT_READY =",
    DB_FULL_GVH_DOF_COUNT_READY
)
print(
    "DB_QUANTIZATION_READY =",
    DB_QUANTIZATION_READY
)
print(
    "DB_WHEELER_DEWITT_GVH_READY =",
    DB_WHEELER_DEWITT_GVH_READY
)
print(
    "HD_NEXT_AUTHORIZED =",
    HD_NEXT_AUTHORIZED
)
```
