## 1.2 Symbolisation complète du témoin

Nous remplaçons maintenant les valeurs numériques par des symboles :

\[
x_{00},v_0,R_c,\omega_{10},\omega_{21},u_2.
\]

Le but n'est **pas** de forcer une EOM inexistante.

Le test demande plutôt :

> La construction cesse-t-elle d'exister sans les valeurs numériques,
> ou devient-elle une famille paramétrique exacte ?

C'est une distinction essentielle. Si la construction reste exacte pour
des symboles arbitraires, alors `.3.3.27.2` ne sélectionne pas les
paramètres : elle montre une possibilité cinématique.


```python
t = sp.symbols("t", real=True)
x00, v0, Rc = sp.symbols("x00 v0 Rc", real=True)
w10, w21, u2 = sp.symbols("omega10 omega21 u2", real=True)

def Rz_sym(theta):
    return sp.Matrix([
        [sp.cos(theta), -sp.sin(theta), 0],
        [sp.sin(theta),  sp.cos(theta), 0],
        [0, 0, 1],
    ])

x0_sym = sp.Matrix([x00 + v0*t, 0, 0])
R10_sym = Rz_sym(w10*t)

x1_target_sym = sp.Matrix([
    Rc*sp.cos(w10*t),
    Rc*sp.sin(w10*t),
    0,
])

T10_sym = sp.simplify(x1_target_sym - R10_sym*x0_sym)
x1_constructed_sym = sp.simplify(T10_sym + R10_sym*x0_sym)
x1_identity_residual = sp.simplify(x1_constructed_sym - x1_target_sym)

R21_sym = Rz_sym(w21*t)
T21_sym = sp.Matrix([0, 0, u2*t])
x2_sym = sp.simplify(sp.trigsimp(T21_sym + R21_sym*x1_target_sym))

x2_expected_sym = sp.Matrix([
    Rc*sp.cos((w10+w21)*t),
    Rc*sp.sin((w10+w21)*t),
    u2*t,
])
x2_identity_residual = sp.simplify(
    sp.trigsimp(x2_sym - x2_expected_sym)
)

G273_X1_ENGINEERED_TARGET_IDENTITY = all(z == 0 for z in x1_identity_residual)
G273_X2_HELIX_SYMBOLIC_IDENTITY = all(z == 0 for z in x2_identity_residual)
G273_SYMBOLIC_PARAMETRIC_FAMILY_MATERIALIZED = all([
    G273_X1_ENGINEERED_TARGET_IDENTITY,
    G273_X2_HELIX_SYMBOLIC_IDENTITY,
])

assert G273_SYMBOLIC_PARAMETRIC_FAMILY_MATERIALIZED

print("T10 symbolic =")
sp.pprint(T10_sym)
print("G273_X1_ENGINEERED_TARGET_IDENTITY =", G273_X1_ENGINEERED_TARGET_IDENTITY)
print("G273_X2_HELIX_SYMBOLIC_IDENTITY =", G273_X2_HELIX_SYMBOLIC_IDENTITY)
print("G273_SYMBOLIC_PARAMETRIC_FAMILY_MATERIALIZED =", G273_SYMBOLIC_PARAMETRIC_FAMILY_MATERIALIZED)
```

Le remplacement des constantes par des symboles **ne détruit pas**

\[

\[

\[
T_{10}+R_{10}x_0=x_1^{\rm target}.

Ainsi, le cercle \(C_1\) est un **témoin d'existence construit**,
pas une sortie dynamique sélectionnée par GVH.


# Niveau 2 — Physique établie

La cinématique standard d'un référentiel rigide mobile donne :

\[
x_P=T+Rx_C,
\]

\[
v_P=\dot T+Rv_C+\dot R\,x_C,
\]

avec :

\[
\dot R\,x_C
=
\boldsymbol\omega\times(Rx_C).
\]

Pour l'hélice obtenue dans `.3.3.27.2` :

\[
r_\perp=R_c,\qquad
\phi(t)=(\omega_{10}+\omega_{21})t,\qquad
z(t)=u_2t.
\]

Donc la cinématique établie implique :

\[
v_\perp=R_c(\omega_{10}+\omega_{21}),
\]

et :

\[
a_\perp
=
R_c(\omega_{10}+\omega_{21})^2.
\]

Mais ces relations **ne fournissent aucune force ni EOM qui sélectionne**
\(\omega_{10},\omega_{21},u_2\).

La physique établie sert donc ici de benchmark cinématique, pas de
destination imposée à GVH.


```python
omega_sum = sp.simplify(w10+w21)

v2_sym = sp.simplify(sp.diff(x2_expected_sym, t))
a2_sym = sp.simplify(sp.diff(v2_sym, t))

v_perp_sq = sp.simplify(v2_sym[0]**2 + v2_sym[1]**2)
a_perp_sq = sp.simplify(a2_sym[0]**2 + a2_sym[1]**2)

EXPECTED_VPERP_SQ = sp.simplify(Rc**2 * omega_sum**2)
EXPECTED_APERP_SQ = sp.simplify(Rc**2 * omega_sum**4)

G273_ESTABLISHED_KINEMATICS_VELOCITY_PASS = (
    sp.simplify(v_perp_sq - EXPECTED_VPERP_SQ) == 0
)
G273_ESTABLISHED_KINEMATICS_ACCELERATION_PASS = (
    sp.simplify(a_perp_sq - EXPECTED_APERP_SQ) == 0
)
G273_ESTABLISHED_KINEMATICS_BENCHMARK_PASS = all([
    G273_ESTABLISHED_KINEMATICS_VELOCITY_PASS,
    G273_ESTABLISHED_KINEMATICS_ACCELERATION_PASS,
])

assert G273_ESTABLISHED_KINEMATICS_BENCHMARK_PASS

print("v2 symbolic =")
sp.pprint(v2_sym)
print("a2 symbolic =")
sp.pprint(a2_sym)
print("G273_ESTABLISHED_KINEMATICS_BENCHMARK_PASS =", G273_ESTABLISHED_KINEMATICS_BENCHMARK_PASS)
```

# Niveau 3 — Test numérique minimal

Nous testons maintenant plusieurs jeux de paramètres différents.

Le but n'est pas de chercher le « meilleur » jeu.

Le test falsifiable est :

> Si la construction est réellement une identité cinématique engineered,
> plusieurs choix arbitraires de \(\omega_{10},\omega_{21},u_2,R_c\)
> doivent tous reproduire ligne/cercle/hélice avec des résidus numériques
> proches de la précision machine.

Si cela se produit, le notebook confirme la cohérence numérique de la
construction **et simultanément** son absence de pouvoir de sélection
dynamique.


```python
def Rz(theta):
    c, s = np.cos(theta), np.sin(theta)
    return np.array([[c,-s,0.0],[s,c,0.0],[0.0,0.0,1.0]])

def run_witness(params):
    Tmax = params["Tmax"]
    N = params["N"]
    x00 = params["x00"]
    v0 = params["v0"]
    Rc = params["Rc"]
    w10 = params["w10"]
    w21 = params["w21"]
    u2 = params["u2"]

    tt = np.linspace(0.0, Tmax, N)
    x0 = np.column_stack([
        x00 + v0*tt,
        np.zeros_like(tt),
        np.zeros_like(tt),
    ])

    x1_target = np.column_stack([
        Rc*np.cos(w10*tt),
        Rc*np.sin(w10*tt),
        np.zeros_like(tt),
    ])

    x1 = np.empty_like(x0)
    T10 = np.empty_like(x0)
    for i,tv in enumerate(tt):
        R10 = Rz(w10*tv)
        T10[i] = x1_target[i] - R10 @ x0[i]
        x1[i] = T10[i] + R10 @ x0[i]

    x2 = np.empty_like(x1)
    T21 = np.column_stack([
        np.zeros_like(tt),
        np.zeros_like(tt),
        u2*tt,
    ])
    for i,tv in enumerate(tt):
        x2[i] = T21[i] + Rz(w21*tv) @ x1[i]

    radius1 = np.sqrt(x1[:,0]**2 + x1[:,1]**2)
    radius2 = np.sqrt(x2[:,0]**2 + x2[:,1]**2)
    phi2 = np.unwrap(np.arctan2(x2[:,1],x2[:,0]))
    fit_phi = np.polyfit(tt, phi2, 1)[0]
    fit_z = np.polyfit(tt, x2[:,2], 1)[0]

    x1_rec = np.empty_like(x1)
    x0_rec = np.empty_like(x0)
    for i,tv in enumerate(tt):
        x1_rec[i] = Rz(w21*tv).T @ (x2[i]-T21[i])
        x0_rec[i] = Rz(w10*tv).T @ (x1_rec[i]-T10[i])

    return {
        "max_C1_radius_error": float(np.max(np.abs(radius1-Rc))),
        "max_C2_radius_error": float(np.max(np.abs(radius2-Rc))),
        "C2_omega_error": float(abs(fit_phi-(w10+w21))),
        "C2_zrate_error": float(abs(fit_z-u2)),
        "max_C0_inverse_error": float(np.max(np.abs(x0_rec-x0))),
    }

parameter_sets = [
    {"name":"canonical","Tmax":20.0,"N":2001,"x00":1.25,"v0":0.18,"Rc":7.0,"w10":0.45,"w21":0.20,"u2":0.35},
    {"name":"set_A","Tmax":12.0,"N":1601,"x00":0.5,"v0":0.08,"Rc":3.5,"w10":0.12,"w21":0.31,"u2":0.07},
    {"name":"set_B","Tmax":8.0,"N":1401,"x00":2.0,"v0":-0.05,"Rc":9.0,"w10":0.70,"w21":-0.15,"u2":0.44},
    {"name":"set_C","Tmax":15.0,"N":1801,"x00":-1.0,"v0":0.22,"Rc":1.8,"w10":-0.25,"w21":0.60,"u2":-0.12},
    {"name":"set_D","Tmax":10.0,"N":1501,"x00":4.0,"v0":0.01,"Rc":12.0,"w10":0.03,"w21":0.05,"u2":1.20},
]

rows = []
for ps in parameter_sets:
    r = run_witness(ps)
    row = {"name":ps["name"], **r}
    row["pass"] = max(r.values()) < 1e-10
    rows.append(row)

sweep = pd.DataFrame(rows)

G273_MINIMAL_NUMERIC_SWEEP_PASS = bool(sweep["pass"].all())
G273_PARAMETER_SET_COUNT = len(sweep)
G273_BENCHMARK_SELECTS_UNIQUE_PARAMETER_SET = False

assert G273_MINIMAL_NUMERIC_SWEEP_PASS
assert not G273_BENCHMARK_SELECTS_UNIQUE_PARAMETER_SET

print(sweep.to_string(index=False))
print("G273_MINIMAL_NUMERIC_SWEEP_PASS =", G273_MINIMAL_NUMERIC_SWEEP_PASS)
print("G273_PARAMETER_SET_COUNT =", G273_PARAMETER_SET_COUNT)
print("G273_BENCHMARK_SELECTS_UNIQUE_PARAMETER_SET =", G273_BENCHMARK_SELECTS_UNIQUE_PARAMETER_SET)
```
