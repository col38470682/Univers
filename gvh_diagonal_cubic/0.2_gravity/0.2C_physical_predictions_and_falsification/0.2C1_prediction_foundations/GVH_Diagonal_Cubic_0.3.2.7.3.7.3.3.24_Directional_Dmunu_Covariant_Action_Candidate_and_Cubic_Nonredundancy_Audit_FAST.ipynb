\[

```python
d1,d2=sp.symbols("d1 d2", real=True)
d3=-d1-d2
J2=sp.expand(d1**2+d2**2+d3**2)
J3=sp.expand(d1**3+d2**3+d3**3)
detD=sp.expand(d1*d2*d3)
J3_EQUALS_3DET_PASS=sp.simplify(J3-3*detD)==0
assert J3_EQUALS_3DET_PASS
print("J2 =",sp.factor(J2))
print("J3 =",sp.factor(J3))
print("J3_EQUALS_3DET_PASS =",J3_EQUALS_3DET_PASS)
```

## Candidat auxiliaire

\[
\mathcal L_D=

À \(\lambda_3=0\) :

\[
D_{\mu\nu}=-\frac{\kappa_D}{M_D^2}\sigma_{\mu\nu},

donc :

\[
\mathcal L_{\rm eff}=
-\frac{\kappa_D^2}{2M_D^2}\sigma^2.

À premier ordre en \(\lambda_3\), un terme \(\mathrm{tr}(\sigma^3)\) apparaît, mais il n'établit toujours pas une signature GVH unique.

## Une vraie signature cubique exige plus qu'un tenseur STF

Un \(D_{ij}\) symétrique possède des axes propres, mais ces axes ne constituent pas automatiquement trois directions physiques privilégiées.

Si la base **Diagonal Cubic** signifie réellement trois axes spatiaux physiques avec symétrie cubique discrète, le candidat naturel est un trièdre :

\[
e_{(A)}^\mu,\quad A=1,2,3,
\]

\[
u_\mu e_{(A)}^\mu=0,\qquad
g_{\mu\nu}e_{(A)}^\mu e_{(B)}^\nu=\delta_{AB},
\]

et le tenseur :

\[
\boxed{
C_{\mu\nu\rho\sigma}
=
\sum_A e^{(A)}_\mu e^{(A)}_\nu e^{(A)}_\rho e^{(A)}_\sigma.
}
\]

```python
A,B=sp.symbols("A B", real=True)
# Isotropic rank-4 tensor T=A δijδkl+B(δikδjl+δilδjk)
T1111=A+2*B
T1122=A
T1212=B
# Cubic tensor C=sum axes e^4
C1111=sp.Integer(1); C1122=sp.Integer(0); C1212=sp.Integer(0)
sol=sp.solve([sp.Eq(T1122,C1122),sp.Eq(T1212,C1212)],[A,B],dict=True)
ISOTROPIC_MATCH_EXISTS=bool(sol) and sp.simplify(T1111.subs(sol[0])-C1111)==0
CUBIC_TENSOR_NOT_ISOTROPIC_PASS=not ISOTROPIC_MATCH_EXISTS
assert CUBIC_TENSOR_NOT_ISOTROPIC_PASS
print("isotropic solution from mixed components =",sol)
print("T1111 after that =",T1111.subs(sol[0]))
print("C1111 =",C1111)
print("CUBIC_TENSOR_NOT_ISOTROPIC_PASS =",CUBIC_TENSOR_NOT_ISOTROPIC_PASS)
```

Ainsi, \(C_{\mu\nu\rho\sigma}\) ne peut pas être réduit à un tenseur isotrope de rang quatre construit seulement avec \(h_{\mu\nu}\).

Si ce tenseur cubique est une structure **physique** et non un simple choix de repère, il contient donc une information directionnelle absente du contenu de champ Einstein-æther minimal \(\{g_{\mu\nu},u^\mu\}\).

## Témoin d'orientation

Considérons :

\[
\mathcal O_C=C^{\mu\nu\rho\sigma}\sigma_{\mu\nu}\sigma_{\rho\sigma}.
\]

Dans le trièdre cubique :

\[
\mathcal O_C=\sigma_{11}^2+\sigma_{22}^2+\sigma_{33}^2.
\]

Un invariant isotrope \(\sigma^2\) ne change pas sous rotation du cisaillement ; \(\mathcal O_C\), lui, change si le trièdre est physiquement fixé.

```python
s=sp.symbols("s", real=True, nonzero=True)
rt=sp.sqrt(2)
Sigma=sp.Matrix([[s,0,0],[0,-s,0],[0,0,0]])
R=sp.Matrix([[1/rt,-1/rt,0],[1/rt,1/rt,0],[0,0,1]])
SigmaR=sp.simplify(R*Sigma*R.T)

def frob2(M):
    return sp.simplify(sum(M[i,j]**2 for i in range(3) for j in range(3)))
def OC(M):
    return sp.simplify(sum(M[i,i]**2 for i in range(3)))

iso0,iso1=frob2(Sigma),frob2(SigmaR)
cub0,cub1=OC(Sigma),OC(SigmaR)

ISOTROPIC_SHEAR_ROTATION_INVARIANT_PASS=sp.simplify(iso0-iso1)==0
CUBIC_OPERATOR_ORIENTATION_SENSITIVE_PASS=sp.simplify(cub0-cub1)!=0
assert ISOTROPIC_SHEAR_ROTATION_INVARIANT_PASS
assert CUBIC_OPERATOR_ORIENTATION_SENSITIVE_PASS

print("Sigma rotated =",SigmaR)
print("sigma^2 before/after =",iso0,iso1)
print("O_C before/after =",cub0,cub1)
print("ISOTROPIC_SHEAR_ROTATION_INVARIANT_PASS =",ISOTROPIC_SHEAR_ROTATION_INVARIANT_PASS)
print("CUBIC_OPERATOR_ORIENTATION_SENSITIVE_PASS =",CUBIC_OPERATOR_ORIENTATION_SENSITIVE_PASS)
```
