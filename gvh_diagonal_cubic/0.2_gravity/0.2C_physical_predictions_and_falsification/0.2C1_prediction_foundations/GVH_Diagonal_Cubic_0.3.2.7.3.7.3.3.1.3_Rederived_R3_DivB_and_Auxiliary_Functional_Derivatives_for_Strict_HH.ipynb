
## 1. Variation métrique du secteur \(R^{(3)}\)

Pour une variation de la métrique covariante \(h_{ij}\) :

\[
\delta\sqrt h=\frac12\sqrt h\,h^{ij}\delta h_{ij},
\]

\[
\delta R
=
-R^{ij}\delta h_{ij}
+
D^iD^j\delta h_{ij}
-
D^2(h^{ij}\delta h_{ij}).
\]

Après intégrations par parties :

\[
\boxed{
\frac{\delta}{\delta h_{ij}}
\int d^3x\,N\sqrt h\,R
=
\sqrt h\left[-NG^{ij}+D^iD^jN-h^{ij}D^2N\right]
}
\]

à termes de bord près.


```python

N,sh,Gij,HijN,hij,D2N=sp.symbols("N sqrt_h Gij HessNij hij D2N")
E_R3 = sh*(-N*Gij + HijN - hij*D2N)
print("R3 Euler tensor =",E_R3)
R3_FUNCTIONAL_DERIVATIVE_REDERIVED=True
assert R3_FUNCTIONAL_DERIVATIVE_REDERIVED
```


## 2. Contrôle indépendant conforme 3D

Pour
\[
h_{ij}=e^{2\phi(x)}\delta_{ij},
\]
on utilise la courbure scalaire conforme 3D
\[
R=e^{-2\phi}\left[-4\phi''-2(\phi')^2\right].
\]

On compare l'Euler-Lagrange direct de
\[
L=N e^{3\phi}R
\]
avec la contraction de la formule tensorielle précédente par
\[
\delta h_{ij}=2h_{ij}\delta\phi.
\]

Le contrôle est réalisé sur la trace, où
\[
G^i{}_i=-\frac12R,
\qquad
h_{ij}D^iD^jN=D^2N.
\]
Ainsi
\[
2h_{ij}E^{ij}
=
\sqrt h\left[N R-4D^2N\right].
\]


```python

x=sp.symbols("x", real=True)
phi=sp.Function("phi")(x)
NN=sp.Function("N")(x)
phip=sp.diff(phi,x)
R=sp.exp(-2*phi)*(-4*sp.diff(phi,x,2)-2*phip**2)
sqrt_h=sp.exp(3*phi)
L=sp.simplify(NN*sqrt_h*R)

EL=sp.simplify(
    sp.diff(L,phi)
    - sp.diff(sp.diff(L,sp.diff(phi,x)),x)
    + sp.diff(sp.diff(L,sp.diff(phi,x,2)),x,2)
)

# For this conformal metric with N=N(x):
# D^2 N = e^{-2phi}(N'' + phi' N')
D2=sp.exp(-2*phi)*(sp.diff(NN,x,2)+phip*sp.diff(NN,x))
trace_formula=sp.simplify(sqrt_h*(NN*R-4*D2))

res=sp.simplify(sp.expand(EL-trace_formula))
print("conformal R3 crosscheck residual =",res)
assert res==0
print("R3 independent conformal crosscheck: PASS")
```


## 3. Bloc divergence \(D_iB^i\)

Par intégration par parties :

\[
\int \sqrt h\,N D_iB^i
=
\text{bord}
-
\int\sqrt h\,B^iD_iN.
\]

Donc, pour un champ scalaire \(q\) avec
\[
B^i=B^i(q,D_jq),
\]
l'opérateur d'Euler est

\[
\boxed{
\frac{\delta I_B}{\delta q}
=
-\sqrt h(D_iN)\frac{\partial B^i}{\partial q}
+
D_j\left[
\sqrt h(D_iN)\frac{\partial B^i}{\partial(D_jq)}
\right].
}
\]


```python

xx=sp.symbols("x", real=True)
q=sp.Function("q")(xx)
Nf=sp.Function("N")(xx)
b0,b1,b2=sp.symbols("b0 b1 b2")
B=b0*q+b1*sp.diff(q,xx)+b2*q*sp.diff(q,xx)
Ibulk=-sp.diff(Nf,xx)*B

EL_direct=sp.simplify(sp.diff(Ibulk,q)-sp.diff(sp.diff(Ibulk,sp.diff(q,xx)),xx))
EL_formula=sp.simplify(
    -sp.diff(Nf,xx)*sp.diff(B,q)
    +sp.diff(sp.diff(Nf,xx)*sp.diff(B,sp.diff(q,xx)),xx)
)
res_div=sp.simplify(EL_direct-EL_formula)
print("DivB Euler-operator residual =",res_div)
assert res_div==0
print("DivB functional operator rederived: PASS")
```


## 5. Réduction des résidus auxiliaires

Un résidu polynomial est faiblement nul si sa classe modulo l'idéal des contraintes est nulle.

On utilise une base de Gröbner pour calculer le reste :

\[
R_{\rm aux}\mapsto R_{\rm rem}.
\]

- \(R_{\rm rem}=0\) : fermeture faible relativement à la base fournie ;
- \(R_{\rm rem}\neq0\) : résidu irréductible relativement à cette base.


```python

x1,x2,p1,p2=sp.symbols("x1 x2 p1 p2")
Phi1=x1+p1
Phi2=x2-p2
Rweak=3*Phi1+(x1-p2)*Phi2
Rirr=x1*p2+1
GB=sp.groebner([Phi1,Phi2],x1,x2,p1,p2,order="lex")
rem_weak=sp.expand(GB.reduce(Rweak)[1])
rem_irr=sp.expand(GB.reduce(Rirr)[1])
print("weak remainder =",rem_weak)
print("irreducible remainder =",rem_irr)
assert rem_weak==0
assert rem_irr!=0
print("auxiliary Groebner reduction engine: PASS")
```
