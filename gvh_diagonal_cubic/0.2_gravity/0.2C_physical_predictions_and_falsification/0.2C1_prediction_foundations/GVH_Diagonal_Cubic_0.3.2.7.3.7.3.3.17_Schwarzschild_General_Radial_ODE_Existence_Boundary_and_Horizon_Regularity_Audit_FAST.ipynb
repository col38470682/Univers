# ODE17.2 — Équation vectorielle asymptotique

À grande distance :

\[
N=1-\frac{M}{r}+O(r^{-2}),
\qquad
A=1+\frac{M}{r}+O(r^{-2}),
\]

et pour une solution asymptotiquement alignée :

\[
|\eta|\ll1.
\]

Au premier ordre en \(\eta\), la contribution \(c_4a^2\) ne modifie pas l'opérateur homogène plat, tandis que :

\[
\mathcal L_{\eta,\infty}^{(2)}
=
-(c_1+c_3)
\left[
r^2(\eta')^2+2\eta^2
\right]
-c_2r^2
\left(
\eta'+\frac{2\eta}{r}
\right)^2.
\]

L'équation d'Euler–Lagrange doit se réduire à :

\[
\boxed{
2(c_1+c_2+c_3)
\left(
r^2\eta''+2r\eta'-2\eta
\right)=0.
}
\]

```python
# ODE17.3 — Derive the asymptotic linear Euler operator exactly
r,M = sp.symbols("r M", positive=True)
c1,c2,c3,c4 = sp.symbols("c1 c2 c3 c4", real=True)
eta = sp.Function("eta")(r)

c13 = c1+c3
c123 = c1+c2+c3
c14 = c1+c4

L_asym_quad = (
    -c13*(r**2*sp.diff(eta,r)**2 + 2*eta**2)
    -c2*r**2*(sp.diff(eta,r) + 2*eta/r)**2
)

Eeta_asym = sp.factor(
    sp.diff(L_asym_quad,eta)
    - sp.diff(sp.diff(L_asym_quad,sp.diff(eta,r)),r)
)

Eeta_asym_expected = sp.factor(
    2*c123*(r**2*sp.diff(eta,r,2) + 2*r*sp.diff(eta,r) - 2*eta)
)

ASYMPTOTIC_LINEAR_VECTOR_EOM_DERIVED = (
    sp.simplify(Eeta_asym-Eeta_asym_expected) == 0
)
assert ASYMPTOTIC_LINEAR_VECTOR_EOM_DERIVED

print("E_eta_asym =", Eeta_asym)
print("ASYMPTOTIC_LINEAR_VECTOR_EOM_DERIVED =", ASYMPTOTIC_LINEAR_VECTOR_EOM_DERIVED)
```

```python
# ODE17.5 — Indicial roots
m = sp.symbols("m", real=True)
indicial = sp.factor(m*(m-1)+2*m-2)
roots = sp.solve(sp.Eq(indicial,0),m)

INDICIAL_POLYNOMIAL = indicial
INDICIAL_ROOTS = roots
ASYMPTOTIC_GROWING_EXPONENT = sp.Integer(1)
ASYMPTOTIC_DECAYING_EXPONENT = sp.Integer(-2)

INDICIAL_ROOTS_PASS = set(roots) == {sp.Integer(1), sp.Integer(-2)}
GENERIC_DECAYING_MODE_IS_R_MINUS_2 = INDICIAL_ROOTS_PASS

assert GENERIC_DECAYING_MODE_IS_R_MINUS_2

print("indicial =", indicial)
print("roots =", roots)
print("GENERIC_DECAYING_MODE_IS_R_MINUS_2 =", GENERIC_DECAYING_MODE_IS_R_MINUS_2)
```

# ODE17.6 — Test de la contrainte radiale \(F=0\)

Sur Schwarzschild :

\[
f=1-\frac{2M}{r},
\qquad
N=\sqrt f,
\qquad
A=\frac1{\sqrt f},
\]

\[
\nu\equiv\frac{N'}N
=
\frac{M}{r(r-2M)}.
\]

Le \(F\) exact de `.3.3.16` est :

\[
\begin{aligned}
F={}&-c_1\left(\eta'^2-\nu^2+\frac{2\sinh^2\eta}{r^2}\right)\\
&-c_2\left(\cosh\eta\,\eta'+\sinh\eta\,\nu+\frac{2\sinh\eta}{r}\right)^2\\
&-c_3\left[
(\cosh\eta\,\eta'+\sinh\eta\,\nu)^2+\frac{2\sinh^2\eta}{r^2}
\right]\\
&+c_4(\sinh\eta\,\eta'+\cosh\eta\,\nu)^2.
\end{aligned}
\]

L'équation provenant de la variation radiale impose nécessairement :

\[
\boxed{F=0}.
\]

On injecte maintenant la seule branche décroissante permise :

\[
\eta=\frac{Q}{r^2}+O(r^{-3}).
\]

```python
# ODE17.7 — Exact leading asymptotic residual of F
Q = sp.symbols("Q", real=True)
x = sp.symbols("x", positive=True)  # x = 1/r, set M symbolic through M*x
# Work directly in r with eta=Q/r^2 and expand at infinity via x=1/r.
eta_trial = Q*x**2
r_x = 1/x
nu_x = M*x**2/(1-2*M*x)
eta_prime_x = -x**2*sp.diff(eta_trial,x)

s = sp.sinh(eta_trial)
ch = sp.cosh(eta_trial)

F_x = (
    -c1*(eta_prime_x**2 - nu_x**2 + 2*s**2*x**2)
    -c2*(ch*eta_prime_x + s*nu_x + 2*s*x)**2
    -c3*((ch*eta_prime_x + s*nu_x)**2 + 2*s**2*x**2)
    +c4*(s*eta_prime_x + ch*nu_x)**2
)

F_series = sp.series(F_x, x, 0, 7).removeO().expand()
F_x4 = sp.factor(F_series.coeff(x,4))
F_x5 = sp.factor(F_series.coeff(x,5))

ASYMPTOTIC_F_LEADING_COEFFICIENT = F_x4
ASYMPTOTIC_F_LEADING_C14_PASS = (
    sp.simplify(F_x4 - c14*M**2) == 0
)
assert ASYMPTOTIC_F_LEADING_C14_PASS

print("F[x^4] =", F_x4)
print("F[x^5] =", F_x5)
print("ASYMPTOTIC_F_LEADING_C14_PASS =", ASYMPTOTIC_F_LEADING_C14_PASS)
```

Le résultat est :

\[
\boxed{
F
=
(c_1+c_4)\frac{M^2}{r^4}
+O(r^{-5}).
}
\]

La constante \(Q\) du mode radial décroissant n'intervient donc pas au premier ordre nécessaire pour annuler ce terme.

Sur la branche faible champ stable :

\[
\boxed{c_1+c_4>0},
\]

on a nécessairement :

\[
F\neq0
\]

pour \(r\) suffisamment grand.

Ainsi, une solution globale satisfaisant simultanément :

\[
\eta\to0,\qquad
c_1+c_2+c_3\neq0,\qquad
c_1+c_4>0

ne peut pas résoudre le système radial nécessaire sur la métrique Schwarzschild exacte.

# ODE17.11 — Régularité au futur horizon

Les coordonnées Schwarzschild sont singulières à \(r=2M\). On passe donc à la coordonnée ingoing :

\[
v=t+r_*,
\qquad
\frac{dr_*}{dr}=\frac1f.
\]

Pour :

\[
u^t=\frac{\cosh\eta}{\sqrt f},
\qquad
u^r=\sqrt f\,\sinh\eta,
\]

on obtient :

\[
u^v
=
u^t+\frac{u^r}{f}
=
\frac{\cosh\eta+\sinh\eta}{\sqrt f}
=
\boxed{\frac{e^\eta}{\sqrt f}}.
\]

On définit :

\[
\boxed{
Y(r)\equiv\frac{e^{\eta(r)}}{\sqrt f}.
}
\]

Alors :

\[
u^v=Y,
\]

et exactement :

\[
\boxed{
u^r=\frac{fY-Y^{-1}}{2}.
}
\]

Une branche future-ingressante régulière à l'horizon requiert :

\[
\boxed{
0<Y_H<\infty.
}
\]

Cela implique :

\[
\eta
=
\frac12\ln f+\ln Y_H+o(1)
\]

près de \(r=2M^+\).

```python
# ODE17.12 — Horizon regularity identity
fY,Y = sp.symbols("f Y", positive=True)
etaY = sp.log(Y*sp.sqrt(fY))

uv = sp.simplify(sp.exp(etaY)/sp.sqrt(fY))
ur = sp.simplify(
    sp.sqrt(fY)*(sp.exp(etaY)-sp.exp(-etaY))/2
)
ur_expected = sp.simplify((fY*Y-1/Y)/2)

HORIZON_UV_IDENTITY_PASS = (sp.simplify(uv-Y) == 0)
HORIZON_UR_IDENTITY_PASS = (sp.simplify(ur-ur_expected) == 0)
HORIZON_REGULARITY_VARIABLE_MATERIALIZED = all([
    HORIZON_UV_IDENTITY_PASS,
    HORIZON_UR_IDENTITY_PASS,
])

assert HORIZON_REGULARITY_VARIABLE_MATERIALIZED

print("u^v =", uv)
print("u^r =", ur)
print("HORIZON_REGULARITY_VARIABLE_MATERIALIZED =", HORIZON_REGULARITY_VARIABLE_MATERIALIZED)
```
