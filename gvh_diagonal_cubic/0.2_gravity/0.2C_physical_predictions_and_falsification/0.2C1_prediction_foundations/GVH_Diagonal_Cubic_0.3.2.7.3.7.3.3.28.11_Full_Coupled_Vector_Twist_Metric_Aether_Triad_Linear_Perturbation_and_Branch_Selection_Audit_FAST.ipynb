# Niveau 2.2 — Orientation du trièdre dans le mode vectoriel

Le cisaillement transverse possède, dans le plan \(xy\), la forme :

\[
\Sigma
=
\begin{pmatrix}
0&s&0\\
s&0&0\\
0&0&0
\end{pmatrix}.
\]

Pour un trièdre tourné d'un angle \(\phi\) dans le plan \(xy\) :

\[
R(\phi)
=
\begin{pmatrix}
\cos\phi&-\sin\phi&0\\
\sin\phi&\cos\phi&0\\
0&0&1
\end{pmatrix},
\]

on définit :

\[
\boxed{
\chi(\phi)
:=
\frac{\mathcal O_C}{\sigma^2}.
}
\]

Le calcul donne :

\[
\boxed{
\chi(\phi)=\sin^2(2\phi)
}
\]

donc :

\[
\boxed{
0\leq\chi\leq1.
}
\]

Deux cas extrêmes :

\[
\chi=0
\]

pour un trièdre où le cisaillement est purement hors-diagonal, et :

\[
\chi=1
\]

dans le trièdre propre qui diagonalise \(\Sigma\).

```python
phi,s = sp.symbols("phi s", real=True)

Sigma = sp.Matrix([
    [0,s,0],
    [s,0,0],
    [0,0,0],
])
R = sp.Matrix([
    [sp.cos(phi),-sp.sin(phi),0],
    [sp.sin(phi), sp.cos(phi),0],
    [0,0,1],
])

Sigma_R = sp.simplify(R.T*Sigma*R)
OC_phi = sp.simplify(
    sum(Sigma_R[i,i]**2 for i in range(3))
)
sigma2_matrix = sp.simplify(
    sum(Sigma[i,j]**2 for i in range(3) for j in range(3))
)
chi_phi = sp.trigsimp(sp.simplify(OC_phi/sigma2_matrix))

G2811_CHI_ORIENTATION_EXACT_PASS = (
    sp.trigsimp(chi_phi-sp.sin(2*phi)**2) == 0
)
G2811_CHI_ZERO_WITNESS_PASS = (
    sp.simplify(chi_phi.subs(phi,0)) == 0
)
G2811_CHI_ONE_WITNESS_PASS = (
    sp.simplify(chi_phi.subs(phi,sp.pi/4)) == 1
)

assert G2811_CHI_ORIENTATION_EXACT_PASS
assert G2811_CHI_ZERO_WITNESS_PASS
assert G2811_CHI_ONE_WITNESS_PASS

print("Sigma_R =")
sp.pprint(Sigma_R)
print("O_C =", OC_phi)
print("sigma^2 =", sigma2_matrix)
print("chi(phi) =", chi_phi)
```

```python
s0obs,s1obs,c13obs,c14obs,cmobs = sp.symbols(
    "s0obs s1obs c13obs c14obs cminus_obs",
    real=True
)

cm_from_s0 = sp.factor(
    2*c14obs*s0obs
    - c13obs/(1-c13obs)
)

C1_from_s1 = sp.factor(
    (2*c14obs*s1obs-cmobs)
    /(1-cmobs+2*c14obs*s1obs)
)

zeta_from_pair = sp.factor(
    2*(c13obs-C1_from_s1)
)

# Exact recovery checks against symbolic model.
cm_recovered = sp.simplify(
    cm_from_s0.subs({
        c13obs:c13,
        c14obs:c14,
        s0obs:s0sq,
    })
)

C1_recovered = sp.simplify(
    C1_from_s1.subs({
        c14obs:c14,
        s1obs:s1sq,
        cmobs:cminus,
    })
)

zeta_recovered = sp.simplify(
    zeta_from_pair.subs({
        c13obs:c13,
        c14obs:c14,
        s1obs:s1sq,
        cmobs:cminus,
    })
)

G2811_TWO_SPEED_CMINUS_RECOVERY_EXACT_PASS = (
    sp.simplify(cm_recovered-cminus) == 0
)
G2811_TWO_SPEED_C1_EFFECTIVE_RECOVERY_EXACT_PASS = (
    sp.simplify(C1_recovered-(c13-zeta/2)) == 0
)
G2811_TWO_SPEED_ZETA_RECOVERY_EXACT_PASS = (
    sp.simplify(zeta_recovered-zeta) == 0
)

assert G2811_TWO_SPEED_CMINUS_RECOVERY_EXACT_PASS
assert G2811_TWO_SPEED_C1_EFFECTIVE_RECOVERY_EXACT_PASS
assert G2811_TWO_SPEED_ZETA_RECOVERY_EXACT_PASS

print("c_minus from s0^2 =")
sp.pprint(cm_from_s0)
print("C1 from s1^2 =")
sp.pprint(C1_from_s1)
print("zeta_C from pair =")
sp.pprint(zeta_from_pair)
```
