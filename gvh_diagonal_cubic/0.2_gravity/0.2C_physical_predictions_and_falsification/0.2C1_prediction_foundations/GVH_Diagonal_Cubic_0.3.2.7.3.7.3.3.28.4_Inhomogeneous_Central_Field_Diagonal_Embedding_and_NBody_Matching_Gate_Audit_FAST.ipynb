## 1.1 Hessien exact du potentiel central

Pour :

\[
\Phi(\mathbf x)
=
-\frac{\mu}{r},
\qquad
r=\sqrt{x^2+y^2+z^2},
\]

l'accélération est :

\[
\mathbf g
=
-\nabla\Phi
=
-\frac{\mu}{r^3}\mathbf x.
\]

La différence d'accélération entre deux géodésiques proches est :

\[
\delta\ddot{\mathbf x}
=
\mathcal T\,\boldsymbol\xi
\]

avec :

\[
\boxed{
\mathcal T
=
\nabla\mathbf g
=
-\nabla\nabla\Phi.
}
\]

Le tenseur attendu est :

\[
\boxed{
\mathcal T_{ij}
=
\frac{\mu}{r^3}
\left(
3n_in_j-\delta_{ij}
\right).
}
\]

Il est :

- symétrique ;
- trace-free hors de la source ;
- diagonal dans la base radiale/tangentielle ;
- spatialement inhomogène via \(r^{-3}\).

```python
x,y,z,mu = sp.symbols("x y z mu", real=True)
r = sp.sqrt(x**2+y**2+z**2)
Phi = -mu/r
coords = sp.Matrix([x,y,z])

gradPhi = sp.Matrix([sp.diff(Phi,q) for q in (x,y,z)])
g = -gradPhi
T = sp.simplify(g.jacobian([x,y,z]))

n = coords/r
T_expected = sp.simplify(mu/r**3 * (3*n*n.T - sp.eye(3)))

G284_CENTRAL_TIDAL_TENSOR_EXACT_PASS = all(
    sp.simplify(T[i,j]-T_expected[i,j]) == 0
    for i in range(3) for j in range(3)
)
G284_CENTRAL_TIDAL_TRACEFREE_PASS = sp.simplify(sp.trace(T)) == 0

# Axis witness x=r0,y=z=0.
r0 = sp.symbols("r0", positive=True, real=True)
T_axis = sp.simplify(T.subs({x:r0,y:0,z:0}))
T_axis_expected = sp.diag(2*mu/r0**3,-mu/r0**3,-mu/r0**3)

G284_CENTRAL_RADIAL_TANGENTIAL_EIGENVALUES_PASS = all(
    sp.simplify(T_axis[i,j]-T_axis_expected[i,j]) == 0
    for i in range(3) for j in range(3)
)

assert G284_CENTRAL_TIDAL_TENSOR_EXACT_PASS
assert G284_CENTRAL_TIDAL_TRACEFREE_PASS
assert G284_CENTRAL_RADIAL_TANGENTIAL_EIGENVALUES_PASS

print("g =", g)
print("T_axis =")
sp.pprint(T_axis)
print("trace(T) =", sp.simplify(sp.trace(T)))
print("G284_CENTRAL_TIDAL_TENSOR_EXACT_PASS =",
      G284_CENTRAL_TIDAL_TENSOR_EXACT_PASS)
print("G284_CENTRAL_RADIAL_TANGENTIAL_EIGENVALUES_PASS =",
      G284_CENTRAL_RADIAL_TANGENTIAL_EIGENVALUES_PASS)
```

## 1.2 Diagonalisation locale et covariance rotationnelle

À tout point \(\mathbf x\neq0\), choisissons une matrice orthogonale \(Q(\mathbf x)\) dont le premier axe est :

\[
\mathbf e_r=\mathbf n.
\]

Alors :

\[
\boxed{
Q^T\mathcal TQ
=
\frac{\mu}{r^3}
\operatorname{diag}(2,-1,-1).
}
\]

La direction radiale est une direction d'étirement tidal, tandis que les deux directions tangentielles sont compressives.

Le caractère diagonal ne dépend donc pas d'un axe cartésien privilégié : il est attaché au repère propre local du champ central.

```python
def central_tidal_tensor(pos, mu=1.0):
    pos = np.asarray(pos, dtype=float)
    rr = np.linalg.norm(pos)
    if rr <= 0:
        raise ValueError("central singularity")
    nn = pos/rr
    return mu/rr**3 * (3*np.outer(nn,nn)-np.eye(3))

rng = np.random.default_rng(284)
eig_errors = []
trace_errors = []

for _ in range(200):
    pos = rng.normal(size=3)
    pos *= rng.uniform(0.5,5.0)/np.linalg.norm(pos)
    muv = rng.uniform(0.2,3.0)
    Tv = central_tidal_tensor(pos,muv)
    rr = np.linalg.norm(pos)
    eig = np.linalg.eigvalsh(Tv)
    target = np.sort(np.array([-muv/rr**3,-muv/rr**3,2*muv/rr**3]))
    eig_errors.append(np.max(np.abs(eig-target)))
    trace_errors.append(abs(np.trace(Tv)))

G284_ROTATION_COVARIANT_EIGENVALUE_SWEEP_PASS = max(eig_errors) < 1e-12
G284_NUMERIC_TRACEFREE_SWEEP_PASS = max(trace_errors) < 1e-12

assert G284_ROTATION_COVARIANT_EIGENVALUE_SWEEP_PASS
assert G284_NUMERIC_TRACEFREE_SWEEP_PASS

print("max eigenvalue error over 200 random directions =", max(eig_errors))
print("max trace error =", max(trace_errors))
print("G284_ROTATION_COVARIANT_EIGENVALUE_SWEEP_PASS =",
      G284_ROTATION_COVARIANT_EIGENVALUE_SWEEP_PASS)
```

\[

\ddot D_J

\[
D_J
=
\operatorname{diag}(d_r,d_t,d_t),
\]

avec :

\[
\ddot d_r
=
2\kappa d_r,
\]

\[
\ddot d_t
=
-\kappa d_t,
\]

où :

\[
\boxed{
\kappa(\mathbf x_0)
=
\frac{\mu}{r_0^3}.
}
\]

Pour les conditions initiales :

\[
D_J(0)=I,
\qquad
\dot D_J(0)=0,

\[
\boxed{
d_r(t)
=
\cosh(\sqrt{2\kappa}\,t)

\]

\[
\boxed{
d_t(t)
=
\cos(\sqrt{\kappa}\,t).
}
\]

Ainsi :

\[
\boxed{
D_J(t,\mathbf x)
=
Q(\mathbf x)
\operatorname{diag}(d_r,d_t,d_t)
Q^T(\mathbf x)
}
\]

est une carte de déformation inhomogène explicite.

```python
kap,t = sp.symbols("kappa t", positive=True, real=True)

dr = sp.cosh(sp.sqrt(2*kap)*t)
dt = sp.cos(sp.sqrt(kap)*t)

dr_res = sp.simplify(sp.diff(dr,t,2)-2*kap*dr)
dt_res = sp.simplify(sp.diff(dt,t,2)+kap*dt)

G284_FROZEN_TIDAL_JACOBI_RADIAL_EXACT_PASS = (dr_res == 0)
G284_FROZEN_TIDAL_JACOBI_TANGENTIAL_EXACT_PASS = (dt_res == 0)

VJ = sp.simplify(dr*dt**2)
VJ_series = sp.series(VJ,t,0,6)
shape_ratio = sp.simplify(dr/dt)

# No t^2 volume change because tr(T)=0.
t2_coeff = sp.expand(VJ_series.removeO()).coeff(t,2)
G284_JACOBI_VOLUME_NO_QUADRATIC_CHANGE_PASS = sp.simplify(t2_coeff) == 0

assert G284_FROZEN_TIDAL_JACOBI_RADIAL_EXACT_PASS
assert G284_FROZEN_TIDAL_JACOBI_TANGENTIAL_EXACT_PASS
assert G284_JACOBI_VOLUME_NO_QUADRATIC_CHANGE_PASS

print("radial ODE residual =", dr_res)
print("tangential ODE residual =", dt_res)
print("D_J volume series =", VJ_series)
print("shape ratio d_r/d_t =", shape_ratio)
print("G284_JACOBI_VOLUME_NO_QUADRATIC_CHANGE_PASS =",
      G284_JACOBI_VOLUME_NO_QUADRATIC_CHANGE_PASS)
```

## 1.4 Cisaillement local induit par la carte de Jacobi

Définissons :

\[
B_J
=
\dot D_JD_J^{-1}.
\]

Dans le repère propre :

\[
B_J
=
\operatorname{diag}(H_r,H_t,H_t),
\]

où :

\[
H_r=\frac{\dot d_r}{d_r},
\qquad
H_t=\frac{\dot d_t}{d_t}.
\]

La partie trace-free est :

\[
\boxed{
S_J
=
B_J-\frac13\operatorname{tr}(B_J)I.
}
\]

Avec deux directions tangentielles égales :

\[
\boxed{
S_r=\frac23(H_r-H_t),
\qquad
S_t=-\frac13(H_r-H_t).
}
\]

Donc :

\[
\boxed{
S_J^2
=
S_r^2+2S_t^2
=
\frac23(H_r-H_t)^2.
}
\]

Pour \(t>0\) dans le domaine où \(d_t\neq0\), le champ central crée donc génériquement un changement de forme radial/tangentiel.

Cette quantité est un **cisaillement de Jacobi / congruence faible-champ**. Elle n'est pas encore déclarée identique au \(\sigma_{\mu\nu}\) complet de l'action GVH hors de cette limite.

```python
Hr = sp.simplify(sp.diff(dr,t)/dr)
Ht = sp.simplify(sp.diff(dt,t)/dt)
DeltaH = sp.simplify(Hr-Ht)

Sr = sp.simplify(sp.Rational(2,3)*DeltaH)
St = sp.simplify(-sp.Rational(1,3)*DeltaH)
SJ2 = sp.simplify(Sr**2+2*St**2)
SJ2_expected = sp.simplify(sp.Rational(2,3)*DeltaH**2)

G284_JACOBI_SHEAR_IDENTITY_EXACT_PASS = sp.simplify(SJ2-SJ2_expected) == 0

kap_num = 0.008
t_num = 1.7
dr_num = math.cosh(math.sqrt(2*kap_num)*t_num)
dt_num = math.cos(math.sqrt(kap_num)*t_num)
Hr_num = math.sqrt(2*kap_num)*math.tanh(math.sqrt(2*kap_num)*t_num)
Ht_num = -math.sqrt(kap_num)*math.tan(math.sqrt(kap_num)*t_num)
SJ2_num = (2/3)*(Hr_num-Ht_num)**2

G284_NONZERO_INHOMOGENEOUS_SHAPE_SHEAR_WITNESS = SJ2_num > 1e-10
G284_WEAK_FIELD_JACOBI_SHEAR_MAP_PASS = all([
    G284_JACOBI_SHEAR_IDENTITY_EXACT_PASS,
    G284_NONZERO_INHOMOGENEOUS_SHAPE_SHEAR_WITNESS,
])
G284_FULL_GVH_SIGMA_IDENTIFICATION_PROVEN = False

assert G284_WEAK_FIELD_JACOBI_SHEAR_MAP_PASS
assert not G284_FULL_GVH_SIGMA_IDENTIFICATION_PROVEN

print("H_r =", Hr)
print("H_t =", Ht)
print("S_J^2 =", SJ2)
print("numeric d_r =", dr_num)
print("numeric d_t =", dt_num)
print("numeric S_J^2 =", SJ2_num)
print("G284_WEAK_FIELD_JACOBI_SHEAR_MAP_PASS =",
      G284_WEAK_FIELD_JACOBI_SHEAR_MAP_PASS)
print("G284_FULL_GVH_SIGMA_IDENTIFICATION_PROVEN =",
      G284_FULL_GVH_SIGMA_IDENTIFICATION_PROVEN)
```

## 2.1 Superposition exacte du tenseur tidal de référence

Pour :

\[
\Phi_N(\mathbf x)
=
-\sum_{a=1}^N
\frac{\mu_a}
{|\mathbf x-\mathbf x_a|},
\]

la linéarité du Hessien donne :

\[
\boxed{
\mathcal T_N(\mathbf x)
=
\sum_{a=1}^N
\frac{\mu_a}
{r_a^3}
\left(
3\mathbf n_a\mathbf n_a^T-I
\right).
}
\]

C'est le matching naturel entre :

\[
\text{source unique}
\]

et :

\[
\text{système }N\text{-corps}.
\]

Dans cette limite, chaque niveau ou sous-système contribue au champ local par son tenseur tidal, et la matrice totale peut être diagonalée localement :

\[
\boxed{
Q^T(\mathbf x)\mathcal T_N(\mathbf x)Q(\mathbf x)
=
\operatorname{diag}
(\lambda_1,\lambda_2,\lambda_3).
}
\]

Le repère diagonal devient donc généralement **position-dépendant**.

```python
def nbody_accel_point(point, positions, mus):
    p = np.asarray(point,dtype=float)
    pos = np.asarray(positions,dtype=float)
    mus = np.asarray(mus,dtype=float)
    a = np.zeros(3)
    for xa,mua in zip(pos,mus):
        d = p-xa
        rr = np.linalg.norm(d)
        if rr <= 0:
            raise ValueError("field point on source")
        a += -mua*d/rr**3
    return a

def nbody_tidal(point, positions, mus):
    p = np.asarray(point,dtype=float)
    pos = np.asarray(positions,dtype=float)
    mus = np.asarray(mus,dtype=float)
    Tsum = np.zeros((3,3))
    for xa,mua in zip(pos,mus):
        d = p-xa
        Tsum += central_tidal_tensor(d,mua)
    return Tsum

positions = np.array([
    [-0.7,0.2,0.1],
    [0.9,-0.1,-0.2],
    [0.1,0.8,0.3],
    [-0.2,-0.6,0.4],
])
mus = np.array([1.2,0.7,0.3,0.5])
field_point = np.array([2.4,-1.7,1.1])

T_analytic = nbody_tidal(field_point,positions,mus)

# Finite-difference Jacobian of acceleration.
h = 1e-5
T_fd = np.zeros((3,3))
for j in range(3):
    e = np.zeros(3); e[j] = 1
    ap = nbody_accel_point(field_point+h*e,positions,mus)
    am = nbody_accel_point(field_point-h*e,positions,mus)
    T_fd[:,j] = (ap-am)/(2*h)

fd_err = float(np.max(np.abs(T_fd-T_analytic)))
trace_nbody = float(abs(np.trace(T_analytic)))

G284_NBODY_TIDAL_SUPERPOSITION_FINITE_DIFFERENCE_PASS = fd_err < 1e-9
G284_NBODY_VACUUM_TRACEFREE_WITNESS_PASS = trace_nbody < 1e-12

assert G284_NBODY_TIDAL_SUPERPOSITION_FINITE_DIFFERENCE_PASS
assert G284_NBODY_VACUUM_TRACEFREE_WITNESS_PASS

print("N-body analytic tidal tensor =")
print(T_analytic)
print("finite-difference max error =", fd_err)
print("trace =", np.trace(T_analytic))
print("G284_NBODY_TIDAL_SUPERPOSITION_FINITE_DIFFERENCE_PASS =",
      G284_NBODY_TIDAL_SUPERPOSITION_FINITE_DIFFERENCE_PASS)
```

```python
cluster_pos = np.array([
    [-0.4,0.0,0.1],
    [0.3,0.2,-0.1],
    [0.1,-0.3,0.0],
    [0.0,0.1,0.2],
])
cluster_mu = np.array([1.0,0.8,0.5,0.4])
Mtot = np.sum(cluster_mu)
Rcm = np.sum(cluster_mu[:,None]*cluster_pos,axis=0)/Mtot

distances = np.array([5.0,10.0,20.0,40.0,80.0])
hier_rows = []

for RR in distances:
    p = Rcm + np.array([RR,0.37*RR,-0.21*RR])
    Tex = nbody_tidal(p,cluster_pos,cluster_mu)
    Tcm = central_tidal_tensor(p-Rcm,Mtot)
    rel = np.linalg.norm(Tex-Tcm)/np.linalg.norm(Tex)
    hier_rows.append({"Rscale":RR,"relative_tidal_error":rel})

df_hier = pd.DataFrame(hier_rows)
errors = df_hier["relative_tidal_error"].to_numpy()

G284_HIERARCHICAL_MONOPOLE_ERROR_DECREASE_PASS = bool(
    np.all(errors[1:] < errors[:-1])
)
G284_HIERARCHICAL_FAR_FIELD_MATCHING_WITNESS_PASS = (
    errors[-1] < errors[0]/10
)

assert G284_HIERARCHICAL_MONOPOLE_ERROR_DECREASE_PASS
assert G284_HIERARCHICAL_FAR_FIELD_MATCHING_WITNESS_PASS

print("cluster barycenter =", Rcm)
print(df_hier.to_string(index=False))
print("G284_HIERARCHICAL_MONOPOLE_ERROR_DECREASE_PASS =",
      G284_HIERARCHICAL_MONOPOLE_ERROR_DECREASE_PASS)
print("G284_HIERARCHICAL_FAR_FIELD_MATCHING_WITNESS_PASS =",
      G284_HIERARCHICAL_FAR_FIELD_MATCHING_WITNESS_PASS)
```

\[

avec :
