
## 3. Candidate minimal covariant action

Use the reference Einstein–Hilbert sector plus a quadratic first-derivative vector sector:

\[
S_{\rm cand}
=
\frac{1}{16\pi G}
\int d^4x\,\sqrt{-g}
\left[
R
-
K^{\alpha\beta}{}_{\mu\nu}
\nabla_\alpha u^\mu
\nabla_\beta u^\nu
+
\lambda(u^\mu u_\mu+1)
\right]
+
S_m.
\]

We parameterize

\[
K^{\alpha\beta}{}_{\mu\nu}
=
c_1 g^{\alpha\beta}g_{\mu\nu}
+
c_2 \delta^\alpha_{\mu}\delta^\beta_{\nu}
+
c_3 \delta^\alpha_{\nu}\delta^\beta_{\mu}
-
c_4 u^\alpha u^\beta g_{\mu\nu}.
\]

The constants \(c_1,c_2,c_3,c_4\) are **not GVH predictions** here. They are free candidate couplings introduced to expose the closure problem.


````python

c1, c2, c3, c4, G = sp.symbols("c1 c2 c3 c4 G", real=True)

coupling_registry = pd.DataFrame([
    {"coupling":"c1", "status":"CANDIDATE_ANSATZ", "derived_by_GVH":False},
    {"coupling":"c2", "status":"CANDIDATE_ANSATZ", "derived_by_GVH":False},
    {"coupling":"c3", "status":"CANDIDATE_ANSATZ", "derived_by_GVH":False},
    {"coupling":"c4", "status":"CANDIDATE_ANSATZ", "derived_by_GVH":False},
])
coupling_registry
````


## 4. Invariant decomposition

At first derivative order, the candidate kinetic scalar can be represented as

\[
\mathcal L_u
=
-c_1(\nabla_\mu u_\nu)(\nabla^\mu u^\nu)
-c_2(\nabla_\mu u^\mu)^2
-c_3(\nabla_\mu u_\nu)(\nabla^\nu u^\mu)
+c_4 a_\mu a^\mu,

\[
a^\mu=u^\nu\nabla_\nu u^\mu.

The important methodological point is not the notation: it is that the minimal covariant vector sector already introduces several independent couplings. Unless GVH supplies relations among them, dynamical closure has merely moved the freedom from \((a_1,a_2,b_1)\) into \((c_1,c_2,c_3,c_4)\).


## 5. Variation with respect to \(\lambda\)

The multiplier equation is immediate:

\[
\frac{\delta S_{\rm cand}}{\delta\lambda}=0
\quad\Rightarrow\quad
u^\mu u_\mu+1=0.
\]

This is a genuine consequence **of the candidate action once the action is assumed**, but it is not evidence that the candidate action is the unique GVH dynamics.


````python

u2 = sp.symbols("u_squared", real=True)
constraint_eq = sp.Eq(u2 + 1, 0)
constraint_solution = sp.solve(constraint_eq, u2)
constraint_solution
````


## 12. Minimal minisuperspace sanity check

A simplified homogeneous control can test the unit constraint without pretending to derive the full theory.

For an FLRW-like metric,

\[
ds^2=-N(t)^2dt^2+a(t)^2d\vec x^2,
\]

take a homogeneous timelike vector

\[
u^\mu=(u^0(t),0,0,0).
\]

The unit constraint gives

\[
-N^2(u^0)^2=-1,
\]

so on the future-directed branch

\[
u^0=\frac{1}{N}.
\]

This is a consistency check of the normalization sector only.


````python

N, u0 = sp.symbols("N u0", positive=True, real=True)
norm_eq = sp.Eq(-N**2 * u0**2, -1)
u0_solutions = sp.solve(norm_eq, u0)
u0_solutions
````
