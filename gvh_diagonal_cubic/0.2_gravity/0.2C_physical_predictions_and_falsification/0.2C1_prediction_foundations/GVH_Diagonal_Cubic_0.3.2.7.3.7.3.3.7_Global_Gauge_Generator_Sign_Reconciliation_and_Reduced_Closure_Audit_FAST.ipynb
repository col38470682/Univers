```python
# GGA7.3 — Exact canonical sign and right-action witness

q1,q2,p1,p2 = sp.symbols("q1 q2 p1 p2")
qs = (q1,q2)
ps = (p1,p2)

def pb(F,G,qs,ps):
    return sp.expand(sum(
        sp.diff(F,q)*sp.diff(G,p)
        - sp.diff(F,p)*sp.diff(G,q)
        for q,p in zip(qs,ps)
    ))

X = sp.Matrix([
    q1 + q1*q2,
    q2 + q1**2,
])
Y = sp.Matrix([
    q2 + q1*q2,
    q1 - q2**2,
])

JX = sp.expand(p1*X[0] + p2*X[1])
JY = sp.expand(p1*Y[0] + p2*Y[1])

right_action_q = sp.Matrix([
    pb(q1,JX,qs,ps),
    pb(q2,JX,qs,ps),
])

left_action_q = sp.Matrix([
    pb(JX,q1,qs,ps),
    pb(JX,q2,qs,ps),
])

brXY = sp.Matrix([
    sp.expand(sum(X[j]*sp.diff(Y[i],qs[j]) - Y[j]*sp.diff(X[i],qs[j]) for j in range(2)))
    for i in range(2)
])
Jbr = sp.expand(p1*brXY[0] + p2*brXY[1])

RIGHT_ACTION_POSITIVE_MAP_PASS = sp.simplify(right_action_q - X) == sp.zeros(2,1)
LEFT_ACTION_POSITIVE_MAP_FAILS = sp.simplify(left_action_q - X) != sp.zeros(2,1)
RIGHT_MAP_ANTI_HOM_PASS = sp.expand(pb(JX,JY,qs,ps) + Jbr) == 0

D_SIGN_CONVENTION_RECONCILED = all([
    RIGHT_ACTION_POSITIVE_MAP_PASS,
    LEFT_ACTION_POSITIVE_MAP_FAILS,
    RIGHT_MAP_ANTI_HOM_PASS,
    UPSTREAM["p334"]["D_positive_momentum_map"],
])

assert D_SIGN_CONVENTION_RECONCILED

print("RIGHT_ACTION_POSITIVE_MAP_PASS =", RIGHT_ACTION_POSITIVE_MAP_PASS)
print("LEFT_ACTION_POSITIVE_MAP_FAILS =", LEFT_ACTION_POSITIVE_MAP_FAILS)
print("RIGHT_MAP_ANTI_HOM_PASS =", RIGHT_MAP_ANTI_HOM_PASS)
print("D_SIGN_CONVENTION_RECONCILED =", D_SIGN_CONVENTION_RECONCILED)
```
