## 4. Rank test for coefficient determination

Let a candidate closure produce equations

\[
F_A(a_1,a_2,b_1;\theta)=0.
\]

A local necessary condition for unique coefficient determination is that the Jacobian
with respect to \((a_1,a_2,b_1)\) have rank 3 at the candidate solution:

\[
{\rm rank}\left[
\frac{\partial(F_1,F_2,F_3,\ldots)}
{\partial(a_1,a_2,b_1)}
\right]=3.
\]

This is necessary locally, not sufficient globally.

```python
def jacobian_rank(equations, variables=(a1,a2,b1), substitutions=None):
    if not equations:
        return 0
    J = sp.Matrix(equations).jacobian(variables)
    if substitutions:
        J = J.subs(substitutions)
    return J.rank()

print("Rank with no supplied dynamical closure equations:", jacobian_rank([]))
print("Required rank:", required_rank)
```

## 5. Control branch: explicit GR recovery

For audit purposes only, impose the restricted recovery relations

\[
b_1=a_1,\qquad a_2=a_1^2,
\]

and a normalization \(a_1=1\). This gives

\[
a_1=a_2=b_1=1,\qquad \gamma=\beta=1.
\]

This demonstrates internal compatibility with the chosen GR-normalized branch.
It does **not** prove that GVH dynamically derives those relations.

```python
F_gr = [
    b1 - a1,
    a2 - a1**2,
    a1 - 1,
]
sol_gr = sp.solve(F_gr, (a1,a2,b1), dict=True)
rank_gr = jacobian_rank(F_gr, substitutions=sol_gr[0])

print("Control solution:", sol_gr)
print("Jacobian rank:", rank_gr)
print("gamma:", gamma_gvh.subs(sol_gr[0]))
print("beta :", beta_gvh.subs(sol_gr[0]))
```

## 8. Symbolic placeholder for a future closure

To keep the next derivation auditable, introduce generic closure equations

## 9. Parameter-transfer test

A common failure mode is **parameter transfer**:

\[

with no reduction in independent freedom.

A valid closure must reduce the number of undetermined physical parameters or derive
relations among them from the theory.

```python
n_original_free = 3
n_transferred_free = 3
parameter_reduction = n_original_free - n_transferred_free

print("Original free coefficients:", n_original_free)
print("Free parameters after placeholder transfer:", n_transferred_free)
print("Net reduction:", parameter_reduction)
assert parameter_reduction == 0
```
