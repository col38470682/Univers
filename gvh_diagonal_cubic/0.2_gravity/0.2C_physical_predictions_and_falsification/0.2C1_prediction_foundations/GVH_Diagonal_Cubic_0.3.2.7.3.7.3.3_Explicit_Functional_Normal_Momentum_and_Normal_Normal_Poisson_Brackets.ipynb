## 1. Rappel du générateur spatial

La contrainte de moment obtenue en 7.7.2.4 est
\[
\mathcal C_i
=
-2h_{ij}D_k\pi^{kj}
+p_sD_is
+p_v^{\,j}D_iv_j
-D_j(p_v^{\,j}v_i).
\]

Sous forme intégrée par parties :
\[
D[\xi]
=
\int d^3x\,
\left[
\pi^{ij}\mathcal L_\xi h_{ij}
+p_s\mathcal L_\xi s
+p_v^{\,i}\mathcal L_\xi v_i
\right].
\]

Il est donc le générateur canonique des difféomorphismes spatiaux sur toutes les variables canoniques.


## 2. Test du secteur normal–momentum

Après 7.7.3.2, \(\mathscr C_N\) est une densité scalaire spatiale de poids 1 construite à partir des variables canoniques et de leurs dérivées spatiales, sans dépendance explicite au lapse.

Pour toute densité scalaire de poids 1 :
\[
\delta_\xi\mathscr C_N
=
\mathcal L_\xi\mathscr C_N
=
\partial_i(\xi^i\mathscr C_N).
\]

Donc :
\[
\begin{aligned}
\{H[N],D[\xi]\}
&=
\int d^3x\,N\,\mathcal L_\xi\mathscr C_N\\
&=
-\int d^3x\,(\mathcal L_\xi N)\mathscr C_N
\end{aligned}
\]
modulo terme de bord.

Ainsi :
\[
\boxed{
\{H[N],D[\xi]\}
=
-H[\mathcal L_\xi N].
}
\]


```python
# Exact 1D symbolic integration-by-parts identity representing the density-weight argument.
x=sp.symbols("x", real=True)
N=sp.Function("N")(x)
xi=sp.Function("xi")(x)
C=sp.Function("C")(x)

lhs_integrand=sp.expand(N*sp.diff(xi*C,x))
rhs_integrand=sp.expand(-sp.diff(N,x)*xi*C)
boundary_remainder=sp.simplify(lhs_integrand-rhs_integrand)

assert sp.simplify(boundary_remainder-sp.diff(N*xi*C,x))==0
print("HD density integration-by-parts identity: PASS")
print("{H[N],D[xi]} = -H[L_xi N] modulo boundary: PASS")
```

## 4. Préparation du secteur normal–normal

Le crochet
\[
\{H[N],H[M]\}
\]
est plus fort.

La densité \(\mathscr C_N\) contient trois familles d'objets :

1. le secteur cinétique canonique construit avec \(Q^{-1}\) ;
2. le potentiel spatial, notamment \(R^{(3)}\) ;
3. le terme divergence \(D_iB^i\).

Le crochet complet nécessite les dérivées fonctionnelles :
\[
\frac{\delta H[N]}{\delta h_{ij}},
\quad
\frac{\delta H[N]}{\delta \pi^{ij}},
\quad
\frac{\delta H[N]}{\delta s},
\quad
\frac{\delta H[N]}{\delta p_s},
\quad
\frac{\delta H[N]}{\delta v_i},
\quad
\frac{\delta H[N]}{\delta p_v^{\,i}}.
\]

La formule fonctionnelle est :
\[
\{H[N],H[M]\}
=
\int d^3x
\left[
\frac{\delta H[N]}{\delta q^A}
\frac{\delta H[M]}{\delta p_A}
-
(N\leftrightarrow M)
\right].
\]


```python
# Register canonical sectors used in the functional bracket.
canonical_pairs=[
    ("h_ij","pi^ij"),
    ("s","p_s"),
    ("v_i","p_v^i"),
]
print("Canonical pairs registered:")
for q,p in canonical_pairs:
    print(" ",q,"<->",p)
```

## 5. Structure antisymétrique attendue du crochet HH

Le résultat doit être antisymétrique sous
\[
N\leftrightarrow M.
\]

Par localité et covariance spatiale, le premier ordre en dérivées des smearings possède nécessairement la structure
\[
N D_iM-MD_iN.
\]

La cible ADM est donc
\[
\boxed{
\{H[N],H[M]\}
=
D[\beta^i]
+\mathcal A[N,M]
}
\]
avec
\[
\beta^i=h^{ij}(ND_jM-MD_jN),
\]
et \(\mathcal A\) un éventuel résidu proportionnel à des contraintes auxiliaires ou une déformation véritable.


```python
# Algebraic antisymmetry check of the expected shift vector.
N1,N2,M1,M2,h11,h12,h22=sp.symbols("N1 N2 M1 M2 h11 h12 h22")
# N1,N2 and M1,M2 stand for spatial derivatives in two directions.
N0,M0=sp.symbols("N M")
hInv=sp.Matrix([[h11,h12],[h12,h22]])
dN=sp.Matrix([N1,N2]); dM=sp.Matrix([M1,M2])

beta=hInv*(N0*dM-M0*dN)
beta_swapped=hInv*(M0*dN-N0*dM)
assert sp.simplify(beta+beta_swapped)==sp.zeros(2,1)
print("Expected HH shift vector antisymmetry: PASS")
```
