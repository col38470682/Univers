## 2.1 Inversion si \(c_-\) devient identifiable

Si une future branche GVH fournit :

\[
c_-=c_1-c_3
\]

de manière indépendante, alors :

\[
\boxed{
c_1=\frac{c_{13}+c_-}{2}
}
\]

\[
\boxed{
c_3=\frac{c_{13}-c_-}{2}
}
\]

\[
\boxed{
c_4=c_{14}-\frac{c_{13}+c_-}{2}
}
\]

et :

\[
\boxed{c_2=c_2}.
\]

Le rang passe alors exactement de :

\[
3\rightarrow4.
\]

Ainsi le problème n'est pas nécessairement de trouver une « loi qui fixe les constantes » ; il peut être résolu en dérivant une **quatrième observable ou solution indépendante** qui soit sensible à \(c_-\).

Le secteur le plus naturel est celui où :

\[
\boxed{
\omega_{\mu\nu}\neq0.
}
\]

```python
c13s,c14s,cmins = sp.symbols(
    "c13 c14 c_minus", real=True
)

inverse_map = {
    "c1": sp.simplify((c13s+cmins)/2),
    "c2": c2,
    "c3": sp.simplify((c13s-cmins)/2),
    "c4": sp.simplify(c14s-(c13s+cmins)/2),
}

# Preferred-frame alpha1 expressed using the missing combination.
alpha1 = sp.simplify(
    -8*(c3**2+c1*c4)
    /(2*c1-c1**2+c3**2)
)

alpha1_cminus = sp.factor(
    alpha1.subs({
        c1:inverse_map["c1"],
        c3:inverse_map["c3"],
        c4:inverse_map["c4"],
    })
)
dalpha1_dcminus = sp.factor(
    sp.diff(alpha1_cminus,cmins)
)

G289_ALPHA1_SENSITIVE_TO_CMINUS = (
    dalpha1_dcminus != 0
)
G289_CMINUS_IS_IDENTIFIABILITY_CLOSING_COMBINATION = all([
    G289_EXTENDED_COMBINATION_MAP_RANK4_PASS,
    G289_ALPHA1_SENSITIVE_TO_CMINUS,
])

assert G289_CMINUS_IS_IDENTIFIABILITY_CLOSING_COMBINATION

print("inverse map =", inverse_map)
print("alpha1(c13,c14,c_minus) =")
sp.pprint(alpha1_cminus)
print("d alpha1 / d c_minus =")
sp.pprint(dalpha1_dcminus)
print("G289_CMINUS_IS_IDENTIFIABILITY_CLOSING_COMBINATION =",
      G289_CMINUS_IS_IDENTIFIABILITY_CLOSING_COMBINATION)
```

# 2.2 Témoin d'identifiabilité : mêmes combinaisons actuelles, preferred-frame différent

Prenons deux jeux internes de couplages reliés exactement par la direction libre :

\[
\delta\mathbf c
=
\lambda(1,0,-1,-1).
\]

Ils ont donc les mêmes :

\[
c_{13},\quad c_2,\quad c_{14},
\]

et passent indistinguables dans les benchmarks actuellement dépendants seulement de ces combinaisons.

Mais ils ont des \(c_-\) différents.

Si leurs \(\alpha_1\) diffèrent, cela démontre concrètement que l'information manquante n'est pas un artefact du calcul de rang.

Ce témoin reste **dimensionless/internal** :

- aucune calibration observationnelle ;
- aucune valeur de couplage n'est promue en constante physique mesurée.

```python
base = {
    c1:0.06,
    c2:-0.02,
    c3:0.01,
    c4:0.04,
}
lam = 0.02
shifted = {
    c1:base[c1]+lam,
    c2:base[c2],
    c3:base[c3]-lam,
    c4:base[c4]-lam,
}

def witness_row(name,sub):
    return {
        "set":name,
        "c1":sub[c1],
        "c2":sub[c2],
        "c3":sub[c3],
        "c4":sub[c4],
        "c13":float((c1+c3).subs(sub)),
        "c14":float((c1+c4).subs(sub)),
        "c_minus":float((c1-c3).subs(sub)),
        "alpha1":float(alpha1.subs(sub)),
    }

witness = pd.DataFrame([
    witness_row("A",base),
    witness_row("B=A+lambda*null",shifted),
])

same_current_combos = all([
    abs(witness.loc[0,"c13"]-witness.loc[1,"c13"]) < 1e-15,
    abs(witness.loc[0,"c14"]-witness.loc[1,"c14"]) < 1e-15,
    abs(witness.loc[0,"c2"]-witness.loc[1,"c2"]) < 1e-15,
])
different_cminus = (
    abs(witness.loc[0,"c_minus"]-witness.loc[1,"c_minus"]) > 1e-12
)
different_alpha1 = (
    abs(witness.loc[0,"alpha1"]-witness.loc[1,"alpha1"]) > 1e-6
)

G289_NULL_DIRECTION_NUMERIC_IDENTIFIABILITY_WITNESS_PASS = all([
    same_current_combos,
    different_cminus,
    different_alpha1,
])

assert G289_NULL_DIRECTION_NUMERIC_IDENTIFIABILITY_WITNESS_PASS

print(witness.to_string(index=False))
print("G289_NULL_DIRECTION_NUMERIC_IDENTIFIABILITY_WITNESS_PASS =",
      G289_NULL_DIRECTION_NUMERIC_IDENTIFIABILITY_WITNESS_PASS)
```
