La superposition ne doit toutefois pas être extrapolée au régime fortement non linéaire.

```python
def phi_nbody(point, positions, mus):
    p = np.asarray(point,dtype=float)
    val = 0.0
    for xa,mua in zip(np.asarray(positions,float),np.asarray(mus,float)):
        d = p-xa
        val += -mua/np.linalg.norm(d)
    return val

def tidal_nbody_from_action(point, positions, mus):
    p = np.asarray(point,dtype=float)
    Tsum = np.zeros((3,3))
    for xa,mua in zip(np.asarray(positions,float),np.asarray(mus,float)):
        d = p-xa
        rr0 = np.linalg.norm(d)
        nn = d/rr0
        Tsum += mua/rr0**3*(3*np.outer(nn,nn)-np.eye(3))
    return Tsum

positions = np.array([
    [-0.8,0.1,0.0],
    [0.6,-0.2,0.2],
    [0.2,0.7,-0.1],
])
mus = np.array([1.1,0.5,0.35])
p0 = np.array([2.7,-1.9,1.4])

T_exact = tidal_nbody_from_action(p0,positions,mus)

h = 1e-5
T_fd = np.zeros((3,3))
for j in range(3):
    ej = np.zeros(3); ej[j] = 1.0
    # g = -grad Phi, compute via central difference in potential,
    # then Jacobian via a nested direct analytic acceleration function.
    pass

def accel_nbody_from_action(point, positions, mus):
    p = np.asarray(point,dtype=float)
    a = np.zeros(3)
    for xa,mua in zip(np.asarray(positions,float),np.asarray(mus,float)):
        d = p-xa
        rr0 = np.linalg.norm(d)
        a += -mua*d/rr0**3
    return a

for j in range(3):
    ej = np.zeros(3); ej[j] = 1.0
    ap = accel_nbody_from_action(p0+h*ej,positions,mus)
    am = accel_nbody_from_action(p0-h*ej,positions,mus)
    T_fd[:,j] = (ap-am)/(2*h)

nbody_fd_err = float(np.max(np.abs(T_fd-T_exact)))
nbody_trace = float(abs(np.trace(T_exact)))

G285_ACTION_WEAK_NBODY_TIDAL_FD_PASS = nbody_fd_err < 1e-9
G285_ACTION_WEAK_NBODY_VACUUM_TRACE_PASS = nbody_trace < 1e-12
G285_WEAK_NBODY_ACTION_MATCHING_CLOSED = all([
    G285_ACTION_WEAK_NBODY_TIDAL_FD_PASS,
    G285_ACTION_WEAK_NBODY_VACUUM_TRACE_PASS,
])

assert G285_WEAK_NBODY_ACTION_MATCHING_CLOSED

print("N-body action-derived tidal FD max error =", nbody_fd_err)
print("N-body tidal trace =", np.trace(T_exact))
print("G285_WEAK_NBODY_ACTION_MATCHING_CLOSED =",
      G285_WEAK_NBODY_ACTION_MATCHING_CLOSED)
```
