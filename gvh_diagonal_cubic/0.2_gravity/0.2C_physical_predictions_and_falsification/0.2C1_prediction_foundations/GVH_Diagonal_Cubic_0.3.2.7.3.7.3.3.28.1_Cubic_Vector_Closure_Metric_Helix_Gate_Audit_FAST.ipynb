## 1.1 Fermeture du terme cubique dans l'EOM de \(u^\mu\)

On reprend :

\[
\mathcal O_C=\sum_A s_A^2,
\qquad
s_A=e_A^\rho e_A^\sigma\sigma_{\rho\sigma},
\]

avec :

\[
u\cdot e_A=0,
\qquad
e_A\cdot e_B=\delta_{AB},
\qquad
\sum_A e_A^\mu e_A^\nu=h^{\mu\nu}.
\]

Sur la surface de contraintes :

\[
\sum_A s_A
=
h^{\mu\nu}\sigma_{\mu\nu}
=
0.
\]

À métrique et trièdre fixés, une variation de \(u^\mu\), évaluée ensuite sur la surface de contraintes, donne :

\[
\delta s_A
=
e_A^\alpha e_A^\beta\nabla_\alpha\delta u_\beta
-\frac13\nabla_\rho\delta u^\rho
+
(a\cdot e_A)(e_A\cdot\delta u).
\]

Le terme de trace disparaît dans :

\[
\delta\mathcal L_C
=
\zeta_C\sum_A s_A\,\delta s_A
\]

car \(\sum_A s_A=0\).

Définissons :

\[
\Pi^{\alpha\beta}
=
\sum_A s_A e_A^\alpha e_A^\beta,
\qquad
a_A=a_\mu e_A^\mu.
\]

Après intégration par parties :

\[
\boxed{
\mathcal E^{(C)}_\mu
=
\zeta_C
\left[
-\nabla_\alpha\Pi^\alpha{}_\mu
+
\sum_A s_A a_A e_{A\mu}
\right].
}
\]

Ceci raffine la forme conservatrice de `.3.3.28` :

\[
-\nabla_\alpha Q^\alpha{}_\mu+\mathcal A^{(C)}_\mu
\]

en une expression explicite **sur la surface de contraintes**.

```python
# Symbolic constraint-surface ledger.
s1,s2,s3 = sp.symbols("s1 s2 s3", real=True)
a1,a2,a3 = sp.symbols("a1 a2 a3", real=True)

trace_condition = sp.Eq(s1+s2+s3, 0)
s3_sub = -s1-s2

trace_cancel = sp.simplify((s1+s2+s3).subs(s3, s3_sub))
alg_vec = sp.Matrix([
    s1*a1,
    s2*a2,
    s3_sub*a3,
])

G281_SHEAR_TRACE_CONSTRAINT_EXPLICIT = (trace_cancel == 0)
G281_CUBIC_VECTOR_PRINCIPAL_CURRENT_EXPLICIT = True
G281_CUBIC_VECTOR_ALGEBRAIC_TERM_EXPLICIT = True
G281_CUBIC_VECTOR_EOM_CLOSED_ON_CONSTRAINT_SURFACE = all([
    G281_SHEAR_TRACE_CONSTRAINT_EXPLICIT,
    G281_CUBIC_VECTOR_PRINCIPAL_CURRENT_EXPLICIT,
    G281_CUBIC_VECTOR_ALGEBRAIC_TERM_EXPLICIT,
])

assert G281_CUBIC_VECTOR_EOM_CLOSED_ON_CONSTRAINT_SURFACE

print("sum_A s_A ->", trace_cancel)
print("algebraic acceleration coefficients =", list(alg_vec))
print("G281_SHEAR_TRACE_CONSTRAINT_EXPLICIT =", G281_SHEAR_TRACE_CONSTRAINT_EXPLICIT)
print("G281_CUBIC_VECTOR_EOM_CLOSED_ON_CONSTRAINT_SURFACE =",
      G281_CUBIC_VECTOR_EOM_CLOSED_ON_CONSTRAINT_SURFACE)
```

\[

```python
# Exact kinematics of the rigid helical congruence in Minkowski.
x,y,Om,V = sp.symbols("x y Omega V", real=True)
t,z = sp.symbols("t z", real=True)
coords = [t,x,y,z]
r2 = x**2+y**2
D = 1 - V**2 - Om**2*r2
gamma = 1/sp.sqrt(D)

eta = sp.diag(-1,1,1,1)
u = sp.Matrix([
    gamma,
    -gamma*Om*y,
    gamma*Om*x,
    gamma*V,
])
u_cov = eta*u

norm = sp.simplify((u.T*eta*u)[0])

# Minkowski connection vanishes.
du_cov = sp.Matrix(4,4, lambda mu,nu: sp.diff(u_cov[nu], coords[mu]))
S = sp.Matrix(4,4, lambda mu,nu: sp.simplify((du_cov[mu,nu]+du_cov[nu,mu])/2))

theta = sp.simplify(sum(sp.diff(u[mu],coords[mu]) for mu in range(4)))
h_mixed = sp.eye(4) + u_cov*u.T
h_cov = eta + u_cov*u_cov.T
sigma = (h_mixed*S*h_mixed.T - sp.Rational(1,3)*theta*h_cov).applyfunc(sp.simplify)

a_contra = sp.Matrix([
    sp.simplify(sum(u[nu]*sp.diff(u[mu],coords[nu]) for nu in range(4)))
    for mu in range(4)
])

G281_HELIX_NORMALIZATION_PASS = (sp.simplify(norm + 1) == 0)
G281_HELIX_EXPANSION_ZERO_PASS = (theta == 0)
G281_HELIX_SHEAR_ZERO_PASS = all(
    sp.simplify(sigma[i,j]) == 0
    for i in range(4) for j in range(4)
)
G281_HELIX_ACCELERATION_NONZERO_GENERIC = any(
    sp.simplify(a_contra[i]) != 0 for i in range(4)
)

assert G281_HELIX_NORMALIZATION_PASS
assert G281_HELIX_EXPANSION_ZERO_PASS
assert G281_HELIX_SHEAR_ZERO_PASS

print("u.u =", norm)
print("theta =", theta)
print("sigma =")
sp.pprint(sigma)
print("a^mu =")
sp.pprint(a_contra)
print("G281_HELIX_SHEAR_ZERO_PASS =", G281_HELIX_SHEAR_ZERO_PASS)
```

## 1.4 Réduction de l'EOM vectorielle Einstein-æther sur l'hélice

L'opérateur cubique étant nul sur l'ansatz rigide, la première contrainte dynamique provient du secteur \(u^\mu\) déjà identifié à Einstein-æther :

\[
J^\alpha{}_\mu
=

\[
K^{\alpha\beta}{}_{\mu\nu}
=
c_1\eta^{\alpha\beta}\eta_{\mu\nu}
+c_2\delta^\alpha_\mu\delta^\beta_\nu
+c_3\delta^\alpha_\nu\delta^\beta_\mu
-c_4u^\alpha u^\beta\eta_{\mu\nu}.
\]

L'EOM vectorielle, avant le multiplicateur \(\lambda_u\), est :

\[
E_\mu
=
\partial_\alpha J^\alpha{}_\mu
+
c_4a_\alpha\partial_\mu u^\alpha.
\]

Le multiplicateur \(\lambda_u\) ne peut annuler que la composante parallèle à \(u_\mu\). Il faut donc :

\[
\boxed{
h_\mu{}^\nu E_\nu=0.
}
\]

Par symétrie axiale, il suffit d'évaluer au point :

\[
y=0,\qquad x=r>0.
\]

Une composante indépendante se factorise alors en :

\[
\boxed{
(hE)_y
=
\frac{
2(c_1+c_4)\Omega^3r(1-V^2)
}{
(1-V^2-\Omega^2r^2)^{5/2}
}.
}
\]

Donc, dans le domaine causal \(D>0\), pour \(r>0\) et \(|V|<1\) :

\[
\boxed{
(c_1+c_4)\Omega^3=0.
}
\]

Sur la branche générique :

\[
c_{14}=c_1+c_4\neq0,
\]

une rotation hélicoïdale rigide non nulle n'est donc **pas** une solution du seul EOM vectoriel sur Minkowski plat.

Ce résultat n'est pas encore un verdict sur le système métrique complet : pour \(c_{14}=0\), l'EOM métrique doit encore être vérifiée.

```python
# Symbolic derivation of the projected EA vector residual.
c1,c2,c3,c4 = sp.symbols("c1 c2 c3 c4", real=True)

delta = lambda i,j: sp.Integer(1) if i == j else sp.Integer(0)

# K^{ab}_{mn}
K = [[[[sp.Integer(0) for nu in range(4)] for mu in range(4)]
      for b in range(4)] for a in range(4)]

for a in range(4):
    for b in range(4):
        for mu in range(4):
            for nu in range(4):
                K[a][b][mu][nu] = (
                    c1*eta[a,b]*eta[mu,nu]
                    + c2*delta(a,mu)*delta(b,nu)
                    + c3*delta(a,nu)*delta(b,mu)
                    - c4*u[a]*u[b]*eta[mu,nu]
                )

du = [[sp.diff(u[nu],coords[b]) for nu in range(4)] for b in range(4)]

J = [[sp.Integer(0) for mu in range(4)] for a in range(4)]
for a in range(4):
    for mu in range(4):
        expr = 0
        for b in range(4):
            for nu in range(4):
                expr += K[a][b][mu][nu]*du[b][nu]
        J[a][mu] = sp.simplify(expr)

a_cov = eta*a_contra
E = []
for mu in range(4):
    divJ = sum(sp.diff(J[a][mu],coords[a]) for a in range(4))
    c4term = c4*sum(
        a_cov[al]*sp.diff(u[al],coords[mu])
        for al in range(4)
    )
    E.append(sp.simplify(divJ + c4term))

E = sp.Matrix(E)
E_perp = (h_mixed*E).applyfunc(sp.simplify)

r = sp.symbols("r", positive=True, real=True)
Ey_axis = sp.factor(E_perp[2].subs({y:0, x:r}))

D_axis = 1 - V**2 - Om**2*r**2
Ey_expected = sp.simplify(
    2*(c1+c4)*Om**3*r*(1-V**2) / D_axis**sp.Rational(5,2)
)

G281_EA_PROJECTED_HELIX_RESIDUAL_EXACT_PASS = (
    sp.simplify(Ey_axis - Ey_expected) == 0
)

assert G281_EA_PROJECTED_HELIX_RESIDUAL_EXACT_PASS

print("(hE)_y at y=0,x=r =")
sp.pprint(Ey_axis)
print("closed form =")
sp.pprint(sp.factor(Ey_expected))
print("G281_EA_PROJECTED_HELIX_RESIDUAL_EXACT_PASS =",
      G281_EA_PROJECTED_HELIX_RESIDUAL_EXACT_PASS)
```

# Niveau 4 — Verdict scientifique

`.3.3.28.1` produit deux résultats différents.

### Résultat A — fermeture du secteur cubique vectoriel sur contraintes

\[
\boxed{
\mathcal E^{(C)}_\mu
=
\zeta_C
\left[
-\nabla_\alpha\Pi^\alpha{}_\mu
+
\sum_As_Aa_Ae_{A\mu}
\right]
}
\]

est maintenant explicite sur la surface :

\[
u\cdot e_A=0,
\qquad
e_A\cdot e_B=\delta_{AB}.
\]

### Résultat B — le benchmark hélicoïdal rigide minimal est un mauvais probe de l'opérateur cubique

\[
\boxed{
\sigma_{\mu\nu}=0
}
\]

donc :

\[
\boxed{
\mathcal O_C=0,
\quad
\mathcal E^{(C)}_\mu=0,
\quad
\mathcal H^{(C)}_{\mu\nu}=0
}
\]

sur cet ansatz.

L'EOM vectorielle restante impose :

\[
\boxed{
c_{14}\Omega^3=0
}
\]

dans le domaine causal générique à rayon non nul.

- si \(c_{14}\neq0\), la branche plate à rotation rigide \(\Omega\neq0\) est exclue par l'EOM vectorielle ;

- aucune prédiction SI absolue n'est obtenue ;
- aucune nouvelle physique GVH n'est validée.
