# G22.6 — Route D exacte en variable \(y\)

Dans le domaine extérieur asymptotiquement plat :

\[
P(y)=1+2y+\frac{c_{14}}2y^2,
\]

\[
ry'=-yP(y),
\]

\[
\frac{d\ln N}{dy}=-\frac1{P(y)}.
\]

Avec la normalisation \(N\to1\) lorsque \(y\to0\) :

\[
\boxed{
I(y;c_{14})
\equiv
\int_0^y \frac{dz}{P(z)}
}
\]

\[
\boxed{
N(y)=e^{-I(y)}
}
\]

et :

\[
\boxed{
\frac{r(y)}{M}
=
\frac{\sqrt{P(y)}}{y}e^{I(y)}.
}
\]

Pour \(0<c_{14}<2\), si :

\[
D=\sqrt{4-2c_{14}},
\]

alors :

\[
\boxed{
I(y)
=
\frac1D
\ln\!\left[
\frac{c_{14}y+2-D}{c_{14}y+2+D}
\frac{2+D}{2-D}
\right].
}
\]

La limite \(c_{14}\to0\) donne :

\[
I(y;0)=\frac12\ln(1+2y).
\]

```python
# G22.7 — Exact parametric solution cross-check
y,c = sp.symbols("y c", positive=True)
P = sp.factor(1+2*y+c*y**2/2)
D = sp.sqrt(4-2*c)

I_closed = (
    sp.log(
        ((c*y+2-D)/(c*y+2+D))
        *((2+D)/(2-D))
    )/D
)

I_DERIVATIVE_PASS = (
    sp.simplify(sp.diff(I_closed,y)-1/P) == 0
)

assert I_DERIVATIVE_PASS

print("P(y) =", P)
print("dI/dy =", sp.factor(sp.diff(I_closed,y)))
print("I_DERIVATIVE_PASS =", I_DERIVATIVE_PASS)
```

# G22.12 — Sweep fort-champ exact

Le sweep suivant utilise les expressions exactes paramétriques, pas le développement 2PN.

Il calcule :

- \(r_{\rm ph}/M\),
- \(b_{\rm crit}/M\),
- \(r_{\rm ISCO}/M\),
- \(M\Omega_{\rm ISCO}\),
- \(E_{\rm ISCO}\),
- \(L_{\rm ISCO}/M\).

Ces observables sont des propriétés de la famille static-æther et ne doivent donc pas être présentées comme signatures propres à GVH.

```python
# G22.13 — High-precision exact strong-field sweep
mp.mp.dps = 60

def I_mp(yv,cv):
    yv = mp.mpf(yv)
    cv = mp.mpf(cv)
    if abs(cv) < mp.mpf("1e-40"):
        return mp.mpf("0.5")*mp.log(1+2*yv)
    Dv = mp.sqrt(4-2*cv)
    return (
        mp.log(
            ((cv*yv+2-Dv)/(cv*yv+2+Dv))
            *((2+Dv)/(2-Dv))
        )/Dv
    )

def strong_field_obs(cv):
    cv = mp.mpf(cv)
    def Pv(v):
        return 1+2*v+cv*v*v/2

    yphv = mp.mpf(1)
    Iph = I_mp(yphv,cv)
    Nph = mp.e**(-Iph)
    rph = mp.sqrt(Pv(yphv))/yphv * mp.e**Iph
    bcrit = rph/Nph

    yis = 2/(4+mp.sqrt(16+2*cv))
    Iis = I_mp(yis,cv)
    Nis = mp.e**(-Iis)
    ris = mp.sqrt(Pv(yis))/yis * mp.e**Iis
    OmM = Nis*mp.sqrt(yis)/ris
    Eis = Nis/mp.sqrt(1-yis)
    LisM = ris*mp.sqrt(yis/(1-yis))
    return rph,bcrit,yis,ris,OmM,Eis,LisM

c_values = ["0","0.1","0.5","1.0","1.5","1.9"]
rows = []
for cv in c_values:
    vals = strong_field_obs(cv)
    rows.append((cv,)+tuple(mp.nstr(v,14) for v in vals))

print("c14 | r_ph/M | bcrit/M | y_ISCO | r_ISCO/M | M Omega_ISCO | E_ISCO | L_ISCO/M")
for row in rows:
    print(" | ".join(row))

gr = strong_field_obs("0")
STRONG_FIELD_GR_LIMIT_NUMERIC_PASS = all([
    abs(gr[0]-3) < mp.mpf("1e-50"),
    abs(gr[1]-3*mp.sqrt(3)) < mp.mpf("1e-50"),
    abs(gr[2]-mp.mpf(1)/4) < mp.mpf("1e-50"),
    abs(gr[3]-6) < mp.mpf("1e-50"),
    abs(gr[4]-1/(6*mp.sqrt(6))) < mp.mpf("1e-50"),
    abs(gr[5]-2*mp.sqrt(2)/3) < mp.mpf("1e-50"),
    abs(gr[6]-2*mp.sqrt(3)) < mp.mpf("1e-50"),
])
assert STRONG_FIELD_GR_LIMIT_NUMERIC_PASS

print("STRONG_FIELD_GR_LIMIT_NUMERIC_PASS =", STRONG_FIELD_GR_LIMIT_NUMERIC_PASS)
```
