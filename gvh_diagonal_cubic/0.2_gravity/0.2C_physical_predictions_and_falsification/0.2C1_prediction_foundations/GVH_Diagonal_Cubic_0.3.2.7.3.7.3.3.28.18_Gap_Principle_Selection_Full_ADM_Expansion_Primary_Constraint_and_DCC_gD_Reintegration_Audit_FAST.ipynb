# G18-B — DCC historique \(\rightarrow(g,D)\)

Le DCC historique possède une dépendance orientationnelle de triade. Dans les audits antérieurs, une rotation de branche pouvait faire varier un facteur du type :

\[
\chi(\phi)=\sin^2(2\phi).
\]

Question :

\[
\boxed{
\text{peut-on reproduire cette information avec seulement }
\operatorname{Tr}(A^2),\operatorname{Tr}(A^3),\operatorname{Tr}(A^4),\ldots ?
}
\]

Toute quantité :

\[
\operatorname{Tr}(A^n)
\]

est invariante sous :

\[
A\rightarrow RAR^{-1}.
\]

Elle dépend du spectre de \(A\), pas de l'orientation d'une base propre relativement à un second objet.

Donc une fonction algébrique :

\[
F(\operatorname{Tr}A^2,\operatorname{Tr}A^3,\ldots)
\]

ne peut pas reproduire une dépendance \(\chi(\phi)\) si le spectre reste fixe pendant la rotation.

```python
phi,a = sp.symbols("phi a", real=True)
R2 = sp.Matrix([
    [sp.cos(phi),-sp.sin(phi)],
    [sp.sin(phi), sp.cos(phi)],
])
A2_0 = sp.diag(a,-a)
A2_phi = sp.simplify(R2*A2_0*R2.T)

I2_phi = sp.trigsimp(sp.trace(A2_phi**2))
I3_phi = sp.trigsimp(sp.trace(A2_phi**3))
I4_phi = sp.trigsimp(sp.trace(A2_phi**4))
chi_phi = sp.simplify(sp.sin(2*phi)**2)

G2818_ALGEBRAIC_TRACE_INVARIANTS_ORIENTATION_BLIND = all([
    sp.simplify(I2_phi-2*a**2)==0,
    sp.simplify(I3_phi)==0,
    sp.simplify(I4_phi-2*a**4)==0,
])

assert G2818_ALGEBRAIC_TRACE_INVARIANTS_ORIENTATION_BLIND

print("Tr(A^2) =",I2_phi)
print("Tr(A^3) =",I3_phi)
print("Tr(A^4) =",I4_phi)
print("chi(phi) =",chi_phi)
```

# G18-C — ADM principal expansion

On utilise :

\[
ds^2
=
-N^2dt^2
+
\gamma_{ij}(dx^i+N^idt)(dx^j+N^jdt),
\]

avec :

\[
\boxed{
K_{ij}
=
\frac{1}{2N}
\left(
\dot\gamma_{ij}
-D_iN_j-D_jN_i
\right)
}
\]

à convention de signe fixée pour ce notebook.

Pour le tenseur mixte :

\[
\nabla_0A
=
\dot A
+
[\Gamma_0,A].
\]

Dans un frame spatial local diagonal pour :

\[
A_{\rm bg}=\operatorname{diag}(\ldots,a_i,a_j,\ldots),
\]

le bloc principal spatial \(ij\) prend :

\[
\Gamma_0^{(ij)}
\sim
\begin{pmatrix}
0&K_{ij}\\
K_{ij}&0
\end{pmatrix}.
\]

On peut donc calculer exactement le mélange entre :

\[
\dot A_{ij}
\]

et :

\[
K_{ij}\sim\dot\gamma_{ij}.
\]

```python
ai,aj,qdot,kij,Q,KS,kappa = sp.symbols(
    "a_i a_j qdot K_ij Q K_S kappa_D",
    real=True
)
d = sp.factor(ai-aj)

Apair = sp.diag(ai,aj)
dotApair = sp.Matrix([[0,qdot],[qdot,0]])
GammaPair = sp.Matrix([[0,kij],[kij,0]])

Bn = sp.simplify(
    dotApair + GammaPair*Apair - Apair*GammaPair
)

trB2 = sp.factor(sp.trace(Bn*Bn))
trAB = sp.factor(sp.trace(Apair*Bn))
commBn = sp.simplify(Apair*Bn-Bn*Apair)
trCommB2 = sp.factor(sp.trace(commBn*commBn))

Cij = sp.factor(KS*Q-kappa*d**2/2)

# Time-kinetic contribution after Lorentz contraction:
Lpair_D = sp.factor(Cij*(qdot**2-d**2*kij**2))

G2818_ADM_SPATIAL_PAIR_IDENTITIES_PASS = all([
    sp.simplify(trB2-2*(qdot**2-d**2*kij**2))==0,
    trAB==0,
    sp.simplify(
        trCommB2+2*d**2*(qdot**2-d**2*kij**2)
    )==0,
])

assert G2818_ADM_SPATIAL_PAIR_IDENTITIES_PASS

print("nabla_n A pair =")
sp.pprint(Bn)
print("Tr(B_n^2) =",trB2)
print("Tr([A,B_n]^2) =",trCommB2)
print("C_ij =",Cij)
print("L_D principal pair =",Lpair_D)
```

## Bloc principal métrique-\(D\)

Le résultat est :

\[
\boxed{

\[

\qquad

Donc :

\[
\boxed{
L_{ij}^{\rm total}
=
C_{ij}\dot A_{ij}^2
+
\left[
\mathcal A_{\rm EH}
-
C_{ij}(a_i-a_j)^2

Dans ce bloc, il n'y a pas de terme croisé \(\dot A_{ij}K_{ij}\).

Les deux conditions principales sont donc :

\[
\boxed{

\[
\boxed{
\mathcal A_{\rm EH}

\[
a_i=a_j,

donc le terme de correction métrique disparaît exactement dans ce canal.

```python
AEH = sp.symbols("A_EH", positive=True, real=True)

Ltotal_pair = sp.expand(
    Cij*qdot**2
    +(AEH-Cij*d**2)*kij**2
)
Hpair = sp.hessian(Ltotal_pair,(qdot,kij))
detHpair = sp.factor(Hpair.det())

G2818_ADM_PRINCIPAL_PAIR_HESSIAN_DERIVED = True
G2818_ADM_PRINCIPAL_PAIR_CROSS_TERM_ZERO = (
    sp.simplify(Hpair[0,1])==0
)
G2818_FULL_ADM_EXPANSION_CLOSED = False
G2818_G18C_GATE_CLOSED = False

assert G2818_ADM_PRINCIPAL_PAIR_HESSIAN_DERIVED
assert G2818_ADM_PRINCIPAL_PAIR_CROSS_TERM_ZERO
assert not G2818_FULL_ADM_EXPANSION_CLOSED

print("principal pair Hessian =")
sp.pprint(Hpair)
print("det =",detHpair)
```
