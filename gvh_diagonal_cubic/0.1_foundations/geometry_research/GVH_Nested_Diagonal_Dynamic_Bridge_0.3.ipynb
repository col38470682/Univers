
# 4. Test de métricité d'une branche dynamique

Soit une branche réelle de vitesse au carré :

\[
y(\mathbf n)=c^2(\mathbf n).
\]

Une métrique riemannienne spatiale unique existe pour cette branche si l'on peut trouver une matrice symétrique positive définie :

\[
C=
\begin{pmatrix}
C_{11}&C_{12}&C_{13}\\
C_{12}&C_{22}&C_{23}\\
C_{13}&C_{23}&C_{33}
\end{pmatrix}
\]

telle que :

\[
\boxed{
y(\mathbf n)
=
\mathbf n^T C\mathbf n
}
\]

pour **toutes** les directions.

Les six coefficients de \(C\) peuvent être reconstruits à partir de six évaluations indépendantes.

Une reconstruction sur quelques directions n'est cependant pas la preuve globale : après reconstruction, il faut vérifier l'identité sur le domaine directionnel complet.

## Cas diagonal visé par la branche géométrique

Pour :

\[
g_{\rm eff}
=
{\rm diag}(1,1,f^2),
\]

la co-métrique vaut :

\[
\boxed{
C=g_{\rm eff}^{-1}
=
{\rm diag}(1,1,f^{-2}).
}
\]

Donc la branche caractéristique correspondante doit satisfaire :

\[
\boxed{
c^2(\mathbf n)
=
n_x^2+n_y^2+f^{-2}n_z^2.
}
\]

Cette formule fournit le test exact qui reliera éventuellement la dynamique GVH à la métrique certifiée de `0.2–0.2.1.1`.


```python

# Exact reconstruction of a symmetric quadratic form from six probes.

c11,c22,c33,c12,c13,c23 = sp.symbols(
    "c11 c22 c33 c12 c13 c23",
    real=True
)

C = sp.Matrix([
    [c11,c12,c13],
    [c12,c22,c23],
    [c13,c23,c33],
])

e1 = sp.Matrix([1,0,0])
e2 = sp.Matrix([0,1,0])
e3 = sp.Matrix([0,0,1])

def qform(M, v):
    return sp.expand((v.T*M*v)[0])

def reconstruct_symmetric_C(y1,y2,y3,y12,y13,y23):
    C11 = sp.simplify(y1)
    C22 = sp.simplify(y2)
    C33 = sp.simplify(y3)
    C12 = sp.simplify((y12-C11-C22)/2)
    C13 = sp.simplify((y13-C11-C33)/2)
    C23 = sp.simplify((y23-C22-C33)/2)
    return sp.Matrix([
        [C11,C12,C13],
        [C12,C22,C23],
        [C13,C23,C33],
    ])

f0 = sp.Rational(2,1)
C_target = sp.diag(
    1,
    1,
    1/f0**2
)

probes = [
    e1,
    e2,
    e3,
    e1+e2,
    e1+e3,
    e2+e3,
]

vals = [
    qform(C_target,v)
    for v in probes
]

C_recovered = reconstruct_symmetric_C(*vals)

GNDG03_METRIC_EXTRACTION_SYNTHETIC_EXACT_PASS = (
    sp.simplify(
        C_recovered-C_target
    ) == sp.zeros(3)
)

assert GNDG03_METRIC_EXTRACTION_SYNTHETIC_EXACT_PASS

print("C_target =")
sp.pprint(C_target)
print("C_recovered =")
sp.pprint(C_recovered)
print(
    "GNDG03_METRIC_EXTRACTION_SYNTHETIC_EXACT_PASS =",
    GNDG03_METRIC_EXTRACTION_SYNTHETIC_EXACT_PASS
)
```


# 5. Test falsifiable : toutes les anisotropies ne sont pas métriques

Une dépendance directionnelle peut être parfaitement régulière sans provenir d'une forme quadratique.

Par exemple, sur la sphère :

\[
\boxed{
y_{\rm nonmetric}(\mathbf n)
=
1+\mu n_x^2n_z^2
}
\]

contient une anisotropie quartique.

On peut toujours ajuster une matrice \(C\) sur six directions, mais une direction indépendante révèle en général un résidu non nul.

Ainsi :

\[
\boxed{
\text{direction-dependence}
\not\Rightarrow
\text{Riemannian metric}.
}
\]

Si une branche GVH échoue à ce test, le pont vers une unique métrique \(g_{\rm eff}\) échoue sous cette forme.

Il faudrait alors considérer, selon ce que la dynamique impose réellement :

- plusieurs métriques effectives ;
- une structure de dispersion plus générale ;
- éventuellement une géométrie de type Finsler.

Aucune de ces extensions n'est autorisée par ce notebook ; elles ne sont mentionnées que comme classification d'un éventuel FAIL.


```python

mu = sp.Rational(1,5)

def ynon(v):
    # homogeneous direction function evaluated after normalization
    vv = sp.Matrix(v)
    norm2 = sp.expand((vv.T*vv)[0])
    nx2 = sp.simplify(vv[0]**2/norm2)
    nz2 = sp.simplify(vv[2]**2/norm2)
    return sp.simplify(1 + mu*nx2*nz2)

probe_vals_nonmetric = [
    ynon(v)
    for v in probes
]

C_fit_nonmetric = reconstruct_symmetric_C(
    *probe_vals_nonmetric
)

vtest = sp.Matrix([1,1,1])
n2test = sp.expand((vtest.T*vtest)[0])

# For phase speed on unit directions, quadratic model is
# (v^T C v)/(v^T v).
pred_test = sp.simplify(
    qform(C_fit_nonmetric,vtest)
    / n2test
)
true_test = ynon(vtest)
nonmetric_residual = sp.simplify(
    true_test-pred_test
)

GNDG03_NONMETRIC_BRANCH_REJECTION_PASS = (
    nonmetric_residual != 0
)

assert GNDG03_NONMETRIC_BRANCH_REJECTION_PASS

print("independent residual =", nonmetric_residual)
print(
    "GNDG03_NONMETRIC_BRANCH_REJECTION_PASS =",
    GNDG03_NONMETRIC_BRANCH_REJECTION_PASS
)
```
