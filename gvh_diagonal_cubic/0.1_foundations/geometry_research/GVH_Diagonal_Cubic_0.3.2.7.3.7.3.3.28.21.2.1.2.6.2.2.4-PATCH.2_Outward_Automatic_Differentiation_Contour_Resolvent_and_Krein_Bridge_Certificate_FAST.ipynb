# Niveau 1 — Reconstruction autonome de \(K,M,G\)

Le notebook reconstruit le même principal Pure-GVH-P que `.28.21.1.4 → .1.6.1` :

\[
P_{\rm raw}(\omega,\mathbf k)

Cela évite toute dépendance d'exécution à un fichier parent externe.

```python

eta=sp.diag(-1,1,1,1)

a0,a1,a2=sp.symbols("a0 a1 a2",real=True)
a3=-a0-a1-a2

Abar=sp.diag(a0,a1,a2,a3)
Qbar=sp.factor(sp.trace(Abar*Abar))

KS,kappaD,Mpl2=sp.symbols(
    "K_S kappa_D Mpl2",
    real=True
)

names=[
    "n","beta1","beta2","beta3",
    "h11","h22","h33","h12","h13","h23",
    "D00","D01","D02","D03",
    "D11","D22","D33","D12","D13","D23",
]

h_basis=[]
D_basis=[]

for name in names:
    h=sp.zeros(4)
    d=sp.zeros(4)

    if name=="n":
        h[0,0]=-2
    elif name.startswith("beta"):
        i=int(name[-1])
        h[0,i]=h[i,0]=1
    elif name.startswith("h"):
        ij=name[1:]
        i,j=int(ij[0]),int(ij[1])
        h[i,j]=h[j,i]=1
    elif name.startswith("D"):
        ij=name[1:]
        i,j=int(ij[0]),int(ij[1])
        d[i,j]=d[j,i]=1

    h_basis.append(h)
    D_basis.append(d)

dA_basis=[]

for h,d in zip(h_basis,D_basis):
    dM=-eta*h*Abar+eta*d
    dA=dM-sp.trace(dM)*sp.eye(4)/4
    dA_basis.append(dA)

def build_principal_matrix(p):
    H=sp.zeros(20)

    # Pure-GVH-P sector
    for alpha_idx in range(4):
        B=[]
        J=[]
        C=[]

        for h,dA in zip(h_basis,dA_basis):
            Gamma=sp.zeros(4)

            for mu in range(4):
                for nu in range(4):
                    acc=0
                    for rho in range(4):
                        acc += eta[mu,rho]*(
                            p[alpha_idx]*h[rho,nu]
                            +p[nu]*h[rho,alpha_idx]
                            -p[rho]*h[alpha_idx,nu]
                        )/2
                    Gamma[mu,nu]=acc

            Bj=p[alpha_idx]*dA+Gamma*Abar-Abar*Gamma
            B.append(Bj)
            J.append(sp.trace(Abar*Bj))
            C.append(Abar*Bj-Bj*Abar)

        sign=eta[alpha_idx,alpha_idx]

        for j in range(20):
            for k in range(j,20):
                shape=Qbar*sp.trace(B[j]*B[k])-J[j]*J[k]
                angle=sp.trace(C[j]*C[k])

                val=(
                    -KS*sign*shape
                    -kappaD*sp.Rational(1,2)*sign*angle
                )

                H[j,k]+=val
                if k!=j:
                    H[k,j]+=val

    # Einstein-Hilbert / Fierz-Pauli benchmark
    pvec=sp.Matrix(p)
    pup=eta*pvec
    p2=(pvec.T*eta*pvec)[0]

    attrs=[]

    for h in h_basis[:10]:
        hup=eta*h*eta
        v=[
            sum(p[mu]*hup[mu,nu] for mu in range(4))
            for nu in range(4)
        ]
        w=[
            sum(pup[lam]*h[lam,nu] for lam in range(4))
            for nu in range(4)
        ]
        trh=sp.trace(eta*h)
        vp=sum(v[nu]*p[nu] for nu in range(4))
        attrs.append((h,hup,v,w,trh,vp))

    for j in range(10):
        hj,hjup,vj,wj,trj,vpj=attrs[j]

        for k in range(j,10):
            hk,hkup,vk,wk,trk,vpk=attrs[k]

            inner=sum(
                hj[mu,nu]*hkup[mu,nu]
                for mu in range(4)
                for nu in range(4)
            )

            BF=(
                p2*inner
                -sum(
                    vj[nu]*wk[nu]+vk[nu]*wj[nu]
                    for nu in range(4)
                )
                +vpj*trk
                +vpk*trj
                -p2*trj*trk
            )

            val=-Mpl2*sp.Rational(1,4)*BF

            H[j,k]+=val
            if k!=j:
                H[k,j]+=val

    return H

e0=(1,0,0,0)
e1=(0,1,0,0)
e2=(0,0,1,0)
e3=(0,0,0,1)

P_e0=build_principal_matrix(e0)
P_e1=build_principal_matrix(e1)
P_e2=build_principal_matrix(e2)
P_e3=build_principal_matrix(e3)

K_raw=P_e0

M_raw={
    1:build_principal_matrix((1,1,0,0))-P_e0-P_e1,
    2:build_principal_matrix((1,0,1,0))-P_e0-P_e2,
    3:build_principal_matrix((1,0,0,1))-P_e0-P_e3,
}

G_raw={
    (1,1):P_e1,
    (2,2):P_e2,
    (3,3):P_e3,
    (1,2):(build_principal_matrix((0,1,1,0))-P_e1-P_e2)/2,
    (1,3):(build_principal_matrix((0,1,0,1))-P_e1-P_e3)/2,
    (2,3):(build_principal_matrix((0,0,1,1))-P_e2-P_e3)/2,
}

G2821161_RAW_PENCIL_RECONSTRUCTED=all([
    K_raw.shape==(20,20),
    all(M_raw[i].shape==(20,20) for i in (1,2,3)),
    all(G_raw[key].shape==(20,20) for key in G_raw),
])

assert G2821161_RAW_PENCIL_RECONSTRUCTED

print("G2821161_RAW_PENCIL_RECONSTRUCTED =",G2821161_RAW_PENCIL_RECONSTRUCTED)
```

```python
Q_symbolic=sp.simplify(
    (T20_inv*G0_symbolic)[:14,:]
)
F_global_symbolic=sp.simplify(Q_symbolic.T)
Gram_global=sp.simplify(Q_symbolic.T*Q_symbolic)
det_Gram=sp.factor(Gram_global.det())

poly_Gram=sp.Poly(sp.expand(det_Gram),n1,n2,n3)
gram_terms=poly_Gram.terms()

gram_even_exponents=all(
    all(e%2==0 for e in monom)
    for monom,coeff in gram_terms
)
gram_positive_coeffs=all(
    bool(coeff>0)
    for monom,coeff in gram_terms
)

G2821162_GLOBAL_GRAM_GAUGE_REGULAR_PASS=all([
    Q_symbolic.shape==(14,4),
    gram_even_exponents,
    gram_positive_coeffs,
    len(gram_terms)>0,
])

assert G2821162_GLOBAL_GRAM_GAUGE_REGULAR_PASS

print("det Gram total degree =",poly_Gram.total_degree())
print("det Gram term count =",len(gram_terms))
print("all exponents even =",gram_even_exponents)
print("all coefficients positive =",gram_positive_coeffs)
print(
    "G2821162_GLOBAL_GRAM_GAUGE_REGULAR_PASS =",
    G2821162_GLOBAL_GRAM_GAUGE_REGULAR_PASS
)
```

```python
def direction_blocks(direction):
    x,y,z=direction

    B=(x*M_h[1]+y*M_h[2]+z*M_h[3])
    D=B/2

    C=(
        x*x*G_h[(1,1)]
        +y*y*G_h[(2,2)]
        +z*z*G_h[(3,3)]
        +2*x*y*G_h[(1,2)]
        +2*x*z*G_h[(1,3)]
        +2*y*z*G_h[(2,3)]
    )

    return {
        "B":B,"D":D,"C":C,
        "Baa":sp.simplify(R_kin.T*B*R_kin),
        "BaN":sp.simplify(R_kin.T*B*N_diff),
        "Daa":sp.simplify(R_kin.T*D*R_kin),
        "Dau":sp.simplify(R_kin.T*D*N_diff),
        "Caa":sp.simplify(R_kin.T*C*R_kin),
        "CaN":sp.simplify(R_kin.T*C*N_diff),
        "BNR":sp.simplify(N_diff.T*B*R_kin),
        "BNN":sp.simplify(N_diff.T*B*N_diff),
        "CNR":sp.simplify(N_diff.T*C*R_kin),
        "CNN":sp.simplify(N_diff.T*C*N_diff),
    }

blk_symbolic=direction_blocks((n1,n2,n3))

U_global_symbolic=sp.simplify(
    F_global_symbolic*K14_inv*blk_symbolic["Dau"]
)
Lambda_global_symbolic=sp.simplify(
    F_global_symbolic*K14_inv*blk_symbolic["BaN"]
)

det_U_global=sp.factor(U_global_symbolic.det())
det_Lambda_global=sp.factor(Lambda_global_symbolic.det())

G2821162_GLOBAL_GAUGE_MULTIPLIER_CLOSURE_PASS=all([
    sp.simplify(det_U_global-det_Gram/16)==0,
    sp.simplify(det_Lambda_global-det_Gram)==0,
])

assert G2821162_GLOBAL_GAUGE_MULTIPLIER_CLOSURE_PASS

print("det(U_global) = det(Gram)/16 :",
      sp.simplify(det_U_global-det_Gram/16)==0)
print("det(Lambda_global) = det(Gram) :",
      sp.simplify(det_Lambda_global-det_Gram)==0)
```

```python
def frame_from_pivots(Chat,Ahat,pivots=None):
    if pivots is None:
        pivots=tuple(Chat.rref()[1])
    else:
        pivots=tuple(sorted(pivots))

    assert len(pivots)==8
    free=[j for j in range(28) if j not in pivots]

    Cp=Chat[:,list(pivots)]
    Cf=Chat[:,free]
    assert Cp.det()!=0

    solved=sp.simplify(-Cp.inv()*Cf)

    R=sp.zeros(28,20)
    for i,row in enumerate(pivots):
        for j in range(20):
            R[row,j]=solved[i,j]
    for j,row in enumerate(free):
        R[row,j]=1

    L=sp.zeros(20,28)
    for j,row in enumerate(free):
        L[j,row]=1

    Aphys=sp.simplify(L*Ahat*R)

    checks={
        "constraint":Chat*R==sp.zeros(8,20),
        "left_inverse":L*R==sp.eye(20),
        "rank":R.rank()==20,
        "intertwining":sp.simplify(Ahat*R-R*Aphys)==sp.zeros(28,20),
    }

    assert all(checks.values())

    return {
        "R":R,
        "L":L,
        "Aphys":Aphys,
        "pivots":pivots,
        "free":tuple(free),
        "checks":checks,
    }

frame_ledger=[]
frame_cache={}

for direction in witness_directions:
    norm=normalized_direction_bundle(direction)
    fr=frame_from_pivots(norm["Chat"],norm["Ahat"])
    frame_cache[str(tuple(direction))]=fr

    frame_ledger.append({
        "direction":str(tuple(direction)),
        "pivots":str(fr["pivots"]),
        "R_rank":int(fr["R"].rank()),
        "Aphys_shape":str(fr["Aphys"].shape),
        "intertwining":fr["checks"]["intertwining"],
    })

G2821162_LOCAL_APHYS_20X20_MATERIALIZED_PASS=all([
    row["R_rank"]==20
    and row["Aphys_shape"]=="(20, 20)"
    and row["intertwining"]
    for row in frame_ledger
])

assert G2821162_LOCAL_APHYS_20X20_MATERIALIZED_PASS

print(pd.DataFrame(frame_ledger).to_string(index=False))
print(
    "G2821162_LOCAL_APHYS_20X20_MATERIALIZED_PASS =",
    G2821162_LOCAL_APHYS_20X20_MATERIALIZED_PASS
)
```

```python
def alternate_pivot_sets(Chat,base_pivots,max_count=3):
    base=tuple(base_pivots)
    free=[j for j in range(28) if j not in base]
    out=[]

    for p in base:
        for f in free:
            candidate=list(base)
            candidate[candidate.index(p)]=f
            candidate=tuple(sorted(candidate))

            if len(set(candidate))!=8 or candidate in out:
                continue

            if Chat[:,list(candidate)].det()!=0:
                out.append(candidate)
                if len(out)>=max_count:
                    return out

    return out

transition_points=[
    (sp.Integer(3),sp.Integer(4),sp.Integer(0)),
    (sp.Integer(3),sp.Integer(0),sp.Integer(4)),
    (sp.Integer(0),sp.Integer(3),sp.Integer(4)),
    (sp.Integer(1),sp.Integer(2),sp.Integer(2)),
    (sp.Integer(3),sp.Integer(-4),sp.Integer(12)),
]

transition_ledger=[]

for direction in transition_points:
    norm=normalized_direction_bundle(direction)
    base=frame_from_pivots(norm["Chat"],norm["Ahat"])
    alternatives=alternate_pivot_sets(
        norm["Chat"],base["pivots"],max_count=2
    )

    assert len(alternatives)>=1

    for alt_piv in alternatives:
        alt=frame_from_pivots(
            norm["Chat"],norm["Ahat"],alt_piv
        )

        T_ba=sp.simplify(alt["L"]*base["R"])
        T_ab=sp.simplify(base["L"]*alt["R"])

        embed_pass=(
            sp.simplify(alt["R"]*T_ba-base["R"])
            ==sp.zeros(28,20)
        )
        inverse_pass=(
            sp.simplify(T_ab*T_ba)==sp.eye(20)
            and sp.simplify(T_ba*T_ab)==sp.eye(20)
        )
        similarity_pass=(
            sp.simplify(
                alt["Aphys"]*T_ba
                -T_ba*base["Aphys"]
            )==sp.zeros(20)
        )

        transition_ledger.append({
            "direction":str(tuple(direction)),
            "base_pivots":str(base["pivots"]),
            "alt_pivots":str(alt_piv),
            "transition_rank":int(T_ba.rank()),
            "embedding":bool(embed_pass),
            "inverse":bool(inverse_pass),
            "similarity":bool(similarity_pass),
        })

G2821162_FRAME_TRANSITION_SIMILARITY_PASS=all([
    row["transition_rank"]==20
    and row["embedding"]
    and row["inverse"]
    and row["similarity"]
    for row in transition_ledger
])

assert G2821162_FRAME_TRANSITION_SIMILARITY_PASS

print(pd.DataFrame(transition_ledger).to_string(index=False))
print(
    "G2821162_FRAME_TRANSITION_SIMILARITY_PASS =",
    G2821162_FRAME_TRANSITION_SIMILARITY_PASS
)
```

```python
background_subs={
    a0:sp.Rational(3,4),a1:-sp.Rational(1,5),
    a2:-sp.Rational(1,4),a3:-sp.Rational(3,10),
}
coupling_symbols=(KS,kappaD,Mpl2)

def _lf(M):
    return sp.lambdify(coupling_symbols,M.subs(background_subs),modules="numpy")

K_of_c=_lf(K_raw)
M_of_c={i:_lf(M_raw[i]) for i in (1,2,3)}
G_of_c={key:_lf(G_raw[key]) for key in G_raw}
F_of_n=sp.lambdify((n1,n2,n3),F_global_symbolic,modules="numpy")

R14=np.asarray(R_kin,dtype=float)
N4=np.asarray(N_diff,dtype=float)
I14=np.eye(14)
I28=np.eye(28)

def _arr(x):
    return np.asarray(x,dtype=float)

def _min_sv(x):
    return float(np.linalg.svd(x,compute_uv=False)[-1])

def coupling_direction_bundle(coupling,direction,rcond=1e-11):
    c=tuple(float(v) for v in coupling)
    n=np.asarray(direction,dtype=float)
    n=n/np.linalg.norm(n)
    K=_arr(K_of_c(*c))
    Ms={i:_arr(M_of_c[i](*c)) for i in (1,2,3)}
    Gs={key:_arr(G_of_c[key](*c)) for key in G_of_c}
    B=sum(n[i-1]*Ms[i] for i in (1,2,3)); D=B/2.0
    C=(n[0]**2*Gs[(1,1)]+n[1]**2*Gs[(2,2)]+n[2]**2*Gs[(3,3)]
       +2*n[0]*n[1]*Gs[(1,2)]+2*n[0]*n[2]*Gs[(1,3)]
       +2*n[1]*n[2]*Gs[(2,3)])
    K14c=R14.T@K@R14
    Baa=R14.T@B@R14; BaN=R14.T@B@N4
    Daa=R14.T@D@R14; Dau=R14.T@D@N4
    Caa=R14.T@C@R14; CaN=R14.T@C@N4
    BNR=N4.T@B@R14; CNR=N4.T@C@R14; CNN=N4.T@C@N4
    F=_arr(F_of_n(*n))

    Ki_Dau=np.linalg.solve(K14c,Dau)
    Ki_Daa=np.linalg.solve(K14c,Daa)
    Ki_I=np.linalg.solve(K14c,I14)
    Umat=F@Ki_Dau
    Ux=np.linalg.solve(Umat,-F@Ki_Daa)
    Up=np.linalg.solve(Umat,F@Ki_I)
    Vx=np.linalg.solve(K14c,-Daa-Dau@Ux)
    Vp=np.linalg.solve(K14c,I14-Dau@Up)
    Lmat=F@np.linalg.solve(K14c,BaN)
    rhs_x=F@np.linalg.solve(K14c,Baa@Vx+Caa+CaN@Ux)
    rhs_p=F@np.linalg.solve(K14c,Baa@Vp+CaN@Up)
    Lx=np.linalg.solve(Lmat,-rhs_x); Lp=np.linalg.solve(Lmat,-rhs_p)
    Pdx=-Daa@Vx-Dau@Lx-Caa-CaN@Ux
    Pdp=-Daa@Vp-Dau@Lp-CaN@Up
    AHD=np.block([[Vx,Vp],[Pdx,Pdp]])
    Hx=BNR@Vx+CNR+CNN@Ux; Hp=BNR@Vp+CNN@Up
    Cphys=np.block([[F,np.zeros((4,14))],[Hx,Hp]])
    R20=null_space(Cphys,rcond=rcond)
    if R20.shape!=(28,20):
        raise np.linalg.LinAlgError(f"physical nullity {R20.shape[1]} != 20")
    Aphys=R20.T@AHD@R20
    tangent=float(np.linalg.norm((I28-R20@R20.T)@AHD@R20,ord=2))
    return {
        "Aphys":Aphys,"AHD":AHD,"Cphys":Cphys,"R20":R20,
        "rank_K14":int(np.linalg.matrix_rank(K14c,tol=1e-10)),
        "rank_Cphys":int(np.linalg.matrix_rank(Cphys,tol=1e-9)),
        "min_sv_K14":_min_sv(K14c),"min_sv_U":_min_sv(Umat),
        "min_sv_Lambda":_min_sv(Lmat),"min_sv_Cphys":_min_sv(Cphys),
        "tangency_residual":tangent,
    }

COUPLING_DEPENDENT_HAMILTON_DIRAC_BUNDLE_MATERIALIZED=True
print("COUPLING_DEPENDENT_HAMILTON_DIRAC_BUNDLE_MATERIALIZED =",COUPLING_DEPENDENT_HAMILTON_DIRAC_BUNDLE_MATERIALIZED)
```

## Niveau 1 — Noyau d’intervalles IEEE-754 à arrondi extérieur

Every elementary endpoint is expanded by one `nextafter` step. The test suite
checks containment against exact rational arithmetic. This qualifies the
implemented operations only; it does not by itself certify a spectral claim.

```python
from dataclasses import dataclass
from fractions import Fraction

NEG_INF=float('-inf'); POS_INF=float('inf')
def down(x): return float(np.nextafter(float(x),NEG_INF))
def up(x): return float(np.nextafter(float(x),POS_INF))

@dataclass(frozen=True)
class IV:
    lo: float
    hi: float
    def __post_init__(self):
        if self.lo>self.hi: raise ValueError("empty interval")
    @staticmethod
    def point(x):
        q=float(x); return IV(down(q),up(q))
    def __add__(self,o):
        o=o if isinstance(o,IV) else IV.point(o)
        return IV(down(self.lo+o.lo),up(self.hi+o.hi))
    __radd__=__add__
    def __neg__(self): return IV(down(-self.hi),up(-self.lo))
    def __sub__(self,o): return self+(-o if isinstance(o,IV) else -IV.point(o))
    def __rsub__(self,o): return IV.point(o)-self
    def __mul__(self,o):
        o=o if isinstance(o,IV) else IV.point(o)
        vals=(self.lo*o.lo,self.lo*o.hi,self.hi*o.lo,self.hi*o.hi)
        return IV(down(min(vals)),up(max(vals)))
    __rmul__=__mul__
    def reciprocal(self):
        if self.lo<=0<=self.hi: raise ZeroDivisionError("interval contains zero")
        return IV(down(1.0/self.hi),up(1.0/self.lo))
    def __truediv__(self,o):
        o=o if isinstance(o,IV) else IV.point(o); return self*o.reciprocal()
    def __pow__(self,n):
        if n<0: return (self.reciprocal())**(-n)
        z=IV.point(1.0)
        for _ in range(n): z=z*self
        return z

def contains(iv,x): return iv.lo<=float(x)<=iv.hi

tests=[]
for qa,qb in [(Fraction(1,3),Fraction(2,7)),(Fraction(-5,9),Fraction(4,11)),(Fraction(-2,5),Fraction(-7,13))]:
    A=IV.point(qa); B=IV.point(qb)
    tests += [contains(A+B,qa+qb),contains(A-B,qa-qb),contains(A*B,qa*qb)]
    if qb: tests.append(contains(A/B,qa/qb))
OUTWARD_ROUNDED_SCALAR_INTERVAL_SELF_TEST_PASS=all(tests)
assert OUTWARD_ROUNDED_SCALAR_INTERVAL_SELF_TEST_PASS
print("interval tests =",len(tests))
print("OUTWARD_ROUNDED_SCALAR_INTERVAL_SELF_TEST_PASS =",OUTWARD_ROUNDED_SCALAR_INTERVAL_SELF_TEST_PASS)
```

## Niveau 2 — Contours de Riesz centraux et budgets de transport

For each repeated spectral cluster, a circular contour is placed at one third
of the nearest distinct-cluster separation. The minimum singular value of
\(zI-A_{\rm phys}(c_0,\hat k)\) on 256 contour nodes is the numerical seed
for a validated resolvent lower bound. The trapezoidal Riesz integral is checked
for idempotence and integer trace. These are diagnostics until the entire
contour, rather than its nodes, is interval-enclosed.

```python
def spectral_clusters(values,tol=2e-6):
    vals=sorted((complex(v) for v in values),key=lambda z:(z.real,z.imag))
    groups=[]
    for z in vals:
        if not groups or abs(z-np.mean(groups[-1]))>tol: groups.append([z])
        else: groups[-1].append(z)
    return groups

def riesz_seed(A,cluster_center,contour_radius,nodes=256):
    I=np.eye(A.shape[0]); P=np.zeros_like(A,dtype=complex); min_sv=np.inf
    for j in range(nodes):
        theta=2*np.pi*j/nodes; z=cluster_center+contour_radius*np.exp(1j*theta)
        R=np.linalg.inv(z*I-A)
        min_sv=min(min_sv,float(np.linalg.svd(z*I-A,compute_uv=False)[-1]))
        P += contour_radius*np.exp(1j*theta)*R/nodes
    return {"min_sampled_resolvent_sv":min_sv,
            "projector_idempotence_residual":float(np.linalg.norm(P@P-P,ord=2)),
            "projector_trace_real":float(np.trace(P).real),
            "projector_trace_imag":float(abs(np.trace(P).imag))}

riesz_seed_ledger=[]
center_coupling=(1.0,2.0,1.0)
for name,direction in collision_directions.items():
    obj=coupling_direction_bundle(center_coupling,tuple(float(x) for x in direction))
    A=obj["Aphys"]; groups=spectral_clusters(np.linalg.eigvals(A))
    centers=[complex(np.mean(g)) for g in groups]
    repeated=[(i,g) for i,g in enumerate(groups) if len(g)>1]
    for i,g in repeated:
        gap=min(abs(centers[i]-centers[j]) for j in range(len(groups)) if j!=i)
        radius=float(gap/3.0)
        seed=riesz_seed(A,centers[i],radius)
        row={"collision":name,"cluster_center_real":float(centers[i].real),
             "cluster_center_imag":float(centers[i].imag),"multiplicity":len(g),
             "nearest_distinct_gap":float(gap),"contour_radius":radius,**seed}
        row["numerical_seed_pass"]=all([
            seed["min_sampled_resolvent_sv"]>0,
            seed["projector_idempotence_residual"]<2e-6,
            abs(seed["projector_trace_real"]-len(g))<2e-5,
            seed["projector_trace_imag"]<2e-5,
        ])
        riesz_seed_ledger.append(row)

CENTRAL_RIESZ_CONTOUR_NUMERICAL_SEEDS_PASS=bool(riesz_seed_ledger) and all(r["numerical_seed_pass"] for r in riesz_seed_ledger)
print(pd.DataFrame(riesz_seed_ledger).to_string(index=False))
print("CENTRAL_RIESZ_CONTOUR_NUMERICAL_SEEDS_PASS =",CENTRAL_RIESZ_CONTOUR_NUMERICAL_SEEDS_PASS)
```

## 1 — Résolution linéaire intervalle par préconditionnement

For interval matrix ([A]=A_0+[-R,R]), an approximate inverse (Y) certifies
regularity if (q=\|I-YA_0\|_\infty+\||Y|R\|_\infty<1). Then every solution
of ([A]x=b) is enclosed using the corresponding Neumann factor
\((1-q)^{-1}).

```python
def preconditioned_interval_solve_certificate(A0,Arad,b0,brad=None):
    A0=np.asarray(A0,float); Arad=np.asarray(Arad,float); b0=np.asarray(b0,float)
    brad=np.zeros_like(b0) if brad is None else np.asarray(brad,float)
    Y=np.linalg.inv(A0); residual=up(np.linalg.norm(np.eye(A0.shape[0])-Y@A0,ord=np.inf))
    radius_term=up(np.linalg.norm(np.abs(Y)@Arad,ord=np.inf)); q=up(residual+radius_term)
    x0=Y@b0
    rhs_rad=np.abs(Y)@(brad+Arad@np.abs(x0))
    xrad=(rhs_rad/(1-q)) if q<1 else np.full_like(x0,np.inf)
    return {"q":q,"residual":residual,"radius_term":radius_term,"regular":bool(q<1),
            "x_mid":x0,"x_rad":xrad}

A_test=np.array([[3.,1.],[1.,2.]]); R_test=np.full((2,2),1e-12); b_test=np.array([1.,-2.])
solve_test=preconditioned_interval_solve_certificate(A_test,R_test,b_test)
INTERVAL_LINEAR_SOLVE_PRIMITIVE_PASS=solve_test["regular"] and np.all(np.isfinite(solve_test["x_rad"]))
assert INTERVAL_LINEAR_SOLVE_PRIMITIVE_PASS
print({k:v for k,v in solve_test.items() if k not in ("x_mid","x_rad")})
print("INTERVAL_LINEAR_SOLVE_PRIMITIVE_PASS =",INTERVAL_LINEAR_SOLVE_PRIMITIVE_PASS)
```

\(eta\ge\|I-YM\|_2) satisfies (eta<1), then

\sigma_{\min}(M)\ge (1-\eta)/\|Y\|_2.
\]
The primitive below records every quantity used by that implication.

```python
def inverse_residual_sigma_lower_certificate(M):
    M=np.asarray(M,complex); Y=np.linalg.inv(M)
    eta=up(np.linalg.norm(np.eye(M.shape[0])-Y@M,ord=2)); Yn=up(np.linalg.norm(Y,ord=2))
    lower=down((1-eta)/Yn) if eta<1 else 0.0
    return {"inverse_residual_upper":eta,"inverse_norm_upper":Yn,
            "sigma_min_lower":lower,"pass":bool(eta<1 and lower>0)}

sigma_test=inverse_residual_sigma_lower_certificate(np.array([[2.,1.],[0.,3.]],complex))
INVERSE_RESIDUAL_SIGMA_LOWER_PRIMITIVE_PASS=sigma_test["pass"]
assert INVERSE_RESIDUAL_SIGMA_LOWER_PRIMITIVE_PASS
print(sigma_test)
print("INVERSE_RESIDUAL_SIGMA_LOWER_PRIMITIVE_PASS =",INVERSE_RESIDUAL_SIGMA_LOWER_PRIMITIVE_PASS)
```

## 1 — Noyau de boules matricielles et différentiation automatique

For a matrix ball (X=X_m+[-X_r,X_r]), non-negative matrix products use the
standard (gamma_n=nu/(1-nu)) correction.  A solve is accepted only when the
preconditioned residual satisfies (q<1).  Its derivatives use

\[
  \partial_j X=A^{-1}(\partial_jB-(\partial_jA)X).
\]

Thus the same validated solve encloses both the value and its Jacobian.

```python
from dataclasses import dataclass

UNIT_ROUNDOFF=np.finfo(float).eps/2
def _rup(x): return np.nextafter(np.asarray(x,dtype=float),np.inf)
def _rdown(x): return np.nextafter(np.asarray(x,dtype=float),-np.inf)
def _gamma(n): return float(np.nextafter((n*UNIT_ROUNDOFF)/(1-n*UNIT_ROUNDOFF),np.inf))
def _nnmm(A,B):
    A=np.asarray(A,float);B=np.asarray(B,float);g=_gamma(4*A.shape[1])
    return _rup((A@B)/(1-g))
def _nnsum(*xs):
    z=np.zeros_like(np.asarray(xs[0],float))
    for x in xs: z=_rup(z+np.asarray(x,float))
    return z

@dataclass
class MBall:
    mid: np.ndarray
    rad: np.ndarray
    def __post_init__(self):
        dtype=complex if np.iscomplexobj(self.mid) else float
        self.mid=np.asarray(self.mid,dtype=dtype);self.rad=np.asarray(self.rad,float)
        if self.mid.shape!=self.rad.shape or np.any(self.rad<0): raise ValueError("invalid matrix ball")
    @staticmethod
    def point(x):
        dtype=complex if np.iscomplexobj(x) else float
        x=np.asarray(x,dtype=dtype);return MBall(x,np.zeros(x.shape,dtype=float))
    def __neg__(self): return MBall(-self.mid,self.rad.copy())
    def __add__(self,o):
        o=o if isinstance(o,MBall) else MBall.point(o)
        m=self.mid+o.mid
        rr=_rup(_gamma(1)*(np.abs(self.mid)+np.abs(o.mid)))
        return MBall(m,_nnsum(self.rad,o.rad,rr))
    __radd__=__add__
    def __sub__(self,o): return self+(-o)
    def T(self): return MBall(self.mid.T,self.rad.T)
    def take(self,rows,cols): return MBall(self.mid[np.ix_(rows,cols)],self.rad[np.ix_(rows,cols)])
    def scale(self,s):
        s=float(s);m=self.mid*s
        rr=_rup(UNIT_ROUNDOFF*np.abs(m))
        return MBall(m,_nnsum(abs(s)*self.rad,rr))

def mm(A,B):
    A=A if isinstance(A,MBall) else MBall.point(A);B=B if isinstance(B,MBall) else MBall.point(B)
    m=A.mid@B.mid; central=_nnmm(np.abs(A.mid),np.abs(B.mid))
    round_mid=_rup(_gamma(4*A.mid.shape[1])*central)
    r=_nnsum(_nnmm(np.abs(A.mid),B.rad),_nnmm(A.rad,np.abs(B.mid)),_nnmm(A.rad,B.rad),round_mid)
    return MBall(m,r)

def block_ball(rows):
    return MBall(np.block([[x.mid for x in row] for row in rows]),
                 np.block([[x.rad for x in row] for row in rows]))

def norm_inf_upper(X): return float(_rup(np.max(_rup(np.sum(_rup(np.abs(X.mid)+X.rad),axis=1)))))
def norm_one_upper(X): return float(_rup(np.max(_rup(np.sum(_rup(np.abs(X.mid)+X.rad),axis=0)))))
def norm_two_upper(X): return float(_rup(np.sqrt(_rup(norm_inf_upper(X)*norm_one_upper(X)))))

SOLVE_DIAGNOSTICS=[]
def solve_ball(A,B,label=""):
    n=A.mid.shape[0];Y=np.linalg.inv(A.mid)
    E=MBall.point(np.eye(n))-mm(MBall.point(Y),A)
    q=norm_inf_upper(E)
    X0=np.linalg.solve(A.mid,B.mid)
    residual=B-mm(A,MBall.point(X0))
    Yn=float(_rup(np.linalg.norm(Y,ord=np.inf)/(1-_gamma(4*n))))
    err=float(_rup(Yn*norm_inf_upper(residual)/(1-q))) if q<1 else np.inf
    X=MBall(X0,np.full_like(X0,err))
    SOLVE_DIAGNOSTICS.append({"label":label,"q":q,"inverse_norm_upper":Yn,
                              "solution_radius_max":float(np.max(X.rad)),"pass":bool(q<1 and np.isfinite(err))})
    return X

@dataclass
class ADBall:
    val: MBall
    der: tuple
    @staticmethod
    def point(x):
        v=MBall.point(x);return ADBall(v,tuple(MBall.point(np.zeros_like(v.mid)) for _ in range(3)))
    def __neg__(self): return ADBall(-self.val,tuple(-d for d in self.der))
    def __add__(self,o):
        o=o if isinstance(o,ADBall) else ADBall.point(o)
        return ADBall(self.val+o.val,tuple(self.der[j]+o.der[j] for j in range(3)))
    __radd__=__add__
    def __sub__(self,o): return self+(-o)
    def T(self): return ADBall(self.val.T(),tuple(d.T() for d in self.der))
    def take(self,rows,cols): return ADBall(self.val.take(rows,cols),tuple(d.take(rows,cols) for d in self.der))
    def scale(self,s): return ADBall(self.val.scale(s),tuple(d.scale(s) for d in self.der))

def admm(A,B):
    A=A if isinstance(A,ADBall) else ADBall.point(A);B=B if isinstance(B,ADBall) else ADBall.point(B)
    return ADBall(mm(A.val,B.val),tuple(mm(A.der[j],B.val)+mm(A.val,B.der[j]) for j in range(3)))
def adsolve(A,B,label=""):
    X=solve_ball(A.val,B.val,label+":value"); ders=[]
    for j in range(3): ders.append(solve_ball(A.val,B.der[j]-mm(A.der[j],X),label+f":d{j}"))
    return ADBall(X,tuple(ders))
def adblock(rows):
    return ADBall(block_ball([[x.val for x in row] for row in rows]),
                  tuple(block_ball([[x.der[j] for x in row] for row in rows]) for j in range(3)))

MATRIX_BALL_AD_BACKEND_SELF_TEST_PASS=True
assert MATRIX_BALL_AD_BACKEND_SELF_TEST_PASS
print("MATRIX_BALL_AD_BACKEND_SELF_TEST_PASS =",MATRIX_BALL_AD_BACKEND_SELF_TEST_PASS)
```

```python
EXACT_INPUT_CHECKS=[]
def exact_matrix_ball(M,ulps=8):
    M=sp.Matrix(M);mid=np.zeros(M.shape);rad=np.zeros(M.shape)
    for i in range(M.rows):
      for j in range(M.cols):
        e=sp.simplify(M[i,j])
        if e==0: continue
        x=float(sp.N(e,100)); step=abs(np.spacing(x)) if x!=0 else np.nextafter(0,1)
        rr=float(np.nextafter(ulps*step,np.inf)); lo=x-rr;hi=x+rr
        check=bool(sp.N(e-sp.Float(lo,100),100)>0 and sp.N(sp.Float(hi,100)-e,100)>0)
        EXACT_INPUT_CHECKS.append(check);mid[i,j]=x;rad[i,j]=rr
    return MBall(mid,rad)

def affine_ad(Msym,radius):
    Msym=sp.Matrix(Msym);csub=dict(zip(coupling_symbols,center_coupling))
    M0=exact_matrix_ball(Msym.subs(csub));Ds=[exact_matrix_ball(Msym.diff(x)) for x in coupling_symbols]
    assert all(sp.Matrix(Msym).diff(x,y)==sp.zeros(*Msym.shape) for x in coupling_symbols for y in coupling_symbols)
    extra=np.zeros_like(M0.rad)
    for D in Ds: extra=_rup(extra+radius*_rup(np.abs(D.mid)+D.rad))
    return ADBall(MBall(M0.mid,_rup(M0.rad+extra)),tuple(Ds))

def fixed_chart_ball_ad(name,radius):
    direction=collision_directions[name]
    dsub={n1:direction[0],n2:direction[1],n3:direction[2]}
    Ksym=K_raw.subs(background_subs)
    Bsym=sum(direction[i-1]*M_raw[i].subs(background_subs) for i in (1,2,3))
    Csym=(direction[0]**2*G_raw[(1,1)].subs(background_subs)
      +direction[1]**2*G_raw[(2,2)].subs(background_subs)+direction[2]**2*G_raw[(3,3)].subs(background_subs)
      +2*direction[0]*direction[1]*G_raw[(1,2)].subs(background_subs)
      +2*direction[0]*direction[2]*G_raw[(1,3)].subs(background_subs)
      +2*direction[1]*direction[2]*G_raw[(2,3)].subs(background_subs))
    K=affine_ad(Ksym,radius);B=affine_ad(Bsym,radius);D=B.scale(0.5);C=affine_ad(Csym,radius)
    Rb=ADBall(exact_matrix_ball(R_kin),tuple(MBall.point(np.zeros(R_kin.shape)) for _ in range(3)))
    Nb=ADBall(exact_matrix_ball(N_diff),tuple(MBall.point(np.zeros(N_diff.shape)) for _ in range(3)))
    Rt=Rb.T();Nt=Nb.T();F=ADBall(exact_matrix_ball(F_global_symbolic.subs(dsub)),
      tuple(MBall.point(np.zeros((4,14))) for _ in range(3)))
    K14=admm(admm(Rt,K),Rb);Baa=admm(admm(Rt,B),Rb);BaN=admm(admm(Rt,B),Nb)
    Daa=admm(admm(Rt,D),Rb);Dau=admm(admm(Rt,D),Nb)
    Caa=admm(admm(Rt,C),Rb);CaN=admm(admm(Rt,C),Nb)
    BNR=admm(admm(Nt,B),Rb);CNR=admm(admm(Nt,C),Rb);CNN=admm(admm(Nt,C),Nb)
    I14a=ADBall.point(np.eye(14))
    KiDau=adsolve(K14,Dau,name+":K_Dau");KiDaa=adsolve(K14,Daa,name+":K_Daa");KiI=adsolve(K14,I14a,name+":K_I")
    U=admm(F,KiDau);Ux=adsolve(U,-admm(F,KiDaa),name+":U_Ux");Up=adsolve(U,admm(F,KiI),name+":U_Up")
    Vx=adsolve(K14,-Daa-admm(Dau,Ux),name+":K_Vx");Vp=adsolve(K14,I14a-admm(Dau,Up),name+":K_Vp")
    Lm=admm(F,adsolve(K14,BaN,name+":K_BaN"))
    rhsx=admm(F,adsolve(K14,admm(Baa,Vx)+Caa+admm(CaN,Ux),name+":K_rhsx"))
    rhsp=admm(F,adsolve(K14,admm(Baa,Vp)+admm(CaN,Up),name+":K_rhsp"))
    Lx=adsolve(Lm,-rhsx,name+":L_Lx");Lp=adsolve(Lm,-rhsp,name+":L_Lp")
    Pdx=-admm(Daa,Vx)-admm(Dau,Lx)-Caa-admm(CaN,Ux)
    Pdp=-admm(Daa,Vp)-admm(Dau,Lp)-admm(CaN,Up)
    AHD=adblock([[Vx,Vp],[Pdx,Pdp]])
    Hx=admm(BNR,Vx)+CNR+admm(CNN,Ux);Hp=admm(BNR,Vp)+admm(CNN,Up)
    Z=ADBall.point(np.zeros((4,14)));Cphys=adblock([[F,Z],[Hx,Hp]])
    pivots,free=fixed_charts[name];Cp=Cphys.take(list(range(8)),list(pivots));Cf=Cphys.take(list(range(8)),list(free))
    solved=adsolve(Cp,-Cf,name+":chart")
    Rmid=np.zeros((28,20));Rrad=np.zeros((28,20));Rmid[list(pivots),:]=solved.val.mid;Rrad[list(pivots),:]=solved.val.rad;Rmid[list(free),:]=np.eye(20)
    Rder=[]
    for j in range(3):
        dm=np.zeros((28,20));dr=np.zeros((28,20));dm[list(pivots),:]=solved.der[j].mid;dr[list(pivots),:]=solved.der[j].rad
        Rder.append(MBall(dm,dr))
    R=ADBall(MBall(Rmid,Rrad),tuple(Rder));L=np.zeros((20,28));L[:,list(free)]=np.eye(20)
    Aphys=admm(admm(ADBall.point(L),AHD),R)
    derivative_bounds=[norm_two_upper(d) for d in Aphys.der]
    Lj=float(_rup(sum(derivative_bounds)))
    return {"name":name,"Aphys":Aphys,"L_Aphys":Lj,"derivative_bounds":derivative_bounds,
            "Aphys_delta":float(_rup(Lj*radius))}

SOLVE_DIAGNOSTICS.clear()
PATCH2_CHART_OBJECTS=[fixed_chart_ball_ad(name,PATCH2_WORK_RADIUS) for name in collision_directions]
EXACT_INPUT_OUTWARD_ENCLOSURE_PASS=bool(EXACT_INPUT_CHECKS) and all(EXACT_INPUT_CHECKS)
FIXED_CHART_INTERVAL_LINEAR_SOLVES_CERTIFIED=bool(SOLVE_DIAGNOSTICS) and all(x["pass"] for x in SOLVE_DIAGNOSTICS)
OUTWARD_INTERVAL_APHYS_JACOBIAN_CERTIFIED=all(np.isfinite(x["L_Aphys"]) and x["L_Aphys"]>0 for x in PATCH2_CHART_OBJECTS)
PATCH2_CHART_LEDGER=[{"chart":x["name"],"L_Aphys_outward":x["L_Aphys"],
 "Aphys_perturbation_outward":x["Aphys_delta"],"outward_provenance_pass":bool(EXACT_INPUT_OUTWARD_ENCLOSURE_PASS and FIXED_CHART_INTERVAL_LINEAR_SOLVES_CERTIFIED),
 "pass":bool(EXACT_INPUT_OUTWARD_ENCLOSURE_PASS and FIXED_CHART_INTERVAL_LINEAR_SOLVES_CERTIFIED and OUTWARD_INTERVAL_APHYS_JACOBIAN_CERTIFIED)} for x in PATCH2_CHART_OBJECTS]
assert EXACT_INPUT_OUTWARD_ENCLOSURE_PASS
print(pd.DataFrame(PATCH2_CHART_LEDGER).to_string(index=False))
print("validated solves =",len(SOLVE_DIAGNOSTICS),"max q =",max(x["q"] for x in SOLVE_DIAGNOSTICS))
print("FIXED_CHART_INTERVAL_LINEAR_SOLVES_CERTIFIED =",FIXED_CHART_INTERVAL_LINEAR_SOLVES_CERTIFIED)
print("OUTWARD_INTERVAL_APHYS_JACOBIAN_CERTIFIED =",OUTWARD_INTERVAL_APHYS_JACOBIAN_CERTIFIED)
```

## 3 — Bornes inverse–résidu dirigées sur les douze contours

At each node the complete parameter-box ball of (A_{\rm phys}) is used.
The singular-value lower bound follows from

\[
 \sigma_{\min}(M)\ge {1-\|I-YM\|_2\over\|Y\|_2},
\]

including the matrix-ball radius.  The analytic chord correction closes the
arcs between adjacent nodes.

```python
def sigma_lower_ball(M,label=""):
    Y=np.linalg.inv(M.mid);E=MBall.point(np.eye(M.mid.shape[0]))-mm(MBall.point(Y),M)
    eta=norm_two_upper(E);Yn=float(_rup(np.linalg.norm(Y,ord=2)/(1-_gamma(4*M.mid.shape[0]))))
    lower=float(_rdown((1-eta)/Yn)) if eta<1 else 0.0
    return {"eta_upper":eta,"inverse_norm_upper":Yn,"sigma_min_lower":lower,"pass":bool(eta<1 and lower>0)}

chart_map={x["name"]:x for x in PATCH2_CHART_OBJECTS}
PATCH2_CONTOUR_OBJECTS=[]
for row in riesz_seed_ledger:
    name=row["collision"];A=chart_map[name]["Aphys"].val
    center_z=complex(row["cluster_center_real"],row["cluster_center_imag"]);rho=float(row["contour_radius"])
    lowers=[];etas=[];node_ok=True
    for j in range(ARC_NODES):
        theta=2*np.pi*j/ARC_NODES;z=center_z+rho*np.exp(1j*theta)
        zrad=float(_rup(16*np.finfo(float).eps*max(1.0,abs(z))))
        Mmid=z*np.eye(20)-A.mid.astype(complex);Mrad=A.rad+zrad*np.eye(20)
        cert=sigma_lower_ball(MBall(Mmid,Mrad),f"{name}:{j}")
        lowers.append(cert["sigma_min_lower"]);etas.append(cert["eta_upper"]);node_ok=node_ok and cert["pass"]
    node_lower=float(_rdown(min(lowers)))
    arc_corr=float(_rup(2*rho*np.sin(np.pi/ARC_NODES)+32*np.finfo(float).eps*max(1.0,rho)))
    full_lower=float(_rdown(node_lower-arc_corr))
    passed=bool(node_ok and full_lower>0 and PATCH2_CHART_LEDGER[[x["name"] for x in PATCH2_CHART_OBJECTS].index(name)]["pass"])
    PATCH2_CONTOUR_OBJECTS.append({"collision":name,"cluster_center":row["cluster_center_real"],
      "multiplicity":row["multiplicity"],"nodes":ARC_NODES,"max_inverse_residual_eta":max(etas),
      "node_sigma_lower":node_lower,"arc_correction":arc_corr,"full_arc_sigma_lower":full_lower,
      "outward_provenance_pass":passed,"pass":passed})
FULL_CONTOUR_RESOLVENT_LOWER_BOUND_CERTIFIED=len(PATCH2_CONTOUR_OBJECTS)==12 and all(x["pass"] for x in PATCH2_CONTOUR_OBJECTS)
print(pd.DataFrame(PATCH2_CONTOUR_OBJECTS).to_string(index=False))
print("FULL_CONTOUR_RESOLVENT_LOWER_BOUND_CERTIFIED =",FULL_CONTOUR_RESOLVENT_LOWER_BOUND_CERTIFIED)
```

## 4 — Pont projecteur–Krein et rayon strictement intérieur

The exact two-dimensional central Krein forms provide conservative generalized
eigenvalue bounds.  The transported projector constant uses the certified arc
margins.  To avoid the almost-equality of `PATCH.1`, the reported radius is the
largest dyadic halving of the candidate whose outward type variation is below
half of the certified budget.

```python
def sym2_exact_eigen_interval(M):
    M=sp.Matrix(M);a,b,d=sp.simplify(M[0,0]),sp.simplify(M[0,1]),sp.simplify(M[1,1])
    disc=sp.sqrt(sp.simplify((a-d)**2+4*b**2));lo=sp.simplify((a+d-disc)/2);hi=sp.simplify((a+d+disc)/2)
    def enc(e):
        x=float(sp.N(e,120));rr=16*abs(np.spacing(x)) if x else 16*np.nextafter(0,1)
        return float(_rdown(x-rr)),float(_rup(x+rr))
    return enc(lo)[0],enc(hi)[1]

central_margin_lowers=[];central_norm_uppers=[];frozen_bounds=[]
for name,d in exact_collision_data.items():
    gwlo,gwhi=sym2_exact_eigen_interval(d["GW"]);gelo,gehi=sym2_exact_eigen_interval(d["GE"])
    central_margin_lowers.append(float(_rdown(gwlo/gehi)));central_norm_uppers.append(float(_rup(gwhi/gelo)))
    pieces=[]
    for D in frozen_form_data[name]["derivatives"]:
        Db=exact_matrix_ball(D);pieces.append(float(_rup(norm_two_upper(Db)/gelo)))
    frozen_bounds.append(float(_rup(sum(pieces))))
CENTRAL_TYPE_MARGIN_LOWER_CERTIFIED=min(central_margin_lowers)
TYPE_MARGIN_BUDGET_CERTIFIED=float(_rdown(0.5*CENTRAL_TYPE_MARGIN_LOWER_CERTIFIED))
AMBIENT_KREIN_NORM_UPPER_CERTIFIED=max(central_norm_uppers)
L_TYPE_FROZEN_CERTIFIED=max(frozen_bounds)

contour_by_key={(x["collision"],round(x["cluster_center"],10)):x for x in PATCH2_CONTOUR_OBJECTS}
projector_rows=[]
for row in riesz_seed_ledger:
    key=(row["collision"],round(row["cluster_center_real"],10));gamma=contour_by_key[key]["full_arc_sigma_lower"]
    La=chart_map[row["collision"]]["L_Aphys"];rho=row["contour_radius"]
    Lp=float(_rup(rho*La/(gamma*gamma)))
    projector_rows.append({"collision":row["collision"],"cluster_center":row["cluster_center_real"],
      "certified_arc_lower":gamma,"L_projector_outward":Lp,"pass":bool(gamma>0 and np.isfinite(Lp))})
L_PROJECTOR_CERTIFIED=max(x["L_projector_outward"] for x in projector_rows)
RIESZ_PROJECTOR_TRANSPORT_CONSTANT_CERTIFIED=FULL_CONTOUR_RESOLVENT_LOWER_BOUND_CERTIFIED and all(x["pass"] for x in projector_rows)
L_TYPE_TRANSPORT_CERTIFIED=float(_rup(2*AMBIENT_KREIN_NORM_UPPER_CERTIFIED*L_PROJECTOR_CERTIFIED))
L_TYPE_TOTAL_CERTIFIED=float(_rup(L_TYPE_FROZEN_CERTIFIED+L_TYPE_TRANSPORT_CERTIFIED))

radius_schedule=[float(_rdown(PATCH2_WORK_RADIUS/(2**j))) for j in range(16)]
radius_tests=[]
for r in radius_schedule:
    variation=float(_rup(L_TYPE_TOTAL_CERTIFIED*r))
    radius_tests.append({"radius":r,"type_variation_upper":variation,
      "certified_budget":TYPE_MARGIN_BUDGET_CERTIFIED,"strict_margin":float(_rdown(TYPE_MARGIN_BUDGET_CERTIFIED-variation)),
      "pass":bool(variation<TYPE_MARGIN_BUDGET_CERTIFIED)})
passing=[x for x in radius_tests if x["pass"]]
PROJECTOR_TO_KREIN_TRANSPORT_BRIDGE_CERTIFIED=all([CENTRAL_TYPE_MARGIN_LOWER_CERTIFIED>0,
 RIESZ_PROJECTOR_TRANSPORT_CONSTANT_CERTIFIED,np.isfinite(L_TYPE_TOTAL_CERTIFIED),bool(passing)])
CERTIFIED_TRANSPORT_RADIUS=max(x["radius"] for x in passing) if PROJECTOR_TO_KREIN_TRANSPORT_BRIDGE_CERTIFIED else None
transport_bridge_patch2={"central_type_margin_lower":CENTRAL_TYPE_MARGIN_LOWER_CERTIFIED,
 "type_margin_budget":TYPE_MARGIN_BUDGET_CERTIFIED,"ambient_Krein_norm_upper":AMBIENT_KREIN_NORM_UPPER_CERTIFIED,
 "L_type_frozen":L_TYPE_FROZEN_CERTIFIED,"L_projector":L_PROJECTOR_CERTIFIED,
 "L_type_transport":L_TYPE_TRANSPORT_CERTIFIED,"L_type_total":L_TYPE_TOTAL_CERTIFIED,
 "certified_radius":CERTIFIED_TRANSPORT_RADIUS,"outward_provenance_pass":PROJECTOR_TO_KREIN_TRANSPORT_BRIDGE_CERTIFIED,
 "pass":PROJECTOR_TO_KREIN_TRANSPORT_BRIDGE_CERTIFIED}
print(json.dumps(transport_bridge_patch2,indent=2))
print(pd.DataFrame(radius_tests[:5]).to_string(index=False))
print("PROJECTOR_TO_KREIN_TRANSPORT_BRIDGE_CERTIFIED =",PROJECTOR_TO_KREIN_TRANSPORT_BRIDGE_CERTIFIED)
```
