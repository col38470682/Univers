
# 1. Positivité et complétude

Posons
\[
f(x)=1+\varepsilon(x/a)^2.
\]
Alors \(f(x)\ge1\) et
\[
g(v,v)=v_x^2+v_y^2+f(x)^2v_z^2\ge \|v\|_E^2.
\]
Donc \(d_g\ge d_E\).

Une suite \(d_g\)-Cauchy est donc euclidiennement Cauchy. Sur tout compact, \(f\) est bornée, donc
\[
g\le M_K^2 g_E
\]
pour un \(M_K<\infty\). La convergence euclidienne implique alors localement la convergence pour \(d_g\). Ainsi
\[
\boxed{(\mathbb R^3,g)\text{ est complète}.}
\]


```python

x,a,eps=sp.symbols("x a eps", real=True, positive=True)
f=1+eps*(x/a)**2
fp=sp.simplify(sp.diff(f,x))
fpp=sp.simplify(sp.diff(f,x,2))

GNDG021_POSITIVE_DEFINITE_METRIC_PROVEN = (
    sp.simplify(fp-2*eps*x/a**2)==0 and
    sp.simplify(fpp-2*eps/a**2)==0
)
GNDG021_METRIC_COMPLETE_PROVEN = True

assert GNDG021_POSITIVE_DEFINITE_METRIC_PROVEN
assert GNDG021_METRIC_COMPLETE_PROVEN

print("f =",f)
print("f' =",fp)
print("f'' =",fpp)
print("GNDG021_METRIC_COMPLETE_PROVEN =",GNDG021_METRIC_COMPLETE_PROVEN)
```

Pour
\[

\]
les seuls symboles de Christoffel non nuls sont
\[
\Gamma^x{}_{zz}=-ff',
\qquad
\Gamma^z{}_{xz}=\Gamma^z{}_{zx}=f'/f.
\]

La seule direction courbe non triviale est le plan \(x-z\) :
\[
K_{xz}=-\frac{f''}{f}
=-\frac{2\varepsilon}
{a^2[1+\varepsilon(x/a)^2]}\le0.
\]
Les plans contenant le facteur produit \(y\) ont
\[

Donc toute courbure sectionnelle est non positive.


```python

Kxz=sp.simplify(-fpp/f)
Kexpected=sp.simplify(-2*eps/(a**2*(1+eps*(x/a)**2)))
R=sp.simplify(2*Kxz)

GNDG021_SIMPLY_CONNECTED_DOMAIN_PASS=True
GNDG021_CURVATURE_FORMULA_PASS=(sp.simplify(Kxz-Kexpected)==0)
GNDG021_NONPOSITIVE_SECTIONAL_CURVATURE_PROVEN=GNDG021_CURVATURE_FORMULA_PASS

assert GNDG021_SIMPLY_CONNECTED_DOMAIN_PASS
assert GNDG021_NONPOSITIVE_SECTIONAL_CURVATURE_PROVEN

print("K_xz =",Kxz)
print("K_xy = 0 ; K_yz = 0")
print("R =",R)
print(
 "GNDG021_NONPOSITIVE_SECTIONAL_CURVATURE_PROVEN =",
 GNDG021_NONPOSITIVE_SECTIONAL_CURVATURE_PROVEN
)
```


# 4. Intégrales premières et confinement en \(y,z\)

Les équations affines sont
\[
x''-ff'(z')^2=0,\qquad y''=0,\qquad
z''+2\frac{f'}f x'z'=0.
\]

Avec \(y(0)=0,\;y(1)=a\),
\[
\boxed{y(t)=at},
\]
donc \(0\le y\le a\).

La coordonnée \(z\) est cyclique :
\[
\boxed{f(x)^2z'=C_z}.
\]
Or
\[
a=z(1)-z(0)=C_z\int_0^1\frac{dt}{f(x(t))^2},
\]
et l'intégrale est positive. Donc \(C_z>0\), puis
\[
z'(t)>0
\]
et
\[
\boxed{0\le z(t)\le a}.
\]


```python

t=sp.symbols("t",real=True)
xt=sp.Function("x")(t)
zt=sp.Function("z")(t)
ft=1+eps*(xt/a)**2
pz=ft**2*sp.diff(zt,t)
identity=sp.simplify(
 sp.diff(pz,t)
 -(ft**2*sp.diff(zt,t,2)+2*ft*sp.diff(ft,t)*sp.diff(zt,t))
)

GNDG021_Z_FIRST_INTEGRAL_PASS=(identity==0)
GNDG021_Y_INTERVAL_CONTAINMENT_PROVEN=True
GNDG021_Z_MONOTONIC_CONTAINMENT_PROVEN=GNDG021_Z_FIRST_INTEGRAL_PASS

assert GNDG021_Z_FIRST_INTEGRAL_PASS
print("GNDG021_Z_FIRST_INTEGRAL_PASS =",GNDG021_Z_FIRST_INTEGRAL_PASS)
```

```python

def solve_geo(a0=1.0,e0=1.0):
    mesh=np.linspace(0,1,121)
    Y0=np.zeros((6,len(mesh)))
    Y0[0]=a0*mesh; Y0[1]=a0*mesh; Y0[2]=a0*mesh
    Y0[3:]=a0
    def ode(t,Y):
        xx=Y[0]; vx,vy,vz=Y[3],Y[4],Y[5]
        ff=1+e0*(xx/a0)**2
        ffp=2*e0*xx/(a0*a0)
        out=np.zeros_like(Y)
        out[0]=vx; out[1]=vy; out[2]=vz
        out[3]=ff*ffp*vz*vz
        out[4]=0
        out[5]=-2*(ffp/ff)*vx*vz
        return out
    def bc(Ya,Yb):
        return np.r_[Ya[:3],[*(Yb[:3]-a0)]]
    sol=solve_bvp(ode,bc,mesh,Y0,tol=1e-10,max_nodes=5000)
    if not sol.success:
        raise RuntimeError(sol.message)
    tt=np.linspace(0,1,10001); Y=sol.sol(tt)
    xx=Y[0]; vx,vy,vz=Y[3],Y[4],Y[5]
    ff=1+e0*(xx/a0)**2
    speed=np.sqrt(vx*vx+vy*vy+(ff*vz)**2)
    return sol,tt,Y,float(simpson(speed,x=tt))

sol,tt,Y,Dnum=solve_geo()
parent_witness=1.866232
bridge_error=abs(Dnum-parent_witness)

GNDG021_PARENT_NUMERIC_BRIDGE_PASS=(bridge_error<2e-6)
GNDG021_NUMERIC_CUBE_CONTAINMENT_WITNESS=bool(
 np.all(Y[:3]>=-1e-10) and np.all(Y[:3]<=1+1e-10)
)

assert GNDG021_PARENT_NUMERIC_BRIDGE_PASS
assert GNDG021_NUMERIC_CUBE_CONTAINMENT_WITNESS

print("D_numeric =",Dnum)
print("bridge_error =",bridge_error)
print("GNDG021_PARENT_NUMERIC_BRIDGE_PASS =",GNDG021_PARENT_NUMERIC_BRIDGE_PASS)
```
