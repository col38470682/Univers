# 1. Limite isotrope

Sur le cube \(0\le x\le a\),
\[
1\le 1+\varepsilon(x/a)^2\le 1+\varepsilon.
\]
Donc
\[
g_E\le g_\varepsilon\le (1+\varepsilon)^2 g_E.
\]
Pour toute courbe,
\[
L_E\le L_{g_\varepsilon}\le (1+\varepsilon)L_E.
\]
Pour les sommets opposés,
\[
\sqrt3\,a\le D^{(g_\varepsilon)}\le (1+\varepsilon)\sqrt3\,a,
\]
donc par encadrement
\[
\boxed{D^{(g_\varepsilon)}\to \sqrt3\,a}
\quad(\varepsilon\to0^+).
\]
À \(\varepsilon=0\), \(g_0=I_3\) exactement, et \(\eta^{(g_\varepsilon)}\) retrouve le diagnostic euclidien.

```python
x,a,eps = sp.symbols("x a eps", positive=True, real=True)
f = 1 + eps*(x/a)**2
gzz = sp.expand(f**2)
GNDG0211_ISOTROPIC_LIMIT_CLOSED = all([
    sp.simplify(gzz.subs(eps,0)-1) == 0,
    sp.simplify(gzz.subs(x,a)-(1+eps)**2) == 0,
])
assert GNDG0211_ISOTROPIC_LIMIT_CLOSED
print("g_zz(eps=0) =", gzz.subs(eps,0))
print("GNDG0211_ISOTROPIC_LIMIT_CLOSED =", GNDG0211_ISOTROPIC_LIMIT_CLOSED)
```

# 2. Secteur anisotrope constant

Pour
\[
g_c=\mathrm{diag}(\alpha^2,\beta^2,\gamma^2),\qquad \alpha,\beta,\gamma>0,
\]
le changement
\[
X=\alpha x,\quad Y=\beta y,\quad Z=\gamma z

transforme exactement \(ds^2\) en métrique euclidienne. Le secteur est donc isométrique à \(\mathbb R^3\) euclidien.

Entre \((0,0,0)\) et \((a,a,a)\),
\[
\boxed{D_c^{(g)}=a\sqrt{\alpha^2+\beta^2+\gamma^2}}.

La ligne droite est l'unique géodésique minimisante.

```python
alpha,beta,gamma = sp.symbols("alpha beta gamma", positive=True, real=True)
Gconst = sp.diag(alpha**2,beta**2,gamma**2)
v = sp.Matrix([a,a,a])
D2 = sp.simplify((v.T*Gconst*v)[0])
expected = sp.simplify(a**2*(alpha**2+beta**2+gamma**2))
GNDG0211_CONSTANT_ANISOTROPIC_SECTOR_CLOSED = (sp.simplify(D2-expected)==0)
assert GNDG0211_CONSTANT_ANISOTROPIC_SECTOR_CLOSED
print("D_g^2 =", D2)
print("GNDG0211_CONSTANT_ANISOTROPIC_SECTOR_CLOSED =", GNDG0211_CONSTANT_ANISOTROPIC_SECTOR_CLOSED)
```

# 3. Invariance par reparamétrisation

Pour un changement admissible \(t=t(s)\) avec \(dt/ds>0\),
\[
\frac{dx^i}{ds}=\frac{dx^i}{dt}\frac{dt}{ds}.
\]
Ainsi
\[
\sqrt{g_{ij}\frac{dx^i}{ds}\frac{dx^j}{ds}}
=
\frac{dt}{ds}\sqrt{g_{ij}\frac{dx^i}{dt}\frac{dx^j}{dt}},
\]
et donc
\[
\boxed{L_g[\gamma\circ t(s)]=L_g[\gamma]}.
\]
Les extrémités physiques étant les mêmes, \(D^{(g)}\) et \(\eta^{(g)}\) sont également invariants.

```python
q,Q = sp.symbols("q Q", positive=True, real=True)
speed_s = sp.sqrt(Q*q**2)
GNDG0211_REPARAMETERIZATION_CLOSED = (sp.simplify(speed_s-q*sp.sqrt(Q))==0)
assert GNDG0211_REPARAMETERIZATION_CLOSED
print("GNDG0211_REPARAMETERIZATION_CLOSED =", GNDG0211_REPARAMETERIZATION_CLOSED)
```

# 4. Covariance de coordonnées

Pour un changement admissible \(x=x(x')\) de jacobien inversible \(J\),
\[
G'=J^T GJ,\qquad v=Jv'.
\]
Alors
\[
\boxed{v'^T G'v'=v^TGv}.
\]
L'élément de longueur et la longueur de toute courbe sont donc invariants. Comme le changement de coordonnées établit une bijection entre les courbes physiques, l'infimum des longueurs est invariant :
\[
\boxed{d_{g'}(A',B')=d_g(A,B)}.
\]
Par conséquent,
\[
\boxed{\eta^{(g')}=\eta^{(g)}}.
\]
Cette fermeture est générale pour un difféomorphisme admissible, et non limitée au témoin linéaire numérique de `0.2`.

```python
j = sp.symbols("j11:14 j21:24 j31:34", real=True)
j11,j12,j13,j21,j22,j23,j31,j32,j33 = j
g11,g22,g33,g12,g13,g23 = sp.symbols("g11 g22 g33 g12 g13 g23", real=True)
u1,u2,u3 = sp.symbols("u1 u2 u3", real=True)
J = sp.Matrix([[j11,j12,j13],[j21,j22,j23],[j31,j32,j33]])
G = sp.Matrix([[g11,g12,g13],[g12,g22,g23],[g13,g23,g33]])
u = sp.Matrix([u1,u2,u3])
Gp = J.T*G*J
residual = sp.simplify((u.T*Gp*u)[0]-((J*u).T*G*(J*u))[0])
GNDG0211_COORDINATE_COVARIANCE_CLOSED = (residual==0)
assert GNDG0211_COORDINATE_COVARIANCE_CLOSED
print("quadratic-form residual =", residual)
print("GNDG0211_COORDINATE_COVARIANCE_CLOSED =", GNDG0211_COORDINATE_COVARIANCE_CLOSED)
```

Donc
\[

\]
et finalement
\[
\boxed{\eta_n^{(g)}=\widetilde L/\widetilde D}

indépendant exactement de \(a_n\) pour la famille autosimilaire.

```python
A,X = sp.symbols("A X", positive=True, real=True)
fscaled = sp.simplify(1 + eps*((A*X)/A)**2)
GNDG0211_NESTED_SCALING_CLOSED = (
    sp.simplify(fscaled-(1+eps*X**2))==0
    and sp.simplify(A**2*fscaled**2-A**2*(1+eps*X**2)**2)==0
)
assert GNDG0211_NESTED_SCALING_CLOSED
print("scaled f =", fscaled)
print("GNDG0211_NESTED_SCALING_CLOSED =", GNDG0211_NESTED_SCALING_CLOSED)
```
