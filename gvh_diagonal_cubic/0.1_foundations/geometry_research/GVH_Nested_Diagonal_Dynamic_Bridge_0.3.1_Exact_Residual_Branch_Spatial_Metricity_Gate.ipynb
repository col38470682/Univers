
# 4. Test exact des branches invariantes sous réflexions

Une co-métrique spatiale générale donne :

\[
c^2(\mathbf n)=n_iC^{ij}n_j.
\]

Si une branche individuelle est invariante sous les réflexions indépendantes :

\[
n_x\to-n_x,\qquad
n_y\to-n_y,\qquad
n_z\to-n_z,
\]

ses termes croisés doivent disparaître.

Elle prend alors la forme :

\[
c^2
=
\alpha n_x^2+
\beta n_y^2+
\gamma n_z^2.
\]

Sur le simplexe :

\[
w=1-u-v,
\]

cela devient :

\[
\boxed{
y=a\,u+b\,v+c,
}
\]

c'est-à-dire une fonction affine en \(u,v\).

On teste donc exactement l'existence d'une racine affine de \(F_2\) et de \(F_3\).

Le critère est algébrique :

\[
F_r(a u+b v+c,u,v)\equiv0.
\]

Si le système des coefficients a pour base de Gröbner :

\[
\boxed{\{1\}},
\]

aucune telle branche n'existe, même sur la clôture algébrique.


```python

a,b,c = sp.symbols("a b c")
L = a*u+b*v+c

def affine_root_groebner_no_solution(F):
    P = sp.Poly(
        sp.expand(F.subs(y,L)),
        u,v
    )
    equations = [
        coeff
        for monom,coeff in P.terms()
    ]
    G = sp.groebner(
        equations,
        a,b,c,
        order="lex"
    )
    return G, (
        len(G.polys)==1
        and G.polys[0].as_expr()==1
    )

G2, no_affine_F2 = affine_root_groebner_no_solution(F2)
G3, no_affine_F3 = affine_root_groebner_no_solution(F3)

GNDG031_F2_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH = no_affine_F2
GNDG031_F3_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH = no_affine_F3

assert GNDG031_F2_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH
assert GNDG031_F3_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH

print("Groebner F2 affine-root system =", G2)
print("Groebner F3 affine-root system =", G3)
print(
    "GNDG031_F2_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH =",
    GNDG031_F2_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH
)
print(
    "GNDG031_F3_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH =",
    GNDG031_F3_NO_REFLECTION_INVARIANT_QUADRATIC_BRANCH
)
```
