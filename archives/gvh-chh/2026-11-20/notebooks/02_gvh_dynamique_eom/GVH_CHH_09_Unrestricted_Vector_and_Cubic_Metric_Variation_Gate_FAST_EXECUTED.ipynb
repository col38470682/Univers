## 1.1 Variation *non restreinte* par variables de premier jet

Write $\mathcal L=\mathcal L_u+\mathcal L_C+\mathcal L_{\rm con}$ and regard the *first jet* $(g_{\mu\nu},u^\mu,e_A^\mu,F_{\alpha\beta})$ as algebraically independent before restoring $F=\nabla u_\flat$. Define **explicit algebraic tensors** by ordinary partial differentiation of the given polynomial/rational scalar:

$$P^{\mu\nu}=\left.\frac{\partial\mathcal L}{\partial g_{\mu\nu}}\right|_{u,e,F},\quad W_\mu=\left.\frac{\partial\mathcal L}{\partial u^\mu}\right|_{g,e,F},\quad Z_{A\mu}=\left.\frac{\partial\mathcal L}{\partial e_A^\mu}\right|_{g,u,F},\quad Q^{\alpha\beta}=\left.\frac{\partial\mathcal L}{\partial F_{\alpha\beta}}\right|_{g,u,e}.$$

**These are not the unresolved functional derivatives from `.3.3.28`:** each is a pointwise partial derivative of an explicitly specified expression, evaluable by automatic differentiation in any local frame.

Because the *independent vector field is $u^\mu$*,

$$\delta F_{\alpha\beta}=g_{\beta\rho}\nabla_\alpha\delta u^\rho+\delta g_{\beta\rho}\nabla_\alpha u^\rho+g_{\beta\rho}u^\sigma\delta\Gamma^\rho{}_{\alpha\sigma},$$
$$\delta\Gamma^\rho{}_{\alpha\sigma}=\frac12g^{\rho\lambda}\left(\nabla_\alpha\delta g_{\lambda\sigma}+\nabla_\sigma\delta g_{\lambda\alpha}-\nabla_\lambda\delta g_{\alpha\sigma}\right).$$

**After integration by parts with compact-support variations**, define

$$H^{\kappa\mu\nu}=\tfrac12\left[Q^{\kappa(\mu}u^{\nu)}+u^\kappa Q^{(\mu\nu)}-Q^{(\mu|\kappa|}u^{\nu)}\right],$$

$$\boxed{\mathcal E^{(u)}_\mu=W_\mu-\nabla_\alpha(Q^{\alpha\beta}g_{\beta\mu})=0},\qquad
\boxed{\mathcal E^{(e_A)}_\mu=Z_{A\mu}=0},$$

$$\boxed{\mathcal E^{(g)\mu\nu}=-G^{\mu\nu}+\tfrac12g^{\mu\nu}\mathcal L+P^{(\mu\nu)}+Q^{\alpha(\mu}\nabla_\alpha u^{\nu)}-\nabla_\kappa H^{\kappa\mu\nu}=0}.$$

The term $-G^{\mu\nu}$ is the Einstein–Hilbert variation with respect to $g_{\mu\nu}$; the displayed Euler tensor uses the overall $16\pi G_0$ normalization. The $\tfrac12 g^{\mu\nu}\mathcal L$ term is essential (metric variation of $\sqrt{-g}$). The connection variation is captured by $-\nabla H$, **not discarded**. These equations are an unrestricted **constitutive Euler–Lagrange representation**. They are not a solved 4D coupled system or a fully expanded component-by-component stress tensor.

## 3.1 Contrôle crucial : trièdre auxiliaire et branche de cisaillement invisible

For any real symmetric **trace-free** $3\times3$ spatial shear $S$, an orthonormal triad exists locally such that **all three diagonal components vanish**:

$$e_A^{\mathsf T}S e_A=0\ (A=1,2,3),\qquad \mathcal O_C=0.$$

Proof outline (constructive): if $S\neq0$, it has eigenvalues $\lambda_-<0<\lambda_+$. Choose $e_1=\sqrt{\lambda_+/(\lambda_+-\lambda_-)}v_-+\sqrt{-\lambda_-/(\lambda_+-\lambda_-)}v_+$. Then $e_1^T S e_1=0$; the restriction to the orthogonal 2D plane also has zero trace. Choose a $45^\circ$ rotation of its eigenbasis to obtain $e_2^TSe_2=e_3^TSe_3=0$. This is a particular case of a classical zero-diagonal/Schur–Horn result.

Conversely, an eigenbasis gives $\mathcal O_C=\mathrm{tr}(S^2)$, and for any orthonormal triad $0\le\mathcal O_C\le\mathrm{tr}(S^2)$. The maximum follows from $\sum_A S_{AA}^2\le\sum_{AB}S_{AB}^2$.

**Critical new audit of the candidate:** because $\mathcal L_C=\zeta_C\sum_A s_A^2/2$, when all $s_A=0$ its *entire first variation* (metric, vector and triad) vanishes, even though the shear $\sigma$ need not vanish. Thus if the null triad can be selected smoothly subject to boundary conditions, the candidate possesses a branch indistinguishable in its local EOM from the $\zeta_C=0$ theory. This is a **possible degeneracy, not proof that the entire theory is redundant**. The triad has no kinetic term to select one physical orientation between neighboring points.

```python
def zero_diag_frame(S):
    """Return orthogonal R (columns are axes) with diag(R.T @ S @ R)=0, for STF 3x3 S."""
    w,V=eigh(S)
    if np.max(abs(w))<1e-13: return np.eye(3)
    assert w[0]<0<w[2]
    a=np.sqrt(w[2]/(w[2]-w[0]));b=np.sqrt(-w[0]/(w[2]-w[0]))
    e1=a*V[:,0]+b*V[:,2]
    eprime=-b*V[:,0]+a*V[:,2]
    Plane=np.column_stack([eprime,V[:,1]])
    A=Plane.T@S@Plane
    wa,Va=eigh(A)
    # traceless 2x2 implies eigenvalues ±w; rotate by 45 degrees
    e2=Plane@(Va[:,0]+Va[:,1])/np.sqrt(2)
    e3=Plane@(Va[:,0]-Va[:,1])/np.sqrt(2)
    R=np.column_stack([e1,e2,e3])
    return R

null_errors=[]; e_basis_maxerr=[]; bound_errors=[]
for _ in range(75):
    A=rng.normal(size=(3,3));S=sym(A);S-=np.trace(S)/3*np.eye(3)
    R=zero_diag_frame(S)
    null_errors.append(float(max(abs(np.diag(R.T@S@R)))))
    e_basis_maxerr.append(float(max(abs((R.T@R-np.eye(3)).ravel()))))
    _,eigenvectors=eigh(S)
    Omax=sum(np.diag(eigenvectors.T@S@eigenvectors)**2)
    Oref=np.trace(S@S)
    bound_errors.append(abs(Omax-Oref))
METRICS['constructive_zero_diagonal_max_abs']=max(null_errors)
check('local_null_triad_constructed_for_random_STF',max(null_errors)<3e-14,max(null_errors))
check('null_triad_is_orthonormal',max(e_basis_maxerr)<3e-14)
check('eigenframe_shear_quadratic_maximum',max(bound_errors)<3e-13)

# Explicit physical 4D field with sigma_ij ≠ 0, yet all s_A=0 and cubic stress/flux=0.
STF=np.array([[.8,.12,-.07],[.12,-.3,.09],[-.07,.09,-.5]])
R=zero_diag_frame(STF)
E_null=torch.tensor(np.column_stack([np.zeros(3),R.T]))
E_eig=torch.tensor(np.column_stack([np.zeros(3),eigh(STF)[1].T]))
U=torch.tensor([1.,0.,0.,0.]);G=ETA.clone(); F=torch.zeros((4,4));F[1:,1:]=torch.tensor(STF)
L_null,grad_null=grads(G,U,E_null,F,'cubic')
L_eig,grad_eig=grads(G,U,E_eig,F,'cubic')
METRICS['shear_norm_sq']=float(np.trace(STF@STF))
METRICS['cubic_null_branch_max_jet_gradient']=max(float(torch.max(abs(t))) for t in grad_null)
check('nonzero_shear_zero_cubic_branch',float(np.trace(STF@STF))>.1 and abs(L_null)<1e-27)
check('nonzero_shear_nonzero_eigenbranch',abs(L_eig-.5*COEFFS['zeta']*np.trace(STF@STF))<1e-13)
check('zero_cubic_branch_entire_first_variation_vanishes',METRICS['cubic_null_branch_max_jet_gradient']<3e-14)
print('Tr(S²)=',METRICS['shear_norm_sq'],'L_C eigenframe=',L_eig,'L_C zero-diag frame=',L_null)
```

## 3.2 Rotation du trièdre **≠** changement de coordonnées

For $S=\mathrm{diag}(1,-1,0)$, an aligned frame has $\mathcal O_C=2$; rotating only $e_{1,2}$ by $45^\circ$ gives $\mathcal O_C=0$, while $\mathrm{tr}S^2=2$ stays fixed. This is **not** a coordinate invariance violation: a passive coordinate transformation rotates all tensor components simultaneously, which was separately verified above. The result shows genuine dependence on the internal triad if those axes are physical.

For a frame rotation in a plane $AB$:
$$\frac{d\mathcal O_C}{d\vartheta_{AB}}\bigg|_0=4(s_A-s_B)M_{AB},$$
so the Euler equation obtained by varying the *orientation* is $\zeta_C(s_A-s_B)M_{AB}=0$. **Both the eigenframe and the zero-diagonal frame are stationary orientation branches.** Selection requires a physical principle, boundary conditions, or a specified triad action. We have not derived such a selection from the current candidate.

```python
S2=sp.diag(1,-1,0)
ang=sp.symbols('theta',real=True)
rot=sp.Matrix([[sp.cos(ang),-sp.sin(ang),0],[sp.sin(ang),sp.cos(ang),0],[0,0,1]])
Mr=sp.simplify(rot.T*S2*rot)
Oc=sum([Mr[i,i]**2 for i in range(3)])
check('orientation_probe_aligned_value_2',sp.simplify(Oc.subs(ang,0)-2)==0)
check('orientation_probe_rotated_value_zero',sp.simplify(Oc.subs(ang,sp.pi/4))==0)
check('orientation_probe_stationary_both_branches',sp.simplify(sp.diff(Oc,ang).subs(ang,0))==0 and sp.simplify(sp.diff(Oc,ang).subs(ang,sp.pi/4))==0)
check('orientation_probe_noninvariant_internal_rotation',sp.simplify(Oc.subs(ang,0)-Oc.subs(ang,sp.pi/4))!=0)

# Check variation of internal SO(3) rotation for a non-diagonal shear with finite differences.
Sgen=np.array([[.8,.12,-.07],[.12,-.3,.09],[-.07,.09,-.5]])
R12=lambda th:np.array([[np.cos(th),-np.sin(th),0],[np.sin(th),np.cos(th),0],[0,0,1.]])
Cfun=lambda th:sum(np.diag(R12(th).T@Sgen@R12(th))**2)
eps=1e-6; fd=(Cfun(eps)-Cfun(-eps))/(2*eps)
expected=4*Sgen[0,1]*(Sgen[0,0]-Sgen[1,1]) # columns rotate R; derivative sign from R^T S R
METRICS['triad_rotation_derivative_residual']=abs(fd-expected)
check('orientation_variation_sign_and_factor',abs(fd-expected)<2e-9,{'fd':fd,'expected':expected})
print('Orientation derivative numeric',fd,'symbolic',expected)
```

## 3.3 Consistency with the preceding CHH-08 reduced equation

This notebook does **not** equate reduced stationarity with unrestricted field equations. We only preserve the exact identity of the earlier reduction:

$$K(\beta)=-c_1-(c_2+c_3)\cosh^2\beta+c_4\sinh^2\beta+\frac{\zeta_C}{3}\cosh^2\beta,$$
$$2K\beta''+K'(\beta)(\beta')^2=0.$$

The null-triad branch described above uses a *different triad choice* than the aligned CHH-08 branch, so it does not contradict the earlier conditional reduction; it demonstrates that the triad has additional stationary branches which must be evaluated before claiming an observable effect.

```python
bet,vel,acc=sp.symbols('beta v a',real=True)
c1,c2,c3,c4,z=sp.symbols('c_1 c_2 c_3 c_4 zeta_C',real=True)
K=sp.expand(-c1-(c2+c3)*sp.cosh(bet)**2+c4*sp.sinh(bet)**2+z*sp.cosh(bet)**2/3)
Delta=c4-c2-c3+z/3
check('CHH08_Kprime_identity',sp.simplify(sp.diff(K,bet)-2*Delta*sp.sinh(bet)*sp.cosh(bet))==0)
EL=sp.expand(sp.diff(K,bet)*vel**2+2*K*acc)
check('CHH08_reduced_EL_first_integral',sp.simplify(vel*EL-(2*K*vel*acc+sp.diff(K,bet)*vel**3))==0)
check('CHH08_prescribed_delta_nonzero',abs(float(Delta.subs({c1:.7,c2:.10,c3:.04,c4:.02,z:.75}))-.13)<1e-12)
```
