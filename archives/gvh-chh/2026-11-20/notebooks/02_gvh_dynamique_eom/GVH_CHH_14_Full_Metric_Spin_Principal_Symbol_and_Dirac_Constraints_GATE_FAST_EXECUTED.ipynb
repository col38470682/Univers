## Niveau 2 — Homogeneous metric+æther+spin reduction with lapse

Choose an **ADM minisuperspace** (Bianchi I), spatially flat and periodic/fiducial cell of unit coordinate volume:

\[
ds^2=-N(t)^2dt^2+\sum_{i=1}^3 e^{2\beta_i(t)}(dx^i)^2,
\qquad u=N^{-1}\partial_t.
\]

Rotate the orthonormal triad in the $1$–$2$ plane through $\alpha(t)$. The congruence has $H_i=\dot\beta_i/N$, $\theta=\sum_i H_i$, $\sigma_i=H_i-\theta/3$; **acceleration and vorticity vanish**. Along $u$ the triad rotates with angular speed $\dot\alpha/N$; its spatial angular connection vanishes on this ansatz, so $\kappa_s$ is not probed here.

Using the ADM Einstein–Hilbert identity up to standard boundary terms, the exact **restricted** action becomes:

\[
L_{\rm hom}=\frac{e^{\beta_1+\beta_2+\beta_3}}{N}\,U(\dot\beta_1,\dot\beta_2,\dot\beta_3,\dot\alpha;\alpha),
\]
\[
U=(1-c_{13})\sum_i\dot\beta_i^2-(1+c_2)(\sum_i\dot\beta_i)^2+\frac{\zeta_C}{2}\sum_{A=1}^3 d_A^2+\frac{\kappa_t}{2}\dot\alpha^2,
\]
\[
\begin{array}{ll}
\delta_i=\dot\beta_i-\frac13\sum_j\dot\beta_j,&
 d_1=\cos^2\alpha\,\delta_1+\sin^2\alpha\,\delta_2,\\
 d_2=\sin^2\alpha\,\delta_1+\cos^2\alpha\,\delta_2,&d_3=\delta_3.
\end{array}
\]

This is a **metric-varying** sector, beyond CHH-13, but still fixes the shift, æther alignment, diagonal spatial metric and one triad rotation. It is not the unrestricted $g,u,e_A$ theory. Derivatives below are tagged `EOM-GVH`, not intrinsic CHH derivatives.

```python
# Construct exact kinetic matrix for 3 gravitational scalefactors and one triad angle.
b1,b2,b3,ad,alpha=sp.symbols('b1 b2 b3 adot alpha',real=True)
c13,c2,z,kt,ks=sp.symbols('c13 c2 z k_t k_s',real=True)
vel=sp.Matrix([b1,b2,b3,ad]); bt=[b1,b2,b3]
th=sum(bt);deltas=[b-th/3 for b in bt]
d1=sp.cos(alpha)**2*deltas[0]+sp.sin(alpha)**2*deltas[1]
d2=sp.sin(alpha)**2*deltas[0]+sp.cos(alpha)**2*deltas[1]
d3=deltas[2];OC=sp.expand_trig(sp.expand(d1**2+d2**2+d3**2))
U=(1-c13)*sum(b*b for b in bt)-(1+c2)*th**2+z*OC/2+kt*ad**2/2
H=sp.simplify(sp.hessian(U,vel));Hprime=sp.simplify(H.diff(alpha))
gate('homogeneous_quadratic_euler_identity',eq((vel.T*H*vel)[0]/2,U))
gate('homogeneous_time_spin_cross_zero',all(eq(H[i,3],0) for i in range(3)))
gate('homogeneous_spin_hessian_kt',eq(H[3,3],kt))
gate('homogeneous_c13_is_c1_plus_c3_only',all(not H[i,j].has(ks) for i in range(4) for j in range(4)))
gate('homogeneous_shear_tracefree',eq(sum(deltas),0))
gate('homogeneous_shear_permutation_alpha_0',eq(OC.subs(alpha,0),sum(x*x for x in deltas)))
gate('homogeneous_theta_independent_of_triad',not th.has(alpha))
# independently compute numerical Hessian from finite differences in the restricted exact action
pv={c13:sp.Rational(3,25),c2:sp.Rational(1,10),z:sp.Rational(1,4),kt:sp.Rational(7,5),ks:sp.Rational(7,10)}
Hfn=sp.lambdify(alpha,H.subs(pv),'numpy');Hpfn=sp.lambdify(alpha,Hprime.subs(pv),'numpy')
for av in [0.0,float(sp.pi/8),float(sp.pi/4)]:
    ev=np.linalg.eigvalsh(np.array(Hfn(av),float))
    gate('minisuperspace_regular_signature_at_'+str(round(av,4)),
         np.sum(ev<0)==1 and np.sum(ev>0)==3 and np.min(abs(ev))>0.3,ev)
# independent centered Hessian of numerical quadratic function at randomly sampled velocity
rng=np.random.default_rng(20261118)
max_hess_err=0.0
for i in range(12):
    av=float(rng.uniform(-np.pi,np.pi));hv=np.array(Hfn(av),float)
    u=rng.normal(size=4); h=2e-4
    f=lambda x: float(.5*x@hv@x)
    recovered=np.empty((4,4))
    for j in range(4):
      for k in range(4):
        ej=np.eye(4)[j]*h; ek=np.eye(4)[k]*h
        recovered[j,k]=(f(u+ej+ek)-f(u+ej-ek)-f(u-ej+ek)+f(u-ej-ek))/(4*h*h)
    max_hess_err=max(max_hess_err,float(np.max(abs(recovered-hv))))
gate('minisuperspace_finite_difference_hessian_12_states',max_hess_err<1e-6,max_hess_err)
observables['minisuperspace_max_hessian_fd_error']=max_hess_err
print('H(alpha=0):');display(H.subs(alpha,0))
```

## Niveau 2 — Dirac–Bergmann algorithm **inside the homogeneous ADM ansatz**

Let $q=(\beta_1,\beta_2,\beta_3,\alpha)$, $v=e^{\sum\beta_i}$, $H_{IJ}(\alpha)=\partial^2U/\partial\dot q^I\partial\dot q^J$, and assume $\det H\ne0$. Then

\[
p_N=0\quad(\text{primary}),\qquad p_I=\frac vN H_{IJ}\dot q^J,\qquad
H_{\rm can}=N\mathcal C,
\]
\[
\boxed{\mathcal C=\frac1{2v}p_I(H^{-1})^{IJ}p_J\approx0}\quad(\text{secondary from }\dot p_N\approx0).
\]

Poisson brackets: $\{p_N,H_T\}=-\mathcal C$ and $\{\mathcal C,\mathcal C\}=0$, so there is no tertiary constraint **within this regular truncated single-lapse model**. Both $(p_N,\mathcal C)$ are first class; formal restricted count is $(10-2\times2)/2=3$ canonical degrees of freedom from the five coordinates $(q,N)$.

**Important:** fixing the shift and æther alignment before variation eliminates constraints, and restricting spatial modes can alter the Dirac algebra. This is **not** a full 4D physical degree-of-freedom count. The homogeneous Hamiltonian contains one negative gravitational direction (as in the usual minisuperspace conformal mode); that fact alone is **not proof of a propagating ghost**.

```python
# Dirac chain in regular homogeneous model. Use symbolic inverse at alpha=0 to cross-check Legendre.
N=sp.symbols('N',positive=True);vol=sp.symbols('v',positive=True)
p=sp.Matrix(sp.symbols('p1 p2 p3 pa',real=True))
H0=sp.simplify(H.subs(alpha,0).subs(pv))
PfromVel=vol/N*H0*vel
Vrecon=N/vol*H0.inv()*PfromVel
gate('legendre_velocity_reconstruction',all(eq(Vrecon[i],vel[i]) for i in range(4)))
C0=sp.factor((p.T*H0.inv()*p)[0]/(2*vol))
Hamiltonian=N*C0
gate('primary_pN_vanishes',not U.has(N))
gate('Hamiltonian_linear_in_lapse',eq(sp.diff(Hamiltonian,N),C0))
gate('secondary_preservation_pN',eq(-sp.diff(Hamiltonian,N),-C0))
gate('no_tertiary_in_single_constraint_minisuperspace',eq(C0-C0,0))
gate('homogeneous_restricted_phase_count_3',(2*5-2*2)//2==3)
# Hamiltonian + Lagrangian Legendre transform exact (for all unrestricted velocities within this ansatz)
Pv=vol/N*H0*vel
energy=(Pv.T*vel)[0]-vol/N*(vel.T*H0*vel)[0]/2
gate('legendre_transform_matches_N_constraint',eq(energy,N*(Pv.T*H0.inv()*Pv)[0]/(2*vol)))
# Reparametrization t -> f(t): qdot -> qdot/r and N -> N/r, dt -> r dt_new
r=sp.symbols('r',positive=True)
gate('homogeneous_lapse_reparametrization',eq((vol/(N/r))*U.subs({b1:b1/r,b2:b2/r,b3:b3/r,ad:ad/r})*r,vol*U/N))
print('C at alpha=0:');display(C0)
```

## Niveau 3 — Constrained coupled evolution (lapse gauge $N=1$)

Choose $N=1$ **after** deriving $p_N$ and $\mathcal C$. Use a zero-energy initial state $U=0$ by solving for the **isotropic expansion**, rather than forcing an artificial constraint. The exact reduced Euler–Lagrange acceleration is

\[
\ddot q=H^{-1}\left(
\underbrace{(U,U,U,\tfrac12\dot q^T H'\dot q)^T}_{(1/v)\,\partial_q L}
-(\dot\beta_1+\dot\beta_2+\dot\beta_3)H\dot q-H'\dot\alpha\dot q
\right),\quad H'=\partial_\alpha H.
\]

Check the Hamiltonian constraint $\mathcal C=U/v$ equivalently $U=0$ in velocity variables, the autonomous conserved energy $E=vU$, and the residual of all **restricted** metric and spin equations. This is a direct evolution of 3 metric variables and spin; there is no externally prescribed shear. No matter source or physical scale calibration.

```python
# Find a genuinely nontrivial constraint-satisfying state analytically.
alpha0=0.33; deviation=np.array([0.11,-0.07,-0.04]);adot0=0.08
assert abs(sum(deviation))<1e-12
Hinitial=np.asarray(Hfn(alpha0),float)
wd=np.r_[deviation,adot0]
wiso=np.array([1.,1.,1.,0.])
AA=.5*wiso@Hinitial@wiso;BB=wiso@Hinitial@wd;CC=.5*wd@Hinitial@wd
roots=np.roots([AA,BB,CC]);realroots=[float(x.real) for x in roots if abs(x.imag)<1e-12]
gate('real_isotropic_expansion_satisfies_constraint',len(realroots)==2,[AA,BB,CC,roots])
Hmean=max(realroots)
q0=np.array([0.,0.,0.,alpha0]);qd0=Hmean*wiso+wd
U0=.5*qd0@Hinitial@qd0
gate('initial_hamiltonian_constraint_exact_numerical',abs(U0)<1e-14,U0)

def ode(t,y):
    q=y[:4];qd=y[4:];a=float(q[3]);h=np.asarray(Hfn(a),float);hp=np.asarray(Hpfn(a),float)
    u=.5*qd@h@qd;theta=float(sum(qd[:3]));src=np.r_[np.full(3,u),.5*qd@hp@qd]
    qdd=np.linalg.solve(h,src-theta*h@qd-qd[3]*(hp@qd))
    return np.r_[qd,qdd]

sol=solve_ivp(ode,(0.,2.),np.r_[q0,qd0],method='DOP853',rtol=5e-12,atol=1e-13,dense_output=True)
gate('homogeneous_ODE_success',sol.success,sol.message)
tm=np.linspace(0.,2.,101);yy=sol.sol(tm);Us=[];Es=[];EL=[];shears=[]
for j in range(len(tm)):
    q=yy[:4,j];qd=yy[4:,j];a=q[3];hv=np.asarray(Hfn(a),float);hp=np.asarray(Hpfn(a),float);Uj=.5*qd@hv@qd;vj=np.exp(sum(q[:3]));Us.append(Uj);Es.append(vj*Uj)
    qdd=ode(tm[j],yy[:,j])[4:]; src=np.r_[np.full(3,Uj),.5*qd@hp@qd]
    EL.append(np.max(abs(hv@qdd+sum(qd[:3])*hv@qd+qd[3]*(hp@qd)-src)))
    shears.append(float(np.std(qd[:3])))
maxC=float(max(abs(x) for x in Us));maxE=float(max(abs(x-Es[0]) for x in Es))
maxEL=float(max(EL))
gate('homogeneous_constraint_preserved',maxC<2e-10,maxC)
gate('homogeneous_energy_conserved',maxE<2e-10,maxE)
gate('homogeneous_metric_and_spin_EOM_residual',maxEL<1e-10,maxEL)
gate('homogeneous_shear_nonzero_without_external_prescription',max(shears)>0.02,max(shears))
gate('homogeneous_spin_evolves',abs(yy[3,-1]-alpha0)>1e-6,float(yy[3,-1]-alpha0))
gate('homogeneous_metric_evolves',np.linalg.norm(yy[:3,-1]-q0[:3])>1e-4)
observables['homogeneous']={'alpha_initial':alpha0,'alpha_final':float(yy[3,-1]),'expansion_rate':Hmean,'max_constraint_abs':maxC,'max_energy_drift_abs':maxE,'max_EOM_residual':maxEL,'max_shear_std':max(shears)}
print('homogeneous initial rates',qd0,'final q',yy[:4,-1]);print('constraint max',maxC)
```

## Niveau 2/3 — Flat-background TT metric + physical triad rotations: principal symbol

Next allow **spacetime-dependent metric perturbations**, with a fixed æther background $u=\partial_t$ and TT wave along $x$:

\[
g_{\mu\nu}=\eta_{\mu\nu}+h_{\mu\nu},\qquad
h_{yy}=h_+(t,x),\quad h_{zz}=-h_+(t,x),\quad h_{yz}=h_\times(t,x),
\]
\[
e_A{}^i=\delta_A{}^i-\tfrac12 h^i{}_A+\psi^i{}_A+O(h^2,\psi^2),\quad \psi_{AB}=-\psi_{BA}.
\]

At first order the actual **Levi-Civita** spin connection is
\[
\boxed{\Omega_{\mu AB}^{(1)}=\partial_\mu\psi_{AB}+\tfrac12(\partial_B h_{A\mu}-\partial_A h_{B\mu})}.
\]
For $h_{0\mu}=0$ and $\partial_y=\partial_z=0$, the temporal spin connection contains $\dot\psi_{AB}$, whereas the transverse spatial components contain derivatives of $h_{+},h_\times$. Thus the metric cannot be frozen when constructing $\mathcal L_{\rm rot}$.

The quadratic TT action **modulo spatial boundary terms** is
\[
\begin{split}
\mathcal L_{\rm TT+spin}^{(2)}={}&\frac12(1-c_{13}+\zeta_C/2)\dot h_+^2+\frac12(1-c_{13})\dot h_\times^2
+\frac{\kappa_t}{2}\sum_{a=1}^3\dot\chi_a^2\\
&-\frac12(1+\kappa_s/2)\big[(\partial_x h_+)^2+(\partial_x h_\times)^2\big]
-\frac{\kappa_s}{2}\sum_{a=1}^3(\partial_x\chi_a)^2.
\end{split}
\]

Here $\chi_a$ are three independent infinitesimal spatial rotations ($\psi_{12},\psi_{13},\psi_{23}$). At this order around zero shear the $\mathcal O_C$ variation is a **polarization-dependent metric kinetic correction**. The special frame aligns the triad with the fixed axes, so a general orientation/direction audit is still needed.

**Scope warning:** This is a **TT gauge-fixed subsector**. It is not an invariant 4D full principal symbol; lapse, shift, æther tilt, scalar/vector perturbations and associated differential constraints have **not** been eliminated here. Positivity of this block is necessary in this sector, not sufficient for the full theory.

```python
# 1D TT decomposition: construct the Levi-Civita spatial spin connection independently.
px,qx=sp.symbols('P_x Q_x',real=True)
chi_x=sp.symbols('X12_x X13_x X23_x',real=True)
chid=sp.symbols('X12_t X13_t X23_t',real=True)
# spatial components h_ab and h'_ab indexed x,y,z = 0,1,2
Hsp=sp.Matrix([[0,0,0],[0,sp.Symbol('P'),sp.Symbol('Q')],[0,sp.Symbol('Q'),-sp.Symbol('P')]])
Hsp_x=Hsp.subs({sp.Symbol('P'):px,sp.Symbol('Q'):qx})
spin_sp={}; spin_t={}
for m in range(3):
    for a,b in [(0,1),(0,2),(1,2)]:
        # partial_b h_{a m} - partial_a h_{b m} (only x derivative)
        mc=( (Hsp_x[a,m] if b==0 else 0) - (Hsp_x[b,m] if a==0 else 0) )/2
        intrinsic=(chi_x[[(0,1),(0,2),(1,2)].index((a,b))] if m==0 else 0)
        spin_sp[(m,a,b)]=sp.expand(intrinsic+mc)
for a,b in [(0,1),(0,2),(1,2)]:
    spin_t[(a,b)]=chid[[(0,1),(0,2),(1,2)].index((a,b))]
# ordered AB is twice the sum over a<b
Qspace=sp.expand(2*sum(x*x for x in spin_sp.values()))
Qtime=sp.expand(2*sum(x*x for x in spin_t.values()))
Qspace_expected=2*sum(w*w for w in chi_x)+px**2+qx**2
Qtime_expected=2*sum(w*w for w in chid)
gate('TT_spin_connection_spatial_quadratic_exact',eq(Qspace,Qspace_expected),Qspace)
gate('TT_spin_connection_temporal_quadratic_exact',eq(Qtime,Qtime_expected))
gate('TT_no_metric_spin_gradient_cross_in_this_gauge',all(eq(sp.diff(Qspace,px,w),0) and eq(sp.diff(Qspace,qx,w),0) for w in chi_x))
gate('TT_linear_spin_connection_spatial_metric_nonzero',spin_sp[(1,0,1)]==-px/2 and spin_sp[(2,0,2)]==px/2)
gate('TT_linear_spin_connection_cross_metric_nonzero',spin_sp[(1,0,2)]==-qx/2 and spin_sp[(2,0,1)]==-qx/2)
# Levi-Civita ADM kinetic Einstein–æther part: sum Kij Kij = 1/2 (P_t^2+Q_t^2); trace K = 0.
pt,qt=sp.symbols('P_t Q_t',real=True)
Ocu=sp.expand((pt/2)**2+(-pt/2)**2)
gate('TT_cubic_shear_orientation',eq(Ocu,pt**2/2))
# Intrinsic 3D spatial curvature: G=I+eps[[P,Q],[Q,-P]] gives
# R^(3) density = 2(P Pxx+Q Qxx)+3/2(Px^2+Qx^2),
# equivalent up to 2 d_x(P Px+Q Qx) to -(Px^2+Qx^2)/2.
P,Q,Pxx,Qxx=sp.symbols('P Q Pxx Qxx',real=True)
scalar_R2=2*(P*Pxx+Q*Qxx)+sp.Rational(3,2)*(px**2+qx**2)
boundary=2*(px**2+P*Pxx+qx**2+Q*Qxx)
gate('TT_3curvature_IBP_gradient',eq(scalar_R2-boundary,-(px**2+qx**2)/2))
# The full quadratic TT+rotation Lagrangian in the isolated principal block.
Ltt=sp.expand((1-c13)*(pt**2+qt**2)/2 + z*Ocu/2 +kt*Qtime/4
              -(px**2+qx**2)/2-ks*Qspace/4)
qtime=sp.Matrix([pt,qt,*chid]);qspace=sp.Matrix([px,qx,*chi_x])
Kt=sp.simplify(sp.hessian(Ltt,qtime));Gx=sp.simplify(-sp.hessian(Ltt,qspace))
Kt_expected=sp.diag(1-c13+z/2,1-c13,kt,kt,kt)
Gx_expected=sp.diag(1+ks/2,1+ks/2,ks,ks,ks)
gate('TT_5x5_time_symbol_from_lagrangian',Kt==Kt_expected)
gate('TT_5x5_space_symbol_from_lagrangian',Gx==Gx_expected)
gate('TT_matrices_symmetric',Kt==Kt.T and Gx==Gx.T)
gate('TT_time_space_derivative_mixed_zero',all(eq(sp.diff(Ltt,qtv,qsv),0) for qtv in qtime for qsv in qspace))
omega,k=sp.symbols('omega k',real=True)
Dsym=sp.diag(*[Kt[i,i]*omega**2-Gx[i,i]*k**2 for i in range(5)])
gate('TT_wave_symbol_factorization',eq(Dsym.det(),sp.prod(Dsym[i,i] for i in range(5))))
gate('TT_spin_triply_degenerate_propagation',eq(Dsym[2,2],Dsym[3,3]) and eq(Dsym[3,3],Dsym[4,4]))
obs_speeds=[sp.factor(Gx[i,i]/Kt[i,i]) for i in range(5)]
gate('TT_plus_speed_expression',eq(obs_speeds[0],(1+ks/2)/(1-c13+z/2)))
gate('TT_cross_speed_expression',eq(obs_speeds[1],(1+ks/2)/(1-c13)))
gate('TT_spin_speed_expression',eq(obs_speeds[2],ks/kt))
numK=np.asarray(Kt.subs(pv),float);numG=np.asarray(Gx.subs(pv),float)
speed2=np.diag(numG)/np.diag(numK)
gate('TT_sample_positive_kinetic',np.min(eigvalsh(numK))>0,np.diag(numK))
gate('TT_sample_positive_gradient',np.min(eigvalsh(numG))>0,np.diag(numG))
gate('TT_sample_real_characteristic_speeds',np.all(speed2>0),speed2)
gate('TT_sample_different_gravitational_polarizations',abs(speed2[0]-speed2[1])>0.05,speed2[:2])
gate('TT_zeta_zero_equal_metric_speeds',eq(obs_speeds[0].subs(z,0),obs_speeds[1].subs(z,0)))
gate('TT_all_spins_equal_speed',all(eq(obs_speeds[2],obs_speeds[i]) for i in [3,4]))
observables['tt']={'kinetic_diagonal':[float(x) for x in np.diag(numK)],'gradient_diagonal':[float(x) for x in np.diag(numG)],'speed_squared':[float(x) for x in speed2], 'speeds':[float(np.sqrt(x)) for x in speed2], 'c13':float(pv[c13]),'c2':float(pv[c2]),'zeta':float(pv[z]),'kt':float(pv[kt]),'ks':float(pv[ks])}
print('K_TT=');display(Kt)
print('G_TT=');display(Gx)
print('speed^2 = ',speed2)
```

## Niveau 3 — Independent metric-connection counterexample and stability boundaries

Cross-check the spatial Levi-Civita connection **without using its first-order formula**, on the exact static metric
\[

\]
with normalized nonrotating triad $e_1=\partial_x$, $e_2=(1+\varepsilon f)^{-1/2}\partial_y$, $e_3=(1-\varepsilon f)^{-1/2}\partial_z$:
\[
\Omega_{y12}=-\frac{\varepsilon f'}{2\sqrt{1+\varepsilon f}},\qquad
\Omega_{z13}=+\frac{\varepsilon f'}{2\sqrt{1-\varepsilon f}}.
\]
Their first-order coefficients agree with the connection in the TT symbol; freezing the metric would miss them.
