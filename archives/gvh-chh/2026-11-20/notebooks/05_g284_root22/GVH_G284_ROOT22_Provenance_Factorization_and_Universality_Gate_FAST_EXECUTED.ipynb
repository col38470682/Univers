## Level 1 — GVH: exact frozen-point relations

On the particular frozen slice \(c_{13}=0,\zeta_C=1\), a localized linearized branch was recorded as

\[\omega_+^2/\kappa^2=(\sqrt{22}-2)/9.\]

After intersection of the compatibility polynomial \(F(c_2,c_{14})=0\) with the reduced Schwarzschild sector \(c_{14}=0\), the historical coupling point was

\[c_2=(\sqrt{22}-4)/9.\]

These are **two linked quantities within the same branch**: not two independent observations of a universal coefficient.


```python
import sympy as sp, json
s22=sp.sqrt(22)
c13=sp.Integer(0)
zetaC=sp.Integer(1)
c14=sp.Symbol('c14',real=True)
c2=sp.Symbol('c2',real=True)
kappa=sp.symbols('kappa',positive=True)
L=(s22-2)/9
C=(s22-4)/9
CHECKS={}
def check(name,assertion):
    v=bool(assertion)
    CHECKS[name]=v
    print(('PASS ' if v else 'FAIL ')+name)
    assert v,name
print('sqrt22 =', sp.N(s22,16))
print('lambda=omega_+^2/kappa^2 =',sp.N(L,16))
print('c2=',sp.N(C,16))
check('coefficient_exact',sp.simplify(C-(s22-4)/9)==0)
check('mode_exact',sp.simplify(L-(s22-2)/9)==0)
check('linked_values',sp.simplify(L-C-sp.Rational(2,9))==0)
```

## Level 2 — Known algebra: reconstruct the imported compatibility polynomial

\(F(c_2,c_{14})\) below is copied **as an input** from archived `.1.5.1`; this audit does not claim to reproduce the full field-equation derivation. On the reduced Schwarzschild condition \(c_{14}=0\), test that its zero is exactly \(C\).


```python
F=8*(
 (126*s22+639)*c14**2*c2
 +(183*s22+876)*c14**2
 +(1494+387*s22)*c14*c2**2
 +(1992+516*s22)*c14*c2
 -(1136+224*s22)*c14
 -(2556+504*s22)*c2
 +96+60*s22
)
F0=sp.factor(F.subs(c14,0),extension=s22)
root=sp.solve(F0,c2)[0]
print('F(c2,0) =',F0)
print('Root F(c2,0)=0:',root)
check('F_root_matches_c2',sp.simplify(root-C)==0)
check('F_is_zero_at_frozen_point',sp.simplify(F.subs({c2:C,c14:0}))==0)
check('F_linear_on_c14_zero',sp.degree(F0,c2)==1)
```

## Level 2 — Minimal polynomials and decompositions

These polynomials are **algebraic identities**, not experimentally established universality relations.

\[27c_2^2+24c_2-2=0,\quad 9\lambda^2+4\lambda-2=0.\]

\(\sqrt{22}=\sqrt2\,\sqrt{11}=9c_2+4=9\lambda+2\) on the studied frozen branch. Multiplying/dividing by freely fitted coefficients would not create a physical law.


```python
z=sp.symbols('z')
poly_c=sp.minpoly(C,z)
poly_l=sp.minpoly(L,z)
print('minpoly c2:',poly_c)
print('minpoly lambda:',poly_l)
check('minimal_polynomial_c2',sp.expand(poly_c-(27*z**2+24*z-2))==0)
check('minimal_polynomial_lambda',sp.expand(poly_l-(9*z**2+4*z-2))==0)
check('factor_sqrt22',sp.simplify(s22-sp.sqrt(2)*sp.sqrt(11))==0)
check('root_from_c2',sp.simplify(s22-(9*C+4))==0)
check('root_from_mode',sp.simplify(s22-(9*L+2))==0)
```

An analogous issue applies to additive offsets and multi-parameter per-system regressions. Any claim that \(\sqrt{22}\) is common to systems must freeze the definitions, units, and fit protocol **before** testing new systems; no retuning per system.
