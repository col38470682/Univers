# VP 0.6 — Consolidation de la dynamique et de la stabilité

On considère deux variables d’état scalaires \(X_A(t)\) et \(X_B(t)\).

Le modèle minimal symétrique est

\[
\dot X_A
=
-\gamma_{\mathrm{VP}}(X_A-X_B),
\]

\[
\dot X_B
=
+\gamma_{\mathrm{VP}}(X_A-X_B),
\]

avec

\[
\gamma_{\mathrm{VP}}
=
\kappa V_{\mathrm{sh}}
\ge 0.
\]

Dans le cas normalisé, on peut écrire

\[
\gamma_{\mathrm{VP}}
=
\kappa_\phi \phi_{AB},
\]

où

\[
\phi_{AB}
=
\frac{V_{AB}^{\mathrm{sh}}}{V_{AB}^{\mathrm{ref}}}
\in[0,1].
\]

Le présent chapitre suppose que
\(\gamma_{\mathrm{VP}}\) est constant dans le temps.
Le système est donc linéaire et autonome.

## Dimensions physiques

Pour assurer l’homogénéité de

\[
\dot X_A
=
-\kappa V_{\mathrm{sh}}(X_A-X_B),
\]

il faut imposer

\[
[\kappa V_{\mathrm{sh}}]=T^{-1}.
\]

Si \(V_{\mathrm{sh}}\) possède la dimension d’un volume,

\[
[\kappa]
=
L^{-3}T^{-1}.
\]

Dans la formulation normalisée utilisant
\(\phi_{AB}\), sans dimension, on écrit

\[
\dot X_A
=
-\kappa_\phi\phi_{AB}(X_A-X_B),
\]

avec

\[
[\kappa_\phi]=T^{-1}.
\]

## Théorème — Conservation de la somme des états

En additionnant les deux équations,

\[
\dot X_A+\dot X_B
=
-\gamma_{\mathrm{VP}}(X_A-X_B)
+
\gamma_{\mathrm{VP}}(X_A-X_B)
=
0.
\]

Ainsi,

\[
X_A(t)+X_B(t)
=
Q_0,
\]

où

\[
Q_0=X_A(0)+X_B(0).
\]

La quantité \(Q_0\) est une charge ou une quantité extensive
conservée du modèle minimal.

Elle ne doit pas être appelée énergie sans définition physique
supplémentaire.

## Réduction du système

La moyenne vérifie

\[
\dot M
=
\frac{\dot X_A+\dot X_B}{2}
=
0.
\]

Elle est donc constante :

\[
M(t)=M_0.
\]

La différence d’état vérifie

\[
\dot\Delta
=
\dot X_A-\dot X_B
=
-2\gamma_{\mathrm{VP}}\Delta.
\]

Sa solution analytique est

\[
\Delta(t)
=
\Delta_0
e^{-2\gamma_{\mathrm{VP}}t}.
\]

## Théorème — Unicité de l’équilibre sous contrainte de conservation

Un équilibre vérifie

\[
\dot X_A=\dot X_B=0.
\]

Pour \(\gamma_{\mathrm{VP}}>0\), cela implique

\[
X_A^\ast=X_B^\ast.
\]

Comme

\[
X_A^\ast+X_B^\ast=Q_0,
\]

on obtient l’unique équilibre compatible avec la condition initiale :

\[
X_A^\ast
=
X_B^\ast
=
\frac{Q_0}{2}.
\]

Il existe donc une famille globale d’équilibres
\((c,c)\), mais un seul équilibre dans chaque classe
définie par \(X_A+X_B=Q_0\).

## Théorème — Stabilité par une fonction de Lyapunov

On définit

\[
\mathcal L(\Delta)
=
\frac12\Delta^2.
\]

Cette fonction vérifie

\[
\mathcal L\ge0,
\]

et

\[
\mathcal L=0
\quad\Longleftrightarrow\quad
\Delta=0.
\]

Sa dérivée temporelle vaut

\[
\dot{\mathcal L}
=
\Delta\dot\Delta.
\]

Comme

\[
\dot\Delta
=
-2\gamma_{\mathrm{VP}}\Delta,
\]

on obtient

\[
\dot{\mathcal L}
=
-2\gamma_{\mathrm{VP}}\Delta^2.
\]

Puisque

\[
\mathcal L=\frac12\Delta^2,
\]

cela donne

\[
\boxed{
\dot{\mathcal L}
=
-4\gamma_{\mathrm{VP}}\mathcal L
}
\]

ou encore

\[
\boxed{
\dot{\mathcal L}
=
-4\kappa V_{\mathrm{sh}}\mathcal L
\le0
}.
\]

## Corollaire — Stabilité asymptotique

Si

\[
\gamma_{\mathrm{VP}}>0,
\]

alors

\[
\dot{\mathcal L}<0
\]

pour tout état tel que

\[
X_A\neq X_B.
\]

Par conséquent,

\[
\Delta(t)\rightarrow0
\quad\text{lorsque}\quad
t\rightarrow\infty.
\]

Ainsi,

\[
X_A(t),X_B(t)
\rightarrow
\frac{Q_0}{2}.
\]

L’équilibre est asymptotiquement stable dans la sous-variété
invariante

\[
X_A+X_B=Q_0.
\]

## Quantité conservée et énergie

La conservation de

\[
X_A+X_B
\]

ne constitue pas automatiquement une conservation de l’énergie.

Une interprétation énergétique exige une fonction explicitement définie,
par exemple

\[
E=C_A X_A+C_B X_B,
\]

où \(C_A\) et \(C_B\) sont des capacités, masses, volumes,
densités ou coefficients physiques adaptés.

Dans le modèle minimal avec

\[
C_A=C_B=1,
\]

la quantité conservée possède la même forme algébrique que la somme
des états, mais son interprétation dépend du système étudié.
