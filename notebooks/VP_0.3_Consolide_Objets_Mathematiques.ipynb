## Consolidation formelle du domaine partagé

On considère deux systèmes définis sur des domaines mesurables
\(\Omega_A\) et \(\Omega_B\).

Le domaine géométriquement commun est

\[
\Omega_{AB}=\Omega_A\cap\Omega_B.
\]

La fonction locale de partage est une application mesurable

\[
\chi_{AB}:\Omega_{AB}\rightarrow[0,1].
\]

Le volume partagé effectif est défini par

\[
V_{AB}^{\mathrm{sh}}
=
\int_{\Omega_{AB}}
\chi_{AB}\,d\mu_{AB}.
\]

Comme \(0\leq\chi_{AB}\leq1\),

\[
0
\leq
V_{AB}^{\mathrm{sh}}
\leq
\mu_{AB}(\Omega_{AB}).
\]

Puisque
\(\Omega_{AB}\subseteq\Omega_A\) et
\(\Omega_{AB}\subseteq\Omega_B\),

\[
V_{AB}^{\mathrm{sh}}
\leq
\min\{\mu(\Omega_A),\mu(\Omega_B)\}.
\]

Cette borne est donc une conséquence des définitions et non un postulat indépendant.

# Théorèmes fondamentaux

## Théorème 1 — Positivité

Comme

\[
\chi_{AB}(x)\ge0
\]

pour presque tout \(x\in\Omega_{AB}\),

on obtient

\[
V_{AB}^{\mathrm{sh}}
=
\int_{\Omega_{AB}}\chi_{AB}\,d\mu
\ge0.
\]

Le volume partagé est donc toujours positif ou nul.

## Théorème 2 — Bornage

Puisque

\[
0\le\chi_{AB}\le1,
\]

on a

\[
V_{AB}^{\mathrm{sh}}
\le
\mu(\Omega_{AB}).
\]

Comme

\[
\Omega_{AB}
=
\Omega_A\cap\Omega_B,
\]

on obtient

\[
V_{AB}^{\mathrm{sh}}
\le
\min\{\mu(\Omega_A),\mu(\Omega_B)\}.
\]

Le volume partagé est donc toujours fini lorsque les domaines sont de mesure finie.

## Théorème 3 — Monotonie

Si

\[
\chi_1(x)\le\chi_2(x)
\]

presque partout,

alors

\[
\int\chi_1\,d\mu
\le
\int\chi_2\,d\mu.
\]

Ainsi,

\[
V_{\mathrm{sh}}(\chi_1)
\le
V_{\mathrm{sh}}(\chi_2).
\]

Le volume partagé croît donc lorsque l'intensité locale de partage augmente.

## Théorème 4 — Cas limites

Si

\[
\chi(x)=0,
\]

alors

\[
V_{\mathrm{sh}}=0.
\]

Si

\[
\chi(x)=1
\]

presque partout,

alors

\[
V_{\mathrm{sh}}
=
\mu(\Omega_{AB}).
\]

Ces deux cas représentent respectivement l'absence totale de partage et le partage maximal.

# Remarque

Le volume partagé est une quantité géométrique pondérée.

La géométrie est portée par la mesure \(\mu\).

L'intensité physique est portée par la fonction \(\chi\).

Cette séparation garantit que les propriétés géométriques et les propriétés physiques demeurent indépendantes dans le formalisme.
